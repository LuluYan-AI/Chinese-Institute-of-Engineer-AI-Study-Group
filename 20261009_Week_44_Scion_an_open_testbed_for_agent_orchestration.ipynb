## 一、Scion 是什麼？

Scion 是 Google 推出的開源多代理協調平台（Multi-Agent Orchestration Platform），讓多個 AI Agent 能夠獨立執行任務、互相溝通、動態分工，並共同完成複雜工作。

它的核心概念是：

> 不再由人類預先寫死每個 AI Agent 的工作流程，而是提供一套執行與溝通機制，讓 AI Agent 自己決定如何分工與合作。

Scion 並不是新的大型語言模型（Large Language Model, LLM），也不負責訓練模型。

它更接近一個 AI Agent 執行與協調環境（Agent Runtime and Orchestration Environment），負責管理 Agent 的啟動、執行、通訊、工作空間與生命週期。

目前 Scion 仍以實驗性平台（Experimental Testbed）為定位，主要用來探索多代理協作的新方法。

https://googlecloudplatform.github.io/scion/overview/

## Scion 是什麼？

Scion 是一套以容器（Container）為基礎的開源多代理協調平台（Multi-Agent Orchestration Platform），可在本機或遠端叢集中，同時執行與管理多個大型語言模型（LLM）驅動的 AI Agent。

每個 Agent 具有獨立的身分、認證與工作空間，能平行處理程式開發、研究、稽核及測試等任務。

Scion 的主要目標是提供一個開放式實驗平台，探索多個 AI Agent 如何自主分工與協作。

## Scion 如何運作？

1\. 容器隔離（Container Isolation）： 每個 Agent 在獨立的 OCI 容器中運行，擁有自己的環境、目錄與認證資訊，避免互相干擾。

2\. 統一操作介面（Harness Abstraction）： 透過 Scion CLI 的 `start`、`stop`、`attach`、`resume` 等指令，管理 Gemini CLI、Claude Code、Codex、ADK、OpenCode 等不同 Agent 工具。

3\. 獨立 Git 工作區（Git Worktree）： 每個 Agent 可使用獨立的 Git Worktree 與開發分支，降低平行修改程式碼時的衝突。

4\. 彈性執行環境（Pluggable Runtimes）： 支援 Docker、Podman、Apple Virtualization Framework 與 Kubernetes，可從個人電腦擴展至遠端叢集。

5\. 分散式管理（Hub + Runtime Broker）： 透過中央 Hub 管理 Agent 狀態、身分驗證與任務協調，由不同 Runtime Broker 執行 Agent，支援跨主機及多人使用。

6\. 代理範本（Agent Templates）： 透過範本設定 System Prompt、工具及執行參數，建立具有不同專長的 Agent，例如安全稽核或 React 開發專家。

7\. 背景執行與即時監控（Detached Execution & Observability）： Agent 可在背景持續執行，使用者能隨時重新連線介入，並透過多層次狀態資訊掌握執行進度。

總結：Scion 不只是同時啟動多個 AI Agent，而是提供一套具備隔離、協調、擴展與監控能力的執行環境，讓 AI Agent 能更有效地平行工作與自主協作。







## 二、影片最重要的五個觀念

### 1. 從單一 Agent 走向多代理自主協作

傳統 AI Agent 通常是由人類指定任務，再依照事先設計的流程執行。

Scion 則允許 Agent 自行啟動其他 Agent，形成動態的協作關係。

```
【傳統方式：固定流程】
   人類指定流程
        │
        ▼
     Agent A
        │
        ▼
     Agent B
        │
        ▼
     Agent C


【Scion：動態協作】

    Orchestrator Agent
     （自主分配任務）
            │
     ┌──────┼──────┐
     ▼      ▼      ▼
  Agent A Agent B Agent C
     │      │      │
     └──────┼──────┘
            ▼
        整合結果

  Agent 可平行執行、互傳訊息
       並動態調整分工

```


關鍵差異是：工作流程不必完全事先定義，而能由 Agent 根據任務需求動態形成。

這種現象稱為湧現式協作（Emergent Collaboration）。

### 2. 每個 Agent 都有獨立的執行環境

Scion 透過容器（Container）隔離不同 Agent，例如使用 Docker、Podman 或 Kubernetes。

每個 Agent 可以具有自己的：

- 工作空間（Workspace）：存放程式碼與任務成果。
- 系統提示與技能（System Prompt and Skills）：定義角色、行為與能力。
- Agent 執行框架（Agent Harness）：例如 Gemini CLI、Claude Code、Codex。
- Git 工作樹（Git Worktree）：讓 Agent 在不同工作目錄中平行修改程式碼。

這樣能降低不同 Agent 同時修改檔案造成的衝突，也方便管理和追蹤工作成果。



### 3. Agent 可以透過 CLI 自主呼叫其他 Agent

這是影片最值得注意的設計。

Scion 提供命令列介面（Command-Line Interface, CLI），而大型語言模型原本就擅長理解與使用命令列。

因此，Agent 可以自己查詢工具指令、建立子代理（Sub-Agent）、等待工作完成，再整合結果。

例如影片中，Orchestrator Agent 透過 Scion CLI 啟動三個 Agent，分別撰寫紅、綠、藍三種顏色的詩。

它不需要人類逐一操作，而是自行完成任務分派、等待、成果彙整與清理。

這代表 CLI 不只是人類操作工具的介面，也成為 Agent 管理其他 Agent 的介面。

### 4. 支援不同模型與不同 Agent 框架

Scion 採取框架無關（Harness-Agnostic）的設計。

例如，可以同時運行 Gemini CLI、Claude Code、Codex、OpenCode 等不同 Agent 工具。

因此，Scion 不要求所有 Agent 都使用相同的模型或相同的執行框架。

這使得不同模型可以依照其專長分工，例如由一個 Agent 負責寫程式，另一個 Agent 負責測試，第三個 Agent 負責程式碼審查。

### 5. 能觀察 Agent 如何合作，而不只看最後答案

Scion 支援收集執行紀錄（Logs）、通訊訊息（Messages）及遙測資料（Telemetry）。

研究者因此可以分析 Agent 之間何時交換資訊、如何分配任務、何時遇到錯誤，以及如何修正策略。

這個能力稱為可觀測性（Observability），對研究多代理系統尤其重要。

## 三、影片展示的兩個實驗

### 實驗一：三個 Agent 平行寫詩

```
       Orchestrator Agent
      （啟動 Agent、分派任務）
                │
       ┌────────┼────────┐
       ▼        ▼        ▼
    Red Agent Green Agent Blue Agent
       │        │        │
       ▼        ▼        ▼
     紅色詩歌   綠色詩歌   藍色詩歌
       │        │        │
       └────────┼────────┘
                ▼
       Orchestrator Agent
       （接收完成通知）
                │
                ▼
        整合三首詩，產生詩集
                │
                ▼
       停止並刪除三個子代理
```

這個實驗證明 Scion 能支援：

- 平行執行（Parallel Execution）
- 任務委派（Task Delegation）
- 事件通知（Event-Driven Notification）
- 結果整合（Result Aggregation）
- 代理生命週期管理（Agent Lifecycle Management）

它展示的是協作機制可行，而不是證明多代理一定比單一 Agent 更有效率。

## 四、Google 的實際研究告訴我們什麼？

第三篇 Google Cloud 官方文章進一步使用 Scion 進行 AI Agent 協作拍片實驗。

https://cloud.google.com/blog/topics/developers-practitioners/what-we-learned-about-agent-teamwork

Google 組織了 10 支 AI Agent 電影製作團隊，每隊包含企劃、技術與剪輯角色，並設置監督角色檢查成果。

這項實驗帶來三個重要發現：

![image](https://www.google.com/s2/favicons?domain=https://cloud.google.com\&sz=32)





第一，檔案比對話更適合保存協作資訊。

Agent 的上下文視窗（Context Window）有限，也可能中斷或重新啟動。將決策與成果寫入檔案，讓其他 Agent 能接續工作，比只依靠聊天紀錄更可靠。

第二，Agent 自主協作仍需要驗證機制。

實驗中曾有 Agent 宣稱完成影片，實際上卻只產生極小的佔位檔案。因此需要驗證關卡（Verification Gates）檢查真正的成果，而非只相信 Agent 的回報。

第三，角色專業化有助於分工，但不保證成果品質。

Agent 能自主產生新做法與分工策略，但最終仍需品質標準、人類回饋與適當監督。

這也說明：自主性（Autonomy）與可靠性（Reliability）必須同時設計。

## 六、總結：Scion 的真正價值

Scion 的重要性不是讓我們同時執行更多 AI Agent，而是提供一個可實驗的環境，研究 AI Agent 如何自主形成團隊、分配任務、交換知識並完成共同目標。

它將研究重點從單一 Agent 的能力，推進到多代理系統的協作能力。

### Herdr：讓多個 AI Agent 平行執行、自主協作的開源工具

## 一、Herdr 是什麼？

Herdr 是一套開源的 AI Agent 終端機多工器（Agent-Aware Terminal Multiplexer），專門用於同時執行、監控及協調多個 AI 編碼代理（AI Coding Agents），例如 Claude Code、Codex、Gemini CLI。

它使用 Rust 開發，採用 Apache 2.0 授權，軟體本身免費，但 AI 模型及雲端運算資源可能產生費用。

核心概念： 從人類逐一操作 Agent，進化到集中管理多個 Agent，甚至讓 Agent 自主呼叫其他 Agent 完成任務。

## 二、Herdr 的六個核心功能

1. 平行執行（Parallel Execution）： 在同一個工作空間（Workspace）中，透過不同終端機窗格（Pane）同時執行多個 Agent。
2. 狀態感知（Agent State Awareness）： 即時顯示 Working（執行中）、Blocked（等待輸入）、Done（完成）、Idle（閒置），快速掌握工作進度。
3. 工作階段持久化（Session Persistence）： 即使關閉終端機或 SSH 斷線，背景 Agent 仍可繼續執行，之後再重新連線（Reattach）。電腦關機則不會繼續執行。
4. 多種 Shell 支援（Multi-Shell Support）： 每個 Pane 可執行 Bash、Zsh、Fish、PowerShell 等 Shell，並不限於 Bash。Windows 可使用 PowerShell。
5. 自主協調（Agent Orchestration）： Agent 可利用 Herdr CLI 或 Socket API 建立其他 Agent、分派任務、等待完成並讀取結果。
6. 工作區隔離（Git Worktree）： 可為不同 Agent 建立獨立的 Git 工作目錄，減少平行修改程式碼時的衝突。

## 三、Herdr 如何讓 Agent 協作？

```
             Herdr Server
                  │
          ┌───────┼───────┐
          ▼       ▼       ▼
       Agent A Agent B Agent C
         Bash  PowerShell  Zsh
          │       │       │
         開發    測試    程式審查
          │       │       │
          └───────┼───────┘
                  ▼
             整合工作成果
```

Agent 之間透過 Herdr CLI 與 Socket API 進行程序間通訊（Inter-Process Communication, IPC）。

底層使用 Unix Domain Socket（Linux／macOS）或 Named Pipe（Windows），並以 JSON 格式交換 API 訊息。

主要交換四種資訊：

- Prompt： 任務指令與描述。
- Terminal Output： Agent 執行後輸出的文字結果。
- Agent State： 執行、等待或完成狀態。
- Control Event： 啟動、停止、按鍵等控制操作。

重要的是，Agent 不會自動共享完整的 LLM Context（上下文）。若需要交換大量資料，可以透過 JSON、Markdown 或共享檔案傳遞資訊。

## 四、遠端與手機操作

Herdr 採用 Client–Server 架構，支援透過 SSH 遠端連線及重新附加既有工作階段。

電腦可使用 Herdr 終端機介面；手機則可透過 SSH App 連線操作。目前主要是終端機操作模式，而非 Web Dashboard。

## 五、總結

Herdr 的真正價值，不只是同時開啟多個 AI Agent，而是提供一套具備狀態監控、持久化執行、跨 Shell 通訊及自主協調能力的開發環境。

對 Python 開發與 AI Agent 研究而言，Herdr 適合快速建立多 Agent 協作環境；若需要容器隔離、大規模分散式運算及更完整的 Agent 執行管理，則可以考慮 Scion。

### Scion vs. Herdr：多 AI Agent 協作平台的技術、架構與可觀測性比較

## 一、核心定位：兩者解決不同層次的問題

Scion 與 Herdr 都能平行執行多個 AI Agent，也都允許 Agent 透過 CLI 呼叫其他 Agent，但它們的系統設計理念不同。

Scion 是 Google Cloud 相關開發者推動的開源多代理協調平台（Multi-Agent Orchestration Platform），著重 Agent 的生命週期、任務委派、容器隔離、跨主機執行，以及多代理協作的可觀測性（Observability）。

Herdr 是為 AI 編碼代理設計的終端機多工器（Agent-Aware Terminal Multiplexer），重點是管理多個持續執行的 Agent 工作階段，提供即時狀態、SSH 遠端控制與 CLI 自動化。

可以簡化成兩句話：

- Herdr：如何讓多個 Agent 容易執行、操作與監控？
- Scion：如何讓多個 Agent 組成團隊，並管理、記錄和分析它們的協作？

這是兩者在架構層次上的差異，而不是說 Herdr 沒有協作能力，或 Scion 無法操作終端機。

## 四、執行架構：Terminal vs. Container

### Herdr：以終端機為中心

Herdr 不限定 Bash，也能使用 PowerShell。它透過背景 Server 保存工作階段，即使 SSH 中斷，Agent 也可以繼續執行。

## 五、Telemetry 與 Observability：最重要的差異

這是兩者在研究用途上最值得注意的差別。

首先要區分三個名詞：

- Logging（日誌）： 記錄執行過程發生什麼事。
- Telemetry（遙測）： 持續收集 Agent 的執行事件、效能與資源使用資料。
- Observability（可觀測性）： 利用 Logs、Metrics（指標）及 Traces（追蹤），分析系統為什麼出現特定行為。

### 1. Herdr：重點是知道 Agent 現在在做什麼

Herdr 的主要可觀測能力來自終端機狀態與輸出。

例如，它可以讓開發者知道：

- 哪個 Agent 正在工作、等待輸入或已完成。
- Agent 在終端機輸出了什麼。
- 哪個 Pane 正在執行哪個程序。
- Agent 何時需要人類介入。

Herdr Socket API 提供 `agent.read`、`agent.wait`、`pane.read`、`events.subscribe` 等方法，可以讓程式讀取終端機內容、等待 Agent 狀態變化，以及訂閱事件。

因此，Herdr 適合建立即時操作監控（Operational Monitoring）。

不過，根據目前官方文件，Herdr 並未提供與 Scion 相同層級、以 OpenTelemetry 為核心的完整內建 Agent Telemetry Pipeline。若要長期儲存 Token、工具呼叫、延遲及協作追蹤資料，通常需要額外整合。

### 2. Scion：重點是記錄 Agent 如何完成任務

Scion 提供較完整的可觀測架構。

Agent 容器內的 `sciontool` 可作為 OpenTelemetry（OTel）資料收集與轉送元件，支援：

- Logs：Agent 與系統執行紀錄。
- Metrics：Token、工具呼叫、執行時間與成本相關指標。
- Traces：Agent 行為、工具操作與事件時間關係。
- Messaging Logs：Agent 之間的訊息路由紀錄。

Telemetry 可轉送到 Google Cloud Logging、Cloud Trace 或其他相容的 OTLP 後端。Scion 也支援針對 Prompt、Tool Output 等欄位進行遮蔽（Redaction），避免敏感資訊進入紀錄系統。

這些能力需要適當設定，而且實際能取得的事件仍取決於 Agent Harness 的支援程度。

## 六、研究實例：分析三個 Agent 如何合作

## 七、持久化、通訊與安全性的差異

Herdr 的持久化（Persistence）是重要特色：終端機或 SSH 斷線後，背景 Server 可以繼續執行 Agent。但主機關機時，原有程序不會存活。

Scion 的持久化偏向 Agent 生命週期與工作狀態管理，例如停止、恢復、重新啟動，以及透過 Hub 儲存專案與 Agent 資訊。

[image](https://www.google.com/s2/favicons?domain=https://herdr.dev\&sz=32)

herdr

+1



通訊方面，Herdr 主要利用 CLI、Socket API 與終端機輸入輸出傳遞 Prompt、執行結果和狀態。不同 Shell（包含 PowerShell）之間也能透過 Herdr Server 協作。

Scion 則提供更明確的 Agent 身分、訊息路由、通知與存取控制機制，適合需要跨容器或跨主機協作的情境。

安全性方面，Scion 的容器隔離比 Herdr 預設的一般終端機程序更適合建立執行邊界；但容器本身也不是絕對安全，仍需要權限、憑證及資源限制。

## 八、規模與遠端管理

Scion 提供 Local、Workstation、Hosted 與高可用性（High Availability）部署模式。可以從單機 Docker 開始，再擴展至多 Runtime Broker 或 Kubernetes。

Herdr 則主要透過 SSH 管理多台電腦中的工作階段，讓開發者能在一個終端機介面集中查看不同主機的 Agent。

Scion 的優勢在於分散式執行與集中管理；Herdr 的優勢在於部署簡單，以及終端機操作體驗。

對手機而言，Scion 有 Web Dashboard；Herdr 主要透過 SSH Client 使用終端機介面。

## 九、研究與實務選擇建議

選擇 Herdr：AI 編碼工作站

適合 3–10 個 Agent 同時撰寫程式、測試、審查，重視即時操作、工作階段持久化與低導入成本的開發情境。

選擇 Scion：多代理協作研究平台

適合研究 Agent 自主分工、動態任務委派、跨 Agent 通訊、大規模執行，以及透過 Telemetry 分析任務效率與協作行為。

## 十、結論

Herdr 與 Scion 最大的差異，不是能執行多少 Agent，而是它們對「Agent 管理」的定義不同。

Herdr 以開發者為中心，強調終端機、即時狀態、持續執行與多 Agent 操作效率。

Scion 以多代理系統為中心，強調 Agent 的執行環境、協作機制、分散式架構，以及可用於研究和營運分析的 Telemetry。

如果目標是讓 AI Agent 協助日常 Python 開發，Herdr 是較直接的起點。

如果目標是建立可重複實驗、可分析協作行為、可測量效能的多代理研究環境，Scion 的原生 Telemetry 與分散式管理能力更具優勢。

兩者並非絕對競爭關係，而是反映 AI Agent 技術的兩個方向：開發者操作效率（Developer Productivity）與多代理系統工程（Multi-Agent Systems Engineering）。

```python
    start_time: str,
    end_time: str,
    limit: int,
    tool_context: ToolContext,
    credential: AuthCredential,

    """List Google Calendar events in a time range."""

        credentials=creds,
    )

            timeMin=start_time,
            timeMax=end_time,

            orderBy="startTime",

            "summary": event.get("summary"),
            "start": event.get("start"),
            "end": event.get("end"),
            "description": event.get("description"),
```

# Step 7：建立「只執行一次」的 OAuth 初始化函式

這個函式的工作是：

```text
觸發 Calendar Tool
↓
ADK 發出 OAuth Request
↓
User Google Login / Consent
↓
Flask 收 callback
↓
把 callback 交回 ADK
↓
ADK Token Exchange
↓
Credential 留在目前 Session
```

```python
import asyncio

from IPython.display import display, HTML


async def initialize_oauth():

    oauth_result.clear()

    # 用一個 harmless read request 觸發 Calendar OAuth
    content = types.Content(
        role="user",
        parts=[
            types.Part(
                text="請查看我今天的 Calendar 行程。"
            )
        ],
    )

    events = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=content,
    )

    auth_call = None
    auth_config_from_adk = None

    async for event in events:

        if not event.content:
            continue

        for part in event.content.parts or []:

            if (
                part.function_call
                and
                part.function_call.name
                == "adk_request_credential"
            ):

                auth_call = part.function_call

                raw_config = (
                    auth_call.args.get("authConfig")
                    or
                    auth_call.args.get("auth_config")
                )

                auth_config_from_adk = (
                    AuthConfig.model_validate(
                        raw_config
                    )
                )

                break

        if auth_call:
            break


    if not auth_call:
        print(
            "✅ OAuth credential already available."
        )
        return


    AUTH_URL = (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_uri
    )


    print("請打開以下網址完成 Google OAuth：\n")
    print(AUTH_URL)

    display(
        HTML(
            f'''
            <p>
            <a href="{AUTH_URL}" target="_blank">
            🔐 Open Google OAuth
            </a>
            </p>
            '''
        )
    )


    print("等待 Google OAuth callback...")


    # 等待 Flask 收到 callback
    for _ in range(300):

        if oauth_result.get("callback_url"):
            break

        await asyncio.sleep(1)


    if not oauth_result.get("callback_url"):
        raise TimeoutError(
            "OAuth callback timeout."
        )


    if oauth_result.get("error"):
        raise RuntimeError(
            f"OAuth error: {oauth_result['error']}"
        )


    # 把 Google callback 交回 ADK
    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_response_uri
    ) = oauth_result["callback_url"]


    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .redirect_uri
    ) = REDIRECT_URI


    auth_content = types.Content(
        role="user",

        parts=[
            types.Part(
                function_response=
                types.FunctionResponse(

                    id=auth_call.id,

                    name="adk_request_credential",

                    response=(
                        auth_config_from_adk
                        .model_dump()
                    ),
                )
            )
        ],
    )


    # Resume 同一個 Session
    events_after_auth = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=auth_content,
    )


    # 必須把 events consume 完
    async for event in events_after_auth:
        pass


    print("✅ Google Calendar OAuth initialized")
```

# Scion Hub 與 Runtime Broker：網路通訊、Port 與註冊機制

Scion 的 Hub 與 Runtime Broker 主要透過 HTTPS + WebSocket（WSS）通訊。最重要的設計是：Runtime Broker 主動連線到 Hub，因此通常不需要在 Broker 所在的電腦開放外部連入 Port。

這個設計很適合讓不同地點的電腦共同執行 AI Agent，例如研究室 Server、家中電腦及 Google Cloud VM。

## 一、Hub 與 Runtime Broker 如何通訊？

Scion Hub

雲端 VM / Linux Server

HTTPS / WSS · TCP 443

Broker A

研究室 PC

Agent 容器

Broker B

家中電腦

Agent 容器

Broker C

雲端 VM

Agent 容器

通訊程序如下：

1. Runtime Broker 啟動後，主動向 Hub 建立 WebSocket 控制通道（Control Channel）。
2. Hub 透過這條連線傳送建立、停止、管理 Agent 等指令。
3. Broker 在自己的機器上啟動容器，執行 Agent。
4. Broker 透過相同通道回報 Agent 狀態、事件及執行結果。
5. WebSocket 保持連線，讓 Hub 能持續管理遠端 Broker。

WebSocket 是建立在 TCP 之上的雙向通訊協定。使用 TLS 加密時稱為 WSS，常使用 TCP Port 443。

Scion 官方確認，這條持久化 WebSocket 通道可以穿越一般 NAT 與防火牆環境。

## 二、需要開放哪些 Port？

假設 Hub 部署在雲端，Runtime Broker 分散在不同機器。

&#x20;    Text                Markdown&#x20;

| 位置              | Port     | 用途                      | 是否需要對外開放               |
| --------------- | -------- | ----------------------- | ---------------------- |
| Hub             | TCP 443  | HTTPS、WebSocket         | 需要讓授權的遠端用戶與 Broker 能連入 |
| Hub             | TCP 8080 | Web Dashboard 預設監聽 Port | 若由反向代理處理 HTTPS，通常不直接公開 |
| Hub             | TCP 9810 | 獨立 Hub API 預設 Port      | 可限制為內部存取               |
| Broker          | TCP 9800 | Broker 本機 API 預設 Port   | 一般不需要公開                |
| Agent Container | 視應用而定    | Python API、Web Server 等 | 通常可透過 Hub 反向通道，不需直接公開  |

&#x20;     CSV          Excel&#x20;

官方設定顯示，Scion 的 Hub API 預設 Port 為 `9810`，Web UI 預設為 `8080`，Runtime Broker 本機 API 通常為 `9800`。若以反向代理提供 HTTPS，對外只需使用 443。

### 實際防火牆設定

如果你的 Hub 位於 Google Cloud VM：

- Hub 防火牆： 允許必要來源連入 TCP 443。
- Broker 防火牆： 允許向 Hub 的 TCP 443 發起連線。
- Broker 路由器： 通常不需要 Port Forwarding。
- Hub 的 9800、9810： 不應因遠端 Broker 連線而直接公開到 Internet。

這代表即使你的 Broker 位於家中 NAT Router 後面，或沒有公開 IP，也通常可以連上 Hub。

但如果公司防火牆封鎖 WebSocket、限制長連線或透過 Proxy 檢查 HTTPS，仍可能需要額外設定。

## 三、Hub 需要執行 Server 嗎？

是的。Hub 本身就是一個需要持續運行的伺服器服務（Server Service）。

它負責：

- 儲存 Broker、Agent、Project 等資訊。
- 管理身分驗證（Authentication）與存取權限。
- 將 Agent 任務派送給適當的 Broker。
- 接收 Broker 回報的狀態與事件。
- 提供 Web Dashboard 與 API。

可以將它理解為 Scion 的中央控制伺服器（Control Plane）。

Hub 可以部署在 Google Cloud VM、一般 Linux Server，或其他支援的雲端環境。

開發測試時，Hub、Broker 與 Web UI 可以在同一台電腦運行；正式部署時則可以分開。

Scion 提供 Single-node 與高可用性（High Availability, HA）部署方式。

## 四、外接的 Broker 如何向 Hub Register？

Scion 有正式的 Runtime Broker 註冊機制，並不是只在 Hub 手動輸入外接電腦的 IP。

目前官方 CLI 使用 `scion runtime-broker` 指令。


Broker 必須能使用已設定的容器執行環境，例如 Docker。

```
hub:
  enabled: true
  endpoint: "https://scion.example.com"
  local_only: false
```

接著執行：

```
scion hub auth login
```

這會讓使用者完成 Hub 的登入驗證。

```
scion runtime-broker register
```

註冊時，Hub 會建立 Broker 紀錄及憑證；Broker 在本機儲存後續連線需要的認證資訊。

&#x20;    Text                Markdown&#x20;

| 通訊資料           | 方向               | 用途        |

## 六、研究室部署建議

假設你的研究室有一台雲端 VM，另外有三台提供 Agent 運算的電腦。

&#x20;    Text                Markdown&#x20;

| 設備              | 建議角色                      | 網路要求             |
| --------------- | ------------------------- | ---------------- |
| Google Cloud VM | Scion Hub + Web Dashboard | TCP 443 Inbound  |
| Linux 工作站       | Runtime Broker A          | TCP 443 Outbound |
| 第二台工作站          | Runtime Broker B          | TCP 443 Outbound |
| 家中電腦            | Runtime Broker C          | TCP 443 Outbound |
| 手機／筆電           | Web Dashboard / CLI       | TCP 443 Outbound |

&#x20;     CSV          Excel&#x20;

這種部署不需要替每台 Broker 設定固定公開 IP，也不需要在每台 Broker 的路由器上做 Port Forwarding。

若使用 Google Cloud，可以將 Hub 放在 VM 上，再透過 Caddy、Nginx 或其他反向代理提供 HTTPS。正式上線時應啟用 OAuth、權限控管及防火牆來源限制。

最後要注意：這裡的 WebSocket 控制通道，是 Hub 與 Runtime Broker 之間的基礎設施通訊，不等同於 AI Agent 之間的語意訊息協定。 Agent 彼此交換任務、結果及通知，屬於 Scion 的另一層應用功能。

## Agent Harness

From Question Answering to Task Completion: A Survey on Agent System and Harness Design https://arxiv.org/html/2606.20683v1


### 一、整篇論文的討論架構

這篇論文以 AI 系統從「回答問題」走向「完成任務」的演進為主軸，將 Agent 系統的發展整理為幾個相互銜接的階段：

> **Prompt Engineering → Context / Workflow Engineering → Harness Engineering → Agent-native Training → Model–Harness Co-evolution**

核心觀點是：當任務從單次問答變成長時間、多步驟、需要與外部環境互動的工作時，系統效能不再只取決於模型本身，而逐漸取決於模型外部的執行架構如何管理資訊、行動、狀態與回饋。

---

### 1. Prompt Engineering：如何讓模型更好地回答

早期大型語言模型的主要使用方式是單次輸入、單次輸出，因此工程重點集中在：

> **如何設計輸入，讓模型更有效地表現既有能力。**

常見方法包括：

* Few-shot prompting
* Chain-of-Thought
* Self-consistency
* Tree of Thoughts
* ReAct prompting

這一階段主要改善的是模型能力的「引出」與「表達」。

但當任務需要多個步驟、外部資訊或實際操作時，單靠 Prompt 會遇到明顯限制，例如：

* 無法持續取得外部環境的新資訊
* 無法可靠保存長期任務狀態
* 缺乏工具執行與結果處理機制
* 缺乏失敗後的恢復能力
* 難以處理跨多個步驟的工作流程

因此，Agent 系統的設計重點開始從單次 Prompt 轉向整個執行過程。

---

### 2. Context / Workflow Engineering：每一步應該讓模型看到什麼

當模型開始參與多步驟任務後，問題不再只是「Prompt 怎麼寫」，而是：

> **在每一次模型推論時，應該提供哪些資訊？**

因此出現了 Context Engineering 與 Workflow Engineering。

系統需要動態組合：

* 使用者目標
* 系統指令
* 過去的互動紀錄
* Retrieval 結果
* Memory
* 工具說明
* 工具輸出
* 中間成果
* 當前任務狀態

Context 可以抽象表示為：

$$
[
C=A(c_1,c_2,\ldots,c_n)
]
$$

其中 (A) 是 context assembly mechanism，負責從多種資訊來源中選擇並組合目前這一步需要的內容。

此時的重點已經從：

```text
How should I prompt the model?
```

轉變成：

```text
What should the model see at this step?
```

這也是 RAG、memory、context compression、context selection 等技術在 Agent 系統中變得重要的原因。

但 Context Engineering 主要處理的是「模型現在看到什麼」，仍然無法完整解決整個任務如何被控制，例如：

* 下一步應該執行什麼？
* 是否應該重試？
* 什麼時候需要重新規劃？
* 如何判定任務已經完成？
* 執行失敗後是否應該 rollback？
* 多個 Agent 或模型之間如何分工？

這些問題進一步導向 Harness Engineering。

---

### 3. Harness Engineering：如何讓整個 Agent 系統持續完成任務

Harness Engineering 關注的不再只是單次模型呼叫，而是：

> **如何把多次模型推論、工具操作、環境回饋與系統控制組合成一個可持續執行的閉環系統。**

Agent 因此可以被理解為：

$$
[
\mathcal A_{\text{LLM}}=
\langle
\mathcal M,\mathcal H
\rangle
]
$$

其中：

$$(\mathcal M)：Model$$
$$(\mathcal H)：Harness$$

也就是：

$$
[
\boxed{\text{Agent System}=\text{Model}+\text{Harness}}
]
$$

模型主要負責推理、判斷與生成行動，而 Harness 則負責管理模型與外部世界之間的整個執行流程。

一個典型的 Agent 執行過程可以概念化為：

```text
Observe
   ↓
Construct Context
   ↓
Model Reasoning
   ↓
Action
   ↓
Environment Feedback
   ↓
Verification
   ↓
Continue / Retry / Replan / Stop
```

因此，Harness 的角色比較接近 Agent 系統的 runtime 或 control layer，而不是單純的工具集合。

在這個階段，Agent Engineering 的核心問題從：

```text
What should the model see?
```

進一步變成：

```text
How should the entire system behave over time?
```

這也是從傳統 LLM application 走向真正 Agent system 的重要分界。

---

### 4. Agent-native Training：哪些 Agent 能力應該直接學進模型

當 Harness 開始承擔大量功能後，會出現另一個問題：

> **哪些能力應該由外部 Harness 實現，哪些能力應該直接成為模型本身學到的行為？**

目前許多 Agent 能力由 Harness 額外實作，例如：

* Planning
* Reflection
* Tool use
* Error recovery
* Verification
* Retry strategy
* Task decomposition

但隨著 reinforcement learning、interactive training 與 agent-environment training 的發展，其中部分能力可能逐漸被內化到模型參數之中。

例如，一個模型如果已經透過訓練學會：

* 何時應該呼叫工具
* 工具失敗後如何修正
* 如何維持長任務的目標
* 如何根據環境回饋改變策略

那麼原本需要由 Harness 明確規定的部分控制邏輯，就可能轉變成模型本身的 learned behavior。

因此，Agent 系統設計開始涉及一個新的邊界問題：

$$
[
\text{What belongs in the Model?}
]
$$

以及：

$$
[
\text{What belongs in the Harness?}
]
$$

這個邊界並不是固定的，而會隨模型能力與訓練方法持續改變。

---

### 5. Model–Harness Co-evolution：模型與 Harness 共同演進

最後一個階段進一步把 Model 與 Harness 視為需要共同最佳化的系統。

傳統做法通常是：

```text
Fixed Model
    +
Human-designed Harness
```

但新的方向開始出現：

```text
Model
  ↕
Harness
  ↕
Evaluation / Feedback
```

Harness 本身也可以被搜尋、學習或自動最佳化。

例如可以最佳化：

* Prompt strategy
* Context selection
* Tool configuration
* Model routing
* Retry policy
* Verification strategy
* Multi-agent orchestration
* Workflow structure

因此系統最佳化的對象，不再只是模型參數：

$$
[
\theta_{\text{model}}
]
$$

也包含 Harness 的設計參數：

$$
[
\phi_{\text{harness}}
]
$$

最終問題變成：

$$
[
(\theta^*,\phi^*)=
\arg\max_{\theta,\phi}
Performance(M_\theta,H_\phi)
]
$$

也就是同時尋找最適合的 Model 與 Harness 組合。

這個方向可以稱為：

> **Model–Harness Co-evolution**

它代表 Agent 系統研究從「使用一個模型」進一步轉向「共同設計與最佳化整個智慧執行系統」。

---

整個討論架構可以濃縮為：

```text
Prompt Engineering
        │
        ▼
如何讓模型更好地回答？
        │
        ▼
Context / Workflow Engineering
        │
        ▼
每一步應該讓模型看到什麼？
        │
        ▼
Harness Engineering
        │
        ▼
整個系統如何持續執行、修正與完成任務？
        │
        ▼
Agent-native Training
        │
        ▼
哪些 Agent 能力應該直接學進模型？
        │
        ▼
Model–Harness Co-evolution
        │
        ▼
模型與執行架構如何共同最佳化？
```

從這個演進可以看出，Agent 系統的工程焦點逐漸從「單次模型輸出品質」，轉向「完整任務執行系統的可靠性與控制能力」。


### 二、Agent Harness 的核心結構

Agent Harness 可以視為模型外部的執行層，負責管理模型如何取得資訊、執行行動、保存狀態，以及驗證結果。

可以抽象表示為：

$$
\mathcal H =
\left\langle
\mathcal I_{\text{obs}},
\mathcal C,
\mathcal L,
\mathcal I_{\text{act}},
\mathcal S,
\mathcal V
\right\rangle
$$

其中包含六個核心部分：

1. Observation Interface
2. Context Manager
3. Control Loop
4. Action Interface
5. State and Artifact Store
6. Verification and Governance

---

### 1. Observation Interface：模型如何看見環境

Observation Interface 負責把外部環境轉換成模型可以理解的資訊。

例如 Coding Agent 可能接收：

```text
Repository
Terminal Output
Git Diff
Test Results
Logs
```

Web Agent 則可能接收：

```text
Screenshot
DOM
Current URL
Previous Actions
```

核心問題是：

> **模型現在需要看到哪些環境資訊，才能做出正確決策？**

Observation 並不是越多越好。資訊過多會增加 context 成本與干擾，因此重點是提供與當前決策相關的狀態。

---

### 2. Context Manager：這一次推論要放什麼資訊

Context Manager 負責決定每一次模型呼叫時，哪些資訊真正進入 context。

可能包括：

* User goal
* System instructions
* Current task state
* Recent observations
* Memory
* Retrieved knowledge
* Tool descriptions
* Intermediate artifacts

可以理解為：

```text
Persistent Information
        ↓
Context Selection
        ↓
Current Context
        ↓
Model
```

Observation Interface 與 Context Manager 的差異是：

* Observation Interface：環境中有哪些資訊可以取得
* Context Manager：這一次模型實際看到哪些資訊

因此核心不是把所有歷史紀錄全部塞入 context，而是動態選擇有用資訊。

---

### 3. Control Loop：整個任務如何被控制

Control Loop 負責 Agent 的 execution flow。

它決定：

* 下一步做什麼
* 是否呼叫工具
* 是否重試
* 是否重新規劃
* 是否交給其他 Agent
* 何時停止

最基本的形式是：

```text
Observe
   ↓
Reason
   ↓
Act
   ↓
Observe
   ↓
...
```

較結構化的形式則可能是：

```text
Plan
 ↓
Execute
 ↓
Verify
 ↓
Repair / Replan
```

因此 Control Loop 決定的是 Agent 的整體執行方式。

其中主要 trade-off 是：

$$
\text{Adaptability}
\leftrightarrow
\text{Stability}
$$

控制權越多交給 Model，彈性越高；Harness 規定的流程越多，通常穩定性越高。

---

### 4. Action Interface：模型可以做什麼

Action Interface 定義模型可以對外部環境執行哪些操作。

例如 Coding Agent：

```text
read_file()
search_code()
edit_file()
run_tests()
execute_shell()
```

Web Agent：

```text
click()
type()
open_url()
scroll()
```

重點不只是提供哪些 Tools，也包括工具的 abstraction level。

例如：

```text
shell(command)
```

提供很高的自由度。

而：

```text
search_code()
edit_function()
run_tests()
```

則限制較多，但通常更容易控制。

因此主要取捨是：

$$
\text{Flexibility}
\leftrightarrow
\text{Controllability}
$$

---

### 5. State and Artifact Store：哪些資訊需要持續保存

State and Artifact Store 負責保存跨多個步驟仍然需要使用的資訊，例如：

* Plans
* Checkpoints
* Memory
* Logs
* Code diffs
* Generated files
* Search results
* Intermediate outputs

其中最重要的概念是：

$$
\text{Context}
\neq
\text{State}
$$

Context 是：

> 模型這一次看到什麼。

State 是：

> 系統長期保存了什麼。

因此 Long-horizon Agent 不應只依賴 conversation history，而需要獨立的 persistent state。

---

### 6. Verification and Governance：如何判斷正確性與控制風險

Verification 負責判斷 Agent 的結果是否正確，例如：

* Unit tests
* Assertions
* Schema validation
* Consistency checks
* Verifier model

基本流程是：

```text
Action
   ↓
Result
   ↓
Verification
   ↓
Pass / Fail
   ↓
Continue / Repair
```

Governance 則負責限制 Agent 的權限與風險，例如：

* Permission control
* Sandbox
* Approval gate
* Retry limit
* Budget limit
* Rollback
* Termination

重要原則是：

> **關鍵限制應由系統機制直接執行，而不是只依靠 Prompt 約束模型。**

例如，比起：

```text
"Do not delete production data."
```

更可靠的是：

```text
DELETE production_data
        ↓
      DENIED
```

---

## 六個元件之間的關係

整體 Harness 可以簡化成：

```text
Environment
    ↓
Observation Interface
    ↓
Context Manager
    ↓
Model
    ↓
Control Loop
    ↓
Action Interface
    ↓
Environment
    ↓
Verification / Governance
    ↓
Continue / Retry / Stop
```

State and Artifact Store 則橫跨整個流程，持續保存重要的任務資訊。

因此：

$$
\text{Harness} = \text{Observation} + \text{Context} + \text{Control} + \text{Action} + \text{State} + \text{Verification}
$$

Harness 並不是單純的一組 Tools，而是負責把模型的推理能力轉換成可持續、可控制、可驗證任務執行的完整 runtime。


### 三、Harness 的幾種主要結構

前一部分說明的是 Harness 由哪些元件組成；這一部分則關注：

> **這些元件在實際 Agent 系統中，如何被組織成不同的執行結構。**

論文並沒有提出一個唯一、封閉的 Harness 類型分類，而是指出不同系統會依照任務特性，採用不同的 **control style**。其中最基本的幾種形式包括：

1. Simple Perceive–Act Loop
2. ReAct-style Loop
3. Plan–Execute–Verify
4. Hierarchical / Multi-Agent Workflow

此外，實際系統中還常見 Fixed Pipeline、Search-guided Workflow、Managed Runtime，以及可被搜尋或最佳化的 Learnable Harness。([arXiv][1])

不同 Harness 的主要差異，在於「多少控制權交給 Model，多少控制權保留在 Harness」。

---

### 1. Simple Perceive–Act Loop：最小型迴圈

最簡單的 Agent Harness 是持續進行：

```text
Observe
   ↓
Model
   ↓
Act
   ↓
Observe
   ↓
...
```

每一輪中：

1. 取得目前環境狀態
2. 交給模型判斷
3. 模型選擇下一個 Action
4. Harness 執行 Action
5. 將新的環境結果送回模型

這種結構的特徵是：

> **Harness 很薄，大部分控制交給 Model。**

優點是簡單、彈性高、runtime overhead 小；缺點是長任務中容易發生重複操作、方向漂移或不知道何時停止。

可以概括為：

$$
\text{Thin Harness}
+
\text{Model-driven Control}
$$

---

### 2. ReAct-style Loop：Reasoning 與 Action 交替

ReAct 在基本迴圈中加入明確的 reasoning：

```text
Observation
     ↓
Reason
     ↓
Action
     ↓
Observation
     ↓
Reason
     ↓
...
```

例如 Coding Agent：

```text
Inspect Error
     ↓
Reason
     ↓
Search Code
     ↓
Observe
     ↓
Edit Code
     ↓
Run Tests
```

這種 Harness 仍然由模型動態決定下一步，因此具有很高的 flexibility。

但如果任務很長，也容易產生：

* 重複 reasoning
* 無效 tool calls
* trajectory drift
* token 消耗增加

所以 ReAct 可以理解成：

$$
\text{Interactive Loop}
+
\text{Model-directed Planning}
$$

---

### 3. Plan–Execute–Verify：把規劃、執行與驗證分開

更結構化的 Harness 會明確區分：

```text
Plan
  ↓
Execute
  ↓
Verify
  ↓
 ┌──────────────┐
 │              │
Pass           Fail
 │              │
Done       Repair / Replan
                │
                ↓
             Execute
```

其中：

* **Planner**：拆解任務與安排步驟
* **Executor**：執行工具與操作
* **Verifier**：判斷結果是否正確

例如 Coding Agent：

```text
Plan Repair
    ↓
Modify Code
    ↓
Run Tests
    ↓
Pass?
 ┌─────┴─────┐
Yes          No
 │            │
Done       Diagnose
              ↓
            Repair
```

與 ReAct 最大的差異是：

```text
ReAct
→ Model 動態決定整個 trajectory

Plan–Execute–Verify
→ Harness 先規定重要 execution stages
```

因此這種架構通常比較穩定，也更容易加入 verification 與 recovery。

---

### 4. Fixed / Structured Pipeline：由 Harness 主導流程

有些任務可以進一步把 workflow 固定下來。

例如：

```text
Localization
     ↓
Repair Generation
     ↓
Patch Selection
     ↓
Validation
```

此時模型負責完成各個 stage，但無法自由改變整體流程。

可以表示為：

$$
\text{Structured Harness}
+
\text{Model as Component}
$$

而不是：

$$
\text{Minimal Harness}
+
\text{Model as Controller}
$$

這類結構的優點是：

* 穩定
* 容易重現
* 成本較容易控制
* 每一階段容易驗證

缺點則是彈性較低。

因此，自主程度越高並不一定代表效果越好。對流程明確、可驗證的任務，Structured Pipeline 可能反而更有效。

---

### 5. Search-guided / Domain-guided Harness：限制搜尋空間

另一種做法是在自由 Agent 與固定 Pipeline 之間加入領域知識。

例如 Coding Agent 可以利用：

```text
Issue
  ↓
Repository Search
  ↓
Code Localization
  ↓
Generate Patch
  ↓
Validation
```

這類 Harness 不讓模型在整個環境中完全自由搜尋，而是利用：

* Code structure
* AST
* Repository search
* Domain-specific tools

縮小可能的 action space。

其概念可以表示為：

$$
\text{LLM Reasoning}
+
\text{Domain Structure}
$$

Harness 在這裡扮演的是 **search-space shaping** 的角色。

---

### 6. Hierarchical / Multi-Agent Harness：多 Agent 分工

當系統包含多個 Agent 時，Harness 還需要決定：

* 誰負責哪一個任務
* 何時 handoff
* 是否平行執行
* 如何整合結果
* 誰負責驗證

典型架構：

```text
                 Orchestrator
              /      |       \
             /       |        \
      Researcher   Coder    Browser
             \       |        /
              \      |       /
                 Verifier
```

常見形式包括：

#### Planner–Executor–Verifier

```text
Planner
   ↓
Executor
   ↓
Verifier
```

#### Specialist Routing

```text
             Router
          /    |     \
       Coding Search  Math
```

#### Sequential Handoff

```text
Agent A
   ↓
Agent B
   ↓
Agent C
```

#### Parallel Execution

```text
             Orchestrator
          /       |       \
      Agent A  Agent B  Agent C
          \       |       /
             Aggregator
```

因此 Multi-Agent 並不是獨立於 Harness 的概念，而可以理解為 Control Loop 的一種擴展：

$$
\text{Multi-Agent Orchestration}
\subset
\text{Harness Control Design}
$$

---

### 7. Managed Runtime：完整的 Agent 執行環境

更完整的 Harness 不只控制單一 workflow，而是提供一整套 runtime，例如：

* Context management
* Tool execution
* File editing
* Browser access
* State persistence
* Verification
* Recovery
* Sandbox

可以概念化為：

```text
             Managed Runtime
      ┌────────────────────────┐
      │        Context         │
      │           ↓            │
      │         Model          │
      │           ↓            │
      │    Action / Tools      │
      │           ↓            │
      │      Environment       │
      │                        │
      │ State   Verification   │
      │ Recovery   Sandbox     │
      └────────────────────────┘
```

這種結構適合：

* Long-horizon tasks
* 多工具環境
* 開放式工作
* 需要保存狀態的任務
* 需要 recovery 的場景

---

### 8. Learnable / Search-Optimized Harness：Harness 本身也可以最佳化

Harness 不一定完全由人類手工設計，也可以把它視為可最佳化的系統：

$$
H_{\phi}
$$

其中 $\phi$ 可以代表：

* Prompt configuration
* Tool selection
* Context policy
* Routing policy
* Retry strategy
* Verification strategy
* Workflow structure

目標變成：

$$
\phi^* =
\arg\max_{\phi}
Performance(M,H_{\phi})
$$

也就是在 Model 固定的情況下，搜尋更好的 Harness。

流程可以是：

```text
Candidate Harness
       ↓
Execute Tasks
       ↓
Collect Traces
       ↓
Evaluate
       ↓
Modify Harness
       ↓
Execute Again
```

這代表 Agent optimization 不只包含模型訓練，也包含 runtime architecture 的最佳化。

---

## Harness 結構可以視為一條 Control Spectrum

上述結構可以放在一條「控制權分配」的光譜上：

```text
更多控制交給 Model

Simple Perceive–Act
        ↓
ReAct
        ↓
Plan–Execute–Verify
        ↓
Search-guided Workflow
        ↓
Structured Pipeline
        ↓
Managed Runtime
        ↓
Hierarchical / Multi-Agent
        ↓
Learnable Harness

更多控制與結構放在 Harness
```

因此真正的設計問題不是：

> 哪一種 Harness 最先進？

而是：

$$
\text{How much control should belong to the Model,}
$$

$$
\text{and how much should belong to the Harness?}
$$

Model 控制越多，通常彈性越高；Harness 控制越多，通常穩定性、可驗證性與可預測性越高。最適合的結構仍取決於任務本身。


### 四、Task Landscape and Harness Configuration：不同任務需要不同的 Harness

不存在一套適合所有 Agent 的最佳 Harness。比較合理的設計方式是先分析任務特性，再決定哪些 Harness 元件需要被強化。

可以把流程簡化為：

```text
Task Property
      ↓
Failure Pressure
      ↓
Harness Bottleneck
      ↓
Harness Configuration
```

也就是：

> **任務的結構，決定系統最容易在哪裡失敗；Harness 則需要針對這些失敗點進行設計。**

---

### 1. 三個主要任務維度

判斷 Harness 如何配置，可以先看三個面向：

1. **Task Horizon**：任務有多長、需要多少步
2. **Environment**：Agent 在什麼環境中工作
3. **Autonomy**：Agent 被允許自主決策到什麼程度

---

### 1.1 Task Horizon：任務長度

任務越長，對 Context、State 與 Control 的要求越高。

| 任務類型         | 例子                                | 主要 Harness 壓力             |
| ------------ | --------------------------------- | ------------------------- |
| Single-step  | Search、Translate                  | Context、Verification      |
| Multi-step   | Form filling、Code generation      | Action、State、Verification |
| Long-horizon | Repo coding、Deep research         | Context、State、Control     |
| Open-ended   | Monitoring、Autonomous exploration | Control、Governance        |

長任務最常見的問題是 **state drift**：

* 忘記原始目標
* 忘記已完成的工作
* 重複嘗試失敗的方法
* 中間結果散落在 context 中

因此 Long-horizon Agent 通常需要：

$$
\mathcal C
+
\mathcal S
+
\mathcal L
$$

也就是：

* Context Manager
* State and Artifact Store
* Control Loop

---

### 1.2 Environment：Agent 在什麼環境工作

不同環境的主要瓶頸不同。

#### Coding / Terminal

Coding Agent 可以取得：

* Repository
* Terminal
* Compiler output
* Tests
* Logs

最大的優勢是存在較強的 verification signal。

典型流程：

```text
Generate
   ↓
Run
   ↓
Test
   ↓
Pass / Fail
   ↓
Repair
```

因此 Coding Agent 通常是 **Verification-dominant**。

---

#### Web / GUI

Web Agent 主要問題是：

> 模型是否正確理解目前畫面，並把 Action 作用到正確的位置。

典型流程：

```text
Web Page
   ↓
Observation
   ↓
Model
   ↓
Action
   ↓
Updated Page
```

因此主要壓力集中在：

$$
\mathcal I_{\text{obs}}
+
\mathcal I_{\text{act}}
+
\mathcal V
$$

也就是 Observation、Action Grounding 與 Verification。

---

#### Research / Knowledge Environment

Research Agent 的主要問題不是操作工具，而是管理大量資訊：

```text
Search
   ↓
Collect Evidence
   ↓
Store Sources
   ↓
Synthesize
   ↓
Check Provenance
```

這類任務通常沒有像 unit test 一樣明確的 oracle，因此主要依賴：

$$
\mathcal C
+
\mathcal S
+
\mathcal V
$$

也就是 Context、State 與 Verification。

---

### 1.3 Autonomy：Agent 有多少自主權

自主程度越高，Harness 所承擔的控制責任也越高。

```text
Human-in-the-loop
        ↓
Semi-autonomous
        ↓
Fully autonomous
```

Human-in-the-loop 可以在高風險操作前加入人工確認：

```text
Agent proposes action
        ↓
Human approval
        ↓
Execute
```

Fully autonomous Agent 則必須自行處理：

* Verification
* Retry
* Logging
* Budget control
* Rollback
* Escalation
* Termination

因此：

$$
\text{Autonomy}
\uparrow
\quad\Rightarrow\quad
\text{Harness Responsibility}
\uparrow
$$

---

## 2. 不同 Domain 有不同 Harness Bottleneck

幾種典型任務可以整理成：

| Domain               | 主要 Bottleneck | 重點 Harness 元件              |
| -------------------- | ------------- | -------------------------- |
| Software Engineering | Verification  | Verification、State         |
| Web / GUI            | Grounding     | Observation、Action         |
| Research             | Synthesis     | Context、State、Verification |
| Medical Assistance   | Safety        | Verification、Governance    |
| Embodied Agent       | Control       | Control Loop、Observation   |

這表示不同 Agent 不應該只是換 Tool，而應該改變整個 Harness configuration。

---

## 3. Oracle Strength 會直接影響 Harness 結構

一個非常重要的變數是：

> **結果是否可以被可靠驗證？**

Coding Agent 通常具有 strong oracle：

```text
Code
 ↓
Run Tests
 ↓
Pass / Fail
```

因此可以建立：

```text
Generate
   ↓
Verify
   ↓
Repair
   ↓
Verify
```

Research Agent 則通常屬於 weak oracle：

```text
Claim
 ↓
沒有明確 Pass / Fail
 ↓
需要 Source、Provenance、Cross-check
```

因此：

$$
\text{Strong Oracle}
\Rightarrow
\text{Verifier Loop}
$$

而：

$$
\text{Weak Oracle}
\Rightarrow
\text{Provenance}
+
\text{Review}
+
\text{Cross-check}
$$

所以 verification 的形式必須依任務本身改變。

---

## 4. Irreversible Action 需要更強的 Governance

有些 Action 做錯後可以重試，例如：

* Search
* Read file
* Generate draft
* Run sandbox code

但有些操作具有 persistent side effect，例如：

* Send email
* Submit form
* Purchase
* Delete file
* Modify production system

這種情況不能只依賴：

```text
Execute
   ↓
Fail
   ↓
Retry
```

而需要在 Action 前加入：

```text
Proposed Action
      ↓
Risk Check
      ↓
Permission Gate
      ↓
Execute
```

因此 action 越不可逆，Verification 與 Governance 就越重要。

---

## 5. Long-horizon Agent 必須把 State 與 Context 分開

長任務不能單純把所有歷史資訊一直塞進 Context Window。

需要明確區分：

$$
\text{Context}
\neq
\text{State}
$$

Context 是：

> 模型這一次需要看到什麼。

State 是：

> 系統之後還必須記得什麼。

例如：

```text
Persistent State
 ├── Plan
 ├── Checkpoints
 ├── Evidence
 ├── Artifacts
 └── Failed Attempts
          ↓
    Context Manager
          ↓
      Current Context
          ↓
         Model
```

這是 Long-horizon Agent 維持一致性的重要條件。

---

## 6. Harness Configuration 的核心原則

整個 Task Landscape 可以濃縮成：

$$
\text{Task Structure}
\rightarrow
\text{Failure Pressure}
\rightarrow
\text{Harness Configuration}
$$

例如：

```text
Long-horizon Research
        ↓
容易遺失證據與推論
        ↓
強化 Context + State
        ↓
Artifacts + Provenance + Checkpoints
```

而：

```text
Coding with Tests
        ↓
結果容易驗證
        ↓
強化 Verification Loop
        ↓
Generate → Test → Repair
```

因此，Harness 設計的核心並不是先選一個 Agent Framework，而是先找出：

> **這個任務最容易在哪裡失敗？**

再把系統結構集中在那些真正的 bottleneck 上。


### 五、Evaluation and Empirical Analysis：Agent 應以 Model–Harness System 評估

Agent 的效能不能只歸因於 Model 本身，而應視為完整系統的結果：

$$
\text{Agent Performance} = f(
\text{Model},
\text{Harness},
\text{Task},
\text{Environment},
\text{Evaluation}
)
$$

也就是說，同一個 Model 搭配不同 Harness，可能產生明顯不同的任務完成率、成本與可靠性。([arXiv][1])

---

### 1. Agent Benchmark 測的是完整執行流程

傳統 LLM benchmark 通常測量：

```text
Question
   ↓
Model
   ↓
Answer
   ↓
Score
```

Agent benchmark 則測量整條 execution trajectory：

```text
Task
  ↓
Model + Harness
  ↓
Observe
  ↓
Reason / Act
  ↓
Environment Feedback
  ↓
Verify / Retry / Continue
  ↓
Final Result
```

因此，Context、Tool Interface、Retry Policy、Verification 與 Stopping Criteria 都會直接影響結果。([arXiv][1])

---

### 2. 不同 Benchmark 測到不同 Harness 能力

不同任務環境會形成不同的 Harness 壓力：

| Benchmark      | 主要環境       | 主要 Harness 能力                  |
| -------------- | ---------- | ------------------------------ |
| SWE-bench      | Repository | Code editing、Verification      |
| Terminal-Bench | Terminal   | Action、State、Verification      |
| WebArena       | Browser    | Observation、Grounding、Action   |
| OSWorld        | Desktop    | GUI interaction、State tracking |

因此 Benchmark score 不應只解讀為：

```text
Model X = 70%
```

而應理解為：

```text
Model X
+
Harness Y
+
Runtime Configuration Z
=
70%
```

([arXiv][1])

---

### 3. 同一個 Model，Harness 可以造成很大的差距

論文整理多個 benchmark 後發現，即使 Model 固定，替換 Harness 仍可能造成顯著效能差異。([arXiv][1])

例如在 Terminal-Bench 2.0 中：

```text
Claude Opus 4.6

Claude Code
58.0%
   ↓
Meta-Harness
76.4%
```

差距為：

```text
76.4%
- 58.0%
-------
  18.4%
```

Gemini 3.1 Pro：

```text
Gemini CLI
59.4%
   ↓
TongAgents
80.2%
```

差距為：

```text
  80.2%
- 59.4%
-----
  20.8%
```

在具有至少三種 Harness 結果的模型中，median within-model spread 約為：

$$
13.6%
$$

這顯示：

$$
\text{Fixed Model}
\not\Rightarrow
\text{Fixed Agent Performance}
$$

([arXiv][1])

---

### 4. Web Agent 的 Harness Effect 更明顯

WebArena 特別依賴：

$$
\mathcal I_{\text{obs}}
+
\mathcal I_{\text{act}}
+
\mathcal V
$$

也就是：

* Observation
* Action Grounding
* Verification

因此即使 Model 具備很強的 reasoning capability，如果無法正確理解網頁狀態或操作 UI，實際 task completion 仍然可能很低。([arXiv][1])

例如 GPT-4o：

```text
Model-only baseline
13.1%
     ↓
Best reported harness
54.6%
```

差距達到：

```text
  54.6%
- 13.1%
--------
  41.5%
```


這說明 Web Agent 的效能很大程度取決於 Model 與 Harness 的共同設計。

---

### 5. Harness 越複雜，不代表越好

Harness 的品質不能直接用：

* Tool 數量
* Agent 數量
* Workflow 複雜度
* Prompt 長度

來衡量。

更重要的是：

```text
Right Observation
       +
Right Context
       +
Right Action Space
       +
Right Verification
```

因此：

$$
\text{Harness Quality}
\neq
\text{Harness Complexity}
$$

有些較精簡的 Harness，也可能得到與大型 Managed Runtime 接近的表現。([arXiv][1])

---

### 6. 評估不能只看 Success Rate

Agent evaluation 不應只測：

$$
\text{Task Success Rate}
$$

還應同時考慮：

* Reliability
* Token / API Cost
* Latency
* Timeout
* Safety
* Recovery capability

例如兩個 Agent 都有：

$$
75%
\text{ Success Rate}
$$

但其中一個只需要少量 token 與幾次操作，另一個可能需要大量 context、長時間執行與多次 retry。

兩者的實際 deployment value 並不相同。

---

### 7. 評估單位應該是 Model–Harness Pair

因此更完整的 benchmark unit 應該寫成：

$$
(
\text{Model},
\text{Harness},
\text{Runtime Configuration}
)
$$

而不是只報告 Model 名稱。

實際比較時至少應記錄：

```text
Model Version
Harness
Context Policy
Tools
Retry Policy
Verifier
Execution Budget
Runtime Environment
```

這樣才能判斷效能提升究竟來自 Model、Harness，還是兩者之間的 compatibility。

---

## 第五部分的核心

整體可以濃縮成：

```text
Model
   +
Harness
   +
Runtime Configuration
   ↓
Execution Trajectory
   ↓
Success
Reliability
Cost
Latency
Safety
```

因此：

$$
\boxed{
\text{Agent Performance}
\text{ is a property of the Model--Harness system}
}
$$

評估 Agent 時，重點不應只是「哪一個 Model 分數最高」，而是：

> **哪一個 Model 與 Harness 的組合，能以合理成本穩定完成任務。**

[1]: https://arxiv.org/html/2606.20683v1 "From Question Answering to Task Completion: A Survey on Agent System and Harness Design"


### 六、Outlook and Future Directions：Agent Harness 的未來方向

前面的討論已經說明，Agent 的能力不是只由 Model 決定，而是來自 Model、Harness 與任務環境的共同作用。未來的發展重點，可以濃縮成三個方向：

1. 從 Score 走向 Value-aware Evaluation
2. 讓 Agent 學會 Verification、Recovery 與 Adaptation
3. 在 General Harness 與 Specialized Harness 之間取得平衡

([arXiv][1])

---

### 1. 從 Score 走向 Value-aware Evaluation

目前多數 Agent benchmark 仍然主要比較 task success，但實際部署時還需要考慮：

* Cost
* Latency
* Reliability
* Safety
* Process Quality
* Trace Auditability

因此評估目標不應只是：

```text
Maximize Success Rate
```

而應該同時考慮：

```text
Task Value
+
Success
+
Reliability
+
Process Quality
-
Cost
-
Latency
-
Risk
```

換句話說，一個 Agent 即使成功率較高，如果需要大量 token、執行時間很長、經常 timeout，或缺乏可靠的 verification，也未必是更好的 deployment choice。

因此未來的 Agent optimization 更接近：

> **在成本、延遲與風險限制下，最大化可靠完成任務所產生的實際價值。**

([arXiv][1])

---

### 2. Agent 不只要學會 Acting，也要學會 Verification 與 Recovery

目前很多 Agent 能力仍由 Harness 提供，例如：

```text
Retry
Verification
Error Recovery
Replanning
Rollback
```

未來的一個重要方向，是讓這些能力逐漸成為模型本身可以學習的行為。

也就是從：

```text
Plan
 ↓
Act
```

進一步變成：

```text
Plan
 ↓
Act
 ↓
Verify
 ↓
Diagnose Failure
 ↓
Recover
 ↓
Adapt
```

Execution trace 在這裡不只是紀錄，也可以成為學習資料。

系統可以從過去執行中收集：

* Success / Failure
* Tool results
* Verifier signals
* Recovery attempts
* Cost
* Safety events

再利用這些 evidence 改進 Model 或 Harness。([arXiv][1])

可以表示成：


$$
(\theta_{t+1},\phi_{t+1})
=
\mathrm{VerifyRetain}
\left(
U(\theta_t,\phi_t,\mathcal E_t)
\right)
$$


其中：

* $\theta$：Model parameters
* $\phi$：Harness configuration
* $\mathcal E_t$：執行過程產生的 evidence
* $U$：更新機制

但新的版本只有在通過驗證後才保留，否則應該 rollback。

---

### 3. 從 Model Optimization 走向 Model–Harness Co-evolution

未來 Agent improvement 不一定只修改模型。

可以同時改善：

```text
Model
  +
Harness
```

例如：

```text
Execution
    ↓
Collect Traces
    ↓
Identify Failure
    ↓
 ┌───────────────┐
 ↓               ↓
Improve Model   Improve Harness
 ↓               ↓
 └───────┬───────┘
         ↓
     Re-evaluate
```

Harness 可以改進：

* Context policy
* Observation format
* Action abstraction
* Memory retrieval
* Retry strategy
* Verifier timing
* Model routing

Model 則可以逐漸內化：

* Tool use
* Verification
* Recovery
* Adaptation

因此兩者會形成循環：

```text
Better Harness
      ↓
Better Execution Traces
      ↓
Better Training Data
      ↓
Better Model
      ↓
Different Harness becomes optimal
      ↓
Better Harness
```

這就是 **Model–Harness Co-evolution**。([arXiv][1])

---

### 4. Harness 不會因為 Model 變強而消失

即使未來模型學會更多 planning、tool use、verification 與 recovery，Harness 仍然有許多不能完全由 Model 取代的責任。

例如：

* Sandbox
* Permission control
* Artifact storage
* Budget control
* Audit logs
* Rollback
* Human approval
* Execution environment

因此未來並不是：

```text
Smarter Model
    ↓
No Harness
```

而比較可能是：

```text
Smarter Model
      +
More Adaptive Harness
```

模型會逐漸內化部分 agentic behavior，而 Harness 則繼續負責 runtime control、environment interface、verification 與 governance。([arXiv][1])

---

### 5. General Harness 與 Specialized Harness 的取捨

另一個重要問題是：

> **Harness 應該做成通用平台，還是針對特定任務最佳化？**

有些能力具有很高的共通性，例如：

```text
Tracing
Sandbox
Permission Control
Artifact Storage
Budget Management
Model Routing
Tool Protocol
```

這些可以形成 reusable infrastructure。

但另外一些元件通常高度依賴任務：

```text
Observation Format
Action Abstraction
Memory Policy
Verifier
Recovery Strategy
```

例如 Coding Agent 與 Web Agent 即使使用同一個 Model，也不應該使用完全相同的 Observation Interface 與 Verifier。

因此比較合理的方向是：

```text
Reusable Core Harness
          +
Task-specific Configuration
```

而不是所有任務共用完全相同的 Harness。([arXiv][1])

---

## 第六部分的核心

整個未來方向可以濃縮成：

```text
Current Agent Systems
        ↓
Model + Human-designed Harness
        ↓
Collect Execution Traces
        ↓
Learn Verification / Recovery / Adaptation
        ↓
Improve Model + Improve Harness
        ↓
Re-evaluate under Cost / Risk / Reliability constraints
        ↓
Model–Harness Co-evolution
```

因此 Agent 系統未來的方向，不只是：

> **讓 Model 更強。**

而是：

> **讓 Model 與 Harness 都能根據實際執行經驗持續改善，同時維持 verification、governance 與可控性。**

最終可以簡化成：

```text
Better Agent
=
Better Model
+
Better Harness
+
Better Evaluation
+
Better Feedback Loop
```

這也是從單純的模型能力競爭，走向完整 Agent System Engineering 的核心轉變。 ([arXiv][1])

[1]: https://arxiv.org/html/2606.20683v1 "From Question Answering to Task Completion: A Survey on Agent System and Harness Design"

### agent.py

```python
"""README improvement harness built with ADK.

A SequentialAgent pipeline that fetches a GitHub repo via MCP,
generates a README using skill-based conventions, and refines
it in a LoopAgent until a critic approves.
"""

import os
import pathlib

from google.adk.agents import Agent, LoopAgent, SequentialAgent
from google.adk.agents.callback_context import CallbackContext
from google.adk.skills import load_skill_from_dir
from google.adk.tools import exit_loop
from google.adk.tools.mcp_tool import McpToolset
from google.adk.tools.mcp_tool.mcp_session_manager import StdioConnectionParams
from google.adk.tools.skill_toolset import SkillToolset
from mcp import StdioServerParameters

SKILLS_DIR = pathlib.Path(__file__).parent / "skills"

# --- Tool Group 1: GitHub MCP (fetch repo contents) ---

github_mcp = McpToolset(
    connection_params=StdioConnectionParams(
        server_params=StdioServerParameters(
            command="npx",
            args=["-y", "@modelcontextprotocol/server-github"],
            env={
                "GITHUB_PERSONAL_ACCESS_TOKEN": os.getenv(
                    "GITHUB_PERSONAL_ACCESS_TOKEN", ""
                ),
            },
        ),
        timeout=30,
    ),
    tool_filter=[
        "get_file_contents",
        "search_code",
        "search_repositories",
        "list_commits",
    ],
)

# --- Tool Group 2: README conventions skill ---

readme_skill = SkillToolset(
    skills=[load_skill_from_dir(SKILLS_DIR / "readme-conventions")],
)

# --- Stage 1: Codebase Analyzer ---

codebase_analyzer = Agent(
    model="gemini-3-flash-preview",
    name="codebase_analyzer",
    instruction="""\
You are a codebase analyst. The user will give you a GitHub repository
(owner/repo format or a full URL).

Your job:
1. Use the GitHub MCP tools to explore the repository.
2. Read the top-level file list and identify key files:
   - README.md (if it exists)
   - package.json, pyproject.toml, requirements.txt, or similar
   - Main source directories (src/, lib/, app/, etc.)
   - Configuration files (.env.example, Dockerfile, etc.)
3. Read the main source files (up to 5 key files) to understand
   what the project does.
4. Read any existing README.md to understand current coverage.

Output a structured analysis with these sections:
- **Project purpose**: One sentence on what this project does.
- **Tech stack**: Languages, frameworks, key dependencies.
- **Key modules**: Main directories and what they contain.
- **Setup requirements**: How to install and configure.
- **API surface**: Main exports, commands, or endpoints.
- **Existing README gaps**: What the current README misses or gets wrong.
  If no README exists, say "No README found."
""",
    tools=[github_mcp],
    output_key="codebase_analysis",
)

# --- Callback: Initialize loop state on first iteration ---


async def init_loop_state(callback_context: CallbackContext) -> None:
    """Set default criticism on first loop pass so {criticism} resolves."""
    if "criticism" not in callback_context.state:
        callback_context.state["criticism"] = (
            "No previous feedback. This is the first draft."
        )


# --- Stage 2a: README Writer ---

readme_writer = Agent(
    model="gemini-3-flash-preview",
    name="readme_writer",
    before_agent_callback=init_loop_state,
    instruction="""\
You are a README writer. Your job is to write or improve a README.md
for the repository described in the codebase analysis below.

**Codebase analysis:**
{codebase_analysis}

**Previous criticism (if any):**
{criticism}

Instructions:
1. Load the readme-conventions skill to learn the required structure.
2. Load the checklist reference for section-by-section quality criteria.
3. Write a complete README.md that covers every required section.
4. If criticism was provided, address every point specifically.
5. Write from a developer's perspective: assume the reader wants to
   clone, install, and use this project within 5 minutes.

Output the full README.md content in markdown format.
""",
    tools=[readme_skill],
    output_key="current_readme",
)

# --- Stage 2b: README Critic ---

readme_critic = Agent(
    model="gemini-3-flash-preview",
    name="readme_critic",
    instruction="""\
You are a README quality reviewer. Review the README below against
the required sections checklist.

**README to review:**
{current_readme}

**Codebase context (for accuracy check):**
{codebase_analysis}

Check each required section:
1. **Title and description**: Clear, one-sentence purpose statement?
2. **Installation**: Step-by-step setup commands that work?
3. **Usage**: At least one code example or CLI command?
4. **Configuration**: Environment variables, config files documented?
5. **API / Key features**: Main capabilities listed?
6. **Contributing**: How to contribute mentioned?
7. **License**: License referenced?

Scoring:
- If ALL sections are present and reasonably complete, call exit_loop.
- If ANY section is missing or clearly incomplete, list what needs
  fixing and do NOT call exit_loop.

Be practical, not pedantic. A section with 2-3 sentences is fine.
A missing section is not.
""",
    tools=[exit_loop],
    output_key="criticism",
)

# --- Stage 2: Refinement Loop ---

refinement_loop = LoopAgent(
    name="refinement_loop",
    sub_agents=[readme_writer, readme_critic],
    max_iterations=3,
)

# --- Root Agent: The Harness ---

root_agent = SequentialAgent(
    name="readme_harness",
    description=(
        "A README improvement harness. Give it a GitHub repo and it "
        "fetches the code, generates a README against a quality checklist, "
        "and refines until the critic approves."
    ),
    sub_agents=[codebase_analyzer, refinement_loop],
)

```

### readme-conventions skill

```text
---
name: readme-conventions
description: >
  Quality conventions for writing developer-focused README files.
  Covers required sections, structure, and quality criteria.
  Load the checklist reference for section-by-section requirements.
---

# README Conventions

## When to Use

Load this skill when generating or improving a README.md file for a code repository.

## Process

1. Read `references/checklist.md` for the full section-by-section quality checklist.
2. Follow the section order defined in the checklist.
3. Write from the perspective of a developer who just discovered this repo
   and wants to clone, install, and use it within 5 minutes.

## Rules

1. Start with the project name as an H1 heading, followed by a one-sentence description.
2. Every section from the checklist must be present. If a section does not apply, include it with a brief note explaining why.
3. Installation instructions must include exact commands (not placeholders).
4. Usage examples must show real code or CLI commands, not abstract descriptions.
5. Keep the total README under 500 lines. Put detailed API docs in separate files if needed.
6. Do not use badges unless the project already has CI/CD configured.
7. Do not include a table of contents for READMEs under 200 lines.
```


/reference/checklist.md

```text
# README Quality Checklist

Each section below is required. The README critic validates against this list.

## 1. Title and Description

- H1 heading with the project name
- One-sentence description of what the project does
- Who it is for (target audience)

## 2. Prerequisites

- Runtime requirements (Python version, Node.js, etc.)
- System dependencies (databases, external services)
- Required accounts or API keys

## 3. Installation

- Clone command
- Dependency installation command
- Any build or compilation steps
- Must be copy-pasteable (no placeholders like "your-path-here")

## 4. Configuration

- Environment variables with descriptions
- Configuration files and their format
- Example .env or config file content

## 5. Usage

- At least one complete example (code snippet or CLI command)
- Expected output or result
- Common use cases (2-3 examples if applicable)

## 6. Project Structure

- Brief description of key directories and files
- Where to find the main entry point
- Where configuration lives

## 7. API / Key Features

- Main capabilities or endpoints listed
- Brief description of each
- Link to detailed docs if they exist

## 8. Contributing

- How to set up a development environment
- How to run tests
- How to submit changes (PR process or contact)

## 9. License

- License type stated
- Link to LICENSE file if it exists
```

1. 現在的程式架構

```text
readme-conventions/
├── SKILL.md
│   └── 定義 README 的部分規則
└── references/
    └── checklist.md
        └── 定義完整 checklist

agent.py
├── readme_writer
│   ├── 使用 readme_skill
│   └── prompt 也寫了一部分規則
│
└── readme_critic
    ├── 只有 exit_loop tool
    └── prompt 又另外寫了一套檢查規則
```

問題是「規則散落在三個地方」：

```text
SKILL.md
checklist.md
writer / critic prompts
```

因此很容易發生規則不一致。

2. 建議修改方式

把：

```text
references/checklist.md
```

設成唯一的 Single Source of Truth。

架構改成：

```text
readme-conventions/
├── SKILL.md
│   └── 只說：請讀 checklist.md
└── references/
    └── checklist.md
        └── 所有 README 規則都集中在這裡
```

然後兩個 Agent 都使用同一個 skill：

```python
readme_writer:
    tools=[readme_skill]

readme_critic:
    tools=[readme_skill, exit_loop]
```

Writer prompt 只需要：

```text
讀取 README skill/checklist，
依照規格產生 README。
```

Critic prompt 只需要：

```text
讀取同一份 README skill/checklist，
依照規格檢查 README。
全部通過就 exit_loop，
否則輸出 criticism。
```

最後的理想架構就是：

```text
              checklist.md
              Single Source
                   │
          ┌────────┴────────┐
          ▼                 ▼
    readme_writer      readme_critic
          │                 │
          └── README ──────►│
                            │
                      pass / criticism
```

核心原則：Prompt 管「要做什麼」，`checklist.md` 管「什麼叫做做對」。

## How to Use

## Output

- **stderr**: Color-coded progress showing each harness stage (analyzer, writer, critic, feedback, approval)
- **stdout**: The final improved README in markdown

## What Happens Behind the Scenes

The script calls an ADK agent harness running as an api_server. The harness is a
SequentialAgent with two stages:

1. **Codebase Analyzer** — Fetches repo via GitHub MCP (file tree, source, deps, existing README)
2. **Refinement Loop** — A LoopAgent cycles between:
   - **Writer** — Generates README using a SkillToolset with conventions checklist
   - **Critic** — Validates against 9 required sections, calls `exit_loop` when approved

The loop runs up to 3 iterations. You will see the progress streamed in the terminal.

- "Improve the README for owner/repo"

- "Fix the README for owner/repo"
- "Analyze owner/repo and write a good README"

```

/script/call_harness.py

```python
#!/usr/bin/env python3
"""Call the README harness api_server and stream progress.

Usage:
    python scripts/call_harness.py owner/repo [--host localhost:8000] [--save output.md]

The script:
1. Creates a session on the running api_server
2. Sends the repo to the README harness
3. Streams progress (which agent is running, tool calls, feedback)
4. Prints the final README to stdout (or saves to file)
"""

import argparse
import json
import sys
import urllib.error
import urllib.request


def create_session(base_url: str) -> str:
    """Create a new session and return its ID."""
    url = f"{base_url}/apps/app/users/cli_user/sessions"
    req = urllib.request.Request(
        url,
        data=b"{}",
        headers={"Content-Type": "application/json"},
        method="POST",
    )
    with urllib.request.urlopen(req) as resp:
        data = json.loads(resp.read())
        return data["id"]


def run_harness(base_url: str, session_id: str, repo: str):
    """Send repo to the harness and stream progress. Returns final README."""
    url = f"{base_url}/run_sse"
    payload = json.dumps(
        {
            "app_name": "app",
            "user_id": "cli_user",
            "session_id": session_id,
            "new_message": {
                "role": "user",
                "parts": [{"text": f"Improve the README for {repo}"}],
            },
            "streaming": False,
        }
    ).encode()

    req = urllib.request.Request(
        url,
        data=payload,
        headers={"Content-Type": "application/json"},
        method="POST",
    )

    final_readme = ""
    current_agent = ""
    iteration = 0

    with urllib.request.urlopen(req, timeout=300) as resp:
        for raw_line in resp:
            line = raw_line.decode("utf-8").strip()
            if not line or line.startswith(":"):
                continue
            if not line.startswith("data: "):
                continue

            data = line[6:]
            try:
                event = json.loads(data)
            except json.JSONDecodeError:
                continue

            content = event.get("content", {})
            author = content.get("author", event.get("author", ""))
            parts = content.get("parts", [])

            # Track agent transitions
            if author and author != current_agent:
                current_agent = author
                if author == "codebase_analyzer":
                    log("STAGE 1", "Analyzing codebase via GitHub MCP...")
                elif author == "readme_writer":
                    iteration += 1
                    if iteration == 1:
                        log("STAGE 2", "Generating README (pass 1)...")
                    else:
                        log("REFINE", f"Improving README (pass {iteration})...")
                elif author == "readme_critic":
                    log("REVIEW", f"Critic reviewing (pass {iteration})...")

            for part in parts:
                if "functionCall" in part:
                    fc = part["functionCall"]
                    name = fc["name"]
                    args = fc.get("args", {})

                    if name == "exit_loop":
                        log("APPROVED", "Critic approved the README!")
                    elif name == "get_file_contents":
                        path = args.get("path", "/")
                        log(
                            "  FETCH",
                            f"{args.get('owner', '')}/{args.get('repo', '')}/{path}",
                        )
                    elif name == "search_code":
                        log("  SEARCH", f"query: {args.get('query', '')[:60]}")
                    elif name == "load_skill":
                        log("  SKILL", f"Loading: {args.get('name', '')}")
                    elif name == "load_skill_resource":
                        log("  SKILL", f"Reading: {args.get('path', '')}")
                    else:
                        log("  TOOL", f"{name}")

                elif "text" in part:
                    text = part["text"]
                    if author == "codebase_analyzer":
                        # Show first line of analysis
                        first_line = text.strip().split("\n")[0][:120]
                        log("ANALYSIS", first_line)
                    elif author == "readme_writer":
                        # Capture the README output
                        final_readme = text
                        lines = text.strip().split("\n")
                        log("DRAFT", f"{len(lines)} lines generated")
                    elif author == "readme_critic":
                        if "exit_loop" not in text.lower() and len(text) > 30:
                            # Show first line of feedback
                            first_line = text.strip().split("\n")[0][:120]
                            log("FEEDBACK", first_line)

    return final_readme


def log(tag: str, message: str):
    """Print a formatted progress line to stderr."""
    # Color codes for terminal
    colors = {
        "STAGE 1": "\033[36m",  # cyan
        "STAGE 2": "\033[33m",  # yellow
        "REFINE": "\033[33m",  # yellow
        "REVIEW": "\033[35m",  # magenta
        "APPROVED": "\033[32m",  # green
        "ANALYSIS": "\033[36m",  # cyan
        "DRAFT": "\033[33m",  # yellow
        "FEEDBACK": "\033[31m",  # red
        "  FETCH": "\033[90m",  # gray
        "  SEARCH": "\033[90m",  # gray
        "  SKILL": "\033[90m",  # gray
        "  TOOL": "\033[90m",  # gray
    }
    reset = "\033[0m"
    color = colors.get(tag, "")
    print(f"{color}[{tag:>10}]{reset} {message}", file=sys.stderr)


def main():
    parser = argparse.ArgumentParser(description="Call the README harness")
    parser.add_argument("repo", help="GitHub repo in owner/repo format")
    parser.add_argument(
        "--host", default="http://localhost:8000", help="api_server URL"
    )
    parser.add_argument("--save", help="Save README to file instead of stdout")
    args = parser.parse_args()

    print(f"\033[1m{'=' * 60}\033[0m", file=sys.stderr)
    print(f"\033[1m README Harness — {args.repo}\033[0m", file=sys.stderr)
    print(f"\033[1m{'=' * 60}\033[0m", file=sys.stderr)

    try:
        session_id = create_session(args.host)
        log("SESSION", f"Created: {session_id[:8]}...")
    except urllib.error.URLError:
        print(
            "ERROR: Cannot connect to api_server. Is it running?",
            file=sys.stderr,
        )
        print("Start it with: adk api_server . --port 8000", file=sys.stderr)
        sys.exit(1)

    readme = run_harness(args.host, session_id, args.repo)

    print(f"\033[1m{'=' * 60}\033[0m", file=sys.stderr)

    if args.save:
        with open(args.save, "w") as f:
            f.write(readme)
        log("SAVED", f"README written to {args.save}")
    else:
        # Print README to stdout (progress went to stderr)
        print(readme)


if __name__ == "__main__":
    main()

```

這個 `cli_harness` 可以把它理解成：「讓 Claude Code / Gemini CLI 可以很方便地呼叫你前面那個 ADK README agent 的包裝層」。

現在架構大概是：

```text
cli_harness/
├── SKILL.md
│   └── 告訴 CLI agent：
│       什麼時候要用這個 skill、
│       要執行哪個 script、
│       怎麼啟動 ADK server
│
└── scripts/
    └── call_harness.py
        └── 真正透過 HTTP 呼叫 ADK api_server
```

資料流是：

```text
Claude Code / Gemini CLI
        │
        │ 讀 cli_harness/SKILL.md
        ▼
call_harness.py
        │
        │ HTTP
        ▼
ADK api_server :8000
        │
        ▼
root_agent
├── codebase_analyzer
└── refinement_loop
    ├── readme_writer
    └── readme_critic
```

`SKILL.md` 本身不是執行 README generation 的邏輯。它比較像 CLI agent 的「操作說明書」。例如使用者說：

```text
Improve the README for google/adk-samples
```

CLI agent 看了 skill 後，就知道應該執行：

```bash
python3 .../call_harness.py google/adk-samples
```

真正重要的 code 是 `call_harness.py`。

它主要做三件事情。

第一步，建立 ADK session：

```python
session_id = create_session(args.host)
```

實際 POST：

```text
/apps/app/users/cli_user/sessions
```

取得：

```text
session_id
```

第二步，把 request 丟給你的 `root_agent`：

```python
"new_message": {
    "role": "user",
    "parts": [
        {"text": f"Improve the README for {repo}"}
    ],
}
```

送到：

```text
POST /run_sse
```

所以 CLI script 本身不知道怎麼寫 README，它只是對 ADK 說：

```text
Improve the README for google/adk-samples
```

接下來全部交給 `agent.py`。

第三步，它讀取 SSE event stream，然後把 agent 執行狀況翻譯成人類容易看的 terminal log。

例如收到：

```text
author = codebase_analyzer
```

就顯示：

```text
[ STAGE 1 ] Analyzing codebase via GitHub MCP...
```

收到：

```text
author = readme_writer
```

就顯示：

```text
[ STAGE 2 ] Generating README...
```

收到：

```text
functionCall = exit_loop
```

就顯示：

```text
[ APPROVED ] Critic approved the README!
```

所以這段：

```python
for raw_line in resp:
```

其實是在「監聽 ADK 的執行事件」。

另一個值得注意的是：

```python
final_readme = text
```

只要 event 是：

```python
author == "readme_writer"
```

它就把 writer 最新輸出保存起來。

因此如果 loop 跑三輪：

```text
Writer pass 1 → final_readme = draft 1
Critic        → feedback

Writer pass 2 → final_readme = draft 2
Critic        → feedback

Writer pass 3 → final_readme = draft 3
Critic        → approved
```

最後 return 的就是最後一次 writer 的版本。

還有一個設計很好理解：

```text
stderr → progress
stdout → final README
```

所以：

```bash
python call_harness.py google/adk-samples
```

terminal progress 走 stderr，而真正 README 走 stdout。

這讓它可以很自然地做 shell pipeline，例如概念上：

```bash
python call_harness.py owner/repo > README.md
```

而不會把 `[STAGE 1]`、`[REVIEW]` 之類的 log 混進 README。

你現在可以先把它記成一句話：

```text
cli_harness/SKILL.md
    = 告訴 CLI agent 怎麼使用 harness

call_harness.py
    = ADK api_server 的 HTTP/SSE client

agent.py
    = 真正執行 README 分析、生成、review 的核心邏輯
```

所以 `cli_harness` 並不是另一套 agent system，它只是前面 `agent.py` 的 CLI adapter。

### DeepSeek Harness：核心架構與設計理念

## 1. 它是什麼

DeepSeek Harness（`dsh`）是一個開源的 **Agent Harness Runtime**。
DeepSeek 對它的核心定義是：

> **Agent = Model + Harness**

Model 負責推理；Harness 負責讓 Model 能使用工具、存取環境、維持 Session、執行 Agent Loop，並持續完成真實世界中的工作。

## 2. 最重要的設計：Everything is a Plugin

DeepSeek Harness 底層使用 **Cordis** plugin framework。
Model adapter、Tools、Skills、Session、Storage、Sandbox，甚至 **Agent Loop 本身都是 plugin**。

```text
                 DeepSeek Harness
                        │
                 ┌──── Cordis ────┐
                 │ Shared Context │
                 └───────┬────────┘
                         │
     ┌──────────┬────────┼──────────┬─────────┐
     ▼          ▼        ▼          ▼         ▼
   Model      Tools    Session    Sandbox    Agent Loop
   Plugin     Plugin    Plugin     Plugin      Plugin
```

因此 Harness 沒有一個必須修改的「特權核心」。
想增加或替換能力，主要方式是 mount / unmount plugin，而不是修改 Harness source code。

## 3. Cordis 的角色

Cordis 可以理解成 **Plugin Runtime / Dependency Container**。

Plugin 把能力註冊到共享的 `ctx`：

```text
ctx.llm        → LLM provider
ctx.tools      → Tool registry
ctx.sessions   → Session system
ctx.agents     → Agent management
ctx.agentLoop  → Agent execution loop
```

其他 plugin 依賴的是這些 service interface，而不是直接依賴某個 implementation。

## 4. Agent 如何真正執行

一次 Agent 工作可以分成：

```text
Session
  └── Turn
       └── Step
            ├── 組裝 Prompt + Tools
            ├── 呼叫 LLM
            ├── LLM 產生 Tool Calls
            ├── 執行 Tools
            └── Tool Result 回到下一個 Step
```

其中一個 **Step = 一次 model request + 該次產生的 tool calls**；
一個 Turn 則可以包含多個 Step。

## 5. Session 是整個系統的重要核心

DeepSeek Harness 使用 **event-sourced session**。

```text
Session = append-only SessionEvent log
```

User message、Model response、Tool call、Tool result、Turn/Step 狀態等，都被記錄成 event。
LLM 的 conversation history 不是另外保存，而是每次由這份 log 重新 derive 出來。

因此：

```text
Session Event Log
       ├── Resume
       ├── Replay
       ├── Fork
       ├── UI / Trajectory
       └── LLM Context reconstruction
```

都使用同一份 source of truth。

## 6. 為什麼這個架構重要

DeepSeek Harness 想解決的不是「怎麼寫一個 Agent」，而是：

**怎麼建立一個可以被自由組合、替換、觀察與擴充的 Agent Runtime。**

它的核心設計可以濃縮成：

```text
Model      = Intelligence
Plugin     = Capability
Cordis     = Composition / Lifecycle
Session    = Durable Truth
Agent Loop = Execution
Harness    = 上述能力組合後形成的 Runtime
```

目前 DeepSeek Harness 仍是 **Developer Preview**，官方明確表示核心 APIs 與 plugins 仍可能有 breaking changes。

這一段其實是在建立 Gemini Enterprise 的「管理平面」。順序是：

```text
Google Cloud 帳戶
→ 選定 Cloud Project
→ 綁定 Billing
→ 啟用必要 API
→ 建立 Gemini Enterprise App
→ 設定使用者身分
→ 建立並掛載資料連接器
→ 開放 Web App 給使用者
```

影片中的介面可能比目前版本稍舊。以下按現在的 Google Cloud 設定方式說明。

## 1. Google Cloud 帳戶、Project 與 Billing 是三件不同的事

「有 Google Cloud 帳戶」只代表你可以登入 Cloud Console。真正的資源都必須建立在一個 Google Cloud Project 裡。

一個 Project 會提供：

```text
Project ID
Project number
IAM 權限邊界
API 啟用狀態
Billing 關聯
Gemini Enterprise Apps
Data Stores
稽核紀錄
配額與費用
```

開始前至少需要：

1. 一個 Google Cloud Project。
2. 該 Project 已綁定 Billing Account。
3. 你的帳號具有啟用 API 和建立 Gemini Enterprise 資源的權限。

Google 目前的 quickstart 要求確認 Project 已啟用 billing，且操作者具備 Gemini Enterprise Admin。啟用 API 則需要 `serviceusage.services.enable` 權限，通常由 Project Owner 或 Service Usage Admin 提供。([Google Cloud Documentation][1])

常見 IAM 角色如下：

| 工作                   | 常見角色                    |
| -------------------- | ----------------------- |
| 建立 Project           | Project Creator         |
| 啟用 API               | Service Usage Admin     |
| 管理 Gemini Enterprise | Gemini Enterprise Admin |
| 建立或修改 Data Store     | Discovery Engine Editor |
| 一般使用 Web App         | Gemini Enterprise User  |

其中 Gemini Enterprise Admin 對應的角色 ID 通常是：

```text
roles/discoveryengine.agentspaceAdmin
```

這個名稱仍保留了舊產品名稱 `agentspace`，所以看到 Agentspace 不代表你進錯產品。

## 2. 搜尋「Gemini Enterprise」

在 Google Cloud Console 頂部搜尋框輸入：

```text
Gemini Enterprise
```

進入的是 Gemini Enterprise 管理頁面，而不是：

* 一般消費者使用的 gemini.google.com
* Google AI Studio
* Vertex AI Studio
* Gemini API 模型頁面
* Gemini Enterprise Agent Platform

這幾個產品名稱很接近，但用途不同。

影片使用的是：

```text
Gemini Enterprise App
```

它是提供企業員工使用的搜尋、聊天和 Agent Web App。

ADK 則更接近開發者層面的 Agent Runtime 和 Agent Platform。影片這一段尚未使用 ADK。

## 3. 「Start 30 days free trial」實際做了什麼

影片按下：

```text
Start 30 days free trial
```

這不是一般 Google Cloud 新帳戶的 300 美元試用金，而是 Gemini Enterprise 產品本身的試用或授權 onboarding。

概念上，這一步可能包含：

```text
Billing account
    ↓
Gemini Enterprise subscription / trial
    ↓
Project license allocation
    ↓
Users receive Gemini Enterprise licenses
```

需要區分兩種試用：

### Google Cloud Free Trial

這是新 Google Cloud 客戶常見的試用金，用於支付 Cloud 資源。

### Gemini Enterprise Free Trial

這是 Gemini Enterprise 的產品授權，通常涉及：

* Subscription
* License count
* 將 licenses 分配到特定 Project
* 將 licenses 指派給使用者

目前 Google 文件指出，Gemini Enterprise 的訂閱通常在 Billing Account 層級管理，但 license 是按 Project 分配；同一使用者若要在不同 Project 使用，可能需要在每個 Project 分別配置。([Google Cloud Documentation][2])

因此，影片中的 free trial 不只是「把 API 打開」，而是在進行產品授權與初始設定。

## 4. 啟用的 Gemini Enterprise API 是什麼

影片畫面說的是：

> activate the required API

Gemini Enterprise 核心使用的 API 是：

```text
Gemini Enterprise (Discovery Engine) API
```

其 service name 是：

```text
discoveryengine.googleapis.com
```

這也是為什麼許多 IAM role、REST resource 和舊文件仍包含 `discoveryengine`。

目前 Google 列出的必要 API 包括：

```text
Gemini Enterprise (Discovery Engine) API
Cloud Storage API
Identity and Access Management API
Vertex AI API
```

不同 onboarding 流程可能只在第一個畫面明確要求啟用 Discovery Engine API，其他 API 可能由精靈自動處理，或在後續功能使用時要求啟用。Google 的安全文件將 Vertex AI、Discovery Engine、Cloud Storage 和 IAM 都列為 Gemini Enterprise 所需 API；目前 quickstart 則明確要求啟用 Discovery Engine、Cloud Storage 和 IAM。([Google Cloud Documentation][3])

你也可以手動確認：

```text
Google Cloud Console
→ APIs & Services
→ Enabled APIs & services
```

搜尋：

```text
Gemini Enterprise
Discovery Engine
Vertex AI
Cloud Storage
Identity and Access Management
```

也可以使用 `gcloud`：

```bash
gcloud services enable \
  discoveryengine.googleapis.com \
  storage.googleapis.com \
  iam.googleapis.com \
  aiplatform.googleapis.com \
  --project=YOUR_PROJECT_ID
```

這段命令只是等價的 CLI 做法。使用 Cloud Console onboarding 時，通常不需要手動執行。

### 啟用 API 的意義

啟用 API 不等於建立了一個 Agent，也不等於產生 API key。

它只是允許這個 Project 建立和呼叫相關服務：

```text
API disabled
→ Project 無法建立 Gemini Enterprise 資源

API enabled
→ Project 可以建立 App、Data Store、Engine 等資源
```

Gemini Enterprise Web App 一般也不是靠你手動建立 API key 來運作。登入者身份、IAM、OAuth 和 Workspace identity 才是主要驗證機制。

## 5. 建立 Gemini Enterprise App

API 啟用後，進入：

```text
Gemini Enterprise
→ Apps
→ Create app
```

現在的建立頁面通常會要求：

```text
App name
App ID
Location
Organization/company display name
```

影片只提到 display name 和 region，是簡化過的說法。Google 目前的文件顯示，建立 App 時可以設定 App name、不可變更的 App ID、multi-region，以及對外顯示的組織名稱。([Google Cloud Documentation][4])

### App name

例如：

```text
Simon AI Workspace
```

這是介面中顯示的名稱，通常可以讓人閱讀。

### App ID

例如系統產生：

```text
simon-ai-workspace_174238
```

App ID 是 API 與資源路徑使用的穩定識別值。

建立後通常不能修改，所以正式環境最好使用有意義的命名，例如：

```text
corp-search-prod
corp-search-dev
ai-assistant-tw
```

不要把 production 和測試環境混在同一 App。

### App 在架構上的含義

App 不是 Agent 本身。它是所有終端使用者功能的容器：

```text
Gemini Enterprise App
├── Web App
├── Assistant
├── Agent Library
├── Agent Designer
├── Identity provider
├── Connected data stores
├── Feature toggles
├── Model settings
└── Access policy
```

因此，同一個 Project 可以有多個 App，例如：

```text
Project: enterprise-ai-prod
├── HR Assistant App
├── Engineering Knowledge App
└── Sales Intelligence App
```

每個 App 可以掛接不同資料來源和開放給不同使用者。

## 6. Region 或 Location 要怎麼選

影片說「choose the region」，但目前 Gemini Enterprise App 主要使用的是 multi-region：

```text
global
us
eu
```

Google 建議，沒有資料駐留需求時使用預設的 `global`；如果公司有法規、合約或資料駐留限制，才考慮 `us` 或 `eu`。([Google Cloud Documentation][4])

三者的概念是：

| Location | 適用情況                             |
| -------- | -------------------------------- |
| `global` | 教學、測試、沒有特定資料駐留要求                 |
| `us`     | 要求資料靜態儲存在美國 multi-region         |
| `eu`     | GDPR 或政策要求資料靜態儲存在歐盟 multi-region |

Location 是重要架構決策，因為它會影響：

* App resource 的位置
* Data Store 可使用的位置
* 部分 connector 的支援情況
* CMEK 設定
* Agent Gateway 或 Agent Registry 相容性
* 資料駐留與合規

通常 App 和所連接的 Data Store 應選擇相容的位置。

教學環境可以選：

```text
global
```

正式企業環境則應先詢問資安、法務或 Cloud 管理團隊。

## 7. 建立 App 後，為什麼要設定 Workforce Identity

App 建立完成後，下一步不是立刻接 Drive，而是設定「誰在使用」。

影片選：

```text
Google Identity
```

這表示 Gemini Enterprise Web App 使用者透過 Google Workspace 或 Cloud Identity 帳號登入。

這一步建立的邏輯是：

```text
Google Workspace user
    ↓ authentication
Gemini Enterprise user identity
    ↓ authorization
App access
    ↓ identity-aware connector query
Google Drive / Gmail / Calendar permissions
```

這裡最重要的是：使用者身份會向下傳遞到資料存取層。

例如：

```text
Alice 可讀 HR-policy.pdf
Bob 不可讀 HR-policy.pdf
```

即使 Alice 和 Bob 使用同一個 Gemini Enterprise App，搜尋結果也應不同：

```text
Alice query → 可以取得 HR-policy.pdf
Bob query   → 不應取得 HR-policy.pdf
```

這叫做 data source access control，而不只是 Web App 登入。

## 8. Google Identity 和 Workforce Identity Federation 的差異

目前主要有兩條路。

### Google Identity

適合：

* 公司已使用 Google Workspace
* 使用者有受管理的 Workspace 帳號
* 要連接 Gmail、Drive、Calendar
* 想快速完成教學

登入身份通常是：

```text
user@company.com
```

並由 Google Workspace 或 Cloud Identity 管理。

### Workforce Identity Federation

適合公司使用：

* Microsoft Entra ID
* Okta
* Ping Identity
* 自建 OIDC
* SAML identity provider

流程大致是：

```text
External IdP
→ Workforce Identity Pool
→ Workforce Identity Provider
→ Gemini Enterprise
```

Workforce Identity Federation 讓外部 IdP 的員工帳號可以登入 Google Cloud 服務，而不必替每個人建立傳統 Google Cloud 使用者身份。([Google Cloud][5])

影片為了簡化，選擇 Google Identity 是合理的。對 Google Workspace Connector 而言，這也是最直接的路徑。

目前介面大致是：

```text
Gemini Enterprise
→ Settings
→ Authentication
→ Add identity provider
```

然後針對 App 所在位置配置 identity provider。([Google Cloud Documentation][6])

## 9. Google Identity 並不等於個人 Gmail

這是一個很容易踩到的問題。

影片說 Google Identity，通常是指組織管理的：

```text
Google Workspace account
或
Cloud Identity account
```

不等於一般消費者帳號：

```text
someone@gmail.com
```

特別是 Google Drive Connector，官方目前要求 Cloud Console 登入帳號必須與要連接的 Google Drive instance 屬於相同的 Google Workspace customer；一般 `@gmail.com` 消費者帳號沒有 Workspace customer ID，因此不受支援。([Google Cloud Documentation][7])

所以做這個教學最好使用：

```text
simon@your-company-domain.com
```

而不是：

```text
simon123@gmail.com
```

即使個人 Gmail 可以建立 GCP Project，也不代表它能完成 Workspace enterprise connector 的完整流程。

## 10. Connected Data Source 實際建立的是什麼

身份設定完成後，影片進入：

```text
Connected data sources
→ Add data source
```

目前的介面更常稱為：

```text
Data stores
→ Create data store
```

Google 文件中的術語關係是：

```text
Data source
    ↓ accessed through
Connector
    ↓ represented inside Gemini Enterprise as
Data Store
    ↓ attached to
App
```

例如：

```text
Google Drive
→ Google Drive connector
→ drive-corporate-docs Data Store
→ Simon AI Workspace App
```

Connector 和 Data Store 有時在介面上會混用，但技術上不是完全相同：

* Connector 定義如何連到來源系統。
* Data Store 是 Gemini Enterprise 中可被 App 使用的資料資源。
* App 決定終端使用者可以搜尋哪些 Data Stores。

Google 將 connectors 定義為把 Google 或第三方資料來源帶入 Gemini Enterprise，並以獨立 Data Stores 管理。([Google Cloud Documentation][8])

## 11. 建立 Google Drive Data Store 的前置條件

對影片使用的 Google Drive connector，開始前應確認：

```text
1. 使用 Google Workspace 帳號
2. Cloud Console 帳號與 Drive Workspace 相符
3. 已設定 Identity Provider
4. 具有建立 Data Store 的 IAM 權限
5. App 與 Data Store 的 location 相容
6. Workspace 管理員沒有封鎖所需功能
```

官方特別要求：

* 使用 Cloud Console 的帳號必須與要連接的 Drive instance 相同。
* 必須是 Google Workspace，而不是一般消費者 Gmail。
* 為了正確執行 access control，應先設定 identity provider。([Google Cloud Documentation][7])

## 12. 建立 Drive Connector 時會填哪些東西

介面大致是：

```text
Gemini Enterprise
→ Data stores
→ Create data store
→ Google Drive
```

接下來通常需要設定：

### 要搜尋的資料範圍

可能包含：

```text
All supported Drive content
Specific folders
Shared drives
Include filters
Exclude filters
```

影片中說：

> Choose what you would like to share

這並不是把檔案權限改成公開，而是定義 Connector 的搜尋範圍。

例如：

```text
Connector scope:
  Include: Shared Drive / AI Study Group
  Exclude: Shared Drive / Finance Confidential
```

實際查詢時仍會再套用終端使用者原本的 Drive 權限。

也就是兩層過濾：

```text
Connector scope
AND
User's Drive permissions
```

某份文件要出現在結果中，通常必須同時滿足：

```text
文件在 Connector 搜尋範圍內
且
登入使用者本來就有權讀取
```

### Data connector name

例如：

```text
company-drive
engineering-shared-drive
workspace-documents-prod
```

這只是管理資源名稱，不是 Workspace 資料夾名稱。

命名最好帶環境：

```text
drive-rnd-dev
drive-rnd-prod
```

### Multi-region

通常應和 App 相同或相容：

```text
App: global
Data Store: global
```

或：

```text
App: eu
Data Store: eu
```

### Encryption

在 `us` 或 `eu` 等配置下，部分 Connector 會讓你選：

```text
Google-managed encryption key
Cloud KMS customer-managed key
```

教學通常選 Google-managed key。企業正式環境若有 CMEK 政策，才使用 Cloud KMS。

## 13. Drive 資料是複製進 Gemini Enterprise 嗎

影片的 Drive 連接通常是 federated retrieval，也就是查詢時從 Drive 取得資料，而不是先將所有 Drive 文件完整複製進 Gemini Enterprise index。

概念上：

```text
User asks question
→ Gemini Enterprise interprets query
→ Connector queries Google Drive
→ Drive returns authorized results
→ Gemini model synthesizes answer
```

官方 Drive 文件指出，這類連接可以直接從 Drive 擷取資訊，不需要你管理另一份資料儲存副本。([Google Cloud Documentation][7])

不過應注意：

* Metadata、設定和必要的服務資料仍可能由 Google Cloud 管理。
* 不同 connector 可能使用 federation 或 ingestion。
* 第三方 connector 未必與 Google Drive 採用完全相同的資料流。

因此不能把所有 Connected Data Sources 都理解成「即時 API 查詢」，也不能全部理解成「複製並建立向量索引」。要看 connector 類型。

## 14. 建立 Data Store 後還要掛到 App

建立 Data Store 和讓 App 使用 Data Store 是兩個動作。

可能的流程有兩種：

```text
先建立 App
→ 在 App 裡新增 Connected Data Source
→ 建立並自動掛載 Data Store
```

或：

```text
先建立 Data Store
→ 開啟 App
→ Connect data store
```

最終需要形成：

```text
App
└── Connected data stores
    └── company-drive
```

如果只建立 Data Store，卻沒有連到目前的 App，Web App 不一定能搜尋它。

可以在 App 的資料來源頁確認：

```text
Data store status: Active
Connected to app: Yes
Connector enabled in web app: Yes
```

## 15. 建議你實際操作時的完整檢查清單

```text
A. Cloud 基礎
[ ] 登入正確的 Google Workspace 帳號
[ ] 選到正確的 GCP Project
[ ] Billing 已啟用
[ ] Project ID 記錄下來

B. IAM
[ ] 有 Service Usage Admin 或等價權限
[ ] 有 Gemini Enterprise Admin
[ ] 有 Discovery Engine Editor
[ ] 測試使用者有 Gemini Enterprise User

C. APIs
[ ] discoveryengine.googleapis.com
[ ] storage.googleapis.com
[ ] iam.googleapis.com
[ ] aiplatform.googleapis.com

D. Gemini Enterprise App
[ ] App name 正確
[ ] App ID 命名合理
[ ] Location 已決定
[ ] 公司顯示名稱正確

E. Identity
[ ] Google Identity 已設定
[ ] 使用的是 Workspace 帳號而非個人 Gmail
[ ] 測試使用者能登入 Web App

F. Google Drive
[ ] 使用相同 Workspace customer
[ ] Drive Data Store 已建立
[ ] Data Store 狀態為 Active
[ ] 搜尋範圍正確
[ ] Data Store 已連接到 App
```

最重要的技術關係可以濃縮成：

```text
Google Cloud Project
├── Billing
├── Enabled APIs
├── IAM roles
└── Gemini Enterprise App
    ├── Identity provider
    ├── Web App
    └── Connected Data Stores
        └── Google Drive Connector
```

所以影片前一分鐘並不是單純「啟用一個 Gemini API」。它是在完成四個不同層級的工作：

```text
產品授權：Gemini Enterprise trial/subscription
服務啟用：Discovery Engine 等 Google Cloud APIs
應用資源：Gemini Enterprise App
企業整合：Identity Provider + Data Store Connector
```

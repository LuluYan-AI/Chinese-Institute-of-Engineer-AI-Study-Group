### RAG 的發展歷史：從資訊檢索到生成式 AI

## 一、RAG 出現之前：從「檢索後閱讀（Retrieve-and-Read）」開始

若要理解 **RAG（Retrieval-Augmented Generation，檢索增強生成）**，其實不應該從向量資料庫開始，而應該回到 **開放領域問答（Open-Domain Question Answering, Open-Domain QA）**。

與一般閱讀理解不同，Open-Domain QA 並不會直接提供一段包含答案的文章，而是要求系統先從龐大的知識庫中找到相關文件，再閱讀內容並回答問題。因此，整個流程自然形成了：

> **檢索（Retrieve） → 閱讀（Read） → 回答（Answer）**

這種架構後來被稱為 **Retrieve-and-Read**，也是 RAG 最重要的思想來源。

### DrQA：建立「先找、再讀」的典範（2017）

2017 年，Facebook AI Research 提出的 **DrQA** 是這個方向的重要代表。

DrQA 使用 **TF-IDF** 從 Wikipedia 搜尋可能相關的文章，再交由神經網路閱讀器（Neural Reader）抽取答案。

雖然當時的**檢索器（Retriever）**與**閱讀器（Reader）**仍是兩個彼此獨立的模組，但它已經建立了日後 RAG 的基本架構：

> **外部知識庫不只是背景資料，而是模型推論過程中的即時知識來源。**

### ORQA：讓檢索開始「學習」（2019）

2019 年，Google 提出的 **ORQA（Open-Retrieval Question Answering）** 更向前邁進一步。

ORQA 將證據文件視為**潛在變數（Latent Variable）**，只利用問題與答案進行訓練，同時學習：

* 如何檢索（Retriever）
* 如何回答（Reader）

這代表檢索不再只是固定的搜尋前處理，而是成為可以隨任務共同最佳化（Joint Learning）的模型元件。

### 歷史定位

從技術演進的角度來看，RAG 並不是「向量資料庫 + ChatGPT」的組合。

它真正的技術源頭，是三條研究路線逐漸匯流：

* **資訊檢索（Information Retrieval, IR）**
* **機器閱讀理解（Machine Reading Comprehension, MRC）**
* **生成式語言模型（Generative Language Models）**

直到 2020 年，研究者才首次將「外部知識檢索」與「大型生成模型」整合成一個統一且可訓練的架構，也就是今日所稱的 **RAG**。

---

# 二、2020 年：RAG 正式誕生

2020 年可說是**可檢索語言模型（Retrieval-Augmented Language Models）**的重要轉捩點。

短短幾個月內，連續出現數項關鍵成果：

* **REALM（Google）**：首次將文件檢索納入模型的預訓練流程。
* **DPR（Dense Passage Retrieval）**：證明密集檢索（Dense Retrieval）能大幅超越傳統 BM25。
* **RAG（Facebook AI Research）**：首次提出 Retrieval-Augmented Generation 的完整架構。

其中，Patrick Lewis 等人發表的論文 **Retrieval-Augmented Generation for Knowledge-Intensive NLP Tasks**，正式定義了 RAG。

## 當時要解決的四大問題

### 1. 參數化知識（Parametric Memory）的限制

大型語言模型會把知識壓縮到模型權重（Parameters）中，因此稱為**參數化記憶（Parametric Memory）**。

然而：

* 知道某件事情
* 能在需要時正確取出並使用

其實是兩件不同的事情。

模型可能記得大量知識，但未必能穩定回答特定問題。

---

### 2. 知識更新成本過高

模型完成訓練後，其知識便幾乎固定。

若世界發生變化，例如：

* 新聞事件
* 法規更新
* 公司內部文件修改

通常必須重新訓練（Retraining）或微調（Fine-tuning）模型，成本十分高昂，也難以及時更新。

---

### 3. 缺乏可信的知識來源

傳統 LLM 幾乎無法回答：

> 「你的答案是根據哪一份文件？」

因此缺乏：

* 引用（Citation）
* 證據（Evidence）
* 可追溯性（Provenance）

RAG 則透過外部文件提供可驗證的知識來源，大幅提升回答的可信度。

---

### 4. 知識密集型任務表現不足

在需要大量事實知識的任務，例如：

* 開放領域問答（Open-Domain QA）
* 事實查核（Fact Verification）
* 知識密集型生成（Knowledge-intensive Generation）

單純依靠模型參數，效果往往不如專門的檢索系統。

RAG 的核心目的，就是讓模型能在推論時即時存取外部知識，而非只依賴記憶。

---

## 原始 RAG 的核心架構

Lewis 等人提出將兩種不同形式的知識結合：

* **參數化記憶（Parametric Memory）**

  * 由預訓練語言模型（BART-large）負責

* **非參數化記憶（Non-parametric Memory）**

  * 由 Wikipedia 建立的向量索引負責

整體流程如下：

| 步驟 | 模組                      | 功能                      |
| -- | ----------------------- | ----------------------- |
| 1  | 問題編碼器（Question Encoder） | 將問題轉成查詢向量               |
| 2  | 密集檢索器（Dense Retriever）  | 從 Wikipedia 找出 Top-K 段落 |
| 3  | 生成器（Generator）          | 將問題與文件一起輸入 BART 產生答案    |
| 4  | 邊際化（Marginalization）    | 綜合多份文件的資訊，而非只依賴單一文件     |

數學上可表示為：

[
P(y|x)=\sum_z p_\eta(z|x)\cdot p_\theta(y|x,z)
]

其中：

* **x**：問題（Question）
* **z**：檢索出的文件（Retrieved Documents）
* **y**：生成答案（Generated Answer）

RAG 的核心思想並不是相信某一份文件，而是綜合多個知識來源，共同推估最佳答案。

### RAG-Sequence 與 RAG-Token

原始論文提出兩種不同版本：

* **RAG-Sequence**

  * 整段回答使用同一份文件。

* **RAG-Token**

  * 每個生成 Token 都可以參考不同文件。

其中 **RAG-Token** 更具彈性，但計算成本也更高。

### 與今日 RAG 最大的不同

值得注意的是，**2020 年的 RAG 並不是把搜尋結果貼進 Prompt**。

它是一個可以**端到端（End-to-End）聯合訓練（Joint Training）**的神經網路架構：

* 問題編碼器會持續學習。
* 生成器也會一起更新。
* 文件索引則保持固定。

因此：

* **2020 年的 RAG** 是一種**可微分（Differentiable）的神經模型架構**。
* **2023 年之後流行的 Naive RAG** 則是一種工程系統架構。

兩者共享「檢索增強生成」的核心理念，但設計思維與訓練方式並不相同。

# 三、2020–2022 年：檢索與模型的深度整合

RAG 提出之後，研究重點仍集中在：

> **如何讓檢索器（Retriever）與生成器（Generator）更緊密地共同學習？**

因此，這一階段通常被視為**檢索增強預訓練（Retrieval-Augmented Pretraining）**的探索時期。

代表性工作包括：

* **REALM**

  * 將文件檢索納入預訓練流程。

* **Fusion-in-Decoder（FiD）**

  * 分別編碼多份文件，再由 Decoder 整合所有證據。

* **RETRO**

  * 將外部知識直接整合進 Transformer 結構，利用大規模文本資料庫輔助生成。

此外，Hugging Face 很快便將 RAG 納入 **Transformers** 套件，使研究者能直接使用 DPR、BART 與 RAG，大幅降低學術重現的門檻。

這一階段也留下許多至今仍存在的重要問題，例如：

* 檢索是否足夠完整（Recall）？
* 更多文件是否一定更好？
* 如何平衡品質、延遲與成本？
* 檢索器與生成器是否應共同最佳化？

---

# 四、2022–2023 年：Prompt-native RAG 與工程化浪潮

2022 年底，ChatGPT 問世，改變了整個 RAG 的發展方向。

由於大型語言模型具備強大的**情境式學習（In-Context Learning）**能力，開發者發現：

> 不一定需要重新訓練模型，只要把檢索結果放進 Prompt，LLM 就能利用這些資訊回答問題。

因此，RAG 從原本的模型訓練方法，逐漸演變成一種工程架構。

典型流程如下：

> 文件切塊（Chunking）
> → 向量化（Embedding）
> → 建立索引（Indexing）
> → 相似度檢索（Retrieval）
> → Prompt 組裝（Prompt Assembly）
> → LLM 生成（Generation）

同一時期，也帶動整個 RAG 生態系快速成熟，包括：

* LangChain
* LlamaIndex
* Pinecone
* Milvus
* Qdrant

企業開始能快速將內部文件接入 LLM，而無須重新訓練模型。

### RAG 不等於向量資料庫

值得強調的是：

**RAG 是一種系統架構，而不是一種資料庫。**

向量資料庫只是其中一種檢索方式。

RAG 同樣可以結合：

* BM25
* SQL
* Knowledge Graph
* Web Search
* API
* 混合式搜尋（Hybrid Search）

### RAG 與 Fine-tuning 並非互斥

兩者解決的是不同問題：

**Fine-tuning**

* 改變模型能力
* 學習新風格
* 學習專業領域

**RAG**

* 提供即時知識
* 保持內容可更新
* 建立可追溯來源

成熟的 AI 系統通常會同時採用兩者。

# 五、2023–2024 年：Advanced RAG 與 Modular RAG

當企業大量導入 Naive RAG 後，新的問題開始浮現：

> 找得到文件，不代表找得到真正有用的知識。

因此，研究重點開始從「能不能檢索」，轉向「如何提升檢索品質」。

主要演進包括：

### 1. 更好的查詢（Pre-Retrieval）

包括：

* 查詢改寫（Query Rewriting）
* 問題拆解（Question Decomposition）
* 多查詢生成（Multi-query）
* HyDE（Hypothetical Document Embeddings）

目的都是讓搜尋查詢更接近真正存在於知識庫中的內容。

### 2. 更好的檢索（Retrieval）

開始大量採用：

* 混合搜尋（Hybrid Search）
* 重排序模型（Reranker）
* Metadata Filter
* 權限控制
* 時間限制

讓真正相關的文件排到前面。

### 3. 更好的上下文（Post-Retrieval）

包括：

* 去除重複內容
* 摘要壓縮
* Context Compression
* 引用對齊
* 拒答機制

目的都是提升 Prompt 品質，降低模型產生幻覺（Hallucination）的機率。

### 長上下文是否會取代 RAG？

2024 年 Gemini 1.5 推出百萬 Token 上下文後，業界一度認為：

> 是否只要把所有文件全部放進模型，就不需要 RAG？

後續研究顯示答案是否定的。

原因包括：

* Token 成本高
* 推論速度慢
* 模型容易受到資訊位置影響（Lost in the Middle）
* 大量無關內容反而降低回答品質

因此，目前較普遍的共識是：

> **長上下文（Long Context）負責提供更大的工作記憶；RAG 則負責挑選真正值得放入工作記憶的資訊。**

兩者屬於互補，而非互相取代。

# 六、2024–2026 年：從檢索系統走向上下文工程

近年的 RAG 已經不只是單純的「搜尋 + 回答」。

研究開始聚焦於如何管理模型的上下文（Context），因此逐漸發展成更廣義的 **上下文工程（Context Engineering）**。

## GraphRAG：加入知識圖譜

傳統 RAG 擅長回答：

> 哪一段文件包含答案？

但較難回答：

> 整份資料集主要討論什麼？

Microsoft Research 提出的 **GraphRAG** 先建立**知識圖譜（Knowledge Graph）**，再利用圖結構進行全域搜尋，使模型能理解跨文件之間的關聯。

---

## Self-RAG 與 CRAG：讓模型自行決定是否檢索

Self-RAG 與 CRAG 進一步加入：

* 是否需要檢索？
* 檢索品質是否足夠？
* 是否應重新搜尋？

模型開始具有一定程度的自我反思（Self-Reflection）能力，也成為後來 Agentic RAG 的重要基礎。

---

## Agentic RAG：檢索成為推理的一部分

在 **Agentic RAG** 中，檢索不再是固定流程，而是推理迴圈（Reasoning Loop）中的一個動作。

典型流程如下：

> 判斷資訊不足
> → 規劃搜尋策略
> → 呼叫工具
> → 評估證據
> → 修正答案
> → 必要時再次檢索

這使 RAG 能處理更複雜的多步推理（Multi-hop Reasoning）與研究型任務，但也增加了成本控制、可觀測性（Observability）與終止條件（Stopping Criteria）等新的工程挑戰。

---

## 多模態 RAG（Multimodal RAG）

現代文件不只有文字。

多模態 RAG 需要同時理解：

* 圖片
* 表格
* 圖表
* PDF 版面
* 音訊
* 影片

因此，研究重點已從 OCR 擴展到跨模態對齊（Cross-modal Alignment）、版面分析（Layout Analysis）與多來源證據融合（Evidence Fusion）。

---

## 時間感知 RAG（Temporal RAG）

真實世界的知識會隨時間改變。

例如：

* 「目前 CEO 是誰？」
* 「過去五年公司的策略如何演變？」

需要完全不同的檢索方式。

因此，新一代 RAG 開始重視：

* 文件版本（Versioning）
* 時間軸（Timeline）
* 歷時分析（Diachronic Analysis）
* 動態知識更新（Dynamic Knowledge）

---

## 2026 年的定位

截至目前為止，RAG 已不再只是「檢索 → 生成」的一條固定流程，而逐漸演變成一套完整的**外部知識管理與上下文工程（Context Engineering）**方法論。

目前包括 **Agentic RAG**、**GraphRAG**、**Multimodal RAG** 與 **Context Engineering** 等方向，仍是快速演進中的研究領域，尚未形成統一的標準架構。因此，較適合將它們視為 RAG 的最新發展趨勢，而非已經定型的最終型態。


### Vector Search 的不足

---

## 一、 核心痛點：為什麼單純的向量搜尋不夠？

向量搜尋（Dense Vector Search）擅長處理同義詞、改寫與語意相近的模糊查詢，但「語意相似不等於可用證據」。單純依賴向量搜尋會面臨 10 大結構性限制：

### 1. 單一向量是一種資訊瓶頸（Information Bottleneck）

* **技術原因與成因：**
傳統 Dense Vector Search 的運作機制，是將變長文字區塊（Chunk）通過 Transformer Encoder（如 BERT 或 Ada-002）壓縮成固定維度（如 $768$、$1530$ 或 $3072$ 維）的單一實數向量。這種「強迫壓縮」會造成極嚴重的資訊失真與維度瓶頸。當一個 Chunk 同時夾雜了多種實體（公司名稱、產品型號）、時間點、條件約束與邏輯判斷時，這些資訊必須在同一個向量空間中共享維度權重。這導致 Embedding 只能捕捉到整段文字的「平均主題語意」（Coarse-grained Semantic），極易遺失細粒度的關鍵資訊。
* **具體失敗情境與解決方案：**
例如「方案 A 支援功能 X，但不支援功能 Y」與「方案 A 支援功能 X，也支援功能 Y」，兩者在空間向量上的餘弦相似度（Cosine Similarity）可能高達 $0.95$ 以上，因為極度相似的主題與詞彙覆蓋了「否定詞」的維度影響，導致向量檢索無法穩定分辨否定、數值與版本差異。為了解決這個瓶頸，**Multi-vector Retrieval** 與 **Late Interaction（延遲交互，如 ColBERT）** 應運而生；ColBERT 不將文件壓縮成單一向量，而是保留每一個 Token 的高維向量表示，檢索時透過 Compute Max-Sim 算子 $\sum_i \max_j q_i^\top d_j$ 進行細粒度的 Token 級局部匹配，確保特定條件或專有名詞不會在壓縮過程中被稀釋。

---

### 2. 語意相似不代表答案相關（Semantic Similarity $\neq$ Answer Relevance）

* **技術原因與成因：**
Dense Retriever 的訓練目標（如 Contrastive Learning / InfoNCE Loss）是將「語意相近」的文本對在向量空間中拉近，把「語意無關」的文本拉遠。然而，「語意相似」與「具備回答能耐」在邏輯上是完全不同的維度。例如查詢「Tesla 2024 年的營業利益是多少？」，雙向編碼器（Bi-Encoder）演算法會尋找包含「Tesla」、「2024」、「營收」、「交車量」或「財務報告」等空間距離極近的文件 Chunk。這些文件雖然主題極度高度相關，但可能只是摘要或歷史趨勢，根本不包含「營業利益」這個具體數據。
* **具體失敗情境與解決方案：**
這種情況下，僅靠餘弦相似度排序會召回大量「看起來很像答案，但實際毫無解答能力」的干擾文件。工業界普遍的解法是採用**兩階段檢索架構（Two-stage Retrieval）**。第一階段以 Dense/Sparse 模型追求高召回率（High Recall，取回 Top 100~200）；第二階段引入 **Cross-encoder Reranker**（如 Cohere Rerank、BGE-Reranker）。Cross-encoder 不再分別對 Query 與 Document 進行獨立編碼，而是將兩者拼接成 `[CLS] Query [SEP] Document` 一起輸入 Transformer 進行全注意力機制（Full Self-Attention）運算，直接輸出 $s_{\text{rerank}}(q,d)=f([q;d])$ 的相關性得分，藉此精準識別文本是否真的具備推理與解答邏輯。

---

### 3. Vector Search 不擅長精確字串與稀有實體（Out-of-Vocabulary & Exact Match Failure）

* **技術原因與成因：**
Dense Embedding 模型嚴重依賴子詞分詞器（Subword Tokenizer，如 BPE 或 WordPiece）與預訓練語料庫的統計分佈。當遇到稀有專有名詞、高度結構化的字串（如系統錯誤碼 `ERR_CONNECTION_RESET`、CVE 漏洞編號 `CVE-2025-12345`、SKU 料號、API 函式名稱或特規版本號）時，Tokenizer 會將這些關鍵字切碎成無意義的碎片子詞（Subwords）。進入神經網路後，模型無法理解這些碎片的精確組合含義，將其映射至模糊的相近向量區間，導致 Exact Match（精確匹配）完全失效。
* **具體失敗情境與解決方案：**
例如搜尋「SKU-8A72F」，向量檢索可能會召回「SKU-8A72E」或「SKU-8A73F」，因為兩者在語意空間中極度接近，但對企業系統而言，錯一個字元就是完全錯誤的資料。因此，傳統的 **BM25 詞彙檢索（Lexical Retrieval）** 依然不可或缺，它基於倒排索引（Inverted Index），精準統計 TF-IDF 與 Token 的精確出現。現代檢索系統更延伸出 **Learned Sparse Retrieval（學習型稀疏檢索，如 SPLADE、BGE-M3 Sparse）**，它兼具倒排索引的高效精確比對能力，並透過神經網路在索引階段自動進行詞彙擴展（Query/Document Expansion，例如將 `Heart Attack` 自動擴展關聯權重至 `Myocardial Infarction`），實現精確字串與語意泛化的雙重優勢。

---

### 4. 向量相似度無法直接表達結構化限制（Inability to Express Hard Constraints）

* **技術原因與成因：**
向量空間（Vector Space）是一個連續且平滑的高維幾何空間，其設計本質是用距離（如 Cosine、Euclidean Distance）來度量軟性的語意相似度。然而，真實企業場景的查詢往往帶有嚴格的「硬性邏輯條件」（Hard Constraints），例如：「找出 **2025 年第二季**、**台灣地區**、**金額超過 100 萬** 且 **使用者權限可存取** 的合約」。這類涉及時間區間、地理位置、數值布林比較以及存取控制列表（ACL）的條件，無法以幾何距離來精準約束。
* **具體失敗情境與解決方案：**
如果直接將這段查詢送入向量資料庫，模型可能會因為「合約」與「100萬」的語意特徵強烈，而找回「2024 年第一季、美國地區、金額 500 萬」的無效文件。更嚴重的隱患是安全性：無權限查看的高密級文件可能僅因語意高度相似而被向量檢索取出，再嘗試交由 LLM 過濾，這會帶來嚴重的資安洩漏風險。正確的工程實務是**混合 Metadata Filtering（元資料過濾）**，在向量比對前（Pre-filtering）或比對過程中（In-filtering），透過 SQL/WHERE 條件或硬性 Filter 鎖定合法候選集，確保檢索結果同時滿足「結構化硬條件」與「語意軟相似」。

---

### 5. Vector Search 不會執行計算、聚合與 Join（No Support for Analytical & Relational Logic）

* **技術原因與成因：**
向量資料庫本質上是「高維空間距離 Nearest Neighbor 搜尋器」，它既沒有關係型資料庫（RDBMS）的代數計算引擎，也不具備邏輯判斷與資料結構聚合能力。當使用者提出需要「數值計算」或「跨表關聯」的問題（如：「過去三年哪一個產品線的平均毛利率成長最快？」），向量檢索只能將文本 Chunk 轉化成向量並尋找相似文字，完全無法執行 `SUM()`、`AVG()`、`GROUP BY`、`ORDER BY` 或多表 `JOIN` 操作。
* **具體失敗情境與解決方案：**
若將財報中的表格硬切成文字 Chunk 放入向量庫，檢索結果可能只是取出幾頁包含「毛利率」字眼的內文，LLM 拿到這些碎片文字後無法保證計算的正確性，甚至會產生幻覺。對於這類資料，產業解決方案是轉向 **Structured Retrieval（結構化檢索，如 Text-to-SQL）** 或 **TableRAG** 技術。系統會先透過 Schema Matching 與 Query Understanding，將使用者的自然語言問題轉譯成強類型的 SQL 查詢指令，直接對資料倉庫（Data Warehouse）或結構化資料庫進行精確的 OLAP 聚合運算，僅在需要解釋數據背景時才輔以文字向量檢索。

---

### 6. Flat Vector Search 難以處理跨文件關係（Lack of Multi-hop Knowledge Graph Structure）

* **技術原因與成因：**
傳統的 Vector RAG 假設所有的 Chunk 在資料庫中都是彼此孤立、平坦（Flat）分佈的個體。然而，真實世界中的知識高度依賴複雜的網路關係（網絡拓撲結構）。當面對多跳推理（Multi-hop Reasoning）問題時（如：「哪一項技術透過被收購的公司，最終整合進了產品 X？」），答案需要串聯『產品 X $\rightarrow$ 收購事件 $\rightarrow$ 被收購公司 $\rightarrow$ 核心技術』等跨越數個不同文件與段落的邏輯鏈條。
* **具體失敗情境與解決方案：**
在平坦的向量搜尋中，沒有任何單一 Chunk 會同時包含問題中的所有實體關鍵字與完整解答路徑。因此，單純計算 Cosine 相似度往往只會召回單一端點的資料，無法拼接完整證據鏈。這催生了 **GraphRAG** 與 **Graph Retrieval（如 HippoRAG、LightRAG）** 的爆發。GraphRAG 在索引階段透過 LLM 提取實體（Entities）與關係（Relations）構建知識圖譜（Knowledge Graph），檢索時透過圖演算法（如 Personalized PageRank、Path Traversal）進行鄰居擴展與子圖抽取，甚至利用圖社群（Communities）產生階層式摘要，從根本上解決了跨文件的關係檢索與全域主題分析問題。

---

### 7. Vector Search 沒有原生時間模型（Absence of Temporal Reasoning & Dynamics）

* **技術原因與成因：**
Embedding 模型在編碼時是「時間盲目（Time-blind）」的，它無法區分文件寫入時間（System Time）、事件發生時間（Event Time）以及事實的有效期限（Valid Time）。語意空間無法表達「隨時間演變的動態事實」。例如查詢「Apple 的 CEO 是誰？」與「2010 年 Apple 的 CEO 是誰？」，兩者在向量空間中的表示極度相似，Embedding 模型無法從字面上感知到時間約束的嚴格排他性。
* **具體失敗情境與解決方案：**
若系統缺乏時間維度處理，向量搜尋非常容易檢索出過期、被廢棄的舊版本政策或歷史資料，將其與最新資料混淆並輸入給 LLM。在處理解析「收購案前後產品策略變化」或「條文修訂歷程」等動態問題時更會徹底失控。為此，**Temporal RAG（時間感知檢索）** 必須引進明確的時間工程：包含查詢端的時間表達式解析（將「去年」、「最近三季」轉為 UTC 時間區間）、Temporal Metadata Filter（時間過濾器）、版本感知索引（Version-aware Index）以及融合語意分與時間衰減分（Temporal Decay Score）的重排機制，確保系統能精準區分歷史事實與當前最新狀態。

---

### 8. Chunking 會切斷上下文（Context Fragmentation & Sub-chunk Isolation）

* **技術原因與成因：**
為了符合 Embedding 模型的輸入長度限制（如 512 Tokens），RAG 預處理管道必須將長篇大論切割成細小的 Chunk。這種硬性的切塊（Chunking）破壞了文件的整體語意結構，導致許多 Chunk 失去了關鍵的上下文脈絡——例如標題 hierarchy、代名詞指涉（如內文中的「該公司」、「其營收」）、表格欄位定義或跨段落的因果邏輯。
* **具體失敗情境與解決方案：**
當一個 Chunk 只寫著「其第二季營收成長了 18%」，獨立看這句話時，Embedding 模型完全無法知道「其」指的是哪家公司。檢索時，即使這段話包含了解答關鍵，也無法與查詢正確匹配。為修復脈絡斷裂問題，業界採用了多種補救機制：
1. **Contextual Retrieval（Anthropic）：** 在索引前調用 LLM 為每個 Chunk 前綴一段全域背景說明（如「本文出自 Acme 公司 2024Q2 財報...」）再建立 Embedding；
2. **Late Chunking：** 先讓長文本模型閱讀整篇文件生成 Token 級 Embedding，最後才進行 Pooling 切塊，讓 Chunk 自然包含全域語意；
3. **Parent-Child Retrieval / RAPTOR：** 採用小 Chunk 作為精準匹配標的，檢索命中後回溯取出較大的父級段落（Parent Block）或摘要樹（Summary Tree）作為上下文供給 LLM 輸入。



---

### 9. 單次查詢無法涵蓋複雜問題（Single-shot Query Imbalance & Incompleteness）

* **技術原因與成因：**
單次向量檢索（Single-shot Dense Search）預設了一個過於理想的假設：使用者的輸入問句已經完美包含了所有檢索所需的關鍵資訊與明確語意。然而在真實場景中，使用者提出的原始問題往往高度口語化、充滿模糊代名詞、受限於對話歷史上下文、或者隱含了錯誤的前提假設。更嚴重的狀況是，複雜問題往往包含多個子任務，無法靠一次資料庫查詢解決。
* **具體失敗情境與解決方案：**
例如面對「開發產品 X 的公司後來被誰收購？該收購方的 CEO 當時是誰？」這類多跳問題，直接用整句去向量庫搜尋只會得到一堆不相關的模糊結果。這催生了**查詢端檢索工程（Query-side Retrieval Engineering）** 的發展：
* **Query Rewriting（改寫）：** 利用輕量 LLM 將口語化與帶有歷史上下文的問題改寫為標準搜尋句；
* **Multi-query / Query Decomposition（拆解）：** 將複合問題拆解成多個獨立的子查詢（Sub-queries）並平行檢索；
* **Iterative / Agentic Retrieval（如 IRCoT、FLARE）：** 讓模型的推理與搜尋交替進行（Interleaved Search & Reasoning），根據第一步獲得的中間結果動態生成下一步的檢索查詢，逐步補齊完整證據鏈。



---

### 10. Approximate Nearest Neighbor (ANN) 本身也有召回取捨（Trade-offs in Vector Indexing）

* **技術原因與成因：**
在面對海量（數百萬至數億維度）的向量資料庫時，進行逐一比對的暴力搜尋（Exhaustive Flat Search）在延遲上是完全不可行的。因此，現代向量資料庫絕大多數採用**近似最近鄰演算法（Approximate Nearest Neighbor, ANN）**，如基於圖的 HNSW（Hierarchical Navigable Small World）或基於聚類的 IVF（Inverted File Index）。ANN 的核心邏輯是以犧牲一定的檢索準確度（Recall）為代價，換取毫秒級的低查詢延遲。
* **具體失敗情境與解決方案：**
這意味著向量檢索的失敗可能源於系統工程層面，而非模型不理解語意：即正確的文件 Embedding 雖然被成功生成了，但 ANN 演算法在圖繪製或索引巡檢時「迷路」，根本沒有走訪到那個節點。此外，當在 HNSW 等圖索引上加上**高選擇性（High Selectivity）的 Metadata 過濾器**時，會大量割裂圖節點之間的連接邊，導致 ANN 的圖搜尋路徑斷裂、召回率（Recall）急劇崩潰。工程上必須精細調校 `efSearch`、`M` 參數、導入向量量化（Quantization, PQ/SQ），或在強過濾條件下改用 Filter-aware Search / Exhaustive KNN，這是在系統量級放大後極易觸發的結構性硬傷。

## 二、 現代工業級 RAG 的多維檢索機制

為了補足上述缺口，產業界整合了多種技術方案：

* **檢索模型進化：** Hybrid Search（BM25 + Dense via RRF 融合）、Learned Sparse（SPLADE）、Multi-vector（ColBERT）、Vision-Language 多模態檢索（ColPali）。
* **結構與動態處理：** Hierarchical（RAPTOR）、Structured/SQL（Bedrock/TableRAG）、GraphRAG、Temporal Filter、Federated Multi-source（跨 CRM/SQL/Wiki 聯邦檢索）。
* **主動與動態檢索（Agentic RAG）：** 如 Self-RAG、Corrective RAG（CRAG）、Adaptive-RAG，由 LLM 根據問題複雜度與結果品質，動態決定是否檢索、重新改寫或觸發網絡搜尋。

### Agent Retrieval (formerly Vector Search 2.0)

# Google Cloud Agent Retrieval：從「向量索引」走向完整的 AI Retrieval 資料層

你提供的頁面原本介紹 **Vector Search 2.0**，目前 Google 已將產品重新命名為 **Agent Retrieval**。這個名稱變化相當重要，因為它反映 Google 不再把這項產品定位成單純的向量最近鄰搜尋服務，而是面向 RAG 與 AI Agent 的完整資料儲存、搜尋、融合與重排系統。該文件最近更新於 **2026 年 7 月 23 日**。([Google Cloud Documentation][1])

## 一、這個產品到底是什麼？

Google 對 Agent Retrieval 的官方描述是：

> 一個自動調校、完全受管理，並且為 AI 應用原生設計的搜尋引擎。

傳統的 Google Cloud Vector Search 主要是一套高效能的 **Approximate Nearest Neighbor, ANN（近似最近鄰）索引服務**。使用者的核心工作是：

1. 準備向量
2. 建立索引
3. 部署索引
4. 管理副本與機器資源
5. 查詢最近鄰

Agent Retrieval 則把資源中心從 **Index（索引）**改成：

> **Collection（集合）中的 Data Objects（資料物件）**

這意味著開發者不再只是把一批向量交給 ANN 引擎，而是把完整的 JSON 資料物件，包括文字、數值、布林值、Metadata 與多個向量欄位，存入一個統一資料層，再對這些物件執行不同形式的 Retrieval。([Google Cloud Documentation][1])

簡單比較如下：

| 面向        | 傳統 Vector Search      | Agent Retrieval            |
| --------- | --------------------- | -------------------------- |
| 核心資源      | Index                 | Collection                 |
| 主要資料      | 向量與有限 Metadata        | 完整 JSON Data Object        |
| 儲存方式      | 常需搭配外部資料庫             | 內建可擴充儲存引擎                  |
| 搜尋能力      | 主要是 ANN Vector Search | Vector、全文、語意、Filter、Hybrid |
| Embedding | 通常自行產生                | 可自動產生或自行提供                 |
| 排名流程      | 向量距離排名                | 多路搜尋、RRF 融合、語意重排           |
| 維運方式      | 需要管理索引與部署             | 自動調校、受管理                   |
| 主要用途      | 大規模向量相似度搜尋            | RAG、Agent Memory、語意資料存取    |

Google 的核心設計目標是把原本分散的兩層：

```text
文件／Metadata 資料庫
          +
向量索引服務
```

整合為：

```text
Agent Retrieval
├── 原始資料
├── Metadata
├── Dense／Sparse Vector
├── Filter
├── Full-text Search
├── Semantic Search
├── ANN Index
└── Reranking
```

官方明確指出，Agent Retrieval 本身提供可複寫、可擴充的儲存引擎，因此可作為 AI 應用的統一資料來源，減少另外維護輔助資料庫的需求。([Google Cloud Documentation][1])

# 二、核心資料模型

## 1. Collection：類似資料庫中的 Table

**Collection（集合）**是一組相關 JSON 物件的容器，概念上類似關聯式資料庫中的資料表。單一資料庫中可以建立多個 Collection。([Google Cloud Documentation][1])

例如，可以分別建立：

```text
product_documents
customer_tickets
agent_memories
research_papers
company_policies
```

每個 Collection 都可以有自己的：

* Schema
* 向量欄位
* Filter 欄位
* Index
* Embedding 設定

## 2. Data Object：真正被搜尋的資料單位

**Data Object（資料物件）**是 Collection 中的一筆 JSON 資料。

例如：

```json
{
  "title": "2026 年差旅政策",
  "department": "Human Resources",
  "effective_year": 2026,
  "language": "zh-TW",
  "content": "員工申請海外差旅時……",
  "access_level": "internal"
}
```

它不只包含文件文字，也可以保存：

* 結構化欄位
* 數值
* 分類
* 時間
* 權限資訊
* 一個或多個向量
* 系統 Metadata

因此，它更接近「可搜尋的業務資料物件」，而不只是傳統向量資料庫中的：

```text
id + vector + text
```

## 3. Collection Schema

Collection Schema 用來定義 Data Object 的結構與限制，而且可以採用嚴格或較寬鬆的 Schema 驗證。([Google Cloud Documentation][1])

一個 Collection 可以同時定義：

```text
data_schema
├── title: string
├── year: number
├── department: string
└── content: string

vector_schema
├── content_embedding: dense vector
├── title_embedding: dense vector
└── lexical_embedding: sparse vector
```

## 4. Collection Index

Collection Index 用來加速特定向量欄位上的 ANN Search。一個 Collection 可以建立多個 Index，例如：

* 內容向量索引
* 標題向量索引
* 圖片向量索引
* 產品特徵向量索引

沒有建立 Index 時，搜尋會進行 Brute-force Scan，也就是逐筆比較；建立 ANN Index 後，才能在大量資料上提供低延遲搜尋。([Google Cloud Documentation][2])

Google 預設可以把字串、數值與布林欄位一同推入 ANN Index，以支援搜尋時的 Inline Filtering。也可以透過 `filter_fields` 與 `store_fields` 控制哪些欄位需要參與過濾，哪些欄位只作為 Payload 儲存，以平衡成本與效能。([Google Cloud Documentation][2])

# 三、它不是只有 Vector Search

這是該產品與傳統向量索引服務最大的不同。

Agent Retrieval 的 Search API 支援：

1. **Vector Search（向量搜尋）**
2. **Full-text Search（全文搜尋）**
3. **Semantic Search（語意搜尋）**
4. **Filtering（條件過濾）**
5. **Hybrid Search（混合搜尋）**
6. **Semantic Reranking（語意重排）**

官方文件明確表示，多個不同類型的搜尋可以同時執行，再合併成 Hybrid Search。([Google Cloud Documentation][3])

這正好呼應前面討論的核心觀念：

> Google 自己也沒有把 Retrieval 定義成單一 Vector Search，而是將 Vector、Text、Semantic、Filter、Fusion 與 Reranking 組合成完整流程。

## 1. Vector Search

Vector Search 可以搜尋 Dense Vector 或 Sparse Vector。

使用者可以自行提供 Query Vector，指定要搜尋哪個向量欄位：

```json
{
  "vector_search": {
    "search_field": "content_embedding",
    "vector": {
      "values": [0.12, 0.38, 0.74]
    },
    "top_k": 10
  }
}
```

若 Collection 有多個向量欄位，便可針對不同資訊面向建立獨立表示，例如：

```text
title_embedding
content_embedding
product_feature_embedding
image_embedding
```

多個 `vector_search` 查詢也可以合併，官方文件指出，在未另行設定時，多個向量搜尋會以相同權重合併。([Google Cloud Documentation][3])

## 2. Full-text Search

全文搜尋適合處理：

* 專有名詞
* 產品代碼
* 錯誤訊息
* 函式名稱
* 法條
* 精確片語
* 人名與組織名稱

也就是 Dense Vector Search 不一定能穩定保留的字面資訊。

這說明 Agent Retrieval 的設計邏輯並不是「向量搜尋取代傳統搜尋」，而是讓兩者共存。

## 3. Semantic Search

Semantic Search 讓使用者直接提交自然語言查詢，由系統根據 Collection 的 Embedding 設定產生查詢向量，而不必在 Client 端先呼叫 Embedding API。

概念上是：

```text
使用者查詢
    ↓
Agent Retrieval 自動產生 Query Embedding
    ↓
搜尋指定的 Embedding 欄位
```

Vector Search 比較接近 Bring Your Own Query Vector；Semantic Search 則把 Query Embedding 的產生也納入受管理流程。

## 4. Metadata Filtering

搜尋請求可以同時加入結構化 Filter。例如只搜尋驚悚片：

```json
{
  "vector_search": {
    "search_field": "plot_embedding",
    "vector": {
      "values": [0.42, 0.57, 0.71]
    },
    "filter": {
      "genre": {
        "$eq": "Thriller"
      }
    },
    "top_k": 5
  }
}
```

官方範例直接將 `genre = Thriller` 條件與向量搜尋放在同一次請求中。([Google Cloud Documentation][3])

在企業 RAG 中，可以用於：

```json
{
  "$and": [
    {"department": {"$eq": "Finance"}},
    {"year": {"$gte": 2025}},
    {"region": {"$eq": "Taiwan"}},
    {"access_level": {"$in": ["public", "internal"]}}
  ]
}
```

這種條件不應該由 Embedding 模糊推測，而應以硬性資料約束執行。

# 四、Hybrid Search 與 RRF

Agent Retrieval 可以透過 `batch_search_data_objects` 平行執行多個搜尋，包括：

* Vector Search
* Text Search
* Semantic Search

然後使用 **Reciprocal Rank Fusion, RRF（倒數排名融合）**合併結果。([Google Cloud Documentation][3])

其概念流程是：

```text
Semantic Search Top 50 ─┐
                         │
Text Search Top 50 ──────┼→ RRF → 統一候選排名
                         │
Vector Search Top 50 ────┘
```

RRF 不必直接比較不同搜尋引擎的原始分數，而是根據每份文件在不同結果清單中的排名進行融合：

[
\operatorname{RRF}(d)
=====================

\sum_i \frac{w_i}{k+\operatorname{rank}_i(d)}
]

這很重要，因為：

* BM25 分數
* Cosine Similarity
* Dot Product
* Semantic Search Score

通常沒有共同尺度，不能直接相加。

RRF 使用排名資訊，因此可以穩定地融合異質 Retriever。

例如查詢：

> 「支援 Gemini API 的 Python 3.13 SDK authentication 錯誤」

可以同時執行：

```text
Text Search
→ 精確找出 Python 3.13、Gemini API、錯誤碼

Semantic Search
→ 找到認證失敗、憑證設定等概念相關文件

Metadata Filter
→ 僅保留 Python、2025 年以後、官方文件

RRF
→ 合併所有候選
```

這比單純依靠 Query Embedding 更接近工業級 Retrieval。

# 五、Server-side Semantic Reranking

Agent Retrieval 也內建 **VertexRanker（語意重排器）**。

完整流程為：

```text
多個 Retriever
    ↓
RRF Candidate Fusion
    ↓
VertexRanker
    ↓
最終 Top-K
```

VertexRanker 會將自然語言 Query 與合併後的候選文件交給專用語意 Ranking Model 重新評分。它不是只看向量距離，而是進一步判斷候選文件是否真正符合使用者問題中的細微需求。([Google Cloud Documentation][4])

官方範例大致如下：

```json
{
  "searches": [
    {
      "semantic_search": {
        "search_text": "running shoes",
        "search_field": "embedding",
        "top_k": 50
      }
    },
    {
      "text_search": {
        "search_text": "running shoes",
        "data_field_names": ["title"],
        "top_k": 50
      }
    }
  ],
  "combine": {
    "top_k": 10,
    "ranker": {
      "rrf": {
        "weights": [1.0, 1.0]
      },
      "vertex_ranker": {
        "model": "semantic-ranker-fast@latest",
        "top_n": 50,
        "text_record_spec": {
          "query": "running shoes",
          "title_template": "{title}",
          "content_template": "{body.text}"
        }
      }
    }
  }
}
```

其意義是：

* Retriever 負責找出高召回候選
* RRF 負責融合不同 Retriever
* VertexRanker 負責提高最終 Precision

官方目前支援的模型是 `semantic-ranker-fast@latest`；每次可以送入最多 1,000 筆候選進行重排。重排完全在 Server 端完成，不需要 Client 端額外發送另一個 API Request。([Google Cloud Documentation][4])

此外，VertexRanker 被設計成 Best-effort。若語意重排暫時失敗，搜尋 RPC 通常仍會返回 RRF 融合結果，並在 Metadata 中加入 Warning，而不是讓整次 Retrieval 失敗。([Google Cloud Documentation][4])

這是相當實務的可用性設計：

```text
最佳狀況：RRF + Semantic Reranking
降級模式：只使用 RRF
```

# 六、自動產生 Embedding

Agent Retrieval 可以在建立 Collection 時設定 Embedding Model 與文字模板。當新的 Data Object 沒有自行提供向量時，系統會根據指定欄位自動建立 Embedding。([Google Cloud Documentation][5])

例如：

```python
"genre_embedding": {
    "dense_vector": {
        "dimensions": 768,
        "vertex_embedding_config": {
            "model_id": "text-embedding-004",
            "text_template": "Movie: {title} Genre: {genre} Year: {year}",
            "task_type": "RETRIEVAL_DOCUMENT",
        },
    }
}
```

這裡值得注意的不是 API 語法，而是 `text_template` 的概念。

系統不只是對單一 `genre` 欄位產生 Embedding，而是先把多個欄位組合成：

```text
Movie: {title} Genre: {genre} Year: {year}
```

再產生向量。

這其實是一種簡化的 **Contextualized Indexing（上下文化索引）**：

> 在建立向量前，先把有助於 Retrieval 的 Metadata 與語意背景組合進輸入。

Agent Retrieval 也允許 **Bring Your Own Embeddings, BYOE（自行提供 Embedding）**，所以開發者可以選擇：

```text
Google 自動產生 Embedding
或
使用外部／自行微調的 Embedding Model
```

官方列出的模型包含文字、多語言、程式碼與新一代多模態 Embedding。文件指出 `gemini-embedding-2` 可把文字、圖片、影片與 PDF 映射到統一的向量空間，最大輸出維度為 3,072，最大序列長度為 8,192 tokens。([Google Cloud Documentation][5])

# 七、Dense Vector 與 Sparse Vector 都能成為欄位

Agent Retrieval 的 Vector Schema 不限於 Dense Vector，也能儲存 Sparse Vector。官方搜尋範例顯示，Sparse Vector 由：

```json
{
  "sparse": {
    "indices": [1, 5, 10],
    "values": [0.1, 0.5, 0.21]
  }
}
```

組成。([Google Cloud Documentation][3])

這代表產品底層並不把「向量」等同於 Dense Embedding。

它可以容納：

* Dense Semantic Embedding
* Sparse Lexical Embedding
* 多個不同用途的 Embedding Field
* 原始文字欄位
* 結構化 Filter 欄位

例如可以建立：

```text
content_dense_embedding
content_sparse_embedding
title_dense_embedding
image_embedding
```

然後依任務執行不同的搜尋或融合。

# 八、它與 Google Cloud Vector Search 1.0 的本質差異

不能把 Agent Retrieval 單純理解為「Vector Search 1.0 的新版 API」。

它實際上代表抽象層級的改變。

## Vector Search 1.0 的抽象

```text
你有一批向量
→ 建立 ANN Index
→ 部署 Index Endpoint
→ 查詢最近鄰
```

它是一套高效能的 **Index-as-a-Service**。

## Agent Retrieval 的抽象

```text
你有一批 AI 應用資料
→ 定義 Collection 與 Schema
→ 儲存完整 Data Object
→ 自動或自行產生 Embedding
→ 建立必要的 ANN Index
→ 執行多種 Search
→ Filter
→ RRF Fusion
→ Semantic Reranking
```

它更接近：

> **AI-native Retrieval Database（AI 原生檢索資料庫）**

而不只是：

> **ANN Index Service（近似最近鄰索引服務）**

Google 仍保留原有 Vector Search 的高效能與大規模擴充能力，但在其上增加資料儲存、Schema、Embedding、全文搜尋、Hybrid Search 與 Reranking。([Google Cloud Documentation][1])

# 九、它在 RAG 架構中的位置

Agent Retrieval 可以負責 RAG Pipeline 的中段：

```text
原始文件
    ↓
Parsing／Chunking
    ↓
建立 Data Objects
    ↓
產生 Embedding
    ↓
Agent Retrieval
├── Text Search
├── Semantic Search
├── Vector Search
├── Metadata Filter
├── RRF
└── VertexRanker
    ↓
Top-K Evidence
    ↓
Prompt／Context Construction
    ↓
Gemini 或其他 LLM 生成答案
```

它主要解決的是：

* 資料與向量分開管理
* Embedding Pipeline 維運
* 多 Retriever 並行
* Filter 與 Vector Search 整合
* 搜尋結果融合
* Semantic Reranking
* ANN Index 部署與擴充

它並沒有涵蓋所有進階 RAG 問題。例如文件解析、最佳 Chunking、GraphRAG、Temporal Reasoning、Query Decomposition、Agent 的多步搜尋策略，以及答案 Grounding 評估，仍需要其他元件處理。

因此，較準確的定位是：

> Agent Retrieval 是一個完整度很高的 Retrieval Backend，但不是完整的 RAG 或 Agent 系統。

# 十、文件中值得注意的產品限制

## 1. Index 仍然是效能關鍵

雖然開發者以 Collection 為主要資源，但大規模 ANN Search 仍需要建立 Collection Index。沒有 Index 時會使用 Brute-force Scan，適合測試或小資料集，不適合生產規模。([Google Cloud Documentation][2])

## 2. 自動 Embedding 受模型配額限制

自動產生 Embedding 會使用客戶專案中的 Embedding Model Quota，主要受到每分鐘輸入 Token 與每分鐘預測請求限制。大量匯入資料前需要先確認 Quota。([Google Cloud Documentation][5])

## 3. Reranking 有額外使用成本

VertexRanker 有免費額度，超過後會依 Ranking API 的計價模式收費。官方文件目前列出每個專案每 30 天有 80,000 Ranking Units 的免費額度。([Google Cloud Documentation][4])

## 4. 地區支援有限

文件目前列出的支援區域為：

* `asia-east1`
* `asia-northeast1`
* `asia-southeast1`
* `europe-north1`
* `europe-west2`
* `europe-west4`
* `us-central1`
* `us-east4`
* `us-west1` ([Google Cloud Documentation][1])

對台灣使用者而言，`asia-east1` 是文件列出的支援區域之一。

## 5. API 仍出現 `v1beta`

官方程式範例目前使用：

```text
vectorsearch.googleapis.com/v1beta/...
```

因此，在實際導入前仍應確認 API 穩定性、功能生命週期、區域支援與升級政策。([Google Cloud Documentation][3])

# 十一、這份文件在 Retrieval 技術演進上的意義

這個產品可以視為 2023 年以來 Retrieval 工業化趨勢的一個具體案例。

早期 Naive RAG 常採用：

```text
Chunk
→ Embedding
→ Vector Database
→ Top-K
→ LLM
```

Agent Retrieval 則內建成：

它仍然以向量檢索作為核心能力之一，但不再把向量相似度當作 Retrieval 的全部。Google 將產品從 **Vector Search 2.0**改名為 **Agent Retrieval**，正好反映這項架構上的轉變。([Google Cloud Documentation][1])

### 使用 Vector Search 2.0 與 ADK 建立 Agentic RAG

這篇文章示範如何使用 **Vertex AI Vector Search 2.0** 與 **Agent Development Kit（ADK）**，建立一個倫敦短期住宿搜尋代理。

系統使用約 2,000 筆 Airbnb 房源資料。使用者可以直接提出包含語意需求與結構化條件的問題，例如：

> 幫我找 Hackney 地區、每晚低於 200 英鎊、可以立即預訂，而且具有藝術工作室氛圍的房源。

系統會把問題拆成兩部分：

```text
語意查詢：
creative artist workspace

結構化條件：
neighborhood = Hackney
price < 200
instant_bookable = true
```

整體分工如下：

```text
使用者問題
    ↓
ADK Agent
├── 理解使用者意圖
├── 抽取價格、地區與訂房條件
└── 呼叫搜尋工具
    ↓
Vector Search 2.0
├── Semantic Search
├── Text Search
├── Metadata Filter
└── RRF 結果融合
    ↓
ADK Agent 整理搜尋結果
    ↓
自然語言回答
```

Vector Search 2.0 負責 **Retrieval Layer（檢索層）**，ADK 則負責 **Reasoning Layer（推理與協調層）**。

## 一、建立 Collection 與自動 Embedding

首先建立一個 Collection，用來保存房源資料。

Collection 包含兩種 Schema：

* **Data Schema**：定義房源的原始資料欄位。
* **Vector Schema**：定義向量欄位以及 Embedding 的產生方式。

```python
from google.cloud import vectorsearch_v1beta

PROJECT_ID = "your-project-id"
LOCATION = "us-central1"
COLLECTION_ID = "london-travel-agent-demo"

admin_client = vectorsearch_v1beta.VectorSearchServiceClient()

collection_config = {
    "data_schema": {
        "type": "object",
        "properties": {
            "name": {"type": "string"},
            "price": {"type": "number"},
            "neighborhood": {"type": "string"},
            "listing_url": {"type": "string"},
            "instant_bookable": {"type": "string"},
            "description": {"type": "string"},
            "neighborhood_overview": {"type": "string"},
        },
    },
    "vector_schema": {
        "description_embedding": {
            "dense_vector": {
                "dimensions": 768,
                "vertex_embedding_config": {
                    "model_id": "gemini-embedding-001",
                    "text_template": (
                        "Description: {description}. "
                        "Neighborhood: {neighborhood_overview}."
                    ),
                    "task_type": "RETRIEVAL_DOCUMENT",
                },
            }
        }
    },
}

request = vectorsearch_v1beta.CreateCollectionRequest(
    parent=f"projects/{PROJECT_ID}/locations/{LOCATION}",
    collection_id=COLLECTION_ID,
    collection=collection_config,
)

operation = admin_client.create_collection(request=request)
operation.result()
```

### 程式說明

`data_schema` 定義每筆房源具有哪些欄位，例如價格、地區、描述與房源網址。

`vector_schema` 則建立名為 `description_embedding` 的 768 維向量欄位。

```python
"text_template": (
    "Description: {description}. "
    "Neighborhood: {neighborhood_overview}."
)
```

這個模板會把房源描述與地區介紹組合後再產生 Embedding。如此一來，向量不只包含房屋本身的資訊，也包含周邊環境與地區特色。

```python
"task_type": "RETRIEVAL_DOCUMENT"
```

這表示目前產生的是文件端的檢索向量。Vector Search 2.0 會自行呼叫 `gemini-embedding-001`，開發者不需要另外建立 Embedding API Pipeline。

## 二、匯入房源資料

建立 Collection 後，將每筆 Airbnb 房源轉換成 Data Object。

```python
from google.cloud import vectorsearch_v1beta

data_client = vectorsearch_v1beta.DataObjectServiceClient()

collection_path = (
    f"projects/{PROJECT_ID}/locations/{LOCATION}"
    f"/collections/{COLLECTION_ID}"
)

batch_requests = []

for _, row in df_demo.iterrows():
    batch_requests.append(
        {
            "data_object_id": str(row["id"]),
            "data_object": {
                "data": {
                    "name": row["name"],
                    "price": float(row["price"]),
                    "neighborhood": row["neighbourhood_cleansed"],
                    "instant_bookable": row["instant_bookable"],
                    "listing_url": row["listing_url"],
                    "description": row["description"],
                    "neighborhood_overview": row[
                        "neighborhood_overview"
                    ],
                },
                "vectors": {},
            },
        }
    )
```

程式刻意把 `vectors` 設為空物件：

```python
"vectors": {}
```

這不是表示資料沒有向量，而是要求 Vector Search 2.0 根據 Collection 中的 `vertex_embedding_config` 自動產生向量。

接著分批寫入資料：

```python
BATCH_SIZE = 100

for start in range(0, len(batch_requests), BATCH_SIZE):
    current_batch = batch_requests[start : start + BATCH_SIZE]

    request = vectorsearch_v1beta.BatchCreateDataObjectsRequest(
        parent=collection_path,
        requests=current_batch,
    )

    data_client.batch_create_data_objects(request=request)
```

每筆資料寫入後，系統會在後端完成以下工作：

```text
讀取 description 與 neighborhood_overview
    ↓
套用 text_template
    ↓
呼叫 gemini-embedding-001
    ↓
使用 RETRIEVAL_DOCUMENT 模式產生向量
    ↓
將原始資料與向量一同保存
```

因此，應用程式不必自行維護原始資料與向量之間的同步。

## 三、建立 Hybrid Search 工具

下一步建立 `find_rentals` 函式。ADK 會把這個 Python 函式註冊成 Agent 可以呼叫的工具。

```python
import json
from typing import Any

from google.cloud import vectorsearch_v1beta


search_client = vectorsearch_v1beta.DataObjectSearchServiceClient()


def find_rentals(
    query: str,
    filter: str = "",
) -> list[dict[str, Any]]:
    """搜尋符合語意描述與結構化條件的倫敦房源。

    Args:
        query:
            房源風格或需求的自然語言描述，例如
            "creative artist workspace"。
        filter:
            JSON 格式的 Metadata Filter，例如
            '{"price": {"$lt": 200}}'。

    Returns:
        符合條件的房源清單。
    """
    filter_dict = json.loads(filter) if filter.strip() else None

    semantic_search = vectorsearch_v1beta.SemanticSearch(
        search_text=query,
        search_field="description_embedding",
        filter=filter_dict,
        task_type="QUESTION_ANSWERING",
        top_k=10,
        output_fields=vectorsearch_v1beta.OutputFields(
            data_fields=[
                "name",
                "price",
                "neighborhood",
                "listing_url",
            ]
        ),
    )

    text_search = vectorsearch_v1beta.TextSearch(
        search_text=query,
        data_field_names=[
            "name",
            "description",
            "neighborhood_overview",
        ],
        filter=filter_dict,
        top_k=10,
        output_fields=vectorsearch_v1beta.OutputFields(
            data_fields=[
                "name",
                "price",
                "neighborhood",
                "listing_url",
            ]
        ),
    )

    request = vectorsearch_v1beta.BatchSearchDataObjectsRequest(
        parent=collection_path,
        searches=[
            vectorsearch_v1beta.Search(
                semantic_search=semantic_search
            ),
            vectorsearch_v1beta.Search(
                text_search=text_search
            ),
        ],
        combine=(
            vectorsearch_v1beta
            .BatchSearchDataObjectsRequest
            .CombineResultsOptions(
                ranker=vectorsearch_v1beta.Ranker(
                    rrf=vectorsearch_v1beta.ReciprocalRankFusion(
                        weights=[0.6, 0.4]
                    )
                )
            )
        ),
    )

    response = search_client.batch_search_data_objects(
        request=request
    )

    results = []

    for item in response.results[0].results:
        data = item.data_object.data

        results.append(
            {
                "name": data.get("name"),
                "price": data.get("price"),
                "neighborhood": data.get("neighborhood"),
                "url": data.get("listing_url"),
            }
        )

    return results
```

### 1. 解析 Metadata Filter

```python
filter_dict = json.loads(filter) if filter.strip() else None
```

ADK 傳入的 `filter` 是 JSON 字串。工具先將它轉換成 Python Dictionary，再交給 Vector Search 2.0。

例如：

```json
{
  "price": {
    "$lt": 200
  }
}
```

或同時指定多個條件：

```json
{
  "$and": [
    {
      "neighborhood": {
        "$eq": "Hackney"
      }
    },
    {
      "price": {
        "$lt": 200
      }
    },
    {
      "instant_bookable": {
        "$eq": "t"
      }
    }
  ]
}
```

這些 Filter 會限制搜尋候選集合，確保最後結果符合價格、地區與預訂方式等硬性條件。

### 2. Semantic Search

```python
semantic_search = vectorsearch_v1beta.SemanticSearch(
    search_text=query,
    search_field="description_embedding",
    filter=filter_dict,
    task_type="QUESTION_ANSWERING",
    top_k=10,
)
```

Semantic Search 會把自然語言查詢轉成向量，再搜尋 `description_embedding`。

例如：

```text
creative artist workspace
```

即使資料中沒有完全相同的字詞，仍可能找到：

* Artist loft
* Creative studio
* Converted warehouse
* Inspiring workspace

搜尋時使用：

```python
task_type="QUESTION_ANSWERING"
```

而建立文件向量時使用：

```python
task_type="RETRIEVAL_DOCUMENT"
```

這是一種非對稱的 Embedding 設計。文件與問題使用不同任務角色，使問題可以匹配語句形式不同、但內容能回答需求的文件。

### 3. Text Search

```python
text_search = vectorsearch_v1beta.TextSearch(
    search_text=query,
    data_field_names=[
        "name",
        "description",
        "neighborhood_overview",
    ],
    filter=filter_dict,
    top_k=10,
)
```

Text Search 負責字面與關鍵詞匹配。

Semantic Search 擅長理解意義，但可能忽略某些精確詞彙。Text Search 則能補足：

* 房源名稱
* 特定地區名稱
* 明確關鍵詞
* 描述中的原始字詞

因此，文章沒有只使用 Vector Search，而是同時執行語意搜尋與文字搜尋。

### 4. 使用 RRF 融合結果

```python
rrf=vectorsearch_v1beta.ReciprocalRankFusion(
    weights=[0.6, 0.4]
)
```

兩種搜尋會各自產生一份排名：

```text
Semantic Search Ranking
Text Search Ranking
```

接著以 **Reciprocal Rank Fusion, RRF（倒數排名融合）**合併。

權重設定為：

```text
Semantic Search：0.6
Text Search：0.4
```

代表系統稍微偏重語意相關性，但仍保留文字匹配的影響。

若同一房源在兩種搜尋中都排名靠前，它在融合後通常也會獲得較高排名。

## 七、文章的核心重點

這篇文章展示的 Agentic RAG，核心不是讓 Agent 自主執行很長的多步研究，而是讓 LLM 位於 Retrieval 之前，先完成搜尋規劃：

```text
理解問題
```

其中：

* Vector Search 2.0 負責資料儲存、自動 Embedding、語意搜尋、文字搜尋、Metadata Filtering 與 RRF。
* ADK 負責理解使用者需求、建立搜尋參數、呼叫工具與生成回答。

因此，文章所說的 Agentic RAG，主要體現在 Agent 不再把使用者的原始問題直接送進搜尋引擎，而是先判斷：

> 哪些內容應該用語意搜尋，哪些內容應該轉換成精確的結構化條件。


### 使用 Task Type Embedding 改善 RAG 的檢索品質

## 一、文章要解決的問題

RAG（Retrieval-Augmented Generation，檢索增強生成）通常會將使用者問題與知識庫文件轉換成向量，再利用 Cosine Similarity（餘弦相似度）找出最接近的文件。

典型流程是：

```text
使用者問題
    ↓
Query Embedding
    ↓
與文件 Embedding 比較
    ↓
取得最相似的 Top-K 文件
```

問題在於，一般 Embedding 模型通常衡量的是：

> 兩段文字表達的語意是否相似。

但 RAG 真正需要的是：

> 某份文件是否能回答這個問題。

這兩個目標並不相同。Google 將這個現象稱為：

> **Question is not the answer，問題本身並不是答案。**

例如：

```text
問題：
Why is the sky blue?

句子 A：
The sky is blue today.

句子 B：
The scattering of sunlight causes the blue color.
```

句子 A 與問題都包含「sky」和「blue」，因此字面與語意非常接近，但它沒有回答問題。

句子 B 與問題的措辭不同，卻真正包含答案。

若 Query 與 Document 都使用一般的 `SEMANTIC_SIMILARITY` Embedding，向量搜尋可能將句子 A 排在句子 B 前面。這是因為模型最佳化的是「文字相似度」，而不是「問答關係」。([Google Cloud][1])

---

## 二、為什麼問題與答案會落在不同的向量分布？

Embedding 模型會把文字映射到向量空間：

$$E(x)\in\mathbb{R}^d$$


一般語意搜尋會使用 Cosine Similarity：

$$\operatorname{sim}(q,d) = \frac{E(q)\cdot E(d)}{\lVert E(q)\rVert\lVert E(d)\rVert}$$

其中：

* (q) 是查詢文字
* (d) 是候選文件
* ($E(\cdot)$) 是 Embedding 模型

如果模型只接受「語意相似度」訓練，下面兩句通常會非常接近：

```text
What is my son's birthday?
My son's birthday is December 13.
```

但以下問答組合未必會接近：

```text
What is the best birthday present for my son?
Lego set.
```

第二組並不是改寫或同義句，而是具有「問題與合理答案」的關係。

因此，傳統 Similarity Embedding 面臨的問題是：

```text
Query Embedding Space
與
Answer／Document Embedding Space
沒有針對檢索關係進行對齊
```

---

## 三、過去如何解決這個問題？

### 1. Dual Encoder

傳統做法是訓練 **Dual Encoder（雙編碼器）**，也稱為 **Two-Tower Model（雙塔模型）**。

它使用兩個不同的編碼路徑：

```text
Query
  ↓
Query Encoder
  ↓
Query Vector
                相似度最大化
Document        ←────────→
  ↓
Document Encoder
  ↓
Document Vector
```

訓練資料通常包含：

* Query
* Relevant Document
* Irrelevant Document

模型會學習：

```text
相關的 Query–Document Pair
→ 向量距離拉近

不相關的 Query–Document Pair
→ 向量距離拉遠
```

這種方法效果很好，但通常需要：

* 大量 Query–Document 配對資料
* Negative Sampling
* 模型訓練
* ML Engineering
* 評估與部署流程

對一般 RAG 開發團隊而言，成本相當高。([Google Cloud][1])

### 2. LLM Query Transformation

另一種方法是先使用 LLM 改寫使用者問題，例如：

* Query Expansion
* Hypothetical Document Embeddings（HyDE）
* 產生可能答案
* 產生多個搜尋查詢

例如，HyDE 會先把：

```text
Why is the sky blue?
```

轉換成一段假設性答案：

```text
The sky appears blue because sunlight is scattered
by molecules in Earth's atmosphere.
```

再用這段答案搜尋真正的文件。

這可以縮小問題與答案之間的語意差距，但每次搜尋前都必須呼叫 LLM，因此會增加：

* 推論延遲
* Token 成本
* 系統複雜度
* 查詢結果的不確定性

相較之下，Embedding API 的成本與延遲通常明顯較低。([Google Cloud][1])

---

## 四、Google 的解法：Task Type Embedding

Google Vertex AI 的 Embedding 模型允許開發者在產生向量時指定 **Task Type（任務類型）**。

Task Type 告訴模型：

> 這段文字將被用在什麼任務中？

例如，同一句問題可以依照不同目的產生不同的向量：

```python
TextEmbeddingInput(
    text="Why is the sky blue?",
    task_type="SEMANTIC_SIMILARITY",
)
```

表示：

> 找出與這句話語意相似的文字。

而：

```python
TextEmbeddingInput(
    text="Why is the sky blue?",
    task_type="QUESTION_ANSWERING",
)
```

表示：

> 找出能回答這個問題的文件。

兩者的文字相同，但模型產生向量時採用不同的最佳化目標。

Google 的說明指出，這些模型透過 LLM Distillation（大型語言模型蒸餾）與 Dual Encoder 訓練，使較小且快速的 Embedding 模型能學到 Query 與 Document 之間的任務關係。([Google Cloud][1])

---

## 五、RAG 中的非對稱 Task Type

在問答式 RAG 中，Query 與 Document 不應使用完全相同的 Task Type。

推薦組合是：

| 資料角色  | Task Type            |
| ----- | -------------------- |
| 使用者問題 | `QUESTION_ANSWERING` |
| 知識庫文件 | `RETRIEVAL_DOCUMENT` |

流程如下：

```text
使用者問題
    ↓
QUESTION_ANSWERING
    ↓
Query Embedding

知識庫文件
    ↓
RETRIEVAL_DOCUMENT
    ↓
Document Embedding
```

兩種向量雖然由不同 Task Type 產生，仍位於相容的向量空間中。

模型經過訓練，使「能回答 Query 的文件」更接近 Query，而不只是讓措辭相似的文字彼此靠近。Google 官方也建議，在問答型 RAG 中對問題使用 `QUESTION_ANSWERING`，對文件使用 `RETRIEVAL_DOCUMENT`。([Google Cloud][1])

# 十一、批次產生問題與答案向量

```python
def generate_embeddings(
    df: pd.DataFrame,
    question_task_type: str,
    answer_task_type: str,
) -> pd.DataFrame:
    """替問題與答案產生 Embedding。"""

    q_embeddings: list[list[float]] = []
    a_embeddings: list[list[float]] = []

    for i in tqdm(range(0, len(df), 5)):
        questions = df.loc[
            i : i + 4,
            "question",
        ].tolist()

        answers: list[Any] = [
            row[0]
            for row in df.loc[
                i : i + 4,
                "answer",
            ]
        ]

        q_embeddings.extend(
            get_embeddings(
                questions,
                question_task_type,
            )
        )

        a_embeddings.extend(
            get_embeddings(
                answers,
                answer_task_type,
            )
        )

    df["q_emb"] = q_embeddings
    df["a_emb"] = a_embeddings

    return df
```

這個函式每次處理五筆資料，以減少 API Request 數量。

最後建立兩個欄位：

```text
q_emb：問題向量
a_emb：答案向量
```

# 十二、建立所有問題與答案的相似度矩陣

```python
def calc_cosine_similarities(
    df: pd.DataFrame,
) -> list[list[float]]:
    """計算所有問題與所有答案的相似度。"""

    q_embeddings = df["q_emb"].tolist()
    a_embeddings = df["a_emb"].tolist()

    return cosine_similarity(
        q_embeddings,
        a_embeddings,
    )
```

若資料集有 1,000 個問題與 1,000 個答案，輸出會是一個：

[
1000\times1000
]

的相似度矩陣。

第 (i) 列代表：

```text
第 i 個問題
與
全部 1,000 個答案
的相似度
```

理想狀況下，第 (i) 個問題對應的正確答案應排在前面。

---

# 十三、取得 Top-100 候選答案

```python
def get_top100_similar_answers(
    answer_similarities: list[float],
) -> list[int]:
    """取得相似度最高的 100 個答案索引。"""

    indexed_scores = list(
        enumerate(answer_similarities)
    )

    ranked_scores = sorted(
        indexed_scores,
        key=lambda item: item[1],
        reverse=True,
    )

    ranked_indices = [
        index
        for index, _ in ranked_scores
    ]

    return ranked_indices[:100]
```

這個函式會：

1. 為每個答案附加索引。
2. 依照相似度由高到低排序。
3. 取前 100 筆結果。

# 十四、計算正確答案的排名

```python
def add_query_ranks(
    df: pd.DataFrame,
) -> pd.DataFrame:
    """記錄每個問題的正確答案排名。"""

    similarities = calc_cosine_similarities(df)

    all_query_ranks: list[list[int]] = []

    for i in tqdm(range(len(similarities))):
        ground_truth = df.loc[i, "answer"][0]

        candidate_indices = (
            get_top100_similar_answers(
                similarities[i]
            )
        )

        query_ranks: list[int] = []

        for index in candidate_indices:
            candidate_answer = (
                df.loc[index, "answer"][0]
            )

            query_ranks.append(
                int(candidate_answer == ground_truth)
            )

        all_query_ranks.append(query_ranks)

    df["query_ranks"] = all_query_ranks

    return df
```

例如：

```python
[0, 0, 1, 0, 0]
```

表示正確答案排在第三名。

# 十五、MRR 是什麼？

Notebook 使用 **Mean Reciprocal Rank, MRR（平均倒數排名）**評估檢索效果。

單一問題的 Reciprocal Rank 為：

[
RR=\frac{1}{r}
]

其中 (r) 是正確答案的名次。

例如：

|      正確答案名次 | Reciprocal Rank |
| ----------: | --------------: |
|           1 |               1 |
|           2 |             0.5 |
|           3 |           0.333 |
|          10 |             0.1 |
| Top-100 未出現 |               0 |

整體 MRR 為：

[
MRR
===

\frac{1}{N}
\sum_{i=1}^{N}
\frac{1}{r_i}
]

Notebook 的實作如下：

```python
def calc_mrr(
    df: pd.DataFrame,
) -> float:
    """計算所有問題的 Mean Reciprocal Rank。"""

    all_query_ranks = (
        df["query_ranks"].tolist()
    )

    reciprocal_ranks: list[float] = []

    for query_ranks in all_query_ranks:
        for rank, relevance in enumerate(
            query_ranks,
            start=1,
        ):
            if relevance == 1:
                reciprocal_ranks.append(
                    1 / rank
                )
                break
        else:
            reciprocal_ranks.append(0)

    return (
        sum(reciprocal_ranks)
        / len(reciprocal_ranks)
    )
```

MRR 越高，代表正確答案平均越靠近搜尋結果前方。

# 十九、Task Type 對 RAG 架構的影響

沒有 Task Type 時：

```text
Question
    ↓
Generic Embedding
    ↓
尋找措辭或主題相似的文件
```

使用 Task Type 後：

```text
Question
    ↓
QUESTION_ANSWERING Embedding
    ↓
尋找具有回答關係的文件
```

因此，Task Type 改善的是 Retrieval 的表示層：

```text
原始問題
→ 如何被映射成 Query Vector

原始文件
→ 如何被映射成 Document Vector
```

它不會改變 Vector Database 的 ANN 演算法，也不會取代：

* Hybrid Search
* BM25
* Metadata Filter
* Reranker
* Query Rewriting
* Graph Retrieval

但它能讓最基礎的 Dense Retrieval 更符合實際任務。

---

# 二十、Task Type 的限制

Notebook 最後特別指出，Task Type 模型主要使用一般性語料訓練。

因此，在下列情況中，通用 Task Type 未必足夠：

* 高度專有的公司內部文件
* 特殊產業術語
* 私有產品名稱
* 通用模型從未接觸的知識領域
* 特殊的 Query–Document 關係

例如，企業內部的：

```text
「Bluebird 流程」
「Project Atlas」
「R7 結算例外」
```

對通用模型可能沒有明確語意。

這時可以考慮 **Embedding Tuning（Embedding 微調）**，使用企業自己的 Query–Document Pair 讓模型學習領域特有的檢索關係。

因此，可以把兩者理解為：

```text
Task Type
→ 使用 Google 預先學習的通用任務關係

Embedding Tuning
→ 學習企業或領域特有的檢索關係
```

# 二十一、整體結論

這篇文章與 Notebook 的核心不是宣稱某個 Embedding 模型具有更高的抽象語言能力，而是指出：

> Embedding 的品質不能脫離任務定義來討論。

RAG 需要的通常不是單純的 Semantic Similarity，而是 Query–Document Relevance。

### Agentic RAG 研究整理與核心結論

如果不被 Google 把 Vector Search 2.0 改名成為 Agent Retrieval 的影響，Agentic RAG 是一段時間以來對 RAG 技術做改進的一個方向。

Agentic RAG 將傳統固定的 RAG 流水線轉化為「動態決策過程」，讓 LLM 具備主動判斷是否搜尋、查何來源、改寫問題與自我修正的能力。

---

### 一、 代表性研究列表（按技術發展演進）

#### 1. 2024 年：適應性檢索與自我修正（動態流程成形）

* **Self-RAG：** 透過訓練模型產生 Reflection Tokens（反思標記），實現按需檢索（不檢索/單次/多次）與生成品質的自我評估。
* **Corrective RAG (CRAG)：** 引入 Retrieval Evaluator 評估文件品質（Correct/Incorrect/Ambiguous），遇到不可靠資訊時觸發過濾、重新搜尋或 Web Search 後備機制。
* **Adaptive-RAG：** 透過小型分類模型判斷問題複雜度，彈性路由至「不檢索（簡單）」、「單次檢索（一般）」或「多步迭代檢索（複雜）」。
* **HippoRAG：** 結合 LLM、知識圖譜與 Personalized PageRank 建立圖結構記憶，大幅提升多跳（Multi-hop）問答效能並降低延遲。

#### 2. 2025 年：多智能體分工、Critic 診斷與強化學習（Policy Learning）

* **Critic-guided RAG (RAG-Critic)：** 建立階層式錯誤分類（如檢索不足、文件不相關、問題理解錯誤等），精準指引系統選擇對應的修正路徑。
* **MAIN-RAG：** 採用多 Agent 共識機制評分與過濾文件，旨在降低高召回率帶來的噪音與無關文件干擾。
* **MA-RAG：** 將任務依認知角色拆解（Planner、Step Definer、Extractor、QA Agent），僅在需要時動態啟動角色。
* **Graph-enhanced Agent (GeAR)：** 將圖擴充（Graph Expansion）作為 Agent 的動態動作，解決跨文件與複雜關係鏈的搜尋。
* **Search-R1 & DecEx-RAG：** 引進強化學習（RL）讓模型學習搜尋策略（何時搜/搜什麼/何時停止）。DecEx-RAG 更進一步引入過程監督（Process Supervision）以改善獎勵稀疏問題。
* **Search Wisely：** 揭露 Agentic RAG 的過度搜尋（Over-search）與不足搜尋（Under-search）問題，指出有 27.7% 的搜尋步驟屬無用消耗。

#### 3. 2026 年：階層式工具、聯合訓練與策略最佳化

* **A-RAG：** 將檢索解構為階層式工具（`keyword_search`、`semantic_search`、`chunk_read`），賦予 Agent 控制搜尋粒度與閱讀深度的權限。
* **Agentic-R：** 破除傳統局部相關性訓練，同時考慮 Passage 對最終答案的全域貢獻，進行 Agent 與 Retriever 的雙向迭代最佳化。
* **Reflective RAG：** 採「SFT 學習反思訊號 + RL 最佳化策略」的兩階段訓練，提升多跳泛化能力與訓練穩定度。
* **SPARKLE：** 採用獨立的 Proxy Model 控制檢索策略，以 Plug-and-play 形式運作，無須微調主生成模型。
* **MASS-RAG & SEMA-RAG：** 多智能體證據整合架構。SEMA-RAG 將其專用化於醫療推理（Interpreter、Explorer、Arbiter），顯著提升診斷語意解讀與證據裁決的準確率。

---

### 二、 研究發掘的核心結論

1. **優勢極度集中於複雜任務：** Agentic RAG 對於多跳問題、跨資料源整合、模糊查詢與需要驗證的任務效果顯著；對簡單 FAQ 或單一事實查詢，傳統 Hybrid RAG 在成本與延遲上更具優勢。
2. **檢索（Retrieval）依然是底層最大瓶頸：** Agent 推理能力再強，也無法對未召回的資料做出正確解答。企業 Deep Search 實驗顯示，無法找齊關鍵證據是導致 Agent 推理失敗的主要原因。
3. **自我評估（Self-Evaluation）有效但不可盲目信任：** 模型的評估器（Grader）本身也會犯錯（如誤判相關性或拒答率低下）。**嚴禁讓同一模型同時擔當規劃者、執行者與最終裁判**，應採用獨立模型、確定性規則或人工覆核。
4. **多智能體（Multi-Agent）並非預設最佳解：** 僅在「任務認知可明確拆分、角色需使用不同工具/資料」時才有價值；若所有 Agent 共享相同的 Prompt 與上下文，只會徒增延遲與 Token 成本。
5. **搜尋預算與停止條件是落地的關鍵：** 系統必須設有最大迭代輪數、工具呼叫上限、成本預算與重複查詢偵測，避免 Agent 陷入無止境的推理迴圈（Reasoning Loop）。
6. **長期記憶（Long-term Memory）非必要核心：** Agentic RAG 核心需要的是良好的**短期狀態管理**（保存已讀文件、已知事實、剩餘預算等）；永久記憶若無妥善維護，反而易引入過期資訊與資安風險。

````python
import json

def find_rentals(query: str, filter: str = "") -> list:
    search_kwargs = {
        "search_text": query,
        "search_field": "description_embedding",
        "task_type": "QUESTION_ANSWERING",
        "top_k": 5,
        "output_fields": vectorsearch_v1beta.OutputFields(
            data_fields=[
                "name",
                "price",
                "neighborhood",
            ]
        ),
    }

    if filter.strip():
        search_kwargs["filter"] = json.loads(filter)

    response = search_client.search_data_objects(                #真正把搜尋送到 Google Cloud Vector Search 的地方
        request=vectorsearch_v1beta.SearchDataObjectsRequest(
            parent=collection_path,
            semantic_search=vectorsearch_v1beta.SemanticSearch(
                **search_kwargs
            ),
        )
    )

    return [
        {
            "name": result.data_object.data.get("name"),
            "price": result.data_object.data.get("price"),
            "neighborhood": (
                result.data_object.data.get("neighborhood")
            ),
        }
        for result in response.results
    ]
````

````python
from typing import Any

import pandas as pd
from google.genai.types import EmbedContentConfig
from sklearn.metrics.pairwise import cosine_similarity
from tqdm import tqdm  # to show the progress bar
````

````python
from google import genai

# 請將這裡替換成您在 Google Cloud Console 頂部看到的「專案 ID」
PROJECT_ID = "my-project-2026-494815"

# 初始化客戶端（指定 vertexai=True 才會扣抵您的每月的 premium 額度）
client = genai.Client(
    vertexai=True,
    project=PROJECT_ID,
    location="us-central1"
)

MODEL_ID = "gemini-embedding-001"


def get_embeddings(contents: list[str], task_type: str) -> list[list[float]]:
    """A wrapper to get text embeddings for a list of texts"""
    response = client.models.embed_content(
        model=MODEL_ID,
        contents=contents,
        config=EmbedContentConfig(task_type=task_type),
    )
    return [e.values for e in response.embeddings]
````

### Calculating Q&A search quality with `SEMANTIC_SIMILARITY`

First, let's test the task type `SEMANTIC_SIMILARITY` as a baseline.

Using the `get_embeddings()` function above, we define another function `generate_embeddings()` that takes a list of texts and generate embeddings with the specified task types. The embeddings will be stored on new columns `q_emb` and `a_emb`.

````python
def generate_embeddings(
    df: pd.DataFrame, question_task_type: str, answer_task_type: str
) -> pd.DataFrame:
    """Generate text embeddings for questions and answers"""
    q_embeddings: list[list[float]] = []
    a_embeddings: list[list[float]] = []

    for i in tqdm(range(0, len(df), 5)):  # bundle 5 items for each API call
        q_embeddings.extend(
            get_embeddings(df.loc[i : i + 4, "question"].tolist(), question_task_type)
        )

        answers: list[Any] = [row[0] for row in df.loc[i : i + 4, "answer"]]
        a_embeddings.extend(get_embeddings(answers, answer_task_type))

    df["q_emb"] = q_embeddings
    df["a_emb"] = a_embeddings
    return df
````

It's ready to start generating the embeddings for semantic similarity task. The following code may take a few minutes to finish generating embeddings for the 1K rows.

### Calculate cosine similarities between questions and answers

Next, we will calculate the cosine similarity for all combinations of the questions and answers using the `cosine_similarity()` function from scikit-learn. Thus, the `calc_cosine_similarities()` function will return a two dimensional array with similarities between 1K questions x 1K answers. By printing the result for the first question, you will see 1K values that represents the cosine similarity between the first question and the 1K answers.

````python
def calc_cosine_similarities(df: pd.DataFrame) -> list[list[float]]:
    """Calculates similarities for all combinations of the questions and answers"""
    q_embeddings: list[list[float]] = df["q_emb"].tolist()
    a_embeddings: list[list[float]] = df["a_emb"].tolist()
    return cosine_similarity(q_embeddings, a_embeddings)


# print the answer similarities for the first question
q_and_a_similarities = calc_cosine_similarities(similarity_df)
q_and_a_similarities[:5]
````

### Calculate MRR

Finally, we will measure the search quality using [Mean Reciprocal Rank (MRR)](https://en.wikipedia.org/wiki/Mean_reciprocal_rank). It is a simple algorithm - the table from the Wikipedia page gives an example:

![](https://storage.googleapis.com/github-repo/embeddings/task_type_embeddings/mrr_example.png)

In this tutorial, we will use the `answer` column of the dataset as the correct response for the `question` column. Thus, if the similarity between the question and answer is ranked as 3rd in the similarity list, reciprocal rank is 1/3.

To do it, we define `get_top100_similar_answers()` function that returns a top 100 list of answer index sorted by the similarity.

````python
def get_top100_similar_answers(answer_similarities: list[float]) -> list[int]:
    """Extracts top 100 similar answers"""
    sim_index: list[tuple[int, float]] = list(enumerate(answer_similarities))
    sorted_sim_index: list[tuple[int, float]] = sorted(
        sim_index, key=lambda x: x[1], reverse=True
    )
    sorted_index: list[int] = [index for index, _ in sorted_sim_index]
    return sorted_index[:100]


# print top 100 similar answers with their index
get_top100_similar_answers(q_and_a_similarities[0])[:10]
````

With this top 100 answer list for each question, we will add `query_ranks` column to the DataFrame that has a list like `[0 0 1 0 0...]`. In this case, the correct response is ranked at 3rd.

````python
# mypy: disable-error-code="index"


def add_query_ranks(df: pd.DataFrame) -> pd.DataFrame:
    """Adds a 'query_ranks' column to the DataFrame"""
    # Calculate similarities for all combinations of questions and answers
    q_and_a_similarities: list[list[float]] = calc_cosine_similarities(df)

    # Get query ranks for all questions
    query_ranks_for_all_questions: list[list[int]] = []
    for i in tqdm(range(len(q_and_a_similarities))):
        # For each question, get ground truth and top 100 similar answers
        ground_truth: Any = df.loc[i, "answer"][
            0
        ]  # Assuming 'answer' column contains lists
        similar_answers: list[int] = get_top100_similar_answers(q_and_a_similarities[i])

        # If an answer equals the ground truth, put "1" in the query ranks, otherwise "0"
        query_ranks: list[int] = []
        for index in similar_answers:
            similar_answer: Any = df.loc[index, "answer"][0]
            if similar_answer == ground_truth:
                query_ranks.append(1)
            else:
                query_ranks.append(0)
        query_ranks_for_all_questions.append(query_ranks)

    # Add the query ranks to the dataframe
    df["query_ranks"] = query_ranks_for_all_questions
    return df
````

````python
# add query_ranks column to the similarity_df
add_query_ranks(similarity_df)
similarity_df.head()
````

With this `query_ranks`, we will calculate MRR value for the dataset.

````python
def calc_mrr(df: pd.DataFrame) -> float:
    """Calculates MRR from the query ranks"""
    query_ranks_for_all_questions: list[list[int]] = df["query_ranks"].tolist()
    reciprocal_ranks: list[float] = []

    for query_ranks in query_ranks_for_all_questions:
        for i, relevance in enumerate(query_ranks):
            if relevance == 1:
                reciprocal_ranks.append(1 / (i + 1))
                break
        else:  # This else is associated with the for loop, executed if no break occurred
            reciprocal_ranks.append(0)

    return sum(reciprocal_ranks) / len(query_ranks_for_all_questions)
````

````python
mrr = calc_mrr(similarity_df)
mrr
````

You should see a value around 0.3 with this task type.

### When to use embedding tuning
One caveat of task type embedding is that, the model is trained with generic corpus on the web. That means, the model may be less effective on use cases such as retrieving company proprietary document or question and answering on topics that the model doesn't know. In such cases, [Tune text embeddings](https://cloud.google.com/vertex-ai/generative-ai/docs/models/tune-embeddings) could provide better option than using the pre-trained model.

```text
FAISS                 → 儲存並搜尋向量
Pandas                → 保存 name、price、neighborhood 等原始資料
```

Sentence Transformers 可以把文件和查詢文字轉成向量；FAISS 則專門做高效率的向量相似度搜尋。([Hugging Face][1])

原本 Vertex 的流程：

```text
description
→ Gemini Embedding
→ Google Cloud Collection
→ search_data_objects()
```

改成本機後：

```text
description
→ Sentence Transformer
→ FAISS index
→ index.search()
```

## 2. 建立本機 Embedding 和向量索引

假設你前面已經準備好 `df`：

```python
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# 本機 Embedding 模型
embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

# 將房源 description 轉成向量
document_embeddings = embedding_model.encode_document(
    df["description"].tolist(),
    normalize_embeddings=True,
)

document_embeddings = np.asarray(
    document_embeddings,
    dtype="float32",
)

# 建立 FAISS 向量索引
dimension = document_embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)
index.add(document_embeddings)

print("Indexed documents:", index.ntotal)
```

這裡：

```python
encode_document(...)
```

是把房源描述當成「被搜尋的文件」；查詢時可以使用 `encode_query(...)`。Sentence Transformers 官方對非對稱語意搜尋也建議分別使用文件與查詢編碼方法。([GitHub][2])

`IndexFlatIP` 使用向量內積搜尋。因為前面使用：

```python
normalize_embeddings=True
```

所以內積可以當作 cosine similarity 使用。

## 3. 建立本機搜尋函式

```python
def find_rentals_local(
    query: str,
    max_price: float | None = None,
    neighborhood: str | None = None,
    top_k: int = 5,
) -> list:
    # 將搜尋文字轉成 query embedding
    query_embedding = embedding_model.encode_query(
        query,
        normalize_embeddings=True,
    )

    query_embedding = np.asarray(
        [query_embedding],
        dtype="float32",
    )

    # 先多取一些，再做 metadata filtering
    search_k = min(len(df), top_k * 10)

    scores, indices = index.search(
        query_embedding,
        search_k,
    )

    results = []

    for score, row_index in zip(
        scores[0],
        indices[0],
    ):
        if row_index == -1:
            continue

        row = df.iloc[row_index]

        if max_price is not None:
            if float(row["price"]) >= max_price:
                continue

        if neighborhood is not None:
            if row["neighbourhood_cleansed"] != neighborhood:
                continue

        results.append(
            {
                "name": str(row["name"]),
                "price": float(row["price"]),
                "neighborhood": str(
                    row["neighbourhood_cleansed"]
                ),
                "score": float(score),
            }
        )

        if len(results) == top_k:
            break

    return results
```

使用方式：

```python
results = find_rentals_local(
    query="cozy workspace",
    max_price=80,
)

for result in results:
    print(result)
```

地區加價格：

```python
results = find_rentals_local(
    query="private room",
    max_price=100,
    neighborhood="Southwark",
)

for result in results:
    print(result)
```

## 保存到本機

FAISS index 可以存成檔案：

```python
faiss.write_index(
    index,
    "london_rentals.faiss",
)
```

DataFrame 另外保存：

```python
df.to_pickle(
    "london_rentals.pkl"
)
```

下次載入：

```python
import pandas as pd
import faiss

df = pd.read_pickle(
    "london_rentals.pkl"
)

index = faiss.read_index(
    "london_rentals.faiss"
)
```

所以本機版本完整概念是：

```text
CSV
→ DataFrame 清理
→ 本機 Embedding 模型
→ FAISS 向量索引
→ query embedding
→ 向量搜尋
→ Pandas metadata filtering
→ 回傳搜尋結果
```

Vertex AI 幫你自動處理 Collection、Embedding、儲存和搜尋；本機版本則是由你自己分別管理：

```text
Embedding model
FAISS index
Pandas metadata
本機檔案保存
```

## A2UI & A2A: Building the Dog Weather App

- https://www.youtube.com/watch?v=jycGNV1hN68
- https://github.com/goabego/adventofagents-day31

這段影片主要想說明：**如何使用 Google ADK 建立 AI Agent，透過 A2A 與 Gemini Enterprise 溝通，再透過 A2UI 將 Agent 的結果呈現成可互動的介面。**

## **一、先理解 4 個核心技術**

| 技術                                                      | 主要用途                                     |
| ------------------------------------------------------- | ---------------------------------------- |
| **ADK（Agent Development Kit，代理程式開發套件）**                 | 建立 AI Agent，定義工具與執行邏輯。                   |
| **A2A（Agent-to-Agent Protocol，代理程式間通訊協定）**              | 讓 Gemini Enterprise 能找到、呼叫並與自訂 Agent 溝通。 |
| **A2UI（Agent-to-User Interface Protocol，代理程式對使用者介面協定）** | 讓 Agent 用 JSON 描述介面，由前端呈現卡片、圖片、按鈕等元件。    |
| **Cloud Run（雲端執行服務）**                                   | 部署與執行 ADK Agent 的後端服務。                   |

## **二、A2A 與 A2UI 如何合作？**

**Gemini Enterprise**

使用者操作的前端平台（Client）

A2A：傳遞請求與回應

**ADK Weather Agent**

部署於 Cloud Run，執行查詢天氣、工具呼叫等邏輯

A2UI：回傳宣告式 UI JSON

**Gemini Enterprise 渲染介面**

天氣卡片

狗狗圖片

互動按鈕

簡單區分：

- **A2A 負責 Agent 之間如何溝通。**
- **A2UI 負責 Agent 的結果如何呈現給使用者。**

兩者可以搭配使用，但解決的是不同問題。

## **三、影片實際展示什麼？**

影片建立兩個版本的天氣 Agent。

| 功能      | V1   | V2           |
| ------- | ---- | ------------ |
| 查詢天氣    | 支援   | 支援           |
| 天氣資訊卡片  | 基本卡片 | 更豐富的卡片       |
| 狗狗圖片    | —    | 支援           |
| 互動按鈕    | —    | 支援           |
| 隨機切換城市  | —    | 支援           |
| 攝氏／華氏轉換 | —    | 支援（ADK Tool） |

**V2 最重要的進步不是卡片變漂亮，而是使用者能直接操作介面。**

例如，按下「隨機城市」按鈕，就會再次呼叫後端 Agent，取得新的天氣資料。

## **四、三個值得記住的技術觀念**

**1. Agent Card（代理程式描述卡）是 A2A 的重要入口。**

它是一份 JSON，描述 Agent 的服務網址、能力（Skills）與相關資訊，讓 Gemini Enterprise 知道如何連接這個 Agent。

**2. A2UI 是 Declarative UI Protocol（宣告式使用者介面協定）。**

Agent 不需要產生 HTML 或 JavaScript，而是輸出描述 UI 結構與互動的 JSON，交由支援 A2UI 的 Client 負責渲染。

它不是前端框架，也不是 CSS 樣式系統，而且不限於 Web，也能支援 iOS、Android 等平台。

**3. Backend-only Updates（僅更新後端）是重要優勢。**

開發者可以修改 Cloud Run 裡的 Agent，更新其 A2UI 輸出。只要前端支援相關元件，就能呈現更新後的介面，不必另外修改前端程式碼。

這不代表所有 UI 需求都完全不需要前端開發，而是減少每次變更 Agent 畫面時的前端修改成本。

## **五、這段影片真正要傳達的核心**

傳統 AI Agent 通常以文字回答問題。

結合 A2A 與 A2UI 後，AI Agent 不僅能被其他平台呼叫，還能提供**結構化、可互動的使用者介面**。

最重要的是記住這個分工：

**ADK 負責建立 Agent；A2A 負責連接 Agent；A2UI 負責描述互動介面；Gemini Enterprise 負責呈現介面。**

## **二、A2UI 最核心的三段 JSON**

來源：兩個版本的 `prompt_builder.py`。

A2UI 並不是直接輸出 HTML，而是讓 Agent 產生結構化 JSON。這個專案使用三種主要訊息組合畫面。

```

[
  {
    "beginRendering": {
      "surfaceId": "weather-surface",
      "root": "root",
      "styles": {
        "primaryColor": "#FF5722",
        "font": "Roboto"
      }
    }
  },
  {
    "surfaceUpdate": {
      "surfaceId": "weather-surface",
      "components": [
        {
          "id": "root",
          "component": {
            "Card": {
              "child": "temperature"
            }
          }
        },
        {
          "id": "temperature",
          "component": {
            "Text": {
              "text": {
                "path": "/weather/temperature"
              },
              "usageHint": "h2"
            }
          }
        }
      ]
    }
  },
  {
    "dataModelUpdate": {
      "surfaceId": "weather-surface",
      "path": "/",
      "contents": [
        {
          "key": "weather",
          "valueStruct": {
            "temperature": "25°C"
          }
        }
      ]
    }
  }
]

```

以上是依照 V2 專案欄位整理的最小化教學範例，不是原始程式的完整 JSON。

三個訊息分別負責：

1. `beginRendering`：建立畫面，指定根元件與基本樣式。
2. `surfaceUpdate`：定義畫面有哪些 UI 元件，以及元件之間的關係。
3. `dataModelUpdate`：提供實際資料，例如溫度、城市、天氣狀況。

其中最值得注意的是：

`"path": "/weather/temperature"` 代表 Text 元件要讀取 Data Model 中的溫度資料。

因此，**UI 的結構和 UI 顯示的資料，可以分開管理。**

## **三、V1 和 V2 的資料綁定比較**

### **V1：使用&#x20;**`valueString`

```

{
  "dataModelUpdate": {
    "surfaceId": "weather-surface",
    "path": "/",
    "contents": [
      {
        "key": "location",
        "valueString": "New York, NY"
      },
      {
        "key": "temperature",
        "valueString": "72°"
      },
      {
        "key": "description",
        "valueString": "Partly Cloudy"
      }
    ]
  }
}

```

V1 的每個資料欄位分開定義，適合簡單卡片。

### **V2：使用&#x20;**`valueStruct`

```

{
  "dataModelUpdate": {
    "surfaceId": "weather-surface",
    "path": "/",
    "contents": [
      {
        "key": "weather",
        "valueStruct": {
          "location": "New York, NY",
          "temperature": "72°",
          "description": "Partly Cloudy",
          "dogImageUrl": "https://images.dog.ceo/breeds/retriever-golden/n02099601_3004.jpg"
        }
      }
    ]
  }
}

```

V2 把天氣資料集中成 `weather` 物件。

例如，UI 可以使用：

```
{
  "text": {
    "path": "/weather/location"
  }
}
```

直接讀取 `New York, NY`。

**重點：** `valueStruct` 更適合管理巢狀資料、圖片及多天天氣預報。

## **四、V2 最重要的功能：互動式按鈕**

這段是原始程式碼中 `btnRandom` 的定義。

```

{
  "id": "btnRandom",
  "component": {
    "Button": {
      "child": "lblRandom",
      "action": {
        "name": "sendText",
        "context": [
          {
            "key": "text",
            "value": {
              "literalString": "@Weather Agent V2 Pick a random city and show its weather."
            }
          }
        ]
      }
    }
  }
}

```

其中三個關鍵：

- `Button`：建立按鈕。
- `action.name: "sendText"`：告訴 Client，按下後要送出文字訊息。
- `context`：指定送給 Agent 的訊息內容。

實際執行流程：

**使用者點擊 Random City**

**Gemini Enterprise 執行 sendText**

**透過 A2A 呼叫 ADK Agent**

**Agent 呼叫 get_random_location() 與 get_weather()**

**回傳新的 A2UI JSON，更新天氣卡片**

這裡有個重要觀念：

**A2UI Button 不直接執行 Python，而是透過 Client 的 Action 機制，再讓 Agent 呼叫後端工具。**

攝氏、華氏按鈕也採用相同方式，送出切換單位的文字指令。

## **五、V2 如何動態產生天氣預報？**

這裡使用 `List` 搭配 `template`。

```

{
  "id": "forecastList",
  "component": {
    "List": {
      "children": {
        "template": {
          "componentId": "forecastItem",
          "dataBinding": "/weather/forecast"
        }
      },
      "direction": "horizontal",
      "alignment": "center"
    }
  }
}

```

核心欄位：

- `List`：建立清單。
- `componentId`：指定每筆資料使用哪個元件範本。
- `dataBinding`：指定清單資料來源。

例如資料模型有三天天氣，Client 就能依範本產生三筆預報項目，不必像 V1 一樣逐一列出 `day-1-row`、`day-2-row`。

## **六、A2UI JSON 最後如何送到 Gemini Enterprise？**

這一段位於 [version-2/part_converters.py](https://github.com/goabego/adventofagents-day31/blob/main/version-2/part_converters.py?utm_source=chatgpt.com)。

專案透過特殊分隔字串識別 A2UI：

```
def _extract_a2ui_json(text: str):    if "---a2ui_JSON---" in text:        parts = text.split("---a2ui_JSON---", 1)        conversational_text = parts[0].strip()        ui_json = parts[1].strip()        return conversational_text, ui_json    return text, None
```

以上為原始函式的簡化片段，省略 Markdown code fence 清理邏輯。

Agent 的輸出大致像這樣：

```

今天台北天氣晴朗，氣溫 25°C。

---a2ui_JSON---

[
  { "beginRendering": { ... } },
  { "surfaceUpdate": { ... } },
  { "dataModelUpdate": { ... } }
]

```

後續轉換程式會將 JSON 解析成 A2A `DataPart`，並標示：

```
metadata={    "mimeType": "application/json+a2ui"}
```

這樣支援 A2UI 的 Client 才能辨識這些結構化資料並渲染畫面。

請特別注意：`---a2ui_JSON---` 是這個專案使用的輸出分隔慣例，並不是所有 A2UI 系統都必須使用的標準格式。

## **七、還有一個容易誤解的地方**

我檢查了 V2 的 `gemini_agent.py`，其中 `get_weather()` 產生的是 **Mock Data（模擬天氣資料）**。

例如：

```
conditions = [    "Sunny",    "Rainy",    "Cloudy",    "Snowy",    "Stormy"]selected_condition = random.choice(conditions)
```

這不是在查詢真實氣象服務，而是隨機產生天氣狀況，方便示範 A2UI。

另外，程式中的溫度轉換是由 `get_weather()` 的 `unit` 參數處理，並非獨立的溫度換算工具。

## **最後整理：這個專案最值得學的 4 個 A2UI 知識**

| 關鍵技術                | 應該理解的觀念               |
| ------------------- | --------------------- |
| `surfaceUpdate`     | 用 JSON 定義 UI 元件與布局    |
| `dataModelUpdate`   | 把資料與 UI 結構分離          |
| `List + template`   | 根據資料筆數動態建立元件          |
| `Button + sendText` | 讓使用者操作 UI，並重新呼叫 Agent |

**整個專案的核心價值：讓 ADK Agent 透過 A2UI JSON 控制畫面的內容與互動，而不是讓開發者每次都重新撰寫前端介面。**

建議優先閱讀 [V2 的 prompt_builder.py](https://github.com/goabego/adventofagents-day31/blob/main/version-2/prompt_builder.py?utm_source=chatgpt.com)。這份檔案集中定義了 A2UI 畫面範本、按鈕 Action，以及要求 LLM 產生 A2UI JSON 的 Prompt，是整個範例最重要的學習入口。

# Python **`import`** 與模組搜尋規則


在 Python 中，當我們寫下：

```
from gemini_agent import GeminiAgent
```

Python 必須知道兩件事：

1. `gemini_agent` 這個模組（Module）在哪裡？
2. 這個模組裡面有沒有 `GeminiAgent` 這個名稱？

理解 Python 如何尋找模組，是學習多檔案程式設計的重要基礎。

## **一、先認識 Module、Package 與 Class**

| 名稱         | 中文     | 意思                  |
| ---------- | ------ | ------------------- |
| Module     | 模組     | 通常是一個 `.py` 檔案      |
| Package    | 套件     | 用來組織多個模組的目錄         |
| Class      | 類別     | 用來定義物件的結構與行為        |
| `sys.path` | 模組搜尋路徑 | Python 尋找模組時使用的路徑清單 |

例如：

```
my_project/
│
├── main.py
│
└── agents/
    ├── __init__.py
    └── gemini_agent.py
```

在 `gemini_agent.py` 中：

```
class GeminiAgent:    pass
```

這時：

- `agents` 是 Package。
- `gemini_agent.py` 是 Module。
- `GeminiAgent` 是 Class。

在 `main.py` 中可以寫：

```
from agents.gemini_agent import GeminiAgent
```

意思是從 `agents` 套件中的 `gemini_agent` 模組，匯入 `GeminiAgent` 類別。

## **二、Python 的 import 搜尋順序**

Python 執行 `import` 時，大致遵循以下程序：

**執行 import**

例如 import gemini_agent

**1. 檢查 sys.modules**

是否已經載入這個模組？

**2. 檢查內建模組**

例如 sys、builtins

**3. 依照 sys.path 搜尋**

搜尋本機檔案、標準函式庫、第三方套件

**找到**

載入模組

**找不到**

ModuleNotFoundError

這是一般情況下的簡化流程。Python 也有 Import Hooks 等進階機制，可以改變模組的尋找方式。

其中最值得初學者理解的是 `sys.path`。

## **三、什麼是 sys.path？**

`sys.path` 是 Python 用來尋找模組的一份路徑清單。

一般情況下包含：

| 搜尋來源            | 說明                  |
| --------------- | ------------------- |
| 程式所在目錄          | 例如 `main.py` 的所在位置  |
| `PYTHONPATH`    | 使用者額外設定的搜尋路徑        |
| Python 標準函式庫    | 例如 `json`、`pathlib` |
| `site-packages` | 透過 pip 安裝的第三方套件     |

Python 會依照 `sys.path` 中的順序尋找模組。

你可以自己檢查：

```
import sysfor index, path in enumerate(sys.path):    print(index, path)
```

輸出可能像這樣：

```
0 /home/user/my_project
1 /usr/local/lib/python3.11
2 /usr/local/lib/python3.11/site-packages
```

要注意，這些路徑不是固定的，會因為作業系統、虛擬環境以及 Python 啟動方式而改變。

另外，`python main.py`、`python -m package.module`、Jupyter Notebook 的初始搜尋路徑也可能不同。

## **四、Python 不會自動搜尋所有子目錄**

這是初學者最容易碰到的問題之一。

假設：

```
my_project/
│
├── main.py
│
└── agents/
    ├── __init__.py
    └── gemini_agent.py
```

如果 `main.py` 寫：

```
from gemini_agent import GeminiAgent
```

一般情況下會找不到，因為 `gemini_agent.py` 位於 `agents` 子目錄中。

Python 不會自動遞迴搜尋：

```
my_project/
my_project/agents/
my_project/agents/其他子目錄/
...
```

正確做法是指定完整的套件路徑：

```
from agents.gemini_agent import GeminiAgent
```

這樣 Python 就知道要從 `agents` 裡尋找 `gemini_agent`。

**記住：**`sys.path`**&#x20;決定從哪裡開始找，Package 路徑決定往哪個套件層級找。**

## **五、**`__init__.py`**&#x20;有什麼作用？**

例如：

```
agents/
├── __init__.py
└── gemini_agent.py
```

`__init__.py` 是傳統 Python Package 的標記檔案，可以是空白的。

它有兩個常見用途：

1. 明確表示這個資料夾是一個 Python Package。
2. 放置套件初始化程式碼，或對外匯出常用名稱。

Python 3.3 之後支援 Namespace Package，因此不是所有 Package 都一定需要 `__init__.py`。

不過，**對初學者而言，自己建立一般 Package 時，建議保留&#x20;**`__init__.py`，讓專案結構更明確。

## **六、**`from ... import ...`**&#x20;到底怎麼看？**

比較下面三種寫法。

### **寫法 A：匯入整個模組**

```
import agents.gemini_agentagent = agents.gemini_agent.GeminiAgent()
```

使用時需要寫出模組路徑。

### **寫法 B：直接匯入類別**

```
from agents.gemini_agent import GeminiAgentagent = GeminiAgent()
```

這是專案中很常見的寫法，使用起來較簡潔。

### **寫法 C：使用別名（Alias）**

```
from agents.gemini_agent import GeminiAgent as MyAgentagent = MyAgent()
```

`as` 可以幫匯入的名稱建立別名，避免名稱太長或發生命名衝突。

另外，`from ... import ...` 後面不一定是 Class，也可能是函式、變數或子模組。

## **七、兩種尋找子目錄的方式**

假設：

```
my_project/
├── main.py
└── agents/
    ├── __init__.py
    └── gemini_agent.py
```

**方法一：使用 Package 路徑（推薦）**

```
from agents.gemini_agent import GeminiAgent
```

不必修改 `sys.path`，只要 `my_project` 已經位於搜尋路徑中即可。

**方法二：手動修改 sys.path**

```
import syssys.path.append("./agents")from gemini_agent import GeminiAgent
```

這樣也可能成功，但不建議當作一般專案的主要做法。

因為 `./agents` 是相對於目前工作目錄，而不一定是 `main.py` 的所在目錄。

不同位置啟動程式時，可能產生不同結果。

## **八、特別注意：相對匯入與絕對匯入**

除了：

```
from agents.gemini_agent import GeminiAgent
```

Python 還支援 Relative Import（相對匯入）。

例如專案結構：

```
my_project/
└── agents/
    ├── __init__.py
    ├── gemini_agent.py
    └── helper.py
```

如果在 `gemini_agent.py` 中需要使用同一個 Package 的 `helper.py`，可以寫：

```
from .helper import some_function
```

其中：

- `.` 表示目前 Package。
- `..` 表示上一層 Package。

但相對匯入依賴 Python 的 Package 執行環境。

例如，在專案根目錄直接執行：

```
python agents/gemini_agent.py
```

如果檔案內使用 `from .helper import some_function`，就可能出現：

```
ImportError: attempted relative import
with no known parent package
```

如果它是需要獨立執行的模組，通常可以從專案根目錄使用：

```
python -m agents.gemini_agent
```

這樣 Python 就會以 Package 模組的方式執行。

## **九、常見的 import 錯誤**

| 錯誤                                                       | 可能原因                      |
| -------------------------------------------------------- | ------------------------- |
| `ModuleNotFoundError`                                    | 找不到模組或其相依模組               |
| `ImportError: cannot import name`                        | 找到模組，但無法取得指定名稱；也可能是循環匯入   |
| `attempted relative import with no known parent package` | 直接執行了使用相對匯入的檔案            |
| 匯入到錯誤的模組                                                 | 不同目錄有相同名稱的模組              |
| pip 安裝後還是找不到                                             | 安裝套件的 Python 環境與執行程式的環境不同 |

有一個特別需要注意的問題：**不要隨意把自己的檔案命名為&#x20;**`json.py`**、**`random.py`**、**`typing.py`**&#x20;等標準函式庫的名稱。**

因為 Python 可能優先找到你的檔案，造成匯入衝突。

如果套件安裝後找不到，可以確認目前執行的 Python：

```
python -m pip show google-adk
```

這種寫法能讓 pip 使用你指定的 Python 環境。

## **十、如何確認 Python 找到的是哪個檔案？**

當 import 行為不符合預期，可以使用以下方法。

```
import importlib.utilmodule_name = "json"spec = importlib.util.find_spec(module_name)if spec is not None:    print("模組位置：", spec.origin)else:    print("找不到模組")
```

這可以協助檢查模組會從哪裡載入。

如果已經成功匯入一般 `.py` 模組，也可以使用：

```
import gemini_agentprint(gemini_agent.__file__)
```

但有些內建模組或特殊模組不一定具有 `__file__`。

## **十一、最重要的 5 個觀念**

1. **Python 的&#x20;**`import`**&#x20;是依照模組名稱及搜尋路徑尋找程式碼，不是隨意掃描所有資料夾。**
2. `sys.path` 決定 Python 從哪些根目錄尋找模組。
3. Package 路徑（例如 `agents.gemini_agent`）決定從根目錄往下找的層級。
4. 一般專案優先使用完整 Package 路徑，不要隨意修改 `sys.path`。
5. 如果 import 失敗，先檢查檔案位置、Package 結構、執行方式及 Python 環境。

最後再回到你的 Google ADK 專案：

```
from gemini_agent import GeminiAgent
```

之所以可以這樣寫，是因為 `main.py` 和 `gemini_agent.py` 位於同一個資料夾，而且在這個專案的啟動方式下，該資料夾位於 Python 的模組搜尋路徑中。

**學習 Python import 的關鍵，就是分清楚三件事：檔案放在哪裡、Python 從哪裡開始找、import 寫了什麼路徑。**

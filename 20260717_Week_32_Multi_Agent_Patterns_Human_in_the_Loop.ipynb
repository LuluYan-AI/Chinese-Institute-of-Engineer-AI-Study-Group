## Multi-Agent Patterns: Human in the Loop

### Human in the Loop

Google ADK 的 Human in the Loop，也就是 HITL，是一種把「人類審批」插入 agent 執行流程中的 Multi Agent / agent workflow 結構。它的核心目的不是提高模型推理能力，而是提供 production safeguard：當 agent 即將執行高風險工具，例如退款、金融交易、刪除資料、寫入 production database、寄出正式 email、修改雲端資源時，系統必須強制暫停，等待人類確認，不能只靠 prompt engineering 叫模型「小心一點」。

這個 pattern 的重點是：敏感 action 不是由 LLM 自行決定是否執行，而是由框架層提供一個 hard breakpoint。agent 可以提出「我想做這件事」，但真正的 tool execution 必須等人類回覆 approve 或 reject 之後才會繼續。ADK 內建支援這類 human input workflow，包括 graph workflow 裡的 `RequestInput` HITL node，以及 LLM agent tool 層級的 confirmation / long-running tool 機制。




#### implement HITL

```python
import asyncio
from typing import Any
from google.adk.agents import Agent
from google.adk.events import Event
from google.adk.runners import Runner
from google.adk.tools import LongRunningFunctionTool
from google.adk.sessions import InMemorySessionService
from google.genai import types


# 1. Define a tool that requires human approval
def ask_for_approval(purpose: str, amount: float) -> dict[str, Any]:
    """Ask a human to approve this action before proceeding."""
    return {"status": "pending", "purpose": purpose, "amount": amount}


# 2. Define the tool that runs after approval
def process_refund(purpose: str, amount: float) -> dict[str, str]:
    """Process the approved refund."""
    return {"status": "success", "message": f"Refunded ${amount} for {purpose}."}


# 3. Create the agent with a LongRunningFunctionTool
agent = Agent(
    name="refund_agent",
    model="gemini-3.1-pro",
    instruction="""You handle refunds. Always call ask_for_approval first.
    If approved, call process_refund. If rejected, tell the user.""",
    tools=[LongRunningFunctionTool(func=ask_for_approval), process_refund],
)


async def main():
    # 4. Set up session and runner
    session_service = InMemorySessionService()
    session = await session_service.create_session(
        app_name="hitl-app", user_id="user1", session_id="s1"
    )
    runner = Runner(agent=agent, app_name="hitl-app", session_service=session_service)

    # 5. Run the agent — it will pause when it hits ask_for_approval
    user_msg = types.Content(
        role="user",
        parts=[types.Part(text="Refund $200 for a duplicate charge on txn_123")],
    )

    # Two-step detection:
    #  - Step A: find the function_call on the event with long_running_tool_ids
    #  - Step B: find the function_response matching that call ID
    long_running_call = None
    pending_response = None

    async for event in runner.run_async(
        session_id=session.id, user_id="user1", new_message=user_msg
    ):
        if event.content and event.content.parts:
            for part in event.content.parts:
                # Step A: capture the long-running function *call*
                if (
                    not long_running_call
                    and part.function_call
                    and event.long_running_tool_ids
                    and part.function_call.id in event.long_running_tool_ids
                ):
                    long_running_call = part.function_call

                # Step B: capture the matching function *response*
                if (
                    long_running_call
                    and part.function_response
                    and part.function_response.id == long_running_call.id
                ):
                    pending_response = part.function_response

            # Print any agent text
            text = "".join(p.text or "" for p in event.content.parts)
            if text:
                print(f"Agent: {text}")

    if not pending_response:
        return

    # 6. Get human decision
    print(f"\n⏸  Pending: {pending_response.response}")
    choice = input("Approve? [Y/n]: ").strip().lower()

    # 7. Send decision back — agent resumes
    updated = pending_response.model_copy(deep=True)
    updated.response = {"status": "approved" if choice != "n" else "rejected"}

    async for event in runner.run_async(
        session_id=session.id,
        user_id="user1",
        new_message=types.Content(
            role="user", parts=[types.Part(function_response=updated)]
        ),
    ):
        if event.content and event.content.parts:
            text = "".join(p.text or "" for p in event.content.parts)
            if text:
                print(f"Agent: {text}")


if __name__ == "__main__":
    asyncio.run(main())
```


在 Day 14 範例裡，HITL 是用 `LongRunningFunctionTool` 實現的。這個工具包裝了一個需要外部介入的 function，例如：

```python
def ask_for_approval(purpose: str, amount: float) -> dict[str, Any]:
    """Ask a human to approve this action before proceeding."""
    return {"status": "pending", "purpose": purpose, "amount": amount}
```

這個 function 本身不真的處理退款。它只是把目前待審批的資訊結構化回傳，例如退款理由與金額。把它包成 `LongRunningFunctionTool(func=ask_for_approval)` 之後，ADK 會知道這不是一個普通同步 tool call，而是一個需要外部 human decision 的暫停點。

真正執行敏感動作的是另一個 tool：

```python
def process_refund(purpose: str, amount: float) -> dict[str, str]:
    """Process the approved refund."""
    return {"status": "success", "message": f"Refunded ${amount} for {purpose}."}
```

agent 的 instruction 則定義了 policy：

```python
instruction="""You handle refunds. Always call ask_for_approval first.
If approved, call process_refund. If rejected, tell the user."""
```

這裡的關鍵不是「模型被要求先問」，而是 `ask_for_approval` 被設計成 long-running breakpoint。模型一旦呼叫它，runner 的 async event stream 會暴露 pending tool call，client 可以攔截這個 event，讓執行流程停住。這是框架級的控制點，比 prompt 裡寫「請先確認」可靠得多。

#### 整個 HITL 流程


上面的 HITL 程式執行可以拆成七步。

第一步，定義 approval tool。這個 tool 的功能是建立一個 pending approval request。它應該回傳足夠的人類判斷資訊，例如 `purpose`、`amount`、交易 ID、目標資源、風險等級。

第二步，定義 approved action tool。這才是真正會修改外部世界的工具，例如 `process_refund`。在 production 裡，這個 tool 可能會呼叫支付 API、資料庫、CRM、ticketing system 或 cloud API。

第三步，建立 agent，並把 approval tool 包成 `LongRunningFunctionTool`。這一步讓 ADK runtime 知道：這個 tool call 不是立即完成，而是需要外部 decision 後才能 resume。

第四步，使用 `Runner.run_async()` 啟動 agent。`run_async` 會產生一串 `Event`。client 不只是等最後答案，而是監看整個 event stream。這也是 ADK HITL 的重點：控制層在 stream 裡觀察 agent 嘗試做什麼，而不是事後才檢查結果。

第五步，event interception。當 agent 嘗試呼叫 long-running tool 時，事件裡會有 `event.long_running_tool_ids`。client 需要找到對應的 `function_call.id`，再找到同一個 call ID 對應的 `function_response`。你的範例中這段邏輯很重要：

```python
if (
    not long_running_call
    and part.function_call
    and event.long_running_tool_ids
    and part.function_call.id in event.long_running_tool_ids
):
    long_running_call = part.function_call
```

這段是在捕捉「哪一個 function call 是需要人類介入的 long-running call」。接著：

```python
if (
    long_running_call
    and part.function_response
    and part.function_response.id == long_running_call.id
):
    pending_response = part.function_response
```

這段是在拿到該 approval request 的 pending response，例如：

```python
{"status": "pending", "purpose": "...", "amount": 200}
```

第六步，人類做 decision。範例用 CLI prompt：

```python
choice = input("Approve? [Y/n]: ").strip().lower()
```

production 裡這個地方可以換成 Slack button、web UI、admin dashboard、email approval link、ticket system workflow，或內部 control plane。重點是 approval 不應該由模型自己產生，而應該來自 authenticated human operator。

第七步，把 decision 注入 runner，讓 agent resume。範例中使用 `model_copy(deep=True)` 複製 pending response，然後把 response 改成：

```python
updated.response = {"status": "approved" if choice != "n" else "rejected"}
```

再把這個 `function_response` 放回新的 user message：

```python
new_message=types.Content(
    role="user", parts=[types.Part(function_response=updated)]
)
```

這一步是 state resumption 的關鍵。runner 不會從零開始重新跑整個對話，而是接著剛才 suspended 的地方繼續。agent 收到 approval 結果後，如果是 approved，就呼叫 `process_refund`；如果是 rejected，就告訴使用者該操作被拒絕。

從架構圖來看，這種 HITL pattern 可以表示成：

```text
User request
    ↓
LLM Agent interprets intent
    ↓
Sensitive action needed?
    ↓
Call approval tool
    ↓
ADK emits long-running tool event
    ↓
Client intercepts event
    ↓
Human approves / rejects
    ↓
Function response injected back
    ↓
Agent resumes
    ↓
Execute sensitive tool or abort
```



#### 角色

這裡的 Multi Agent 意義在於：agentic system 不再是單一 LLM 自主閉環，而是由三個角色共同構成控制系統。

第一個角色是 LLM agent，負責理解使用者意圖、規劃工具呼叫、根據 approval 結果繼續推理。

第二個角色是 ADK runner / event loop，負責執行狀態管理、暫停、事件攔截、resume。這是 orchestration layer。

第三個角色是 human approver，負責對高風險行為做授權決策。這個 human 不一定在 chat 裡，也可能在外部 UI 或審批系統裡。

所以 HITL 不是「讓使用者多問一句」。它是一種 control-plane architecture：LLM 可以提出 action proposal，但 side-effecting operation 必須通過人工 gate。

ADK 其實提供兩類 HITL 思路。第一類是 workflow graph 層級的 human input node。ADK 文件中 `RequestInput` 可以在 graph workflow 裡暫停節點執行，等待人類輸入，再把輸入傳給下一個 node。例如一個 workflow 可以先產生旅遊 itinerary，再用 `RequestInput` 問使用者哪些活動感興趣，最後根據回覆 finalize itinerary。這種適合「人類提供資料、偏好、回饋」的場景。fileciteturn1file0

第二類是 tool-confirmation / long-running tool 層級的 HITL。這種適合「LLM agent 要執行某個敏感 tool，但必須先取得 yes/no approval」的場景。你提供的 refund agent 就屬於這一類。ADK 文件也區分了 graph HITL nodes 和 LLM-agent-level tool confirmation：前者是 workflow node，後者是工具執行前的審批點。fileciteturn1file0

兩者差別可以這樣理解：

```text
RequestInput:
    Human supplies missing input or feedback inside a workflow graph.

LongRunningFunctionTool / tool confirmation:
    Human approves or rejects a specific sensitive tool call inside an LLM agent.
```

在 production 設計中，HITL 最重要的是不要只把 approval 當成 UI 功能，而要把它當成安全邊界。也就是說，`process_refund` 這類工具本身最好也檢查 approval state，而不是只相信 agent instruction。更嚴格的設計會把 approval token、approver identity、timestamp、scope、amount limit、audit log 都寫入外部系統。agent 即使 hallucinate，也不能繞過這個 policy gate。

用 Python 工程角度看，這個範例最值得注意的是 async event stream 的處理。`runner.run_async()` 不是單純回傳最後文字，而是逐步產生 events。你要在 events 中同時看 `function_call`、`function_response` 和 `event.long_running_tool_ids`。這種設計讓 client 可以建立自己的 approval UI，而不需要改 agent 本體。agent 定義 policy，runner 暴露 execution events，client 決定如何呈現與回填 human decision。

這種 HITL pattern 適合以下場景：退款、匯款、下單、取消訂單、刪除資料、修改 production config、發送外部郵件、建立雲端資源、授權權限、執行 SQL mutation、觸發 CI/CD deploy。共通點是：action 有外部副作用，而且錯誤成本高。

一句話總結：Google ADK 的 Human in the Loop 是一種把人類審批變成 agent runtime 一部分的安全架構。LLM agent 負責推理與提案，ADK runner 負責暫停與恢復，human approver 負責授權敏感操作。透過 `LongRunningFunctionTool`、event interception 和 function response resumption，agent 可以在不中斷 session state 的情況下安全等待人工決策，然後從原本的位置繼續執行。

#### 1. LongRunningFunctionTool 的核心定位

Google ADK 的 `LongRunningFunctionTool` 是用來處理「不能在同一輪 agent execution 裡立即完成」的工具呼叫。它的重點不是讓 Python function 在本地同步跑很久，而是讓 ADK runtime 知道：這個 tool call 啟動後，結果可能要等外部流程、人類審批、背景 job、callback 或 polling 才會回來。因此，它提供的是一個 pause / resume 的協調機制，而不是單純的非同步執行工具。

一般 `FunctionTool` 的預設假設是：agent 呼叫 tool，tool 執行完，立刻回傳結果，LLM 根據結果繼續推理。但 `LongRunningFunctionTool` 的假設不同：agent 呼叫 tool 後，tool 可能只先回傳一個初始狀態，例如 `{"status": "pending", "job_id": "..."}`。這個 response 代表任務已經被提交或等待審批，但尚未完成。接下來，ADK runner 可以暫停目前流程，等待 client 或外部系統在未來把最終 `FunctionResponse` 補回來。

因此，`LongRunningFunctionTool` 最適合被理解成「跨 invocation 的 continuation point」。它讓 agent 的推理流程可以在某個 function call 上停住，保留上下文，等到外部結果到達後，再從原本的位置繼續。這對 production agent 很重要，因為很多真實任務不是幾秒內能完成的，例如退款審批、資料庫長查詢、報表生成、批次資料處理、模型訓練、檔案轉換、部署流程，或任何需要人工確認的敏感操作。

它也不是 `async def` 的替代品。`async def` 解決的是 Python event loop 裡的非阻塞 I/O；`LongRunningFunctionTool` 解決的是 agent runtime 層級的長流程狀態管理。換句話說，前者處理的是「這個 function 怎麼等待」，後者處理的是「agent 如何在這個 tool call 尚未完成時暫停，並在未來拿到結果後繼續」。

所以，`LongRunningFunctionTool` 的核心價值是把 tool call 拆成兩個階段：第一階段是啟動長任務並回傳 pending 狀態；第二階段是由人類、client、worker 或外部系統把最終結果送回 ADK runner，讓 agent 接續推理。這使得 agent 不需要卡在長時間任務上，也不需要失去原本的對話狀態。


#### 2. 基本使用方式與執行流程

`LongRunningFunctionTool` 的基本使用方式很簡單：先定義一個普通 Python function，然後用 `LongRunningFunctionTool` 包起來。這個 function 的責任通常不是完成整個任務，而是「啟動」任務，並回傳一個可以追蹤後續進度的初始結果。

例如，一個需要人工審批的退款流程可以這樣設計：

```python
from typing import Any
from google.adk.tools import LongRunningFunctionTool

def ask_for_approval(purpose: str, amount: float) -> dict[str, Any]:
    """Ask a human to approve this action before proceeding."""
    return {
        "status": "pending",
        "purpose": purpose,
        "amount": amount,
    }

approval_tool = LongRunningFunctionTool(func=ask_for_approval)
```

這裡的 `ask_for_approval()` 並沒有真的執行退款。它只是產生一個 pending 狀態，告訴 ADK runtime：這個 tool call 已經啟動，但還需要外部決策。實際上，這個 function 也可以建立一張 approval ticket、送出 Slack 通知、寫入資料庫、提交背景 job，或呼叫某個外部 workflow system。

接著，把這個 tool 放進 agent：

```python
from google.adk.agents import Agent

agent = Agent(
    name="refund_agent",
    model="gemini-3.1-pro",
    instruction="""
    You handle refunds.
    Always call ask_for_approval before processing a refund.
    If approved, continue.
    If rejected, tell the user the refund was rejected.
    """,
    tools=[
        LongRunningFunctionTool(func=ask_for_approval),
    ],
)
```

當使用者提出請求，例如「Refund $200 for a duplicate charge」，LLM 會根據 instruction 呼叫 `ask_for_approval`。因為它是 `LongRunningFunctionTool`，ADK 會把這次 tool call 視為一個長時間任務，並在 event stream 中標記它。tool function 回傳的 pending result 會被包成 `FunctionResponse`，讓 agent 或 client 可以知道目前狀態。

整體流程可以簡化成：

```text
User request
    ↓
Agent decides to call long-running tool
    ↓
LongRunningFunctionTool starts external task
    ↓
Tool returns pending response
    ↓
ADK exposes the pending call in the event stream
    ↓
Runner pauses or ends the current turn
    ↓
External system or human later provides final result
    ↓
Client sends FunctionResponse back to runner
    ↓
Agent resumes from the original tool call
```

這個流程的關鍵是：第一次 tool call 回傳的不是最終答案，而是「任務已經被啟動」的狀態。真正的完成結果會在之後透過同一個 function call 的 `FunctionResponse` 補回來。

因此，在設計 `LongRunningFunctionTool` 的回傳值時，最好包含足夠的追蹤資訊，例如：

```python
{
    "status": "pending",
    "operation_id": "refund_123",
    "message": "Refund approval is pending.",
    "estimated_time_seconds": 300,
}
```

其中 `operation_id` 可以是 `job_id`、`ticket_id`、`task_id` 或任何外部系統能用來查詢狀態的 ID。這個 ID 很重要，因為後續 client、worker 或 approval system 需要用它找到原本那個尚未完成的任務。

簡單說，`LongRunningFunctionTool` 的基本流程不是「呼叫工具並立刻拿到答案」，而是「呼叫工具以啟動長任務，先回傳 pending，之後再把完成結果送回 agent」。這種兩階段設計讓 agent 可以處理長時間任務，同時保留原本的對話上下文與推理狀態。


#### 3. event stream：如何偵測 long-running call

`LongRunningFunctionTool` 真正發揮作用的地方，不只是在 agent 裡把 function 包起來，而是在 client 端的 event stream 中偵測它。ADK 的 `Runner.run_async()` 會逐步產生事件，每個事件可能包含 agent text、function call、function response，或其他執行狀態。當 agent 呼叫的是 `LongRunningFunctionTool` 時，ADK 會在 event 裡標記 `long_running_tool_ids`，讓 client 知道這個 tool call 需要外部介入。

也就是說，client 不能只等待最後的文字回答。它必須監看整個 async event stream，找出哪一個 function call 是 long-running call。這個偵測點通常長這樣：

```python
long_running_call = None #用來保存「agent 想呼叫的那個 long-running tool call」
pending_response = None #用來保存「這個 long-running tool 初次回傳的 pending response」

async for event in runner.run_async(
    session_id=session.id,
    user_id="user1",
    new_message=user_msg,
):
    if event.content and event.content.parts:
        for part in event.content.parts:
            if (
                not long_running_call
                and part.function_call
                and event.long_running_tool_ids
                and part.function_call.id in event.long_running_tool_ids
            ):
                long_running_call = part.function_call

            if (
                long_running_call
                and part.function_response
                and part.function_response.id == long_running_call.id
            ):
                pending_response = part.function_response
```

這段程式做了兩件事。第一，它找到 long-running 的 `function_call`。`part.function_call` 代表 agent 想呼叫某個 tool；`event.long_running_tool_ids` 則代表這個 event 裡有哪些 tool call 被 ADK 標記為 long-running。只有當 `part.function_call.id` 出現在 `event.long_running_tool_ids` 裡，client 才應該把它視為需要暫停與外部介入的 call。

第二，它找到對應的 `function_response`。這個 response 通常就是 long-running tool 第一次回傳的 pending 狀態，例如：

```python
{
    "status": "pending",
    "purpose": "duplicate charge",
    "amount": 200,
    "message": "Approve refund of $200?"
}
```

這個 `pending_response` 非常重要，因為它保留了原本 function call 的 identity。後面要 resume agent 時，不是隨便建立一個新的 response，而是要複製這個 `pending_response`，修改其中的 `response` payload，然後用同一個 `function_response.id` 送回 runner。這樣 ADK 才知道：這個回覆是針對剛才那個 long-running function call 的結果。

更清楚地說，偵測 long-running call 時有三個 ID 概念要分清楚：

```text
function_call.id
    ADK / model 產生的 tool call ID。
    用來對應 function call 與 function response。

event.long_running_tool_ids
    ADK 在 event 裡標記的 long-running tool call ID 清單。
    如果 function_call.id 出現在這裡，代表這個 call 需要外部介入。

operation_id / task_id / ticket_id
    你的外部系統產生的任務 ID。
    用來查詢 job、approval ticket、背景任務或 callback 狀態。
```

`function_call.id` 是 ADK 對話與 resume 機制用的；`operation_id` 是你的業務系統用的。兩者都要保存，但用途不同。實務上，你通常會建立一張 mapping 表，把 `function_call.id`、`session_id`、`user_id`、`operation_id`、目前狀態、建立時間都存下來。這樣當外部任務完成時，client 才知道要把結果送回哪個 session，以及要用哪個 `FunctionResponse.id` 來 resume。

可以把 event interception 想成這個流程：

```text
runner.run_async()
    ↓
event contains function_call
    ↓
check event.long_running_tool_ids
    ↓
capture long_running_call
    ↓
capture matching function_response
    ↓
render UI / wait for external result
    ↓
later resume with updated function_response
```

這也是為什麼 `LongRunningFunctionTool` 不會自動產生 UI。它只是在 event stream 中暴露出「這裡有一個 pending long-running call」。真正的 UI、CLI prompt、Slack approval、webhook handler 或 polling mechanism，都是 client 端在偵測到這個 event 後自己決定怎麼處理。


#### 4. state resumption：如何把結果送回 runner

偵測到 long-running call 之後，下一個關鍵步驟是 **state resumption**。所謂 resumption，就是把外部取得的結果重新送回 ADK runner，讓 agent 從剛才暫停的位置繼續推理。這一步的重點不是送一個普通文字訊息，而是送回一個對應原始 tool call 的 `FunctionResponse`。

在前一段中，我們已經從 event stream 裡抓到了兩個東西：

```python
long_running_call = part.function_call
pending_response = part.function_response
```

其中 `long_running_call.id` 是原本 tool call 的 ID，而 `pending_response.id` 應該和它相同。這個 ID 是 ADK 用來對齊「之前哪個 tool call 還沒完成」的關鍵。因此 resume 時，最安全的做法是複製原本的 `pending_response`，只替換它的 `response` payload，不要自己從零建立一個新的 response。

例如，人類審批後選擇 approve：

```python
updated = pending_response.model_copy(deep=True)
updated.response = {
    "status": "approved",
}
```

如果是 reject：

```python
updated = pending_response.model_copy(deep=True)
updated.response = {
    "status": "rejected",
}
```

如果是外部 job 完成，也可以回填完整結果：

```python
updated = pending_response.model_copy(deep=True)
updated.response = {
    "status": "completed",
    "job_id": "job_123",
    "result": {
        "rows": 1000,
        "message": "Query finished successfully",
    },
}
```

接著，把更新後的 `FunctionResponse` 包成一個新的 `types.Content`，再送回 `runner.run_async()`：

```python
from google.genai import types

async for event in runner.run_async(
    session_id=session.id,
    user_id="user1",
    new_message=types.Content(
        role="user",
        parts=[
            types.Part(function_response=updated)
        ],
    ),
):
    if not event.content or not event.content.parts:
        continue

    text = "".join(p.text or "" for p in event.content.parts)
    if text:
        print(f"Agent: {text}")
```

這段看起來像是使用者又送了一個新訊息，但它不是一般文字訊息，而是一個 function response message。ADK runner 會看到這個 `function_response.id`，並把它對回先前那個 long-running tool call。然後，agent 就可以根據這個結果繼續原本的推理流程。

如果 approval 結果是：

```python
{"status": "approved"}
```

agent 可能會繼續呼叫真正的敏感工具，例如 `process_refund()`。

如果結果是：

```python
{"status": "rejected"}
```

agent 則應該停止後續動作，並告訴使用者操作已被拒絕。

如果結果是外部 job 的完成資料：

```python
{
    "status": "completed",
    "result": {
        "records": [...]
    }
}
```

agent 就可以把結果整理成自然語言回覆，或根據結果呼叫下一個 tool。

整個 resumption 的流程可以想成：

```text
pending_response captured
    ↓
human / worker / external system produces result
    ↓
copy pending_response
    ↓
replace updated.response
    ↓
send types.Part(function_response=updated) to runner
    ↓
runner matches function_response.id to original function_call.id
    ↓
agent resumes reasoning
```

這裡最常見的錯誤是自己建立一個新的 response，卻沒有保留原本的 `function_response.id`。如果 ID 對不上，ADK 無法知道這個結果屬於哪個 pending tool call，resume 就可能失敗，或被視為一個普通的新訊息。

另一個常見錯誤是把外部結果當成文字訊息送回去，例如：

```python
types.Content(
    role="user",
    parts=[types.Part(text="approved")]
)
```

這樣 agent 可能可以用語意猜到使用者批准了，但它不會被視為原本 long-running tool call 的正式完成結果。正確做法是送回 `function_response`，而不是普通 text。

因此，`LongRunningFunctionTool` 的 resume 核心原則是：**用同一個 function call ID，把外部結果包回 `FunctionResponse`，再交給 runner。** 這樣 ADK 才能維持 agent execution 的連續性，而不是重新開始一段不相關的對話。


#### 5. 兩種典型模式：HITL approval 與 external async job

`LongRunningFunctionTool` 最常見的使用模式有兩種。第一種是 **Human-in-the-Loop approval**，也就是人類審批。第二種是 **external async job**，也就是把長時間任務交給外部系統執行。兩者的共同點是：agent 呼叫 tool 之後，結果不會立刻完成，而是先回傳 pending 狀態，之後再由外部 decision 或外部 job result 回填給 runner。

第一種模式是 **HITL approval**。這種情境常出現在高風險操作中，例如退款、付款、刪除資料、修改 production database、發送正式 email、部署服務、變更雲端權限等。agent 可以判斷「需要執行某個敏感動作」，但它不能自己直接執行，而是必須先呼叫 approval tool。

概念上會長這樣：

```text id="6weymw"
User asks for refund
    ↓
Agent decides refund may be needed
    ↓
Agent calls ask_for_approval as LongRunningFunctionTool
    ↓
Tool returns {"status": "pending", "amount": 200, "purpose": "..."}
    ↓
Client renders approval UI
    ↓
Human approves or rejects
    ↓
Client sends FunctionResponse back to runner
    ↓
Agent resumes
    ↓
If approved, call process_refund
    ↓
If rejected, tell user the action was rejected
```

這裡的 `LongRunningFunctionTool` 不負責真的退款，而是負責建立一個 approval breakpoint。真正退款的 tool 應該是另一個普通 tool，例如 `process_refund()`。這樣做的好處是把「要求審批」和「執行敏感動作」分開，安全邊界更清楚。

範例結構可以是：

```python id="y33c3u"
from typing import Any
from google.adk.tools import LongRunningFunctionTool

def ask_for_approval(purpose: str, amount: float) -> dict[str, Any]:
    """Ask a human to approve this action before proceeding."""
    return {
        "status": "pending",
        "purpose": purpose,
        "amount": amount,
        "message": f"Approve refund of ${amount} for {purpose}?",
    }

def process_refund(purpose: str, amount: float) -> dict[str, str]:
    """Process the approved refund."""
    return {
        "status": "success",
        "message": f"Refunded ${amount} for {purpose}.",
    }
```

agent 的 instruction 則應該明確要求先審批：

```python id="cbe34y"
instruction = """
You handle refunds.
Always call ask_for_approval before calling process_refund.
If the approval response status is approved, call process_refund.
If the approval response status is rejected, tell the user the refund was rejected.
"""
```

這種 HITL 模式的重點是：人類的 approve / reject 不是一般文字，而是應該被包成原本 long-running call 的 `FunctionResponse` 回填。agent 看到正式的 function response 後，才進入下一步。

第二種模式是 **external async job**。這種情境適合長時間運算或外部工作，例如大型 SQL 查詢、批次資料處理、報表生成、資料匯出、embedding index 建立、模型訓練、影片轉檔、文件 OCR、雲端部署等。這些任務通常不適合在 agent process 裡同步等待，應該交給 worker、queue、job server 或 cloud service 處理。

概念流程如下：

```text id="1bs4cl"
User asks for a long report
    ↓
Agent calls submit_report_job as LongRunningFunctionTool
    ↓
Tool submits job to external backend
    ↓
Tool returns {"status": "pending", "job_id": "..."}
    ↓
Runner pauses / ends current turn
    ↓
Worker processes job in background
    ↓
Client polls job status or receives webhook
    ↓
Client sends final FunctionResponse back to runner
    ↓
Agent resumes and summarizes result
```

在這種模式裡，long-running tool 的 function 通常只做一件事：提交 job，然後立刻回傳 `job_id`。

```python id="0p9zor"
from typing import Any

def submit_report_job(report_type: str, user_id: str) -> dict[str, Any]:
    """Submit a long-running report generation job."""
    job_id = enqueue_report_job(
        report_type=report_type,
        user_id=user_id,
    )

    return {
        "status": "pending",
        "job_id": job_id,
        "message": "Report generation has started.",
        "estimated_time_seconds": 300,
    }
```

這裡的 `enqueue_report_job()` 代表你自己的外部 job system，例如 Celery、Cloud Tasks、Pub/Sub、Cloud Run Jobs、Kubernetes Job、RabbitMQ、Redis Queue，或資料庫裡的一張 jobs table。重點是不要在 `submit_report_job()` 裡真的同步產生整份報表，否則就失去 `LongRunningFunctionTool` 的意義。

external async job 模式通常需要一個 mapping table。因為你要把 ADK 的 tool call 和外部 job 關聯起來：

```text id="5pfwh2"
function_call_id  ↔  job_id  ↔  session_id  ↔  user_id
```

其中：

```text id="s7bsip"
function_call_id
    ADK 用來 resume 原本 tool call 的 ID。

job_id
    外部 job backend 用來追蹤任務的 ID。

session_id
    ADK 用來找到原本對話狀態。

user_id
    用來確認這個結果應該回到哪個使用者。
```

這個 mapping 很重要。當 worker 完成任務時，它通常只知道 `job_id`。但要把結果送回 ADK runner，你還需要知道原本的 `function_call_id` 和 `session_id`。因此 production 系統不應該只保存 job 狀態，也要保存 agent resume 所需的 metadata。

可以把兩種模式做一個簡單對照：

```text id="a9x5uw"
HITL approval:
    等待的是人類 decision。
    回填的是 approved / rejected。
    常用於高風險 side effect。

External async job:
    等待的是外部任務完成。
    回填的是 completed / failed / partial result。
    常用於長時間運算或外部系統操作。
```

雖然兩種模式看起來不同，但在 ADK 的機制上非常相似：都是先產生一個 pending `FunctionResponse`，讓 runner 暫停；再由 client 或外部系統把更新後的 `FunctionResponse` 送回 runner，讓 agent 繼續推理。

因此，`LongRunningFunctionTool` 的核心不是「任務本身是人工還是自動」，而是「這個 tool call 的結果是否需要跨時間、跨系統、跨 invocation 才能取得」。只要答案是 yes，這個 pattern 就適合使用。


#### 6. 與 FunctionTool / RequestInput / async task 的差異

`LongRunningFunctionTool` 很容易和其他幾種機制混在一起，尤其是一般 `FunctionTool`、workflow 裡的 `RequestInput`，以及 Python 自己的 `async task`。它們都可能和「等待」有關，但等待的層級完全不同。要正確使用 `LongRunningFunctionTool`，最重要的是先分清楚：你是在等待一個普通 function 完成，等待使用者提供 workflow input，還是在等待一個跨 invocation 的 tool call 結果。

一般 `FunctionTool` 適合一次呼叫就能完成的工具。agent 呼叫 tool，tool 執行，runner 把結果包成 `FunctionResponse` 回給 LLM，LLM 繼續推理。整個流程發生在同一輪 agent execution 裡。適合的例子包括查詢天氣、做單次 API call、查一筆資料、執行簡短計算、格式轉換、讀取小型檔案等。

```text id="v3xi7a"
FunctionTool:
    Agent calls tool
        ↓
    Tool returns result immediately
        ↓
    Runner sends FunctionResponse to LLM
        ↓
    Agent continues in the same run
```

`LongRunningFunctionTool` 則適合結果不能立刻取得的工具。它的第一次回傳通常只是 pending 狀態，例如 `job_id`、`ticket_id`、`operation_id`。runner 會把這個 long-running call 暴露給 client，client 之後再把最終結果送回來。換句話說，它把一次 tool call 拆成「啟動」和「完成」兩個階段。

```text id="zlxhew"
LongRunningFunctionTool:
    Agent calls tool
        ↓
    Tool starts external process
        ↓
    Tool returns pending response
        ↓
    Runner pauses / yields control to client
        ↓
    Human or external system returns final result later
        ↓
    Client sends FunctionResponse back to runner
        ↓
    Agent resumes
```

所以，如果一個任務只是慢一點，但仍然可以在同一輪 execution 裡合理等待完成，那它不一定需要 `LongRunningFunctionTool`。但如果任務需要人類審批、外部 callback、背景 worker、job queue、webhook、或跨 session 的 resume，那就應該考慮 `LongRunningFunctionTool`。

`RequestInput` 是另一種不同層級的 HITL 機制。它通常出現在 ADK workflow graph 裡，用來讓某個 workflow node 暫停，等待使用者輸入。它的語意是：「這個 workflow 步驟需要人類提供資料或回饋，才能進入下一個 node。」例如旅遊規劃 workflow 先產生 itinerary，再用 `RequestInput` 問使用者喜歡哪些活動，接著把使用者回覆傳給下一個 node。

```text id="d0uhpx"
RequestInput:
    Workflow reaches human input node
        ↓
    Node emits input request
        ↓
    User provides text / structured input
        ↓
    Workflow resumes to next node
```

因此，`RequestInput` 比較像 workflow graph 裡的人工輸入節點；`LongRunningFunctionTool` 則是 LLM agent tool call 層級的延遲完成機制。前者的中心是「workflow node 需要 input」，後者的中心是「某個 tool call 的結果暫時還沒完成」。

兩者也可以這樣區分：

```text id="6yc46r"
RequestInput:
    問使用者：「請提供資料或偏好。」
    例如：你想去哪個城市？你喜歡哪些行程？請輸入審核意見。

LongRunningFunctionTool:
    告訴使用者或外部系統：「這個 tool call 還在等待結果。」
    例如：退款正在等待批准；報表正在生成；SQL 查詢正在執行。
```

Python 的 `async task` 又是另一回事。`async def`、`await`、`asyncio.create_task()` 解決的是 Python runtime 裡的 concurrency 問題。它們讓程式在等待 I/O 時不阻塞 event loop。但它們不會自動幫 agent 保存 tool call identity，也不會自動建立 ADK 的 pause / resume 語義。

例如：

```python id="jmgutn"
async def fetch_data():
    result = await call_api()
    return result
```

這是一個 Python 非同步 function。它可以避免 blocking，但從 agent 的角度看，它仍然是一次普通 tool call：開始、等待、完成、回傳。只要它在同一輪 execution 裡完成，就不需要 `LongRunningFunctionTool`。

相反地，`LongRunningFunctionTool` 處理的是這種情境：

```text id="2ipfkv"
現在先提交任務。
結果可能 5 分鐘後、1 小時後，甚至明天才會回來。
中間可能需要 human approval 或 external callback。
agent 現在不能一直卡著等。
但結果回來時，agent 必須知道這是先前哪個 tool call 的結果。
```

這就是 `async task` 無法單獨解決的問題。你可以在 `LongRunningFunctionTool` 背後使用 async task、worker 或 queue，但 `LongRunningFunctionTool` 負責的是 agent-level state continuation，而不是 Python-level concurrency。

實務上可以用這個判斷：

```text id="q6uemw"
用 FunctionTool：
    工具能在同一輪 agent run 內完成。
    不需要人類審批。
    不需要外部 callback。
    不需要保存 pending state。

用 LongRunningFunctionTool：
    工具只能先啟動任務。
    結果需要稍後回填。
    需要 function_call.id 和 FunctionResponse.id 配對。
    需要 client / worker / human 參與 resume。

用 RequestInput：
    workflow graph 某個節點需要使用者輸入。
    重點是收集資料、偏好、意見或人工填寫內容。
    不一定代表有一個外部 job 正在跑。

用 async task：
    只是 Python 程式內部的非阻塞執行。
    解決 event loop blocking。
    不等於 ADK 的 agent pause / resume。
```

所以，`LongRunningFunctionTool` 的獨特性在於它橫跨了 LLM tool call、ADK event stream、client UI、外部 job system 和 state resumption。它不是單純的 function wrapper，而是一個讓 agent 能處理長生命週期 operation 的 coordination primitive。


#### 7. production 設計原則與常見陷阱

`LongRunningFunctionTool` 在 demo 裡看起來很簡單：包一個 function、回傳 pending、之後再把結果送回 runner。但在 production 系統裡，真正困難的部分不是 `LongRunningFunctionTool(func=...)` 這一行，而是你如何管理長任務的狀態、身分對應、錯誤處理、安全審批與重試機制。

第一個原則是：**不要把真正耗時的工作直接放在 long-running function 裡同步執行。** 這個 function 最好只負責提交任務，例如建立 job、寫入 queue、建立 approval ticket、呼叫外部 workflow，然後立刻回傳 pending 狀態。真正耗時的工作應該由 worker、job backend、Cloud Run Job、Celery、Pub/Sub、Cloud Tasks、Kubernetes Job 或其他外部執行系統處理。

比較好的設計是：

```text id="sdcc08"
LongRunningFunctionTool
    ↓
submit job / create ticket
    ↓
return {"status": "pending", "job_id": "..."}
    ↓
external worker processes job
    ↓
client / webhook receives final result
    ↓
send FunctionResponse back to runner
```

比較不好的設計是：

```text id="0mrr7w"
LongRunningFunctionTool
    ↓
time.sleep(600)
    ↓
run expensive SQL synchronously
    ↓
return final result after 10 minutes
```

後者會讓 agent process 被長時間占住，也失去 long-running tool 的主要價值。

第二個原則是：**一定要保存 identity mapping。** `LongRunningFunctionTool` 會牽涉至少兩種 ID：ADK 的 `function_call.id`，以及外部系統的 `job_id`、`task_id`、`ticket_id` 或 `operation_id`。前者用來 resume 原本的 tool call；後者用來查詢外部任務狀態。production 系統應該把這些資訊存在 durable store，而不是只放在 memory。

常見 mapping 會長這樣：

```text id="hn6j44"
function_call_id
function_name
session_id
user_id
operation_id
status
created_at
updated_at
expires_at
```

如果只保存 `job_id`，但沒有保存 `function_call_id`，等 worker 完成任務時，你會不知道要用哪個 `FunctionResponse.id` 回填。如果只保存 `function_call_id`，但沒有保存 `session_id`，你會不知道結果該送回哪個 agent session。這些 mapping 是 resume 能不能成功的關鍵。

第三個原則是：**pending state 必須可恢復。** 在 demo 裡用全域變數 `running_tasks = {}` 或 `callback_registry = {}` 很方便，但 production 不應該依賴 process memory。因為 server 可能重啟，container 可能被 autoscaling 移除，worker 可能失敗，webhook 可能重送。如果 pending state 只存在記憶體裡，任務還在跑，但 agent 已經失去上下文，最後結果就無法回填。

更可靠的做法是使用：

```text id="cfnqjt"
PostgreSQL
Redis
Firestore
DynamoDB
Cloud SQL
任務 queue 自帶的 metadata store
```

至少要能保存：任務狀態、外部 job ID、ADK function call ID、session ID、使用者 ID、建立時間、最後更新時間、錯誤訊息與最終結果。

第四個原則是：**要設計 idempotency。** 長任務系統一定會遇到重試。client 可能重送同一個 approval response，webhook 可能因 timeout 重送，worker 可能重跑同一個 job，runner 也可能在網路錯誤後再次接收相同結果。如果你的系統沒有 idempotency，同一筆退款可能被處理兩次，同一封 email 可能被寄出兩次，同一個 production change 可能被套用兩次。

因此，每個 side-effecting operation 都應該有 idempotency key，例如：

```text id="so729p"
operation_id = "refund_txn_123_attempt_1"
idempotency_key = function_call_id 或 business_operation_id
```

真正執行副作用的 backend，例如 refund API、email sender、database mutator，也應該檢查這個 key，確保同一個 operation 只執行一次。

第五個原則是：**approval 不應該只靠 agent instruction。** 如果是 HITL approval 場景，不能只在 prompt 裡寫「Always ask for approval before refund」。這對模型行為有幫助，但不是安全邊界。真正的安全邊界應該在工具或 backend 層檢查 approval record。

比較安全的流程是：

```text id="m1t0c5"
Agent requests approval
    ↓
Approval service creates signed approval record
    ↓
Human approves in authenticated UI
    ↓
Backend stores approver, timestamp, scope, amount limit
    ↓
Agent receives approved FunctionResponse
    ↓
process_refund checks approval record before executing
```

也就是說，即使 agent 因為錯誤、prompt injection 或 hallucination 直接呼叫 `process_refund()`，backend 也應該拒絕沒有合法 approval token 的操作。

第六個原則是：**response schema 要穩定。** 不要讓 long-running tool 回傳隨意格式。最好定義清楚的狀態欄位，例如：

```python id="rv9e8d"
{
    "status": "pending",
    "operation_id": "job_123",
    "message": "Report generation has started.",
    "estimated_time_seconds": 300,
    "next_check_after_seconds": 30,
}
```

最終完成時：

```python id="1nncuo"
{
    "status": "completed",
    "operation_id": "job_123",
    "result": {
        "url": "https://...",
        "summary": "Report completed successfully."
    },
}
```

失敗時：

```python id="va27fo"
{
    "status": "failed",
    "operation_id": "job_123",
    "error": {
        "code": "TIMEOUT",
        "message": "The report job exceeded its time limit.",
        "retryable": True,
    },
}
```

HITL approval 則可以設計成：

```python id="aa8w62"
{
    "status": "approved",
    "approval_id": "approval_456",
    "approver": "manager@example.com",
    "approved_at": "2026-07-17T12:00:00Z",
    "scope": {
        "amount": 200,
        "purpose": "duplicate charge refund"
    },
}
```

這樣 agent 比較容易根據 `status` 做正確決策，client 和 backend 也比較容易維護。

第七個原則是：**要處理 timeout 與 expiration。** 不是所有 pending task 都會完成。人類可能不回覆，worker 可能 crash，外部 API 可能沒有 callback。因此每個 pending operation 都應該有 `expires_at` 或 timeout policy。過期後，client 應該能回填：

```python id="fmk5y0"
{
    "status": "expired",
    "operation_id": "approval_456",
    "message": "Approval request expired without a decision."
}
```

agent 收到後應該停止等待，並告訴使用者任務已過期或需要重新發起。

第八個原則是：**要做 audit log。** `LongRunningFunctionTool` 經常用在高風險流程，所以需要可追蹤性。至少應該記錄：

```text id="h95je3"
誰提出請求
agent 產生了什麼 tool call
tool call arguments 是什麼
何時建立 pending operation
誰批准或拒絕
批准範圍是什麼
何時執行真正副作用
最後結果是成功還是失敗
```

這些資料不只是 debug 用，也可能是 compliance、資安審計、事故調查需要的紀錄。

常見陷阱可以總結成幾個：

```text id="lvn4bc"
1. 在 long-running function 裡同步等待很久。
2. 沒有保存 function_call_id，導致無法 resume。
3. 把 pending state 放在 memory，server 重啟後遺失。
4. 用文字訊息回覆 approved，而不是 FunctionResponse。
5. 沒有 idempotency，導致副作用重複執行。
6. 只靠 prompt 要求 approval，backend 沒有真正檢查。
7. 沒有 timeout，pending operation 永遠掛著。
8. response schema 不穩定，agent 難以可靠判斷狀態。
```

所以，production 版的 `LongRunningFunctionTool` 應該被視為一個分散式工作流設計問題，而不只是 ADK 裡的一個 tool wrapper。你需要同時設計 agent 行為、client event interception、外部 job backend、狀態儲存、resume protocol、安全檢查與審計紀錄。這些配套做好之後，`LongRunningFunctionTool` 才能真正安全地支援長生命週期任務。


#### 8. 總結

`LongRunningFunctionTool` 是 Google ADK 裡用來處理「長生命週期 tool call」的核心機制。它解決的不是單純的 Python 非同步問題，而是 agent runtime 的 pause / resume 問題：當 agent 呼叫某個工具，但這個工具的結果無法在同一輪 execution 裡立刻取得時，ADK 可以先暫停，等到外部結果回來後，再讓 agent 接著原本的上下文繼續推理。

它最重要的設計思想是把一次 tool call 拆成兩個階段。第一階段是啟動任務，例如建立 approval request、提交 background job、建立 ticket、呼叫外部 workflow，然後回傳 pending 狀態。第二階段是由 human、client、worker、webhook 或其他外部系統，把最終結果包成同一個 tool call 對應的 `FunctionResponse`，再送回 runner。

因此，`LongRunningFunctionTool` 的關鍵不只是：

```python id="zq1m6m"
LongRunningFunctionTool(func=my_tool)
```

而是整個 surrounding architecture：

```text id="ckq0dv"
agent instruction
    ↓
long-running tool design
    ↓
event stream interception
    ↓
function_call.id / FunctionResponse.id mapping
    ↓
human UI or external job system
    ↓
state resumption
    ↓
final agent response
```

在 Human-in-the-Loop 場景中，它可以讓 agent 在退款、付款、刪除資料、發送 email、修改 production system 前強制等待人工審批。在 external async job 場景中，它可以讓 agent 啟動長時間任務，例如大型 SQL 查詢、報表生成、批次處理、模型訓練或資料匯出，然後在結果完成後接續推理。

與一般 `FunctionTool` 相比，`LongRunningFunctionTool` 的獨特點是它允許 tool call 跨時間完成。與 `RequestInput` 相比，它不是 workflow node 層級的人工輸入，而是 LLM tool call 層級的延遲完成。與 Python `async task` 相比，它不是只解決 event loop blocking，而是解決 agent 如何記住「剛才那個 tool call 還沒完成」並在未來恢復執行的問題。

實作時最重要的技術細節是：client 必須在 `runner.run_async()` 的 event stream 中偵測 `event.long_running_tool_ids`，保存對應的 `function_call.id` 和初始 `pending_response`。等到外部結果出現時，不要用普通文字訊息回覆，而是要複製原本的 `pending_response`，更新它的 `response` payload，再以 `types.Part(function_response=updated)` 的形式送回 runner。這樣 ADK 才能把結果對回原始 tool call。

在 production 中，`LongRunningFunctionTool` 應該搭配 durable state store、idempotency key、timeout、retry、audit log、approval record 和清楚的 response schema。不要把長時間工作直接塞在 tool function 裡同步執行，也不要只依賴 process memory 保存 pending state。真正可靠的設計應該把長任務交給外部 job backend，把 agent process 保持成 orchestration layer。

一句話總結：`LongRunningFunctionTool` 讓 ADK agent 能夠安全處理「現在啟動、稍後完成」的工具呼叫。它把 agent 從單輪同步 tool execution 擴展成可暫停、可恢復、可人工審批、可接外部 job system 的長流程 agent 架構。

````python
import time
import uuid
from typing import Any

from google.adk.agents import Agent
from google.adk.runners import Runner
from google.adk.tools import LongRunningFunctionTool
from google.adk.sessions import InMemorySessionService
from google.genai import types

APP_NAME = "multi-hitl-refund-demo"
USER_ID = "user1"
SESSION_ID = "s1"

execution_path: list[str] = []
manager_approvals: dict[str, dict[str, Any]] = {}
compliance_reviews: dict[str, dict[str, Any]] = {}
refund_records: dict[str, dict[str, Any]] = {}
````

### Cell 2：log helper

````python
def log_step(message: str) -> None:
    timestamp = time.strftime("%H:%M:%S")
    line = f"[{timestamp}] {message}"
    execution_path.append(line)
    print(line)
````

### Cell 3：第一個 **LongRunningFunctionTool，主管審批**

````python
def request_manager_approval(
    transaction_id: str,
    customer_id: str,
    amount: float,
    reason: str,
) -> dict[str, Any]:
    """Request manager approval before processing a high-value refund."""
    approval_id = f"mgr_{uuid.uuid4().hex[:8]}"

    payload = {
        "status": "pending",
        "approval_type": "manager_approval",
        "approval_id": approval_id,
        "transaction_id": transaction_id,
        "customer_id": customer_id,
        "amount": amount,
        "reason": reason,
        "message": f"Manager approval required for refund ${amount:.2f}.",
        "created_at": time.strftime("%Y-%m-%d %H:%M:%S"),
    }

    manager_approvals[approval_id] = payload

    log_step(
        f"TOOL request_manager_approval -> pending {approval_id} "
        f"for ${amount:.2f} on {transaction_id}"
    )

    return payload
````

這個 tool 代表「第一個人工 gate」。它不退款，只建立一個主管 approval request，回傳 pending。

### Cell 4：第二個 LongRunningFunctionTool，合規審查

````python
def request_compliance_review(
    transaction_id: str,
    customer_id: str,
    amount: float,
    reason: str,
) -> dict[str, Any]:
    """Request compliance review before processing suspicious refunds."""
    review_id = f"comp_{uuid.uuid4().hex[:8]}"

    payload = {
        "status": "pending",
        "approval_type": "compliance_review",
        "review_id": review_id,
        "transaction_id": transaction_id,
        "customer_id": customer_id,
        "amount": amount,
        "reason": reason,
        "message": (
            "Compliance review required because the refund appears "
            "suspicious or high-risk."
        ),
        "created_at": time.strftime("%Y-%m-%d %H:%M:%S"),
    }

    compliance_reviews[review_id] = payload

    log_step(
        f"TOOL request_compliance_review -> pending {review_id} "
        f"for {transaction_id}"
    )

    return payload
````

這個 tool 代表「第二個人工 gate」。它和主管審批不同，所以 client 必須根據 function name 分別處理。

### Cell 5：真正有 side effect 的普通 tool

 side effect 意思是：一個 function 執行後，不只是回傳一個值，而是改變了外部世界或系統狀態。

````python
def process_refund(
    transaction_id: str,
    customer_id: str,
    amount: float,
    reason: str,
    manager_approval_id: str,
    compliance_review_id: str,
) -> dict[str, Any]:
    """Process the refund only after all required approvals are granted."""
    manager_record = manager_approvals.get(manager_approval_id)
    compliance_record = compliance_reviews.get(compliance_review_id)

    if not manager_record or manager_record.get("status") != "approved":
        return {
            "status": "blocked",
            "message": f"Manager approval {manager_approval_id} is not approved.",
        }

    if not compliance_record or compliance_record.get("status") != "approved":
        return {
            "status": "blocked",
            "message": f"Compliance review {compliance_review_id} is not approved.",
        }

    refund_id = f"refund_{uuid.uuid4().hex[:8]}"

    record = {
        "status": "success",
        "refund_id": refund_id,
        "transaction_id": transaction_id,
        "customer_id": customer_id,
        "amount": amount,
        "reason": reason,
        "manager_approval_id": manager_approval_id,
        "compliance_review_id": compliance_review_id,
        "processed_at": time.strftime("%Y-%m-%d %H:%M:%S"),
        "message": f"Refunded ${amount:.2f} for {reason}.",
    }

    refund_records[refund_id] = record

    log_step(
        f"TOOL process_refund -> processed {refund_id} "
        f"after manager + compliance approval"
    )

    return record
````

這裡故意讓 process_refund() 同時檢查兩個 approval records。這比只靠 agent instruction 更接近 production，因為真正的 side-effect tool 自己也會防止未授權執行。

這個 tool 代表真正執行退款的動作。它不是單純問人類「要不要批准」，而是實際改變系統狀態。

### Cell 6：建立 agent

````python
agent = Agent(
    name="refund_agent",
    model="gemini-2.5-flash",
    instruction="""
You are a refund operations agent.

You must follow this strict step-by-step policy.

Definitions:
- request_manager_approval is a long-running approval tool.
- request_compliance_review is a long-running approval tool.
- process_refund is the final side-effecting tool.

Critical execution rule:
You may call at most ONE long-running approval tool per assistant turn.

Refund policy:
1. If the refund amount is $100 or more, call request_manager_approval first.
2. After calling request_manager_approval, if its response status is "pending",
   stop immediately. Do not call request_compliance_review. Do not call process_refund.
   Tell the user that manager approval is pending.
3. Only after you receive a later function_response where manager approval status is
   "approved", decide whether compliance review is required.
4. If the case involves suspicious, fraud, abuse, chargeback, or high-risk reasons,
   call request_compliance_review.
5. After calling request_compliance_review, if its response status is "pending",
   stop immediately. Do not call process_refund.
   Tell the user that compliance review is pending.
6. Only after you receive a later function_response where compliance review status is
   "approved", call process_refund.
7. If either approval is "rejected", do not call any more tools. Tell the user the refund
   cannot be processed.
8. Never call process_refund unless both required approvals are approved.
9. Keep transaction_id, customer_id, amount, and reason consistent across all calls.

Important:
A "pending" response means WAIT. It does not mean continue to the next tool.
""",
    tools=[
        LongRunningFunctionTool(func=request_manager_approval),
        LongRunningFunctionTool(func=request_compliance_review),
        process_refund,
    ],
)
````

這裡有兩個 LongRunningFunctionTool。重點是 client 端不能假設只有一個 pending call。

### Cell 7：event parsing helpers

````python
def extract_text(event) -> str:
    if not event.content or not event.content.parts:
        return ""

    return "".join(part.text or "" for part in event.content.parts)
````

````python
def debug_event_parts(event) -> None:
    if event.long_running_tool_ids:
        log_step(f"EVENT long_running_tool_ids = {list(event.long_running_tool_ids)}")

    if not event.content or not event.content.parts:
        return

    for part in event.content.parts:
        if part.function_call:
            log_step(
                f"EVENT function_call -> "
                f"name={part.function_call.name} "
                f"id={part.function_call.id}"
            )

        if part.function_response:
            log_step(
                f"EVENT function_response -> "
                f"name={part.function_response.name} "
                f"id={part.function_response.id} "
                f"response={part.function_response.response}"
            )
````

````python
def capture_long_running_items(event) -> list[dict[str, Any]]:
    if not event.content or not event.content.parts:
        return []

    if not event.long_running_tool_ids:
        return []

    captured = []

    for part in event.content.parts:
        if (
            part.function_call
            and part.function_call.id in event.long_running_tool_ids
        ):
            captured.append(
                {
                    "kind": "call",
                    "id": part.function_call.id,
                    "name": part.function_call.name,
                    "object": part.function_call,
                }
            )

        if (
            part.function_response
            and part.function_response.id in event.long_running_tool_ids
        ):
            captured.append(
                {
                    "kind": "response",
                    "id": part.function_response.id,
                    "name": part.function_response.name,
                    "object": part.function_response,
                }
            )

    return captured
````

這個 helper 是整個新版設計的核心。它不再只回傳一個 long_running_call，而是回傳一批 long-running items。每個 item 都有：

```python
kind: call 或 response
id: function call id
name: function name
object: 原始 function_call 或 function_response
```

這樣後面就可以根據 name dispatch。

### Cell 8：pending registry 與 event 收集

````python
async def run_until_first_pending_call(
    runner,
    session_id: str,
    user_id: str,
    message,
):
    pending_calls: dict[str, dict[str, Any]] = {}

    async for event in runner.run_async(
        session_id=session_id,
        user_id=user_id,
        new_message=message,
    ):
        debug_event_parts(event)

        if event.content and event.content.parts:
            # Pass 1: capture long-running function calls.
            for part in event.content.parts:
                if (
                    part.function_call
                    and event.long_running_tool_ids
                    and part.function_call.id in event.long_running_tool_ids
                ):
                    call_id = part.function_call.id

                    pending_calls[call_id] = {
                        "id": call_id,
                        "name": part.function_call.name,
                        "function_call": part.function_call,
                        "function_response": None,
                    }

                    log_step(
                        f"CLIENT -> captured long-running call "
                        f"name={part.function_call.name} id={call_id}"
                    )

            # Pass 2: capture the matching function response.
            for part in event.content.parts:
                if not part.function_response:
                    continue

                response_id = part.function_response.id

                if response_id in pending_calls:
                    pending_calls[response_id]["function_response"] = (
                        part.function_response
                    )

                    log_step(
                        f"CLIENT -> captured pending response "
                        f"name={part.function_response.name} id={response_id}"
                    )

                    # Critical:
                    # Stop at the first complete pending long-running call.
                    # Do not keep consuming the stream, otherwise the agent may
                    # call another long-running tool in the same runner round.
                    return pending_calls[response_id]

                if part.function_response.name in LONG_RUNNING_RESOLVERS:
                    log_step(
                        f"CLIENT -> captured long-running response by name "
                        f"name={part.function_response.name} id={response_id}"
                    )

                    return {
                        "id": response_id,
                        "name": part.function_response.name,
                        "function_call": None,
                        "function_response": part.function_response,
                    }

        text = extract_text(event)
        if text:
            print(f"\nAgent: {text}\n")

    return None
````

這裡會收集本次 run 裡所有 long-running call。即使目前 ADK 實務上常常是一個 long-running call 就暫停，這個寫法也能支援多個 call，並且可以跨不同 long-running function 做 dispatch。

### Cell 9：依 function name 做不同處理

````python
def resolve_manager_approval(function_response):
    payload = function_response.response

    print("\n" + "=" * 72)
    print("MANAGER APPROVAL REQUIRED")
    print("=" * 72)
    print(f"Approval ID : {payload.get('approval_id')}")
    print(f"Transaction : {payload.get('transaction_id')}")
    print(f"Customer    : {payload.get('customer_id')}")
    print(f"Amount      : ${payload.get('amount')}")
    print(f"Reason      : {payload.get('reason')}")
    print(f"Message     : {payload.get('message')}")
    print("=" * 72)

    choice = input("Manager approve? [Y/n]: ").strip().lower()
    decision = "approved" if choice != "n" else "rejected"

    approval_id = payload["approval_id"]
    manager_approvals[approval_id]["status"] = decision
    manager_approvals[approval_id]["decided_by"] = "manager_colab"
    manager_approvals[approval_id]["decided_at"] = time.strftime("%Y-%m-%d %H:%M:%S")

    updated = function_response.model_copy(deep=True)
    updated.response = dict(manager_approvals[approval_id])

    log_step(f"HUMAN manager -> {decision} {approval_id}")

    return updated
````

````python
def resolve_compliance_review(function_response):
    payload = function_response.response

    print("\n" + "=" * 72)
    print("COMPLIANCE REVIEW REQUIRED")
    print("=" * 72)
    print(f"Review ID   : {payload.get('review_id')}")
    print(f"Transaction : {payload.get('transaction_id')}")
    print(f"Customer    : {payload.get('customer_id')}")
    print(f"Amount      : ${payload.get('amount')}")
    print(f"Reason      : {payload.get('reason')}")
    print(f"Message     : {payload.get('message')}")
    print("=" * 72)

    choice = input("Compliance approve? [Y/n]: ").strip().lower()
    decision = "approved" if choice != "n" else "rejected"

    review_id = payload["review_id"]
    compliance_reviews[review_id]["status"] = decision
    compliance_reviews[review_id]["decided_by"] = "compliance_colab"
    compliance_reviews[review_id]["decided_at"] = time.strftime("%Y-%m-%d %H:%M:%S")

    updated = function_response.model_copy(deep=True)
    updated.response = dict(compliance_reviews[review_id])

    log_step(f"HUMAN compliance -> {decision} {review_id}")

    return updated
````

這兩個 resolver 是不同的「人工介面」。在真實系統裡，一個可能是 manager approval UI，另一個可能是 compliance dashboard。Colab 裡用 input() 模擬。

### Cell 10：dispatcher

````python
LONG_RUNNING_RESOLVERS = {
    "request_manager_approval": resolve_manager_approval,
    "request_compliance_review": resolve_compliance_review,
}
````

````python
def resolve_pending_call(pending_call: dict[str, Any]):
    tool_name = pending_call["name"]
    function_response = pending_call["function_response"]

    resolver = LONG_RUNNING_RESOLVERS.get(tool_name)

    if not resolver:
        raise ValueError(f"No resolver registered for long-running tool: {tool_name}")

    log_step(f"CLIENT -> dispatching resolver for {tool_name}")

    return resolver(function_response)
````

這就是你要的「capture 是哪一個 function，然後才能做對應事情」。不是所有 long-running call 都用同一個 Approve? [Y/n]，而是先看 tool name，再選對應 resolver。

### Cell 11：main loop，支援多次 long-running pause

````python
async def main():
    execution_path.clear()
    manager_approvals.clear()
    compliance_reviews.clear()
    refund_records.clear()

    session_service = InMemorySessionService()
    session = await session_service.create_session(
        app_name=APP_NAME,
        user_id=USER_ID,
        session_id=SESSION_ID,
    )

    runner = Runner(
        agent=agent,
        app_name=APP_NAME,
        session_service=session_service,
    )

    next_message = types.Content(
        role="user",
        parts=[
            types.Part(
                text=(
                    "Refund $249.99 for customer cust_8821. "
                    "The duplicate charge is on transaction txn_ABC123. "
                    "The case is suspicious because the customer reported possible fraud."
                )
            )
        ],
    )

    log_step("USER -> requested suspicious high-value refund")

    max_rounds = 6

    for round_index in range(1, max_rounds + 1):
        log_step(f"CLIENT -> starting runner round {round_index}")

        pending_call = await run_until_first_pending_call(
            runner=runner,
            session_id=session.id,
            user_id=USER_ID,
            message=next_message,
        )

        if not pending_call:
            log_step("CLIENT -> no pending long-running call; workflow complete")
            break

        updated_response = resolve_pending_call(pending_call)

        next_message = types.Content(
            role="user",
            parts=[
                types.Part(function_response=updated_response)
            ],
        )

        log_step(
            f"CLIENT -> prepared FunctionResponse for resume "
            f"name={pending_call['name']} id={pending_call['id']}"
        )

    print("\n" + "=" * 72)
    print("FINAL EXECUTION PATH")
    print("=" * 72)
    for i, step in enumerate(execution_path, start=1):
        print(f"{i:02d}. {step}")

    print("\n" + "=" * 72)
    print("MANAGER APPROVALS")
    print("=" * 72)
    print(manager_approvals)

    print("\n" + "=" * 72)
    print("COMPLIANCE REVIEWS")
    print("=" * 72)
    print(compliance_reviews)

    print("\n" + "=" * 72)
    print("REFUND RECORDS")
    print("=" * 72)
    print(refund_records)
````

#### event stream 偵測程式碼詳解

這段程式的目的，是在 `runner.run_async()` 產生的事件流裡，找出兩個東西：

```python
long_running_call = None
pending_response = None
```

`long_running_call` 用來保存「agent 想呼叫的那個 long-running tool call」。

`pending_response` 用來保存「這個 long-running tool 初次回傳的 pending response」。

兩者必須配對起來，後面才能把人類輸入或外部任務結果送回 runner，讓 agent resume。

---

#### 第一部分：初始化狀態

```python
long_running_call = None
pending_response = None
```

這兩個變數一開始都是 `None`，因為此時我們還沒有看到任何 event。

`long_running_call` 最後會長得像這樣：

```python
FunctionCall(
    id="call_abc123",
    name="ask_for_approval",
    args={
        "purpose": "duplicate charge on txn_123",
        "amount": 200
    }
)
```

它代表模型決定要呼叫某個 tool，而且這個 tool 是 long-running tool。

`pending_response` 最後會長得像這樣：

```python
FunctionResponse(
    id="call_abc123",
    name="ask_for_approval",
    response={
        "status": "pending",
        "purpose": "duplicate charge on txn_123",
        "amount": 200
    }
)
```

注意兩者的 `id` 相同：

```text
function_call.id      == "call_abc123"
function_response.id  == "call_abc123"
```

這個相同的 ID 是整個 pause / resume 機制的核心。它讓 ADK 知道：「這個 response 是針對剛才那個 tool call 的回覆。」

---

#### 第二部分：開始讀取 event stream

```python
async for event in runner.run_async(
    session_id=session.id,
    user_id="user1",
    new_message=user_msg,
):
```

`runner.run_async()` 不是一次回傳最終答案，而是逐步產生一連串 `Event`。

一個 event 可能包含很多種內容，例如：

```text
agent text
function call
function response
long-running tool metadata
error
state update
```

所以這裡使用：

```python
async for event in ...
```

代表我們逐一讀取 agent execution 過程中的事件。

這一點很重要：`LongRunningFunctionTool` 的偵測不是發生在最後答案，而是發生在 streaming event 裡。

---

#### 第三部分：確認 event 有內容

```python
if event.content and event.content.parts:
```

這行是在做防呆。

不是每個 event 都一定有 `content`，也不是每個 `content` 都一定有 `parts`。所以在讀取 `event.content.parts` 前，先確認它存在。

`event.content.parts` 是一個 list。每個 `part` 可能是文字、function call、function response，或其他型態。

概念上它可能像這樣：

```python
event.content.parts = [
    Part(function_call=...),
    Part(function_response=...),
    Part(text="...")
]
```

所以後面要逐一檢查。

---

#### 第四部分：逐一檢查 event 裡的每個 part

```python
for part in event.content.parts:
```

這行代表：這個 event 裡可能同時包含多個部分，我們要一個一個看。

每個 `part` 可能有：

```python
part.text
part.function_call
part.function_response
```

但通常同一個 part 只會有其中一種主要內容。

例如：

```python
part.function_call
```

代表 agent 想呼叫 tool。

```python
part.function_response
```

代表 tool 已經回傳結果。

---

#### 第五部分：偵測 long-running function call

這段是第一個核心判斷：

```python
if (
    not long_running_call
    and part.function_call
    and event.long_running_tool_ids
    and part.function_call.id in event.long_running_tool_ids
):
    long_running_call = part.function_call
```

拆開來看。

第一個條件：

```python
not long_running_call
```

意思是：我們還沒有捕捉到 long-running call。

這是為了避免重複覆蓋。因為 event stream 可能有多個 event，也可能有多個 part。只要我們已經找到第一個 long-running call，就不再重複設定。

第二個條件：

```python
part.function_call
```

意思是：這個 part 必須是一個 function call。

也就是 agent 正在說：「我要呼叫某個 tool。」

例如：

```python
ask_for_approval(
    purpose="duplicate charge on txn_123",
    amount=200
)
```

第三個條件：

```python
event.long_running_tool_ids
```

意思是：這個 event 裡必須有 ADK 標記的 long-running tool call ID。

普通 `FunctionTool` 不會出現在這裡。只有 `LongRunningFunctionTool` 會被 ADK 放進 `event.long_running_tool_ids`。

它可能長這樣：

```python
event.long_running_tool_ids = {"call_abc123"}
```

或類似 list / set 的結構：

```python
event.long_running_tool_ids = ["call_abc123"]
```

第四個條件：

```python
part.function_call.id in event.long_running_tool_ids
```

這是最關鍵的一行。

它的意思是：目前這個 function call 的 ID，是否被 ADK 標記為 long-running？

例如：

```python
part.function_call.id = "call_abc123"
event.long_running_tool_ids = ["call_abc123"]
```

那麼條件成立。

於是：

```python
long_running_call = part.function_call
```

我們就把這個 function call 存起來。

---

#### 第六部分：為什麼不能只檢查 `part.function_call`

不能只寫：

```python
if part.function_call:
    long_running_call = part.function_call
```

因為 agent 可能同時有很多普通 tools。

例如 refund agent 可能有：

```text
ask_for_approval     # long-running tool
process_refund       # ordinary tool
lookup_transaction   # ordinary tool
```

`part.function_call` 只能告訴你「agent 要呼叫 tool」，但不能告訴你「這個 tool 是否需要暫停」。

真正區分 long-running tool 的是：

```python
event.long_running_tool_ids
```

所以這個判斷必須同時檢查：

```python
part.function_call.id in event.long_running_tool_ids
```

這樣才知道它是需要外部介入的 call。

---

#### 第七部分：捕捉 matching function response

第二個核心判斷是：

```python
if (
    long_running_call
    and part.function_response
    and part.function_response.id == long_running_call.id
):
    pending_response = part.function_response
```

這段是在找「剛才那個 long-running tool call 的初始 response」。

拆開來看。

第一個條件：

```python
long_running_call
```

意思是：我們已經找到 long-running function call。

如果還沒找到 call，就不應該找 response。因為我們還不知道要配對哪一個 ID。

第二個條件：

```python
part.function_response
```

意思是：這個 part 必須是一個 function response。

也就是 tool 已經回傳了一個結果。

對 `LongRunningFunctionTool` 來說，這通常是 pending 狀態：

```python
{
    "status": "pending",
    "purpose": "duplicate charge on txn_123",
    "amount": 200
}
```

第三個條件：

```python
part.function_response.id == long_running_call.id
```

這是配對檢查。

它確認這個 response 是不是屬於剛才那個 long-running call。

例如：

```python
long_running_call.id = "call_abc123"
part.function_response.id = "call_abc123"
```

成立。

於是：

```python
pending_response = part.function_response
```

我們就把 pending response 存下來。

---

#### 第八部分：為什麼需要保存 `pending_response`

後面 resume 的時候，需要這個 `pending_response`。

不是因為它的 `response` payload 很重要，而是因為它保存了這些 metadata：

```text
function_response.id
function_response.name
原始 function call 對應關係
```

resume 時通常會這樣做：

```python
updated = pending_response.model_copy(deep=True)
updated.response = {
    "status": "approved"
}
```

也就是：

1. 複製原本的 `pending_response`
2. 保留原本的 `id`
3. 保留原本的 `name`
4. 只改 `response`
5. 再送回 runner

這樣 ADK 才知道：

```text
這個 approved response 是針對 call_abc123 的回覆
```

如果你自己從零建立一個 response，卻沒有正確設定 ID，就可能無法 resume。

---

#### 第九部分：整段程式的語意

整段程式可以翻譯成人話：

```text
我開始跑 agent。

每收到一個 event，我就檢查它有沒有內容。

如果 event 裡有 function call，
而且這個 function call 的 ID 被 ADK 標記為 long-running，
我就把它存成 long_running_call。

接著我繼續看 event 裡有沒有 function response。

如果某個 function response 的 ID
和剛才 long_running_call 的 ID 一樣，
我就把它存成 pending_response。

之後我就可以把 pending_response 顯示給使用者，
讓使用者 approve / reject，
再把更新後的 function_response 送回 runner。
```

假設使用者說：

```text

```

agent 產生 function call：

```python
part.function_call = FunctionCall(

        "purpose": "duplicate charge on txn_123",
        "amount": 200

```

同一個 event 或後續 event 裡，ADK 標記：

```python
event.long_running_tool_ids = ["call_abc123"]
```

所以這段成立：

```python
part.function_call.id in event.long_running_tool_ids
```

於是：

```python
long_running_call = part.function_call
```

接著 tool 回傳：


因為：

#### 第十一部分：更清楚的版本

原始寫法比較精簡。如果要教學，我會改寫成幾個 helper functions，語意會更清楚：

```python
def is_long_running_function_call(event, part) -> bool:
    return (
        part.function_call is not None
        and event.long_running_tool_ids is not None
        and part.function_call.id in event.long_running_tool_ids
    )
```

這個 function 專門判斷目前的 `part` 是否是一個 long-running function call。

```python
def is_matching_function_response(part, function_call) -> bool:
    return (
        function_call is not None
        and part.function_response is not None
        and part.function_response.id == function_call.id
    )
```

這個 function 專門判斷目前的 `part` 是否是對應那個 call 的 response。

然後主流程變成：

```python
long_running_call = None
pending_response = None

async for event in runner.run_async(
    session_id=session.id,
    user_id="user1",
    new_message=user_msg,
):
    if not event.content or not event.content.parts:
        continue

    for part in event.content.parts:
        if long_running_call is None and is_long_running_function_call(event, part):
            long_running_call = part.function_call

        if is_matching_function_response(part, long_running_call):
            pending_response = part.function_response
```

這樣可讀性更高，也比較適合放進 production code。

核心邏輯沒有變：先找 long-running `function_call`，再找同 ID 的 `function_response`。


#### `if event.content and event.content.parts:`

不建議只寫：

```python
if event.content.parts:
```

因為如果某個 `event` 沒有 `content`，這行會直接噴錯：

```text
AttributeError: 'NoneType' object has no attribute 'parts'
```

也就是說，當：

```python
event.content is None
```

時，你不能再往下取：

```python
event.content.parts
```

所以原本這行：

```python
if event.content and event.content.parts:
```

比較安全。它利用 Python 的 short-circuit evaluation：

```python
if event.content and event.content.parts:
```

Python 會先檢查：

```python
event.content
```

只有當 `event.content` 不是 `None` 時，才會繼續檢查：

```python
event.content.parts
```

所以這行等價於比較完整的寫法：

```python
if event.content is not None and event.content.parts:
```

如果你想寫得更清楚，我會建議用這種版本：

```python
if not event.content or not event.content.parts:
    continue

for part in event.content.parts:
    ...
```

這種 guard clause 寫法在 production code 裡通常更乾淨。意思是：如果這個 event 沒有 content，或 content 沒有 parts，就跳過這個 event。

所以結論是：

```python
if event.content.parts:
```

只有在你非常確定每個 `event` 都一定有 `content` 時才安全。但在 ADK event stream 裡，不能做這個假設。建議保留：

```python
if event.content and event.content.parts:
```

或改成更清楚的：

```python
if not event.content or not event.content.parts:
    continue
```


### LongRunningFunctionTool RequestInput async task


用四個彼此正交的維度來看：

1. 執行是否跨越單次 invocation
2. continuation state 由誰保存
3. 等待的是資料、操作結果，還是 CPU／I/O
4. 故障後是否必須恢復

真正的核心不是快或慢，而是：

> 這個 operation 是否具有獨立於目前 Python process 和目前 agent invocation 的生命週期？

## 一、先修正一個容易造成架構誤判的地方

`LongRunningFunctionTool` 不應被理解成「ADK 幫你執行背景工作」。

它比較像：

> ADK 對 LLM 和 client 宣告：這個 function call 目前只有暫時結果，之後還會有與原始 call 對應的補充結果。

官方文件描述的行為是：工具啟動 operation，回傳 operation ID 等初始結果；runner 將其包成 `FunctionResponse`，讓模型可以回報目前狀態，之後結束或暫停這次 agent run。client 再決定輪詢、等待或提交後續的 intermediate／final response。([Adk][1])

因此它解決的是 **agent protocol continuation**，而不是：

* job queue
* worker scheduling
* retry
* webhook delivery
* distributed locking
* timeout management
* operation storage
* exactly-once execution

這些仍然是 application infrastructure 的責任。

更精確的架構是：

```text
LLM / Agent
    │
    │ function_call(call_id)
    ▼
LongRunningFunctionTool adapter
    │
    │ create operation
    ▼
Application Operation Service
    │
    ├── database
    ├── queue
    ├── worker
    ├── approval system
    └── external provider
```

`LongRunningFunctionTool` 是上方 agent protocol 與下方 durable operation system 之間的 adapter。

它本身不應成為你的 job system。

---

## 二、四種機制其實分屬不同 abstraction layer

可以把它們放在四層：

```text
┌───────────────────────────────────────────┐
│ Workflow control layer                    │
│ RequestInput                              │
├───────────────────────────────────────────┤
│ Agent-tool protocol layer                 │
│ FunctionTool / LongRunningFunctionTool    │
├───────────────────────────────────────────┤
│ Application operation layer               │
│ job, approval, order, report, ticket      │
├───────────────────────────────────────────┤
│ Runtime concurrency layer                 │
│ coroutine, task, thread, process          │
└───────────────────────────────────────────┘
```

它們並不是競爭方案。

一個 production system 很可能同時使用全部四種：

```text
RequestInput
    使用者選擇報表範圍
        ↓
FunctionTool
    驗證參數並估算成本
        ↓
LongRunningFunctionTool
    建立報表 generation operation
        ↓
async worker
    非同步查詢資料並生成報表
```

所以問題不應該是：

> FunctionTool、LongRunningFunctionTool、RequestInput、async task 要選哪一個？

而應該是：

> 在每一個 abstraction layer，我需要哪一種 primitive？

---

# 三、它們對 session、state、memory 的影響

這裡要先明確區分四種「記憶」。

## 1. Python process memory

例如：

```python
pending_jobs = {}
tasks = set()
```

這是最弱的記憶。

它只存在於目前 process：

* process restart 後消失
* 多 instance 之間不共享
* serverless scale-to-zero 後消失
* rolling deployment 後可能消失

因此，不應使用 Python dictionary 或 `asyncio.Task` 作為 long-running operation 的 source of truth。

```python
task = asyncio.create_task(generate_report())
```

這個 task 只對目前 event loop 有意義。

它沒有 durable identity，也沒有跨 process recovery。

---

## 2. ADK session events

ADK session 的 `events` 是該 conversation thread 的歷史，包含 user message、agent response、tool call、tool result、state change 和 control signal。([Adk][2])

對 long-running tool 而言，event history 的用途是保留：

```text
Agent called generate_report
call_id = call_123

Tool initially responded:
operation_id = op_456
status = pending

Later:
operation op_456 completed
```

這讓 agent 能重建 conversational continuity。

但 event history 不應被當成 operation database。

原因包括：

* 查詢特定 operation 不一定有效率
* operation 可能跨 session
* business lifecycle 可能比 conversation 長
* event replay 不一定等同 business-state reconstruction
* session rewind 可能讓 conversational timeline 與真實外部 side effect 分離

因此：

> ADK event 是 conversation audit trail，不應是 business operation 的唯一 source of truth。

---

## 3. ADK session state

ADK 將 session state 定位為 conversation-specific scratchpad；它可以跨同一 session 的多個 turn 保存資料，也是 workflow node、tool 和 callback 之間的重要資料共享機制。([Adk][3])

你可以在 state 裡保存：

```python
{
    "active_report_operation_id": "op_456",
    "active_report_status": "pending",
    "requested_report_type": "sales_summary"
}
```

但要注意：session state 比較適合保存 **reference 和 conversation-local projection**，而不是 operation 本體。

建議：

```text
Session state:
    active_operation_id
    latest_known_status
    user-facing summary
    next expected action
```

不要保存：

```text
完整 report rows
worker checkpoint
retry count
provider callback payload
approval audit log
binary report content
```

完整資料應放在 application database、object storage 或 artifact service。

---

## 4. ADK MemoryService

ADK 官方對 memory 的定義，是從 session/state 這種短期 conversation context 中抽取可搜尋的長期知識；應用程式可以把 session 加入 MemoryService，供未來 session 搜尋。([Adk][4])

這表示 long-running operation 的 pending state 通常**不應直接進 long-term memory**。

例如以下內容沒有長期知識價值：

```text
report op_456 is currently 37% complete
refund approval is pending
worker retry count is 2
```

如果把這些都寫進 memory，會產生：

* stale memory
* retrieval pollution
* token waste
* model confusion
* privacy retention 問題
* 過期 operation 被錯誤引用

較合理的是 operation 完成後，選擇性寫入具長期價值的結果：

```text
使用者通常偏好 PDF 報表
使用者核准了採購上限為 50,000 元
本次分析確認資料來源 A 不可靠
該客戶的退款申請已批准
```

也就是說：

> Operational state 屬於 operation store；conversation state 屬於 session；可重用知識才屬於 memory。

---

# 四、推薦的狀態分層

我會採用下面的 ownership model：

| 資料                     | Source of truth                   |
| ---------------------- | --------------------------------- |
| function call history  | ADK session events                |
| conversation 下一步       | ADK session state                 |
| operation status       | application database              |
| worker execution       | queue / worker runtime            |
| large result           | object storage / artifact store   |
| durable user knowledge | MemoryService                     |
| UI pending indicator   | client state，加上 server projection |

例如：

```text
ADK Session
    session_id: s_123
    state:
        active_operation_id: op_456
        active_call_id: call_789
        active_operation_status: pending

Operation Database
    operation_id: op_456
    type: generate_report
    status: running
    requested_by: user_123
    session_id: s_123
    function_call_id: call_789
    progress: 0.42
    result_uri: null
    version: 7
```

這個分離非常重要。

因為 session 可能被刪除，但 business operation 仍可能需要 audit。

反過來，operation 可能已完成或過期，但 conversation history 仍應可讀。

# 五、FunctionTool 對系統的影響

`FunctionTool` 是最簡單的同步 request-response model：

```text
invocation lifetime
┌────────────────────────────────────┐
tool call ─── execution ─── result
└────────────────────────────────────┘
```

## 優點

* 控制流簡單
* transaction boundary 清楚
* 容易測試
* failure 立即可見
* 不需要 operation correlation
* 不需要 resume API
* 不需要額外 persistence

## 缺點

* 長時間佔用 invocation
* request timeout 風險
* client connection 可能中斷
* worker capacity 被長請求占用
* retry 可能重複執行 side effect
* LLM 無法在等待期間做出別的互動

## 適用判準

不要只看執行時間，而要看：

```text
P99 duration < request timeout budget
且
operation 不需要外部 callback
且
operation 可以安全地在目前 invocation 生命週期完成
```

例如一個 20 秒 operation：

* 若平台 timeout 是 60 秒
* API 穩定
* 失敗可以安全 retry
* 使用者願意等待

那仍可能適合普通 `FunctionTool`。

但一個 2 秒就能「提交成功」、實際付款結果稍後由 provider webhook 回報的支付操作，反而應被視為 long-running operation。

---

# 六、async task 對系統的影響

`async` 處理的是 cooperative concurrency。

官方文件指出，async function 可讓 ADK 的 function tools 在 event loop 中並行執行，避免同步工具阻塞其他工具；但它仍然是在目前 agent execution 內等待完成。([Google GitHub][5])

```text
async ≠ durable
async ≠ background job
async ≠ resumable
async ≠ cross-invocation
```

它解決的是：

```text
等待 network I/O 時，event loop 可以執行其他 coroutine
```

它不解決：

```text
server restart 後如何恢復？
callback 如何找到原始 call？
結果如何送回原 agent？
重複 callback 如何去重？
operation 是否已超時？
```

## `asyncio.create_task()` 的危險

例如：

```python
asyncio.create_task(generate_report())
return {"status": "pending"}
```

這在 local demo 很誘人，但 production 中非常脆弱：

1. task 沒有 durable persistence
2. request 結束後 process 可能被終止
3. task exception 可能沒有被妥善處理
4. 多 instance 無法查詢它
5. deployment 時 task 消失
6. 無法可靠 retry
7. client 不知道最後結果在哪裡

因此 production 上應改成：

```python
operation = await operation_repository.create(...)
await queue.publish(operation.id)
return {
    "status": "pending",
    "operation_id": operation.id,
}
```

worker 才負責實際執行。

# 七、RequestInput 的真正定位

`RequestInput` 是 graph workflow 的 interrupt primitive。

官方文件中，workflow node yield `RequestInput` 後會暫停並等待 human input；它支援 message、payload 和 response schema。([Adk][6])

它最適合表達：

```text
workflow 無法決定下一條 edge，
直到取得一份指定格式的 human-provided data。
```

例如：

```text
draft proposal
    ↓
RequestInput:
    approve / reject / revise
    ↓
conditional routing
```

它的優勢是 workflow deterministic：

* 暫停點由 graph 定義
* 下一個 node 由 graph 決定
* input schema 可以明確
* 不需要 LLM 自己重新推理「現在要做什麼」

這對高可靠性流程非常重要。

## RequestInput 與 LongRunningFunctionTool 的交集

人工審批既可以被建模為：

```text
RequestInput:
    workflow 正在等待 reviewer 的決策
```

也可以被建模為：

```text
LongRunningFunctionTool:
    request_approval tool 尚未完成
```

兩者都合理，但設計意義不同。

### 使用 RequestInput

當 approval 是 workflow 本身的顯式節點：

```text
Generate → Review → Approve → Execute
```

這時 graph 擁有流程控制。

### 使用 LongRunningFunctionTool

當 approval 是某個 domain operation 的完成條件：

```text
Agent calls submit_refund()
Refund service creates approval ticket
External approval system later resolves it
```

這時 tool/domain service 擁有流程控制。

簡化成一句話：

> RequestInput 是 workflow-owned interrupt；LongRunningFunctionTool 是 operation-owned continuation。

# 八、LongRunningFunctionTool 對系統架構的真正成本

引入它後，你其實進入了 distributed systems 問題。

至少要處理以下狀態：

```text
CREATED
QUEUED
RUNNING
WAITING_FOR_HUMAN
SUCCEEDED
FAILED
CANCELLED
EXPIRED
```

而不是只有：

```text
pending / complete
```

## 1. Correlation

至少需要保存：

```text
operation_id
session_id
invocation_id
function_call_id
user_id
tool_name
```

官方文件特別要求後續 long-running response 使用正確的 invocation identity，以便與原始 invocation 對應。([Google GitHub][7])

但在你自己的 application model 中，不應把 ADK call ID 當唯一 primary key。

建議：

```text
operation_id       application identity
function_call_id   agent protocol correlation
invocation_id      invocation correlation
provider_job_id    external system correlation
```

這幾個 ID 要分開。

---

## 2. Idempotency

可能發生：

```text
client 沒收到提交結果
    ↓
重新呼叫 tool
    ↓
建立兩個退款操作
```

因此啟動 operation 時要有 idempotency key：

```python
idempotency_key = (
    f"{user_id}:{session_id}:{function_call_id}:{tool_name}"
)
```

資料庫對它設 unique constraint。

但也要注意，某些 model 或 retry flow 可能生成新的 function call ID，因此更高階的 business idempotency key 可能要來自：

```text
order_id + requested_action + semantic version
```

例如：

```text
refund:order_123:full:v1
```

---

## 3. Duplicate completion

webhook、queue、client retry 都可能重複提交 final result。

因此 transition 必須具備 compare-and-set 語意：

```sql
UPDATE operations
SET status = 'SUCCEEDED',
    result_uri = :result_uri,
    version = version + 1
WHERE operation_id = :operation_id
  AND status IN ('QUEUED', 'RUNNING', 'WAITING_FOR_HUMAN');
```

若 update count 為 0，代表 operation 已完成、取消或過期。

---

## 4. Out-of-order response

可能先收到：

```text
100% complete
```

後收到：

```text
70% complete
```

因此 progress update 要帶：

```text
sequence_number
version
occurred_at
```

並拒絕舊版本。

---

## 5. Session 已不存在

operation 可能明天完成，但：

* session 被刪除
* user 開了新 session
* original invocation 無法 resume
* tenant policy 已過期

這時你必須決定：

```text
A. completion 被丟棄
B. 寫入 notification inbox
C. 建立新的 session message
D. 只更新 operation center
E. 寫入 domain record，等使用者下次查詢
```

不能假設所有 long-running operations 都必然回到原 session。

---

# 九、對 LLM context window 的影響

這是一個經常被忽略的問題。

如果每次 operation progress 都被保存成 event 並送回模型：

```text
10%
20%
30%
...
90%
```

會造成：

* event history 膨脹
* prompt token 增加
* inference latency 增加
* model 注意力被 status noise 佔據
* context compression 更頻繁

因此不要把 worker 的每個 progress event 都傳給 agent。

建議分層：

```text
Operation database:
    保存完整 progress events

Client UI:
    即時顯示 1% 細粒度更新

ADK session:
    只記錄語意上重要的 transition
```

例如只記錄：

```text
pending
waiting_for_approval
completed
failed
```

而非每個百分比。

對 LLM 而言，一般只需要：

```json
{
  "operation_id": "op_456",
  "status": "running",
  "progress": 72,
  "summary": "資料查詢完成，正在生成 PDF"
}
```

甚至只需要 final result。

# 十、我會推薦的最終設計

最穩定的方向是建立一個與 ADK 解耦的 `Operation Service`。

```text
┌───────────────────┐
│ ADK Agent         │
└────────┬──────────┘
         │ start tool
         ▼
┌───────────────────┐
│ Tool Adapter      │
│ FunctionTool or   │
│ LongRunningTool   │
└────────┬──────────┘
         │
         ▼
┌─────────────────────────────────────┐
│ Operation Service                   │
│                                     │
│ create_operation()                  │
│ get_operation()                     │
│ complete_operation()                │
│ cancel_operation()                  │
│ expire_operation()                  │
└───────┬──────────────┬──────────────┘
        │              │
        ▼              ▼
   Database         Queue/Worker
                         │
                         ▼
                   External systems
```

domain object 可以像：

```python
@dataclass
class Operation:
    id: str
    operation_type: str
    status: OperationStatus

    user_id: str
    session_id: str | None
    invocation_id: str | None
    function_call_id: str | None

    idempotency_key: str
    input_payload: dict
    result_payload: dict | None
    error: dict | None

    progress: float | None
    version: int

    created_at: datetime
    updated_at: datetime
    expires_at: datetime | None
```

關鍵思想是：

> ADK metadata 是 operation 的 correlation metadata，不是 operation model 本身。

如此一來，即使未來換掉 ADK，domain operation system 仍然可用。

# 十一、工具選擇的實際 decision tree

```text
這一步是否需要 LLM/tool？
    │
    ├─ 否 → 普通 application function
    │
    └─ 是
        │
        ├─ 是否能在目前 invocation 可靠完成？
        │      │
        │      ├─ 是 → FunctionTool
        │      │          └─ I/O concurrency → async def
        │      │
        │      └─ 否
        │          │
        │          ├─ 等待的是 workflow 所需的人類資料？
        │          │      └─ 是 → RequestInput
        │          │
        │          └─ 等待的是 operation completion？
        │                 └─ LongRunningFunctionTool
        │
        └─ 是否必須跨 process/restart？
               ├─ 否 → 可考慮 local task，但僅限 prototype
               └─ 是 → durable operation DB + queue/worker
```

---

# 十二、

第一，不要寫成：

> 任務只是慢一點但仍可以同一輪等待，就不一定需要 LongRunningFunctionTool。

應改得更強：

> 執行時間不是主要判準。主要判準是 operation 是否能安全地綁定於目前 invocation 的生命週期，以及是否需要 durable continuation。

第二，不要讓讀者以為：

> client 把最終結果送回 runner，runner 就自然從原本 Python stack 繼續。

比較準確的是：

> 原 invocation 的執行已經結束或暫停；後續 response 透過 preserved session/event identity 和 invocation/function-call correlation，觸發 agent continuation。它不是恢復原 coroutine stack。

這是一個非常重要的 distinction：

```text
不是 stack continuation
而是 state-based continuation
```

第三，應加入：

> LongRunningFunctionTool 不會替你保存外部 job 的 durable state。production 系統仍需要 operation store、idempotency、retry、timeout、cancellation 和 callback deduplication。

否則讀者很容易把它誤當成 Celery、Temporal、Cloud Tasks 或 durable workflow engine。

---

最終可以用一句架構原則概括：

```text
async 管執行效率，
FunctionTool 管單次 tool request-response，
RequestInput 管 workflow interrupt，
LongRunningFunctionTool 管 agent protocol continuation，
Operation Service 管真正的 durable business lifecycle，
Session 管對話連續性，
Memory 管跨對話仍有價值的知識。
```

#### UI

**`LongRunningFunctionTool` 本身不會自動產生使用者輸入介面**。它只是讓 ADK runner 知道「這個 tool call 需要暫停，等待外部輸入或外部結果」。真正的「產生介面讓 user input」是由 **agent client / UI layer** 負責，例如 ADK Web UI、CLI prompt、Slack bot、React 前端、AG-UI、或你自己寫的 client loop。

設定上分成兩種情況。

#### 情況 A：用 ADK 內建 Web UI 的 tool confirmation

如果你的需求只是「工具執行前跳出確認框，讓使用者 approve / reject」，ADK 有一個更直接的機制叫 **Action confirmations**。使用 ADK Web UI 時，confirmation feature 會顯示一個 dialog box 要求使用者輸入確認。官方文件說明，使用 ADK Web UI 搭配 confirmation feature 時，agent workflow 會向使用者顯示確認對話框；也可以透過 ADK server REST API 傳入遠端回應。([Adk][1])

這種情況比較像：

```text
Agent wants to call sensitive tool
    ↓
ADK Web UI shows confirmation dialog
    ↓
User clicks approve / reject
    ↓
Tool executes or is cancelled
```

這適合「yes/no approval」型場景，例如退款、訂票、刪除資料、發送 email。

#### 情況 B：用 LongRunningFunctionTool 自己攔截 event，自己產生 UI

如果你用的是 `LongRunningFunctionTool`，流程會更底層一點。ADK 的官方說法是：當 long-running tool 被呼叫後，agent runner 會暫停 agent run，讓 agent client 決定要繼續、等待任務完成，或回傳中間 / 最終 response。也就是說，**UI 是 client 的責任**，不是 tool 自己產生的。([Adk][2])

設定上的核心是這三步：

第一，在 agent 裡把需要人類輸入的 function 包成 `LongRunningFunctionTool`：

```python
from typing import Any
from google.adk.agents import Agent
from google.adk.tools import LongRunningFunctionTool

def ask_for_approval(purpose: str, amount: float) -> dict[str, Any]:
    """Ask a human to approve this action before proceeding."""
    return {
        "status": "pending",
        "purpose": purpose,
        "amount": amount,
        "message": f"Approve refund of ${amount} for {purpose}?",
    }

agent = Agent(
    name="refund_agent",
    model="gemini-3.1-pro",
    instruction="""
    You handle refunds.
    Always call ask_for_approval before processing a refund.
    If approved, continue with the refund.
    If rejected, tell the user the refund was rejected.
    """,
    tools=[
        LongRunningFunctionTool(func=ask_for_approval),
    ],
)
```

這裡的 `message`、`purpose`、`amount` 就是你要拿去渲染 UI 的 payload。它可以被 CLI 印出來，也可以被 Web UI 顯示成 modal。

第二，在 client 的 `runner.run_async()` event loop 裡攔截 long-running tool call：

```python
long_running_call = None
pending_response = None

async for event in runner.run_async(
    session_id=session.id,
    user_id="user1",
    new_message=user_msg,
):
    if event.content and event.content.parts:
        for part in event.content.parts:
            if (
                not long_running_call
                and part.function_call
                and event.long_running_tool_ids
                and part.function_call.id in event.long_running_tool_ids
            ):
                long_running_call = part.function_call

            if (
                long_running_call
                and part.function_response
                and part.function_response.id == long_running_call.id
            ):
                pending_response = part.function_response
```

這段就是「讓上面那段發生效果」的關鍵。`event.long_running_tool_ids` 代表 ADK 告訴 client：這個 tool call 是長任務，需要外部介入。client 一旦看到它，就可以產生 input interface。

第三，把 `pending_response.response` 拿去顯示給使用者，然後把使用者輸入包回 `FunctionResponse`：

```python
print(f"Pending approval: {pending_response.response}")
choice = input("Approve? [Y/n]: ").strip().lower()

updated = pending_response.model_copy(deep=True)
updated.response = {
    "status": "approved" if choice != "n" else "rejected",
}
```

最後，把這個 response 塞回 runner：

```python
from google.genai import types

async for event in runner.run_async(
    session_id=session.id,
    user_id="user1",
    new_message=types.Content(
        role="user",
        parts=[types.Part(function_response=updated)],
    ),
):
    ...
```

這就是 resume。ADK runner 會把這個 `FunctionResponse` 對回原本那個 long-running function call，agent 就能繼續推理。

#### 所以「介面」到底在哪裡設定？

不是在 `LongRunningFunctionTool` 裡設定，而是在 client layer 設定。

可以是 CLI：

```python
choice = input("Approve? [Y/n]: ")
```

也可以是 Web UI：

```text
pending_response.response
    ↓
render modal:
  title: "Approval required"
  body: purpose, amount, message
  buttons: Approve / Reject
    ↓
onClick -> send FunctionResponse back to runner
```

也可以是 Slack：

```text
pending_response.response
    ↓
send Slack message with buttons
    ↓
Slack interaction callback
    ↓
send FunctionResponse back to ADK runner
```

一句話說清楚：`LongRunningFunctionTool` 負責「讓 runner 暫停並暴露 pending function call」；`runner.run_async()` 的 client loop 負責「偵測 pending call」；你的 UI / CLI / Slack layer 負責「產生讓 user input 的介面」；最後再由 client 把 user input 包成同一個 `FunctionResponse` 送回 runner。

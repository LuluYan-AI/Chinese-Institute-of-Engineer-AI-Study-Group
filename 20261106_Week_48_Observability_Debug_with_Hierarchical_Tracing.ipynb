### Agent Observability

AI 代理（AI Agents）與傳統微服務最大的差異在於其非確定性（Nondeterministic）本質：相同的輸入（Input）在每次執行時，都可能產生截然不同的輸出與決策路徑。這場由 Sirisha Karra 主講的「Advent of Agents」系列分享，透過實際案例展示如何結合開源標準與工具，解決多代理系統（Multi-Agent System）在追蹤與除錯上的痛點。

**為什麼 AI Agent 需要專屬的可觀測性（Agent Observability）？**
傳統軟體監控（Monitoring）主要關注系統是否存活（Uptime）、延遲（Latency）與 HTTP 狀態碼；但在 AI 代理系統中，就算 API 回傳 200 OK，內容仍可能是嚴重幻覺。因此，代理可觀測性必須深入追蹤模型內部的運作：

* **決策鏈路與推理過程（Step-by-step Thought Process）：** 代理如何拆解任務與規劃步驟。
* **工具呼叫（Tool Calls）：** 代理選擇了什麼外部工具、帶入什麼參數，以及回傳結果是否有效。
* **回應品質與有效性（Response Quality & Validity）：** 輸出是否符合事實、有無偏離指令。

**範例架構：Catalyst 多代理金融分析系統**
為了示範真實場景，講者使用 **Google ADK（Agent Development Kit）** 打造了一個名為「Catalyst」的金融分析應用，並透過 **OpenTelemetry** 與 **Arize Phoenix**（包含 Phoenix SDK、`genai instrument`、`adk instrument`）實現自動化埋點（Auto-instrumentation），無需大幅改動業務邏輯即可收集遙測數據。系統架構分為三個層級：

1. **入口驗證層 — 接待代理（Greeting Agent）：** 接收使用者輸入的公司名稱，並驗證該公司是否於證券交易所合法上市。
2. **平行探索層 — 代理群（Discovery Layer / Swarm）：** 採用扇出（Fan-out）架構，同時平行執行三個子代理（Sub-agents）：
* **企業剖析代理（Corporate Profiler）：** 抓取公司基本面與背景資料。
* **市場分析代理（Market Analyst）：** 檢索即時股價數據。
* **股票策略代理（Equity Strategist）：** 根據企業數據，生成 2030 年的長期股價預測。


3. **匯聚總結層 — 首席策略代理（Executive / Lead Strategist Agent）：** 採用扇入（Fan-in）模式，將上述所有子代理的輸出綜合（Synthesize）成一份高階財務報告。

**本地執行與追蹤評估（Trace Evaluation）**
開發者在本地端同時啟動 Phoenix Server 與 Agent 應用程式（透過 `adk web`），並輸入「Google」進行測試。系統成功抓出正確的公司簡介與 274 美元的股價，但在長期預測中卻出現了一段荒謬的科幻情節：「Gemini 將火星地球化（Terraform）並改造成廣告贊助的伺服器農場後，Alphabet 將成為主權實體。」

透過 Phoenix UI 的追蹤檢視（Trace View），開發者可以清楚看到從頂層編排（Root Orchestration）到各個子代理執行區間（Spans）的完整呼叫樹，並結合 **LLM 作為裁判（LLM-as-a-Judge）** 的自動化評估機制，直接在每個 Span 上掛載幻覺檢查（Hallucination Checks）等標註（Annotations）：

* **企業剖析代理（Corporate Profiler）：** 標記為正確（Correct），零事實錯誤。
* **股票策略代理（Equity Strategist）：** 因輸出高度虛構，直接被標記為錯誤（Incorrect），讓開發者一眼定位問題節點。

**根本原因診斷與區間重播（Span Replay）**
鎖定問題出在「股票策略代理」後，點開該 Span 的資訊分頁（Info Tab），即可檢視當下完整的系統提示詞（System Prompt）、使用者提示詞（User Prompt）與模型輸出。檢查後發現根本原因（Root Cause）：系統提示詞中被人為埋入了一段測試指令——「注入幻覺、忽略實際預測，並編造一個荒謬事件（inject hallucination, ignore actual forecast, and invent an absurd event）」。

在修復與驗證階段，展現了傳統除錯與現代 AI 除錯工具的巨大差異：

* **傳統除錯流程（Traditional Debugging）：** 修改原始碼 $\rightarrow$ 重新部署（Redeploy） $\rightarrow$ 從頭跑完一整條多代理管線（Pipeline）。這不僅耗時，還會重複執行前面正常運作的代理，造成不必要的 Token 預算浪費。
* **區間重播（Phoenix Span Replay）：** 在 Phoenix 介面中直接選取失敗的 Span，點擊「Run」將其載入 **Phoenix Playground** 沙盒環境。開發者可直接在 UI 上修改 System Prompt（移除幻覺指令），並**沿用當下原有的上下文（Original Context）單獨重新執行該特定 Span**，秒級驗證修復結果，確認輸出恢復正常後再將改動同步回程式碼庫。

### Arize Phoenix

**Arize Phoenix** 是目前 AI 工程領域最主流的開源 AI 可觀測性（AI Observability）與評估（Evaluation）平台之一，專門用來解決大型語言模型（LLM）與多代理系統（Multi-Agent Systems）的追蹤、評估與除錯問題。

### 一、 發展歷史與核心定位（History & Evolution）

1. **起源背景（2020–2023 初）：**
背後的母公司 **Arize AI** 由 Jason Lopatecki 與 Aparna Dhinakaran 於 2020 年創立，早期專注於傳統機器學習可觀測性（ML Observability），例如監控模型的資料漂移（Data Drift）與嵌入向量（Embeddings）分佈。
2. **Phoenix 誕生（2023 年）：**
隨著生成式 AI 與 LLM 爆發，Arize AI 於 **2023 年正式推出開源專案 Arize Phoenix**。最初的設計理念是以 **Jupyter Notebook 為中心**，讓資料科學家能在本地端視覺化分析 Embedding 叢集、檢視 RAG（檢索增強生成）效果並進行基礎評估。
3. **擁抱 OpenTelemetry 與定義 OpenInference 標準（2024–2025 年）：**
隨著產業重心轉向複雜的 AI Agents，Phoenix 脫離單純的 Notebook 工具，轉型為完整的全端可觀測性伺服器（支援 SQLite 與 PostgreSQL 儲存）。為了避免開發者被特定廠商綁定（Vendor Lock-in），Arize 主導並開源了 **OpenInference** 規範——這是一套建立在 **OpenTelemetry（OTel）** 之上的語意慣例（Semantic Conventions），專門用來標準化記錄 LLM 的 Prompt、Token 消耗、Tool Calls 與 Agent Spans。
4. **近期動態與生態系擴展（2025–2026 年）：**
* **產品線雙軌制：** Arize 將產品明確劃分為開源、可自架（Self-hosted，採用 Elastic License 2.0）的 **Arize Phoenix**，以及企業級雲端託管平台 **Arize AX**，兩者底層完全共用 OpenTelemetry / OpenInference 埋點標準。
* **併購里程碑：** 2026 年 8 月雲端可觀測性巨頭 **Dynatrace** 宣布以 9.15 億美元收購 Arize AI（於 2026 年 10 月完成交割），並承諾持續維護與支援 Phoenix 開源專案及生態系。
* **Agent 工具鏈整合：** 近期版本更加入了 Prompt Playground、Span Replay（區間重播），以及支援 Claude Code / Cursor 等開發代理直接調用的 `@arizeai/phoenix-cli` 與內建遠端 **MCP Server（Model Context Protocol）**。

### 二、 如何安裝與執行（How to Run）

Phoenix 的架構非常輕量，支援在本地端開發環境、Jupyter Notebook 或透過容器化部署在伺服器上，預設會啟動在本地端的 **`6006` 連接埠（Port 6006）**（接收 OTLP gRPC 追蹤則預設支援 `4317` 埠）。

#### 方式 1：透過 CLI 在終端機啟動（最推薦的本地開發方式）

你可以透過 `pip` 或 `uv` 安裝主程式並直接啟動常駐伺服器：

```bash
# 安裝 Phoenix 主平台套件
pip install arize-phoenix

# 啟動 Phoenix 伺服器
phoenix serve

```

*(若使用 `uv`，也可以直接執行 `uvx --from arize-phoenix phoenix serve` 免預先安裝直接啟動。)*
啟動後，打開瀏覽器前往 **`http://localhost:6006`** 即可進入 Phoenix UI 儀表板。

#### 方式 2：透過 Docker 容器啟動（適合團隊共用或隔離環境）

```bash
docker run -p 6006:6006 -p 4317:4317 arizephoenix/phoenix:latest

```

*(在生產環境或長期測試中，可以掛載環境變數連接外部 PostgreSQL 資料庫，取代預設的本地 SQLite。)*

#### 方式 3：在 Jupyter Notebook / Python 腳本中動態啟動

如果你正在寫 Notebook 實驗，可以直接在程式碼內拉起背景服務：

```python
import phoenix as px

# 在背景啟動 Phoenix 並回傳 session 物件
session = px.launch_app()
# 可在 Notebook 中直接顯示 UI 連結或嵌入畫面
print(session.url)

```

### 三、 如何使用與整合進 AI 專案（How to Use）

Phoenix 的核心使用流程分為三步：**「自動埋點（Auto-instrumentation） $\rightarrow$ 觀測與除錯（Tracing & Replay） $\rightarrow$ 自動化評估（LLM Evals）」**。

#### 步驟 1：安裝輕量化 OTel 埋點套件與對應框架插件

在你的 AI 應用程式專案中，通常不需要安裝龐大的 `arize-phoenix` 主體，只需安裝輕量的 OpenTelemetry 封裝套件 `arize-phoenix-otel` 與你所使用的框架對應的 **OpenInference Instrumentor**（支援 Google GenAI、Google ADK、OpenAI、LangChain / LangGraph、LlamaIndex、Anthropic、CrewAI 等）：

```bash
pip install arize-phoenix-otel
# 依照你使用的框架安裝對應的自動埋點套件，例如：
pip install openinference-instrumentation-google-adk
pip install openinference-instrumentation-google-genai
pip install openinference-instrumentation-openai

```

#### 步驟 2：在程式碼中加入「一行式自動埋點（One-line Instrumentation）」

透過 `phoenix.otel` 提供的 `register()` 函式，將參數 `auto_instrument=True` 打開，它就會自動偵測環境中已安裝的 OpenInference 套件並掛載追蹤器（Tracer）：

```python
from phoenix.otel import register

# 初始化 OpenTelemetry 並自動攔截所有已安裝框架的呼叫
tracer_provider = register(
    project_name="catalyst-demo",          # 在 Phoenix UI 中顯示的專案名稱
    endpoint="http://localhost:6006/v1/traces", # 指向本地或遠端的 Phoenix 伺服器
    auto_instrument=True                   # 自動啟用環境中安裝的 OpenInference 埋點
)

# 接著直接撰寫你原本的 Agent / LLM 業務邏輯即可，完全不需修改既有程式碼！

```

#### 步驟 3：在 Phoenix UI 上執行三大核心任務

當你的 Agent 開始執行後，所有的遙測資料（Telemetry Data）就會即時串流到 `http://localhost:6006`：

1. **全鏈路追蹤檢視（Trace & Span Inspection）：**
* 每次使用者請求會形成一條 **Trace**，內部的每個步驟會被拆解並標記為不同類型的 **Span Kind**（如 `AGENT`、`CHAIN`、`LLM`、`TOOL`、`RETRIEVER`、`GUARDRAIL`）。
* 點擊任一節點，即可查看當下的 System Prompt、User Prompt、傳入 Tool 的 JSON 參數、延遲時間（Latency）與精確的 Token 成本計算。


2. **提示詞沙盒與區間重播（Prompt Playground & Span Replay）：**
* 如你前一篇筆記所述，當某個子代理（Sub-agent）發生幻覺或解析錯誤時，可在 UI 點選該 Span 並載入 **Playground**，直接修改 Prompt 或切換模型（如從 Gemini Flash 切換成 Pro），利用當時的歷史 Context 單獨重跑該步驟。


3. **LLM 當裁判與資料集實驗（LLM-as-a-Judge & Experiments）：**
* 可搭配 `arize-phoenix-evals` 套件，使用獨立的 LLM 對收集到的 Spans 進行批次檢查（例如：Hallucination 幻覺檢測、QA 正確性、RAG 檢索相關性），並將結果寫回 Span 的 **Annotations（標註）** 上，在 UI 上以紅綠燈標示健康度。
* 你也能將真實發生的錯誤 Spans 一鍵匯出成 **Dataset（測試資料集）**，在每次改版後自動跑回歸測試（Experiments），量化比較不同版本的分數變化。

#### 官方文件 What is Arize Phoenix?
https://arize.com/docs/phoenix

這份 **Arize Phoenix 官方文件首頁（What is Arize Phoenix?）** 定義了該平台的核心工作流與功能架構。Phoenix 的設計目標是提供一套從「除錯單次執行」到「以數據驅動品質優化」的閉環開發流程，底層完全建立在 **OpenTelemetry（OTLP）** 與 **OpenInference** 埋點標準之上。

四大核心功能模組與最新特色：

**1. 鏈路追蹤（Tracing）**

* **運作方式：** 逐步記錄 AI 應用程式在單次執行（Single Run）中發生的所有細節，包含模型呼叫（Model Calls）、檢索（Retrieval）、工具調用（Tool Use）與自訂邏輯，用來除錯行為並分析耗時瓶頸。
* **跨生態系整合：** 透過 OpenTelemetry（OTLP）接收 Traces，並提供主流框架（LlamaIndex、LangChain、DSPy、Mastra、Vercel AI SDK、Google ADK）、模型供應商（OpenAI、Anthropic、Bedrock、Gemini）與多語言（Python、TypeScript、Java）的自動埋點（Auto-instrumentation）。

**2. 品質評估（Evaluation）**
協助量化評估應用程式的輸出品質，持續追蹤效能並抓出錯誤或退化（Regressions）：

* **雙模式執行：** 支援透過 SDK 在本地端跑 **Client-Side Evals**，或直接在網頁介面上設定自動執行的 **Server-Side Evals（UI）**。
* **多元評分機制：**
* **LLM-as-a-Judge（LLM 當裁判）：** 執行內建或自訂的 LLM 評估器。
* **程式碼規則與第三方整合：** 支援自訂程式碼檢查，或直接接入 **Ragas、DeepEval、Cleanlab** 等外部評估框架。
* **人工標註（Human Annotations）：** 可直接在 UI 介面上為 Trace 或 Span 貼上真實標籤（Ground Truth Labels）。



**3. 提示詞工程與重播（Prompt Engineering & Playground）**
讓開發者直接使用生產環境（Production）中真實發生的案例來迭代優化 Prompt：

* **Prompt Management & Prompts in Code：** 集中管理提示詞的版本（Versioning）、儲存與部署，並透過 SDK 在不同環境與程式碼之間同步。
* **Prompt Playground：** 並排比較（Side-by-side）不同 Prompt 變體與不同模型的輸出結果。
* **Span Replay（區間重播）：** 直接選取歷史紀錄中的 LLM 呼叫，替換不同輸入或修改 Prompt 後單獨重跑除錯，確認效果後再正式上線。

**4. 資料集與實驗（Datasets & Experiments）**
提供系統化的回歸測試（Regression Testing）機制，確保每次改版都有數據佐證：

* **建立資料集（Create Datasets）：** 直接從生產環境收集到的問題 Traces 轉換為測試集，或由 CSV / 程式碼上傳。
* **自動化實驗比較（Run Experiments）：** 將同一批輸入資料集跑過不同版本的應用程式或 Prompt，自動掛載 **Dataset Evaluators（資料集評估器）** 作為自動化測試案例，量化比較改版前後的分數差異，也可匯出用於模型微調（Fine-tuning）。

**5. 最新進階功能與企業版對照**

* **內建 AI 除錯助手 — PXI Agent（Beta）：** Phoenix 內部直接內建了一個名為 **PXI** 的代理程式，能根據當下的上下文協助開發者直接分析 Traces 並迭代 Prompts。
* **Coding Agents 輔助設定：** 支援讓開發者的編碼代理（Coding Agent，如 Claude Code、Cursor 等）直接讀取官方結構（支援 `/llms.txt` 索引）自動幫既有專案完成 Phoenix 埋點設定。
* **生產環境治理（Settings）：** 內建角色權限控制（RBAC）、API Keys 管理、資料保留策略（Data Retention）與自訂 AI 供應商（Custom AI Providers）。
* **Arize Phoenix vs. Arize AX：** 官方明確區分開源可自架（Self-hosted）的 **Phoenix** 與企業級託管平台 **Arize AX**，兩者共享相同的 OpenTelemetry / OpenInference 開放標準，方便團隊從本地開發無縫擴展至企業級架構。

### OpenTelemetry

**OpenTelemetry（簡稱 OTel）** 是雲端原生計算基金會 （Cloud Native Computing Foundation，CNCF）旗下活躍度僅次於 Kubernetes 的旗艦級開源專案，也是當今全球軟體工程與 AI 系統中，用來收集、處理與匯出遙測數據（Telemetry Data）的**業界絕對標準（De facto Standard）**。

要理解為什麼連 Arize Phoenix、Google ADK 甚至各大雲端與監控巨頭都選擇建立在 OpenTelemetry 之上，可以從它解決的歷史痛點與四個核心演進階段來看：

---

### 一、 史前時代的痛點與「雙雄分裂」（2010–2018 年）

在微服務（Microservices）與容器化架構興起後，一個使用者請求可能會橫跨數十個服務。當系統變慢或出錯時，傳統單機 Log 已經無法定位問題，因此催生了分散式追蹤（Distributed Tracing，最早源自 Google 2010 年發表的《Dapper》論文）。

然而，早期企業面臨嚴重的廠商綁定（Vendor Lock-in）問題：如果你用 Datadog、New Relic 或 Dynatrace，你就必須在程式碼裡埋入他們專屬的 SDK。一旦想更換監控後端，就得重寫所有程式碼的埋點。為了解決這個問題，開源界出現了兩大陣營：

1. **OpenTracing（2016 年加入 CNCF）：**
專注於制定一套**廠商中立的追蹤 API 規範（Vendor-neutral API）**。它只定義介面（Interface），不包含具體實作與傳輸代理，也不涵蓋指標（Metrics）。
2. **OpenCensus（2018 年由 Google 開源）：**
脫胎自 Google 內部的 Census 系統（與微軟等大廠合作），不僅涵蓋 **Traces 與 Metrics**，還直接提供了完整的 SDK 實作與資料收集器（Collector）。

**當時的困境：** 兩個標準互相競爭，導致整個開源生態系與第三方套件維護者（如資料庫驅動、Web 框架作者）無所適從，不知道該支援哪一套標準。

---

### 二、 世紀大合併與 OTel 誕生（2019–2021 年）

為了終結分裂，OpenTracing 與 OpenCensus 兩大社群決定放下分歧，於 **2019 年 5 月正式宣布合併，共同創立 OpenTelemetry**（最初以 Sandbox 沙盒專案加入 CNCF）。

* **吸收兩家之長：** 結合了 OpenTracing 優雅的 API 設計與解耦哲學，以及 OpenCensus 強大的語言 SDK 實作與 Collector 架構。
* **核心架構定型：**
* **OTLP（OpenTelemetry Protocol）：** 定義了統一、高效能（基於 gRPC / HTTP 與 Protobuf）的遙測資料傳輸協定。
* **OTel Collector：** 推出獨立於應用程式之外的代理服務，負責接收（Receive）、處理過濾（Process）並轉發（Export）資料到任何後端（如 Prometheus、Jaeger、Arize Phoenix、Datadog 等）。
* **語意慣例（Semantic Conventions）：** 統一規定了各種屬性的命名標準（例如 HTTP 狀態碼一律叫做 `http.response.status_code`），讓所有工具看得懂彼此的資料。


* **2021 年里程碑：**
* **2021 年初：** 發布 **Tracing（追蹤）1.0 GA 穩定版**，承諾生產環境的向後相容性（Backwards Compatibility）。
* **2021 年 8 月：** 正式晉升為 CNCF **孵化級專案（Incubating Project）**。隨後 OpenTracing（2022 年）與 OpenCensus（2023 年）完成歷史任務，正式封存退役（Archived）。



---

### 三、 可觀測性三大支柱的統一與擴展（2022–2024 年）

早期 OTel 雖然名字叫 Telemetry，但大家主要只拿它來做 Tracing。在 2022 到 2024 年間，OTel 逐步完成了可觀測性「三大支柱（Three Pillars of Observability）」的大一統，並跨足第四支柱：

1. **Metrics（指標）穩定化（2022 年）：** 推出穩定的 Metrics API/SDK，並實現與 Prometheus 生態系的深度互通。
2. **Logs（日誌）穩定化（2023 年）：** 推出 Logs Bridge API，將傳統應用程式日誌與 Trace ID / Span ID 自動關聯（Correlation），讓工程師點擊一條錯誤的 Trace 就能直接看到當下那幾毫秒產生的 Log。
3. **跨入第四支柱 — Continuous Profiling（持續效能剖析，2024–2026 年）：** 接收了由 Elastic 捐贈的 eBPF 剖析代理程式，將 **Profiles（CPU / 記憶體程式碼級別效能剖析）** 納入 OTel 成為第四種官方訊號（Signal），並在 2026 年 3 月進入公開 Alpha 階段。

---

### 四、 AI 與 GenAI 時代的演進與正式畢業（2024–2026 年）

當生成式 AI（GenAI）與多代理系統（Multi-Agent Systems）成為主流後，工程界發現：**LLM 與 Agent 的呼叫鏈路，本質上就是一種非確定性的分散式系統追蹤（Distributed Tracing）**。這讓 OpenTelemetry 直接成為 AI 可觀測性的底層基石：

1. **GenAI 語意慣例（GenAI Semantic Conventions，2024–2026 年）：**
OTel 官方成立了專門的 **GenAI 工作小組（GenAI SIG）**，將原本各自為政的 AI 埋點標準化。定義了標準的 `gen_ai.*` 屬性規範（例如：`gen_ai.request.model` 記錄模型名稱、`gen_ai.usage.input_tokens` 記錄 Token 消耗、以及 Prompt、Tool Calls 與 Agent 狀態）。
* 這也是為什麼前一則討論中的 **Arize Phoenix（透過 OpenInference）**、LangSmith、Datadog 等平台，在 2025–2026 年都全面對齊並原生支援 OTel GenAI 標準——開發者只要用 OTel 埋點一次，就能自由把 Agent 的 Trace 送到任何支援 OTel 的評估平台。


2. **生態系工程化工具成熟（2025–2026 年）：**
* **OTel Weaver 與宣告式配置（Declarative Configuration）：** 2026 年 3 月正式穩定化了跨語言的 JSON/YAML 宣告式設定檔（`OTEL_CONFIG_FILE`），並透過 OTel Weaver 協助企業自動驗證與生成符合語意慣例的遙測 Schema。
* **OTel Arrow 與 OpAMP：** 結合 Apache Arrow 大幅壓縮龐大的遙測資料傳輸頻寬（特別適合資料量極大的 AI Agent 追蹤），並透過 OpAMP 協定遠端管理成千上萬個 Collector。


3. **正式從 CNCF 畢業（2026 年 5 月）：**
在經歷了長達七年的演進、每月超過 150 間企業與上千名開發者共同維護後，**OpenTelemetry 於 2026 年 5 月正式從 CNCF 畢業（Graduated Project）**，與 Kubernetes、Prometheus 並列為雲端原生最高成熟度等級的基礎設施標準。

---

### 總結：OpenTelemetry 的演進脈絡

| 時期 | 階段名稱 | 關鍵演進重點 |
| --- | --- | --- |
| **2016–2018** | **雙雄割據期** | OpenTracing（純 API 標準）與 OpenCensus（Google 主導的 API+SDK）各自發展。 |

# 收費細節(五款主流 AI Tracing 工具)

這五款工具在計費時最大的陷阱在於「計價單位不同」：一次複雜的 Google ADK Agent 執行（稱作 1 個 **Trace**），底層通常會拆解出 10 到 50 個不等的子步驟（稱作 **Span / Observation / Unit / Event / Record**）。因此看免費額度與超額單價時，必須把這個倍數算進去。

五家工具的**免費額度上限、起跳月費與超額計費單價細節**如下：

**1. AgentOps**

* **計費單位**：**Event / Span**（每一次 LLM 呼叫、Tool 執行或 Agent 動作都算 1 個 Event）。
* **免費額度（Basic 方案）**：**$0 / 月**，每月包含 **5,000 個 Events**（歷史紀錄保留期有限，適合個人原型開發）。
* **付費方案與計價細節**：
* **Pro 方案**：底價 **$40 / 月**起跳（內含基礎用量，一般約 50,000 至 100,000 Events，解鎖無上限 Log 保留期、5 個團隊席次與資料匯出）。
* **超額單價**：採 Pay-as-you-go（依用量階梯計費），官網未公開固定單一超額費率表，需在後台透過用量滑桿（Calculator）依預估 Event 數量動態計價。
* **Enterprise 方案**：客製化報價（支援私有雲/AWS/GCP 自架、SOC2/HIPAA 與自訂保留期）。



**2. Phoenix Cloud 雲端託管版（現與 Arize AX 整合）**

* **計費單位**：**Span**（加上儲存容量 GB 上限）。
* **免費額度**：
* **自架開源版（Self-hosted Phoenix）**：**完全免費、無任何額度上限**（自行負擔伺服器成本）。
* **雲端免費版（Phoenix Cloud / Arize AX Free）**：**$0 / 月**，每月包含 **25,000 個 Spans** 與 **1 GB 儲存量**，資料保留 **15 天**，**不限使用者席次（Unlimited Users）**。


* **付費方案與計價細節**：
* **Pro 方案（AX Pro）**：固定月費 **$50 / 月**，每月包含 **50,000 個 Spans** 與 **10 GB 儲存量**，資料保留 **30 天**，同樣**不限使用者人數**。
* **超額機制（需特別注意）**：Pro 方案的 50,000 Spans 是**硬上限（Hard Cap）**，官方未提供公開的按量超額單價（No published overage rate）；一旦單月超過 5 萬個 Spans，必須升級至客製化的 **Enterprise 方案**，否則會暫停接收新資料。



**3. Langfuse (Langfuse Cloud)**

* **計費單位**：**Unit（即 Observation）**，代表 1 個 Trace + 裡面的每個 Span + 每筆評估分數（Score）的總和（例如 1 次 Agent 執行包含 6 個步驟，就消耗 7 個 Units）。
* **免費額度（Hobby 方案）**：**$0 / 月**，每月包含 **50,000 個 Units**，最多 **2 位使用者**，資料保留 **30 天**。（若選擇自架 MIT 開源版則完全免費無上限）。

* **超額單價（階梯式累進降價，Core 與 Pro 相同）**：

* 100 萬 ～ 1,000 萬 Units：每 100,000 Units 收費 **$7.00**。
* 1,000 萬 ～ 5,000 萬 Units：每 100,000 Units 收費 **$6.50**。

# 名詞解釋

**1. CNCF（Cloud Native Computing Foundation，雲端原生計算基金會）**

* **它是什麼：** 成立於 2015 年、隸屬於 Linux 基金會旗下的非營利開源組織。它的核心使命是推動「雲端原生（Cloud-Native）」技術的標準化與普及，確保這些關鍵開源專案保持**廠商中立（Vendor-neutral）**，不會被單一科技巨頭壟斷。
* **成熟度分級機制：** CNCF 會根據專案的穩定度、安全審計與企業採用率，將專案分為三個等級：**沙盒級（Sandbox，早期實驗）** $\rightarrow$ **孵化級（Incubating，成長成熟中）** $\rightarrow$ **畢業級（Graduated，最高成熟度標準，可放心用於大型生產環境）**。
* **為什麼重要：** 當一個專案成為「CNCF 畢業專案」（例如 Kubernetes、Prometheus 與 OpenTelemetry），就代表它已經通過全球各大科技大廠的嚴格檢驗，成為業界公認的基礎設施標準。

**2. Kubernetes（常簡稱為 K8s，容器編排系統）**
*(名稱源自希臘語的「舵手」，因字母 K 與 s 之間有 8 個字母而簡稱 K8s)*

* **它是什麼：** 由 Google 根據內部運行十多年的 Borg 系統經驗重新設計，並於 2014 年開源、2015 年作為「第 1 號種子專案」捐贈給 CNCF（因此催生了 CNCF 的成立）。它是目前全球最主流的**容器編排平台（Container Orchestration Platform）**。
* **解決什麼痛點：** 當工程師用 Docker 把應用程式（或多個 AI Agents）打包成一個個「容器（Containers）」後，如果只有兩三個容器可以手動啟動；但在企業環境中往往有成千上萬個容器跨越多台伺服器運行。
* **核心功能：** K8s 就像一位全自動的「機房總指揮」，負責自動部署容器、自動擴縮容（Auto-scaling，流量暴增時自動複製容器、沒人用時自動縮減）、負載平衡（Load Balancing），以及自我修復（Self-healing，某台機器或容器掛掉時，瞬間在別台機器上重啟一個新的）。

**3. Prometheus（普羅米修斯，開源監控與告警系統）**

* **它是什麼：** 2012 年由音樂串流平台 SoundCloud 的前 Google 工程師開發（靈感源自 Google 內部的 Borgmon 監控系統），並於 2016 年加入 CNCF，成為繼 Kubernetes 之後**第 2 個從 CNCF 畢業**的專案。它是一套專門處理時間序列資料庫（Time-Series Database, TSDB）與指標監控（Metrics Monitoring）的系統。
* **解決什麼痛點：** 在 K8s 這種容器隨時會生滅、IP 隨時會變動的動態環境裡，傳統綁定固定伺服器 IP 的監控工具完全失效。Prometheus 能自動發現（Service Discovery）K8s 裡新長出來的容器，並主動去拉取（Pull / Scrape）它們的健康數據。
* **核心功能：** 它專門收集隨時間變化的數值指標（Metrics），例如：CPU 與記憶體使用率、每秒請求數（RPS）、API 延遲時間或 LLM 的 Token 消耗速率。工程師會使用它專屬的查詢語言 **PromQL** 來分析數據、搭配 Grafana 畫出即時監控儀表板，並透過 Alertmanager 在系統異常時自動發送告警通知。

**三者與 OpenTelemetry / AI 系統的關係總結**

| 名稱 | 定位類型 | 在現代 AI / 雲端系統中的具體職責 |
| --- | --- | --- |
| **CNCF** | **中立組織與標準制定者** | 託管並培育 Kubernetes、Prometheus、OpenTelemetry 等專案，確保開源生態系健康且不綁定單一雲端大廠。 |
| **Kubernetes (K8s)** | **運算與容器調度層** | 負責在底層伺服器叢集上，穩定運行與調度你的 AI Agent 服務、資料庫與 Arize Phoenix 伺服器。 |
| **Prometheus** | **宏觀指標監控（Metrics）** | 告訴你「系統現在健不健康」（例如：某個 Agent 容器記憶體爆了、API 錯誤率飆高）。 |
| **OpenTelemetry / Phoenix** | **微觀鏈路追蹤（Traces）** | 當 Prometheus 發現錯誤率飆高時，讓你深入拆解「為什麼會錯」（例如：看到是哪一個 Sub-agent 的 Prompt 導致了幻覺或 Tool 呼叫失敗）。 |


# Claude Code, Codex, ChatGPT 執行紀錄

這三款工具在執行時確實會在本機留下執行紀錄、對話歷史或快取檔案（大多位於隱藏資料夾中）。以下為各工具在 macOS、Linux 與 Windows 系統上的詳細存放路徑：

**1. Claude Code (CLI 終端機工具)**
Claude Code 會將你的每一次對話、讀取過的檔案、執行的終端機指令與工具呼叫結果，以純文字的 `.jsonl` 格式完整記錄在本機：

* **對話與操作過程紀錄 (Session Logs)：**
* **macOS / Linux：** `~/.claude/projects/`
* **Windows：** `C:\Users\<你的使用者名稱>\.claude\projects\`
* *說明：* 進入該資料夾後，會看到以你「專案絕對路徑」命名的子資料夾（例如 `-Users-name-myproject`），裡面的 `.jsonl` 檔案就是完整的過程追蹤紀錄，可用一般文字編輯器直接打開檢視。


* **全域設定與狀態檔：**
* **路徑：** `~/.claude.json` 以及 `~/.claude/settings.json`（包含偏好設定、權限允許清單與專案統計）。


* **MCP 與系統除錯日誌 (若結合 Claude Desktop 使用)：**
* **macOS：** `~/Library/Logs/Claude/`
* **Windows：** `%APPDATA%\Claude\logs\`（即 `C:\Users\<使用者名稱>\AppData\Roaming\Claude\logs\`）



**2. OpenAI Codex (Codex CLI & 桌面端代理模式)**
當你使用 OpenAI 的 Codex CLI 或是桌面版內建的 Codex 模式時，它會在本機建立 `.codex` 隱藏目錄來存放執行日誌與狀態：

* **執行日誌 (Logs)：**
* **macOS / Linux：** `~/.codex/log/`（例如 `codex-tui.log`）
* **Windows：** `C:\Users\<你的使用者名稱>\.codex\log\`


* **歷史工作階段與狀態追蹤 (Sessions & State)：**
* **歷史紀錄：** `~/.codex/sessions/`（Windows 為 `%USERPROFILE%\.codex\sessions\`）
* **狀態資料庫：** `~/.codex/` 目錄下的 `.sqlite` 檔案（例如 `state_5.sqlite`），用來追蹤本機連線與工作階段狀態。
* **設定檔：** `~/.codex/config.toml`（以及各專案根目錄下的 `.codex/config.toml`）。



**3. ChatGPT (網頁版 vs. 桌面版 App)**

* **網頁版 (Browser)：** 不會在本機硬碟產生獨立的 log 檔。對話紀錄儲存在 OpenAI 雲端伺服器，本機僅在瀏覽器的 `IndexedDB` 與 `Local Storage` 暫存認證 Token 與介面快取。
* **桌面版應用程式 (Desktop App)：** 會在系統的應用程式資料夾中存放執行期快取、狀態資料庫與系統日誌（對話內容本身經加密同步至雲端，不會以純文字明碼存放）：
* **macOS 路徑：**
* 應用程式資料與快取：`~/Library/Application Support/com.openai.chat/`
* 系統與崩潰日誌：`~/Library/Logs/com.openai.chat/`（或 `~/Library/Caches/com.openai.chat/`）


* **Windows 路徑：**
* 設定與狀態資料：`%APPDATA%\ChatGPT\`（即 `C:\Users\<使用者名稱>\AppData\Roaming\ChatGPT\`）
* 本機快取與執行檔：`%LOCALAPPDATA%\ChatGPT\` 或 `%LOCALAPPDATA%\Packages\` 下以 `OpenAI.ChatGPT-Desktop` 開頭的資料夾（若從 Microsoft Store 安裝）。





**如何快速前往查看？**

* **macOS：** 打開 Finder，按下快捷鍵 `Cmd + Shift + G`，直接貼上上述包含 `~` 的路徑即可前往（或在一般資料夾中按 `Cmd + Shift + .` 顯示隱藏檔案）。
* **Windows：** 按下 `Win + R` 開啟「執行」視窗，輸入 `%USERPROFILE%\.claude`、`%USERPROFILE%\.codex` 或 `%APPDATA%` 即可直接打開對應的隱藏目錄。

## Authentication: End-User Identity Propagation
- https://adventofagents.com/2026/03/26




### End-User Authentication with ADK

https://www.youtube.com/watch?v=NzQM4GihwQU

在實際部署 AI Agent 時，常見需求之一是讓 Agent 代表使用者存取私人資源，例如 Google Calendar、Gmail 或 Google Drive。

這類操作不能只依賴 Agent 自己的系統憑證，因為 Agent 必須取得「目前使用者」授權的存取權限。實務上通常透過 **OAuth 2.0** 完成，而不需要把使用者密碼交給 Agent。

## OAuth 2.0 的基本流程

在這個流程中，主要有四個角色：

- User：使用者
- Client：例如 ADK Web
- Authorization Server：例如 Google OAuth Server
- Agent Backend：執行工具與 API 呼叫的 Agent


假設使用者要求：

> 取得我今天的 Calendar Events。

Client 不會直接要求使用者提供 Google 密碼，而是將使用者導向 Google 的授權頁面。

使用者同意授權後，Authorization Server 會先回傳一個 **Authorization Code**。

Client 再使用 Authorization Code 換取：

- Access Token
- Refresh Token

Agent 使用 Access Token 呼叫 Google Calendar API。

Access Token 通常具有較短的有效期限；當 Token 過期時，可以使用 Refresh Token 取得新的 Access Token。

整體流程可以表示為：

$$
\text{User}
\rightarrow
\text{Authorization}
\rightarrow
\text{Authorization Code}
\rightarrow
\text{Access Token}
\rightarrow
\text{API}
$$

因此，Agent 實際取得的是有限期限、有限權限的 Token，而不是使用者帳號密碼。

## 在 ADK 中建立 Calendar Agent

Agent 本身可以定義為一個 Calendar Assistant，例如：

- 查詢 Calendar Events
- 取得單一 Event 的詳細資料

此外，也可以透過 `before_agent_callback` 在每次 Agent 執行前執行 Python Function，例如取得目前時間，再將結果提供給 Agent。

概念上：

```python
root_agent = Agent(
    model=...,
    instruction=...,
    tools=[...],
    before_agent_callback=...
)
```

`before_agent_callback` 適合處理 Agent 執行前需要動態產生的 Context，例如：

```text
Current Time
User Context
Runtime Metadata
```

## 使用 ADK 內建 Tool

Google Calendar 是 ADK 支援的內建工具之一，因此可以直接建立 Calendar Toolset。

通常需要提供：

```text
client_id
client_secret
tool_filter
```

其中 `client_id` 和 `client_secret` 用於 OAuth 2.0 Authentication。

`tool_filter` 則用來限制 Agent 可以使用哪些 Tool。

例如只允許：

```text
list events
get event
```

而不提供：

```text
create event
update event
delete event
```

這樣即使使用者要求 Agent 刪除 Calendar Event，Agent 也沒有對應的 Tool 可以執行。

這是 Agent 系統中非常重要的權限設計原則：

$$
\text{Agent Capability}
\subseteq
\text{Authorized Tools}
$$

Agent 能做什麼，不應只由 Prompt 決定，而應由實際可使用的 Tool 與 Permission 決定。

## Custom Tool 的 Authentication

如果 API 沒有 ADK 內建工具，可以自己建立 Python Function，例如：

```python
def list_calendar_events(...):
    ...
```

這個 Function 可以直接呼叫 Google Calendar API 或其他第三方 API。

但此時需要告訴 ADK：

> 這個 Function 在執行之前需要 Authentication。

可以使用 `AuthenticatedFunctionTool` 包裝原本的 Python Function。

概念上：

```python
AuthenticatedFunctionTool(
    func=list_calendar_events,
    auth_scheme=...,
    auth_credential=...
)
```

其中有兩個重要部分。

第一個是 `auth_scheme`，描述 Authentication Protocol，例如：

```text
OAuth 2.0
API Key
OIDC
```

如果使用 OAuth 2.0，通常還需要設定：

```text
Authorization URL
Token URL
Scopes
```

第二個是 `auth_credential`，其中包含 OAuth Client 的資訊，例如：

```text
client_id
client_secret
```

## Scope 與最小權限原則

OAuth 2.0 的一個重要概念是 **Scope**。

Scope 定義 Access Token 可以執行哪些操作。

例如 Calendar Read-Only Scope：

```text
calendar.readonly
```

代表 Agent 可以：

```text
Read
List
Get
```

但不能：

```text
Create
Update
Delete
```

因此使用者在 OAuth Consent Screen 上看到的，也只會是「查看 Calendar」相關權限。

這符合安全系統中的 **Principle of Least Privilege**：

$$
\text{Granted Permission}
=
\text{Minimum Permission Required}
$$

Agent 不應取得超過任務所需的權限。

---

## Tool Permission 與 Agent 行為

假設 Agent 只有：

```text
list_events
get_event
```

當使用者要求：

> Create a lunch event tomorrow at 1 PM.

Agent 無法完成這項操作，因為系統沒有提供 `create_event` Tool。

因此，一個安全的 Agent 系統不應只依靠 Prompt：

```text
Do not delete events.
```

而應直接從 Tool Layer 限制能力：

```text
沒有 delete_event Tool
→ Agent 無法刪除 Event
```

Tool Availability 本身就是 Agent 的 Permission Boundary。

## 建立 OAuth Client

如果要存取 Google Calendar，需要先在 Google Cloud Console 建立 OAuth Client。

基本流程：

```text
Google Cloud Console
→ OAuth Client
→ Web Application
→ Redirect URI
```

Redirect URI 必須加入 ADK Web 使用的 Callback URI。

建立完成後會取得：

```text
Client ID
Client Secret
```

開發環境可以透過 Environment Variables 載入。

例如：

```text
GOOGLE_CLIENT_ID
GOOGLE_CLIENT_SECRET
```

但在 Production Environment 中，不應把 Secret 直接寫在 Source Code 或一般設定檔。

應使用 Secret Management System，例如：

```text
Google Secret Manager
AWS Secrets Manager
HashiCorp Vault
```

## 核心概念

ADK 的 End-User Authentication 可以理解為：

$$
\text{User Identity}
+
\text{OAuth Authorization}
+
\text{Restricted Tool Access}
$$

OAuth 2.0 解決的是：

> Agent 如何安全地代表使用者呼叫外部 API。

而 Tool Filter 與 OAuth Scope 解決的是：

> Agent 被允許執行哪些操作。

因此，在 Production Agent 中，Authentication 與 Authorization 不只是登入問題，而是 Agent Tool Architecture 的核心安全邊界。

整個流程可以理解成：

1. 使用者要求 Agent 存取私人資料，例如「列出我的 Calendar Events」。

2. Agent 發現目前沒有使用者的授權，因此不能直接呼叫 Google Calendar API。

3. ADK Web 將使用者導向 Google 的 OAuth 授權頁面。

4. 使用者在 Google 登入並同意權限，例如只允許讀取 Calendar。

5. Google 將 **Authorization Code** 傳回 ADK Web。

6. ADK Web 使用 Authorization Code 向 Google 換取：

```text
Access Token
Refresh Token
```

7. Agent 使用 **Access Token** 呼叫 Google Calendar API。

8. Access Token 過期後，可以利用 **Refresh Token** 取得新的 Access Token。

核心流程就是：

$$
\text{User Request}
\rightarrow
\text{OAuth Authorization}
\rightarrow
\text{Authorization Code}
\rightarrow
\text{Access Token}
\rightarrow
\text{API Call}
$$

最重要的概念是：

**Agent 不需要取得使用者密碼，而是取得使用者授權後的 Access Token。**

## Authenticating with tools

https://adk.dev/tools-custom/authentication/?utm_source=adventofagents&utm_medium=web&utm_campaign=adventofagents_s2_2026&utm_content=day26#api-key

### Part 1：ADK Authentication 整體架構

在 ADK 中，Agent 常需要透過 Tool 存取外部服務，例如：

```text
User
  ↓
Agent
  ↓
Tool
  ↓
External API
```

如果外部 API 是受保護的資源，例如 Calendar、Email、Database 或 Cloud Service，Tool 在呼叫 API 前必須先取得有效的 Credential。

因此，Authentication 的核心問題是：

> Agent 的 Tool 要用什麼方式與 Credential，才能合法呼叫受保護的 API？

---

## 1. Authentication 在 Agent 架構中的位置

可以先把整體流程理解成：

```text
User Request
    ↓
Agent
    ↓
Select Tool
    ↓
Tool needs protected API
    ↓
Authentication
    ↓
API Call
    ↓
Tool Result
    ↓
Agent Response
```

Authentication 不屬於 LLM 本身，而是在：

```text
Tool
↕
External API
```

之間建立存取權限。

Agent 負責決定「要不要使用 Tool」；Authentication 則負責確認「這個 Tool 有沒有資格呼叫 API」。

---

## 2. ADK 的兩個核心 Authentication 元件

ADK 使用兩個主要物件描述 Authentication：

```text
AuthScheme
AuthCredential
```

### AuthScheme

`AuthScheme` 描述：

> API 要求使用什麼方式進行 Authentication。

例如：

```text
API Key
HTTP Bearer
OAuth 2.0
OpenID Connect
```

可以簡單理解成：

```text
AuthScheme
→ How to authenticate?
→ 要怎麼認證？
```

例如某個 API 可能要求：

```http
Authorization: Bearer <token>
```

另一個 API 則可能要求：

```text
?apikey=YOUR_API_KEY
```

因此，`AuthScheme` 描述的是 Authentication 的規則，而不是 Credential 本身。ADK 的 Authentication Scheme 與 OpenAPI 3.0 的 Authentication 定義相容。

---

### AuthCredential

`AuthCredential` 描述：

> 執行或啟動 Authentication 所需要的 Credential。

例如：

```text
API Key
→ API Key Value

HTTP Bearer
→ Bearer Token

OAuth 2.0
→ Client ID + Client Secret

Service Account
→ Service Account Credential

OpenID Connect
→ Client ID + Client Secret
```

所以兩者可以簡化成：

```text
AuthScheme
→ 怎麼認證

AuthCredential
→ 拿什麼認證
```

ADK 也會透過 `auth_type` 區分不同 Credential 類型，例如：

```text
API_KEY
OAUTH2
SERVICE_ACCOUNT
OPEN_ID_CONNECT
```

## 3. OpenAPIToolset 的角色

如果外部 API 有 OpenAPI Specification，可以使用：

```python
OpenAPIToolset(
    spec_str=...,
    spec_str_type=...,
    auth_scheme=...,
    auth_credential=...
)
```

這裡有三個核心部分。

```text
OpenAPI Spec
→ API 有哪些 Endpoint
→ 參數是什麼
→ 可以執行哪些 Operation

AuthScheme
→ API 要怎麼 Authentication

AuthCredential
→ 實際使用什麼 Credential
```

因此可以理解成：

```text
OpenAPI Spec
     ↓
定義 API 能做什麼

AuthScheme
     ↓
定義 API 怎麼認證

AuthCredential
     ↓
提供認證所需 Credential

OpenAPIToolset
     ↓
建立 Agent 可以使用的 Tools
```

對 OpenAPI-based Toolset 而言，`auth_scheme` 與 `auth_credential` 會在 Toolset 初始化時提供，並套用到所產生的 Tools。

---

## 4. Authentication 不是 Agent 啟動時一定立刻執行

設定：

```python
auth_scheme = ...
auth_credential = ...
```

不代表程式啟動時就一定立即執行 Authentication。

通常是在 Agent 真正選擇需要受保護 API 的 Tool 時，才會需要 Authentication。

```text
User
↓
提出任務

Agent
↓
選擇 Tool

Tool
↓
需要 Protected API

ADK
↓
檢查 Credential
```

如果已有有效 Credential：

```text
Tool
→ API Call
```

如果沒有 Credential，而且需要 User Interaction：

```text
Tool Call
↓
Authentication Required
↓
User Login / Consent
↓
取得 Credential
↓
Retry Tool Call
↓
API Call
```

OAuth 2.0 與 OpenID Connect 常屬於這類 Interactive Authentication。ADK 會暫停目前執行流程，通知 Agent Client 完成 Authentication，再繼續原本的 Tool Call。

## 5. Interactive 與 Non-interactive Authentication

從 Agent Runtime 的角度，可以先把 Authentication 分成兩類。

### Non-interactive Authentication

通常不需要 User 在執行過程中登入。

例如：

```text

```

概念上：

```text

↓

↓
Attach Credential
↓
API Call
```

---

### Interactive Authentication

需要 User 參與 Authentication Flow。

例如：

```text

OpenID Connect
```

概念上：

```text

↓
Authentication Required
↓
User Login / Consent
↓
Token
↓
Tool Retry
↓
API Call
```

這個差異很重要，因為它會直接影響 Agent Runtime 是否需要暫停與恢復執行。

整體架構可以整理成：

```text

                      ↓

               /            \
```


```text id="7qof53"

```


# 2. HTTP Bearer

HTTP Bearer 的核心概念是：

> Token 已經存在，Client 只需要把 Token 放進 HTTP Authorization Header。

Request 格式是：

```http id="uf6fmc"
Authorization: Bearer <token>
```

例如：

```http id="h2lv6d"
Authorization: Bearer eyJhbGciOi...
```

---

## Credential

Credential 是：

```text id="zm81y7"
Bearer Token
```

這個 Token 可能來自：

```text id="iy9qic"
OAuth 2.0
OIDC
Frontend Login
另一個 Authentication Service
External Identity Provider
```

HTTP Bearer 本身不負責產生 Token。

它只是定義：

> 已經有 Token 之後，要怎麼把它送給 API。

因此要特別區分：

```text id="7n5cvs"
OAuth 2.0
→ How to obtain a token

HTTP Bearer
→ How to send an existing token
```

---

## ADK 程式碼

目前 ADK 的 HTTP Credential 可以用：

```python id="l6ajpq"
from fastapi.openapi.models import HTTPBearer

from google.adk.auth.auth_credential import (
    AuthCredential,
    AuthCredentialTypes,
    HttpAuth,
    HttpCredentials,
)

auth_scheme = HTTPBearer()

auth_credential = AuthCredential(
    auth_type=AuthCredentialTypes.HTTP,
    http=HttpAuth(
        scheme="bearer",
        credentials=HttpCredentials(
            token="YOUR_EXISTING_TOKEN"
        ),
    ),
)
```

`HTTPBearer()` 定義 API 使用：

```text id="zk75u0"
HTTP Authorization: Bearer
```

而：

```python id="3h40ck"
HttpCredentials(
    token="YOUR_EXISTING_TOKEN"
)
```

保存實際 Token。

目前 ADK 的 `HttpAuth` 明確支援 `bearer` scheme，而 `HttpCredentials` 具有 `token` 欄位；`AuthCredentialTypes.HTTP` 則表示這是一組 HTTP Authentication Credential。

接著仍然可以交給：

```python id="zkjl9q"
toolset = OpenAPIToolset(
    spec_str=...,
    spec_str_type="yaml",
    auth_scheme=auth_scheme,
    auth_credential=auth_credential,
)
```

---

## Runtime

```text id="ar8nyl"

↓

↓

Authorization: Bearer TOKEN

↓
API Call
```

沒有 Authorization Code Flow。

## 什麼時候使用？

典型情況：

```text id="qxvhm1"

↓
User 已經登入
↓
Frontend 已經取得 Token
↓
Token 傳到 Agent Backend
↓
Agent 呼叫其他 API
```

因此 HTTP Bearer 很適合：

> Authentication 已經由其他系統完成，Agent 只需要使用現有 Token。

ADK 文件也將 HTTP Credential 描述為可以直接使用已取得的 Bearer Token，不需要額外 Credential Exchange。

# 3. OAuth 2.0

OAuth 2.0 的目的主要是：

> 讓 Application 在 User 授權後，代表 User 存取資源。

```text id="k6fn08"

↓
```

# 五種方法放在一起比較

最重要的差異可以整理成：

| 方法 | Credential 從哪裡來 | 主要代表 | 是否需要 Login |
|---|---|---|---|
| API Key | API Provider | Application / Project | 否 |
| HTTP Bearer | 已存在的 Token | 依 Token 而定 | 否 |
| OAuth 2.0 | OAuth Provider | User Authorization | 通常是 |
| Service Account | Cloud / Service Identity | Application / Service | 否 |
| OpenID Connect | Identity Provider | End User Identity | 通常是 |

最簡單的判斷方式是：

```text id="0hcsam"
只有固定 Key
→ API Key

已經有 Token
→ HTTP Bearer

需要 User 授權資料
→ OAuth 2.0

Backend 自己呼叫 Backend
→ Service Account

需要確認登入 User 身分
→ OpenID Connect
```

其中最容易混淆的是：

```text id="s1of96"
HTTP Bearer
vs
OAuth 2.0
```

因為 OAuth2 最後也經常使用：

```http id="1skkke"
Authorization: Bearer <access_token>
```

但兩者不在同一層：

```text id="iqswgy"
OAuth 2.0
→ Token 如何被取得

HTTP Bearer
→ Token 如何被送進 HTTP Request
```

同樣地：

```text id="v21mm5"
OAuth 2.0
vs
OpenID Connect
```

主要差異是：

```text id="e2k29b"
OAuth 2.0
→ Authorization
→ 可以做什麼

OpenID Connect
→ Identity
→ 你是誰
```

因此這五種方法可以先形成以下 Mental Model：

```text id="qdpgf6"
API Key
→ Fixed Secret

HTTP Bearer
→ Existing Token

OAuth 2.0
→ Delegated Authorization

Service Account
→ Machine Identity

OpenID Connect
→ User Identity
```


### Part 3：ADK Authentication Runtime Flow

前兩段已經介紹 Authentication 的整體架構，以及不同 Authentication 方法的用途。

這一段只討論一件事：

> 當 Agent 執行到需要互動式 Authentication 的 Tool 時，ADK Runtime 實際怎麼暫停、取得 Credential，再恢復原本的 Tool Call？

這裡主要以 OAuth 2.0 / OpenID Connect 這類 Interactive Authentication 為例。

---

## 1. Agent Client 先執行 Agent

Agent 通常透過 `Runner` 執行：

```python
events_async = runner.run_async(
    session_id=session.id,
    user_id="user",
    new_message=content,
)
```

`runner.run_async()` 不只回傳最後答案，而是產生一連串 `Event`。

因此 Client 會逐一檢查：

```python
async for event in events_async:
    ...
```

如果 Agent 執行過程不需要額外 Authentication，就正常執行完成。

但如果某個 Tool 需要 User Authentication，ADK 會產生一個特殊事件：

```text
adk_request_credential
```

這代表：

> 目前 Agent 無法繼續，Client 必須先完成 Authentication。



---

## 2. `adk_request_credential`

Client 可以檢查 Event 中是否出現：

```python
part.function_call.name == "adk_request_credential"
```

例如：

```python
async for event in events_async:
    if (
        auth_request_function_call :=
        get_auth_request_function_call(event)
    ):
        auth_request_function_call_id = (
            auth_request_function_call.id
        )

        auth_config = get_auth_config(
            auth_request_function_call
        )

        break
```

這裡需要保存兩個重要資訊：

```text
auth_request_function_call_id
auth_config
```

### `auth_request_function_call_id`

這是 Authentication Request 的 ID。

之後 Authentication 完成時，Client 必須用同一個 ID 回覆 ADK。

可以把它理解成：

```text
Authentication Request A
        ↓
        ID = 123
        ↓
Authentication Response
        ↓
        ID = 123
```

ADK 才知道：

> 這個 Authentication Response 是在回答哪一個 Authentication Request。

---

### `auth_config`

`auth_config` 包含目前 Authentication Flow 所需要的設定。

例如：

```text
Authorization URI
Auth Scheme
Credential information
OAuth configuration
```

Client 接下來會使用這個資訊，把 User 導向 Authentication Provider。



---

## 3. 將 User Redirect 到 Authorization Endpoint

ADK 產生 Authentication Request 後，Client 會取得：

```python
base_auth_uri = (
    auth_config
    .exchanged_auth_credential
    .oauth2
    .auth_uri
)
```

接著加入：

```text
redirect_uri
```

例如：

```python
redirect_uri = "http://localhost:8000/callback"

auth_request_uri = (
    base_auth_uri
    + f"&redirect_uri={redirect_uri}"
)
```

然後 Browser 被導向：

```text
Authorization Provider
```

例如 Google、Okta 或其他 Identity Provider。

`redirect_uri` 的作用是：

> Authentication 完成後，Provider 要把 User 導回哪裡。

而且這個 URI 必須事先註冊在 OAuth Provider。



---

## 4. 接收 Callback

User 完成 Login / Consent 後，Provider 會將 Browser Redirect 回：

```text
redirect_uri
```

例如：

```text
http://localhost:8000/callback?code=ABC123
```

這個 Callback URL 中通常包含：

```text
authorization_code
state
scope
```

Client 需要取得完整 Callback URL。

例如：

```python
auth_response_uri = (
    "http://localhost:8000/callback"
    "?code=ABC123"
)
```

這一步通常發生在 Web Application 的 Callback Route，而不是 Agent 本身。



---

## 5. 把 Authentication Result 放回 `AuthConfig`

取得 Callback URL 後，需要更新之前保存的 `auth_config`：

```python
auth_config.exchanged_auth_credential.oauth2.auth_response_uri = (
    auth_response_uri
)

auth_config.exchanged_auth_credential.oauth2.redirect_uri = (
    redirect_uri
)
```

這裡的角色可以理解為：

```text
auth_response_uri
→ Provider 回傳 Authentication 結果的位置

redirect_uri
→ 原本使用的 Callback Endpoint
```

此時 `auth_config` 已經包含 ADK 完成後續 Token Exchange 所需要的資訊。

---

## 6. 用 `FunctionResponse` 回覆 ADK

Authentication 完成後，不能直接重新問 Agent 一次原本的問題。

Client 必須回覆之前的：

```text
adk_request_credential
```

因此建立：

```python
auth_content = types.Content(
    role="user",
    parts=[
        types.Part(
            function_response=types.FunctionResponse(
                id=auth_request_function_call_id,
                name="adk_request_credential",
                response=auth_config.model_dump(),
            )
        )
    ],
)
```

這裡有三個關鍵欄位：

```text
id
→ 對應原本的 Authentication Request

name
→ 必須是 adk_request_credential

response
→ 更新後的 AuthConfig
```

其中：

```python
auth_config.model_dump()
```

把 `AuthConfig` 轉成可以放入 `FunctionResponse` 的資料格式。



---

## 7. 再次呼叫 `runner.run_async()`

Authentication Result 準備完成後，再呼叫一次：

```python
events_async_after_auth = runner.run_async(
    session_id=session.id,
    user_id="user",
    new_message=auth_content,
)
```

注意這裡使用的是：

```text
同一個 session_id
```

因為這不是新的 Conversation，而是在恢復剛才中斷的 Authentication Flow。

整個概念是：

```text
第一次 runner.run_async()
↓
Agent 執行
↓
需要 Authentication
↓
暫停

User 完成 Authentication

第二次 runner.run_async()
↓
送回 Authentication Result
↓
繼續原本流程
```

---

## 8. Token Exchange 由 ADK 完成

收到 `adk_request_credential` 的 `FunctionResponse` 後，ADK 會從 Callback URL 中取得 Authorization Code。

接著向：

```text
Token Endpoint
```

執行 Token Exchange。

例如：

```text
Authorization Code
↓
Token Endpoint
↓
Access Token
↓
可能還有 Refresh Token
```

這部分通常不需要 Agent Developer 自己重新寫 HTTP Request。

ADK 會處理 Token Exchange，讓 Authentication Credential 可以被後面的 Tool 使用。

---

## 9. ADK 自動 Retry 原本的 Tool Call

Authentication 完成後，ADK 不會要求 LLM 重新決定一次要不要呼叫 Tool。

它會重新執行先前因 Authentication 被中斷的 Tool。

因此流程實際上是：

```text
Original Tool Call
↓
Authentication Missing
↓
Pause
↓
Authentication
↓
Credential Available
↓
Retry Original Tool Call
↓
API Result
```

這一點很重要。

Authentication Flow 並不是一個新的 Agent Task，而是原本 Tool Execution 的一部分。

ADK 完成 Token Exchange 後，會讓原本的 Tool 使用新的 Credential，再重新執行。

# 10. Custom FunctionTool 的 Authentication

如果使用的是 `OpenAPIToolset`，許多 Authentication 行為可以由 Framework 處理。

但如果自己寫：

```python
FunctionTool
```

就需要在 Python Function 裡處理更多 Authentication State。

例如：

```python
from google.adk.tools import FunctionTool, ToolContext

def my_authenticated_tool(
    query: str,
    tool_context: ToolContext,
) -> dict:
    ...
```

這裡最重要的是：

```python
tool_context: ToolContext
```

ADK 會自動注入 `ToolContext`。

它讓 Tool 可以：

```text
讀取 Session State
取得 Authentication Result
要求 Authentication
保存 Credential
```



---

## 11. 先檢查 Cached Credential

Custom Tool 不應該每次執行都重新 Authentication。

因此第一步通常是查看：

```python
tool_context.state
```

例如：

```python
TOKEN_CACHE_KEY = "my_tool_tokens"

cached_token_info = tool_context.state.get(
    TOKEN_CACHE_KEY
)
```

如果已有 Credential，就先檢查：

```text
Token 是否存在
Token 是否有效
Token 是否過期
是否有 Refresh Token
```

如果 Token 仍有效：

```text
直接呼叫 API
```

如果 Access Token 過期，但有 Refresh Token：

```python
creds.refresh(Request())
```

取得新的 Access Token。



---

## 12. `tool_context.get_auth_response()`

如果沒有 Cached Credential，Tool 可以檢查：

```python
exchanged_credential = tool_context.get_auth_response(
    AuthConfig(
        auth_scheme=auth_scheme,
        raw_auth_credential=auth_credential,
    )
)
```

如果 Authentication 剛完成：

```python
exchanged_credential
```

就會包含 ADK 已經交換完成的 Credential。

例如：

```python
access_token = (
    exchanged_credential.oauth2.access_token
)

refresh_token = (
    exchanged_credential.oauth2.refresh_token
)
```

也就是：

```text
Client 完成 User Authentication
↓
ADK 完成 Token Exchange
↓
Tool 透過 get_auth_response()
取得 Token
```



---

## 13. `tool_context.request_credential()`

如果：

```text
沒有 Cached Credential
+
沒有 Authentication Response
```

代表現在必須正式要求 User Authentication。

Tool 可以呼叫：

```python
tool_context.request_credential(
    AuthConfig(
        auth_scheme=auth_scheme,
        raw_auth_credential=auth_credential,
    )
)
```

接著回傳：

```python
return {
    "pending": True,
    "message": "Awaiting user authentication.",
}
```

ADK 看到 Credential Request 後，就會產生前面介紹的：

```text
adk_request_credential
```

並暫停目前執行流程。



---

# 14. Custom Tool 的完整 Runtime Logic

Custom Authentication Tool 可以濃縮成：

```text
Tool starts
↓
Check cached credential
↓
Valid?
├── Yes
│   ↓
│   API Call
│
└── No
    ↓
Check get_auth_response()
    ↓
Credential returned?
├── Yes
│   ↓
│   Cache Credential
│   ↓
│   API Call
│
└── No
    ↓
request_credential()
    ↓
Pause Tool
```

等 User 完成 Authentication 後：

```text
Tool runs again
↓
get_auth_response()
↓
取得 Token
↓
Cache
↓
API Call
```

這就是 Custom `FunctionTool` 和 ADK Authentication Runtime 的主要整合方式。

# 15. Credential Cache

Token 取得後，可以暫存在：

```python
tool_context.state
```

例如：

```python
tool_context.state[TOKEN_CACHE_KEY] = (
    json.loads(creds.to_json())
)
```

目的是避免：

```text
每一次 Tool Call
↓
重新 Login
↓
重新 Consent
```

而是：

```text
第一次
Authentication

後續
Reuse Credential
```



---

# 16. Token Expiration

Access Token 通常不是永久有效。

因此 Runtime 需要處理：

```text
Valid Token
Expired Token
Revoked Token
Invalid Scope
```

例如：

```python
if not creds.valid and creds.expired and creds.refresh_token:
    creds.refresh(Request())
```

如果 API 回傳：

```text
401 Unauthorized
```

可能代表：

```text
Token invalid
Token expired
Token revoked
```

而：

```text
403 Forbidden
```

通常表示：

```text
Credential 有效
但是沒有足夠權限
```

因此 Authentication Runtime 不只是「取得 Token」，還包括：

```text
Token Lifecycle Management
```

---

# 17. Production 中不要把 Credential 當普通 Session Data

在 Demo 中，可以把 Credential 暫存在 Session State。

但 Production 系統需要特別注意：

```text
Access Token
Refresh Token
Client Secret
Service Account Credential
```

尤其 Refresh Token 通常生命週期較長，一旦洩漏可能造成較大風險。

ADK 文件建議 Production 優先使用 Authentication Manager 或 Secret Manager；如果自行管理 Interactive OAuth / OIDC Credential，則應將 Token 與已驗證的 User ID 關聯，並保存於加密儲存系統。

因此：

```text
Development
→ In-memory storage 可以方便測試

Production
→ Authentication Manager
→ Secret Manager
→ Encrypted Credential Storage
```

ADK 的 `InMemorySessionService` 比較適合 Early Development 與 Testing，而不應被視為 Production Credential Store。

---

# 18. 完整 Runtime Mental Model

把這一段濃縮後，ADK Interactive Authentication 的 Runtime 是：

```text
User Request
↓
Agent Runtime
↓
Tool Execution
↓
Credential Missing
↓
adk_request_credential
↓
Agent execution pauses
↓
Agent Client redirects User
↓
User completes Authentication
↓
Callback URL
↓
FunctionResponse
↓
runner.run_async()
↓
ADK Token Exchange
↓
Credential available
↓
Retry original Tool
↓
Protected API
↓
Tool Result
↓
Agent continues
```

而對 Custom `FunctionTool` 而言，最重要的三個 Runtime API 是：

```python
tool_context.state
```

負責保存目前 Session 的 Authentication State。

```python
tool_context.get_auth_response(...)
```

負責取得完成 Authentication 後的 Credential。

```python
tool_context.request_credential(...)
```

負責要求 ADK 啟動 Interactive Authentication。

因此 Part 3 的核心不是 Authentication Protocol 本身，而是：

$$
\text{Pause}
\rightarrow
\text{Authenticate}
\rightarrow
\text{Resume}
\rightarrow
\text{Retry Tool}
$$

這是 ADK 將 Authentication 整合進 Agent Runtime 的主要機制。

```python
      events = list_calendar_events(
          start_time='2024-09-17T06:00:00',
```

```python
def list_calendar_events(
```

```python
credential
```

# 5. 從 `AuthCredential` 取得 Token

```python
creds = Credentials(
    token=credential.oauth2.access_token,
    refresh_token=credential.oauth2.refresh_token,

```

這裡的：

```python
credential.oauth2.access_token
credential.oauth2.refresh_token
```

不是 Developer 自己去 Google Token Endpoint 手動取得的。

而是：

```text
ADK OAuth Flow
↓
Google Authorization
↓
Token Exchange
↓
AuthCredential
↓
Custom Tool
```

# 19. 這個 Sample 最重要的兩種實作模式

### Pattern B：Authenticated Custom Function

```text
  Developer 自己寫
```

### ngrok

ngrok 的主要用途，是把你電腦或 Colab 裡「只能本機存取的服務」，暫時暴露成一個可以從 Internet 存取的公開 HTTPS URL。

例如你在 Colab 裡跑：

```text

localhost:8081
```

正常情況下，Google OAuth 無法直接連到這個 `localhost`。

```text

   ↓

   ↓

   ↓
你的 Flask / API / Web App
```

所以可以把它理解成：

> ngrok = 把 local server 暫時變成 public server。

### Step 4：建立 Google Calendar Tools

先建立一個 Helper，把 ADK Credential 轉成 Google API Client 可以使用的 Credential。

````python
from google.oauth2.credentials import Credentials
from googleapiclient.discovery import build

from google.adk.tools.tool_context import ToolContext


def get_google_credentials(
    credential: AuthCredential,
) -> Credentials:

    return Credentials(
        token=credential.oauth2.access_token,
        refresh_token=credential.oauth2.refresh_token,

        token_uri=(
            "https://oauth2.googleapis.com/token"
        ),

        client_id=oauth_client_id,
        client_secret=oauth_client_secret,

        scopes=[
            CALENDAR_SCOPE
        ],
    )
````

#### Tool 1：查詢 Calendar

````python
def list_calendar_events(
    start_time: str,
    end_time: str,
    limit: int,
    tool_context: ToolContext,
    credential: AuthCredential,
) -> list[dict]:
    """List Google Calendar events in a time range."""

    creds = get_google_credentials(
        credential
    )

    service = build(
        "calendar",
        "v3",
        credentials=creds,
    )

    result = (
        service.events()
        .list(
            calendarId="primary",
            timeMin=start_time,
            timeMax=end_time,
            maxResults=limit,
            singleEvents=True,
            orderBy="startTime",
        )
        .execute()
    )

    return [
        {
            "id": event.get("id"),
            "summary": event.get("summary"),
            "start": event.get("start"),
            "end": event.get("end"),
            "description": event.get("description"),
        }
        for event in result.get("items", [])
    ]
````

#### Tool 2：建立 Calendar Event

````python
def create_calendar_event(
    summary: str,
    start_time: str,
    end_time: str,
    description: str,
    tool_context: ToolContext,
    credential: AuthCredential,
) -> dict:
    """Create an event in the user's primary Google Calendar."""

    creds = get_google_credentials(
        credential
    )

    service = build(
        "calendar",
        "v3",
        credentials=creds,
    )

    event_body = {
        "summary": summary,
        "description": description,

        "start": {
            "dateTime": start_time,
            "timeZone": "Asia/Taipei",
        },

        "end": {
            "dateTime": end_time,
            "timeZone": "Asia/Taipei",
        },
    }

    event = (
        service.events()
        .insert(
            calendarId="primary",
            body=event_body,
        )
        .execute()
    )

    return {
        "status": "success",
        "id": event.get("id"),
        "summary": event.get("summary"),
        "start": event.get("start"),
        "end": event.get("end"),
        "htmlLink": event.get("htmlLink"),
    }
````

### Step 7：建立「只執行一次」的 OAuth 初始化函式

這個函式的工作是：

```text
觸發 Calendar Tool
↓
ADK 發出 OAuth Request
↓
User Google Login / Consent
↓
Flask 收 callback
↓
把 callback 交回 ADK
↓
ADK Token Exchange
↓
Credential 留在目前 Session
```

````python
import asyncio

from IPython.display import display, HTML


async def initialize_oauth():

    oauth_result.clear()

    # 用一個 harmless read request 觸發 Calendar OAuth
    content = types.Content(
        role="user",
        parts=[
            types.Part(
                text="請查看我今天的 Calendar 行程。"
            )
        ],
    )

    events = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=content,
    )

    auth_call = None
    auth_config_from_adk = None

    async for event in events:

        if not event.content:
            continue

        for part in event.content.parts or []:

            if (
                part.function_call
                and
                part.function_call.name
                == "adk_request_credential"
            ):

                auth_call = part.function_call

                raw_config = (
                    auth_call.args.get("authConfig")
                    or
                    auth_call.args.get("auth_config")
                )

                auth_config_from_adk = (
                    AuthConfig.model_validate(
                        raw_config
                    )
                )

                break

        if auth_call:
            break


    if not auth_call:
        print(
            "✅ OAuth credential already available."
        )
        return


    AUTH_URL = (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_uri
    )


    print("請打開以下網址完成 Google OAuth：\n")
    print(AUTH_URL)

    display(
        HTML(
            f'''
            <p>
            <a href="{AUTH_URL}" target="_blank">
            🔐 Open Google OAuth
            </a>
            </p>
            '''
        )
    )


    print("等待 Google OAuth callback...")


    # 等待 Flask 收到 callback
    for _ in range(300):

        if oauth_result.get("callback_url"):
            break

        await asyncio.sleep(1)


    if not oauth_result.get("callback_url"):
        raise TimeoutError(
            "OAuth callback timeout."
        )


    if oauth_result.get("error"):
        raise RuntimeError(
            f"OAuth error: {oauth_result['error']}"
        )


    # 把 Google callback 交回 ADK
    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_response_uri
    ) = oauth_result["callback_url"]


    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .redirect_uri
    ) = REDIRECT_URI


    auth_content = types.Content(
        role="user",

        parts=[
            types.Part(
                function_response=
                types.FunctionResponse(

                    id=auth_call.id,

                    name="adk_request_credential",

                    response=(
                        auth_config_from_adk
                        .model_dump()
                    ),
                )
            )
        ],
    )


    # Resume 同一個 Session
    events_after_auth = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=auth_content,
    )


    # 必須把 events consume 完
    async for event in events_after_auth:
        pass


    print("✅ Google Calendar OAuth initialized")
````

### Step 9：建立正式 Demo 使用的 `ask_agent()`

````python
async def ask_agent(prompt: str):

    content = types.Content(
        role="user",
        parts=[
            types.Part(text=prompt)
        ],
    )

    events = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=content,
    )

    final_text = None

    async for event in events:

        if not event.content:
            continue

        for part in event.content.parts or []:

            if part.text:
                final_text = part.text

            # OAuth 不應該再次出現
            if (
                part.function_call
                and
                part.function_call.name
                == "adk_request_credential"
            ):
                raise RuntimeError(
                    "OAuth credential is missing or expired. "
                    "Run await initialize_oauth() again."
                )


    if final_text:
        print(final_text)
````

````python
# ============================================================
# ADK Google Calendar Agent
# Assumes:
# - OAUTH_CLIENT_ID / OAUTH_CLIENT_SECRET already loaded
# - REDIRECT_URI already created by ngrok
# ============================================================

from datetime import datetime
from zoneinfo import ZoneInfo

from fastapi.openapi.models import (
    OAuth2,
    OAuthFlowAuthorizationCode,
    OAuthFlows,
)

from google.adk.agents.callback_context import CallbackContext
from google.adk.agents.llm_agent import Agent

from google.adk.auth.auth_credential import (
    AuthCredential,
    AuthCredentialTypes,
    OAuth2Auth,
)

from google.adk.auth.auth_tool import AuthConfig

from google.adk.tools.authenticated_function_tool import (
    AuthenticatedFunctionTool,
)

from google.adk.tools.tool_context import ToolContext

from google.oauth2.credentials import Credentials
from googleapiclient.discovery import build


# ============================================================
# 1. OAuth Configuration
# ============================================================

CALENDAR_SCOPE = "https://www.googleapis.com/auth/calendar"

auth_config = AuthConfig(
    auth_scheme=OAuth2(
        flows=OAuthFlows(
            authorizationCode=OAuthFlowAuthorizationCode(
                authorizationUrl=(
                    "https://accounts.google.com/o/oauth2/v2/auth"
                ),
                tokenUrl=(
                    "https://oauth2.googleapis.com/token"
                ),
                scopes={
                    CALENDAR_SCOPE:
                    "Read and write Google Calendar"
                },
            )
        )
    ),

    raw_auth_credential=AuthCredential(
        auth_type=AuthCredentialTypes.OAUTH2,
        oauth2=OAuth2Auth(
            client_id=oauth_client_id,
            client_secret=oauth_client_secret,
            redirect_uri=REDIRECT_URI,
        ),
    ),
)


# ============================================================
# 2. Convert ADK OAuth Credential to Google Credential
# ============================================================

def get_google_credentials(
    credential: AuthCredential,
) -> Credentials:

    return Credentials(
        token=credential.oauth2.access_token,
        refresh_token=credential.oauth2.refresh_token,
        token_uri="https://oauth2.googleapis.com/token",
        client_id=oauth_client_id,
        client_secret=oauth_client_secret,
        scopes=[CALENDAR_SCOPE],
    )


# ============================================================
# 3. Calendar Tool: List Events
# ============================================================

def list_calendar_events(
    start_time: str,
    end_time: str,
    limit: int,
    tool_context: ToolContext,
    credential: AuthCredential,
) -> list[dict]:

    creds = get_google_credentials(credential)

    service = build(
        "calendar",
        "v3",
        credentials=creds,
    )

    result = (
        service.events()
        .list(
            calendarId="primary",
            timeMin=start_time,
            timeMax=end_time,
            maxResults=limit,
            singleEvents=True,
            orderBy="startTime",
        )
        .execute()
    )

    return [
        {
            "id": event.get("id"),
            "summary": event.get("summary"),
            "start": event.get("start"),
            "end": event.get("end"),
            "description": event.get("description"),
        }
        for event in result.get("items", [])
    ]


# ============================================================
# 4. Calendar Tool: Create Event
# ============================================================

def create_calendar_event(
    summary: str,
    start_time: str,
    end_time: str,
    description: str,
    tool_context: ToolContext,
    credential: AuthCredential,
) -> dict:

    creds = get_google_credentials(credential)

    service = build(
        "calendar",
        "v3",
        credentials=creds,
    )

    event_body = {
        "summary": summary,
        "description": description,
        "start": {
            "dateTime": start_time,
            "timeZone": "Asia/Taipei",
        },
        "end": {
            "dateTime": end_time,
            "timeZone": "Asia/Taipei",
        },
    }

    event = (
        service.events()
        .insert(
            calendarId="primary",
            body=event_body,
        )
        .execute()
    )

    return {
        "status": "success",
        "id": event.get("id"),
        "summary": event.get("summary"),
        "start": event.get("start"),
        "end": event.get("end"),
        "htmlLink": event.get("htmlLink"),
    }


# ============================================================
# 5. Runtime Context
# ============================================================

def update_time(
    callback_context: CallbackContext,
):

    now = datetime.now(
        ZoneInfo("Asia/Taipei")
    )

    callback_context.state["_time"] = (
        now.isoformat()
    )


# ============================================================
# 6. Calendar Agent
# ============================================================

root_agent = Agent(
    name="calendar_planning_agent",

    model="gemini-2.5-flash",

    instruction="""
You are a personal Google Calendar planning assistant.

Current time:
{_time}

Timezone:
Asia/Taipei.

You can:

1. Inspect existing calendar events using
   list_calendar_events.

2. Create new calendar events using
   create_calendar_event.

Convert expressions such as:
- today
- tomorrow
- next Monday
- this afternoon

into concrete ISO 8601 datetime values.

When creating an event:
- determine the title
- determine start and end time
- use Asia/Taipei timezone
- call create_calendar_event

Never claim that an event was created unless
the tool returns status="success".
""",

    tools=[
        AuthenticatedFunctionTool(
            func=list_calendar_events,
            auth_config=auth_config,
        ),

        AuthenticatedFunctionTool(
            func=create_calendar_event,
            auth_config=auth_config,
        ),
    ],

    before_agent_callback=update_time,
)

print("✅ Calendar Agent Ready")
````

```python
    start_time: str,
    end_time: str,
    limit: int,
    tool_context: ToolContext,
    credential: AuthCredential,

        credential

    service = build(

    result = (
        service.events()
        .list(
            calendarId="primary",
            timeMin=start_time,
            timeMax=end_time,
            maxResults=limit,
            singleEvents=True,
            orderBy="startTime",

        .execute()
    )

    return [
        {
            "id": event.get("id"),
            "summary": event.get("summary"),
            "start": event.get("start"),
            "end": event.get("end"),
            "description": event.get("description"),
        }
        for event in result.get("items", [])
    ]
```

# Step 7：建立「只執行一次」的 OAuth 初始化函式

這個函式的工作是：

```text
觸發 Calendar Tool
↓
ADK 發出 OAuth Request
↓
User Google Login / Consent
↓
Flask 收 callback
↓
把 callback 交回 ADK
↓
ADK Token Exchange
↓
Credential 留在目前 Session
```

```python
import asyncio

from IPython.display import display, HTML


async def initialize_oauth():

    oauth_result.clear()

    # 用一個 harmless read request 觸發 Calendar OAuth
    content = types.Content(
        role="user",
        parts=[
            types.Part(
                text="請查看我今天的 Calendar 行程。"
            )
        ],
    )

    events = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=content,
    )

    auth_call = None
    auth_config_from_adk = None

    async for event in events:

        if not event.content:
            continue

        for part in event.content.parts or []:

            if (
                part.function_call
                and
                part.function_call.name
                == "adk_request_credential"
            ):

                auth_call = part.function_call

                raw_config = (
                    auth_call.args.get("authConfig")
                    or
                    auth_call.args.get("auth_config")
                )

                auth_config_from_adk = (
                    AuthConfig.model_validate(
                        raw_config
                    )
                )

                break

        if auth_call:
            break


    if not auth_call:
        print(
            "✅ OAuth credential already available."
        )
        return


    AUTH_URL = (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_uri
    )


    print("請打開以下網址完成 Google OAuth：\n")
    print(AUTH_URL)

    display(
        HTML(
            f'''
            <p>
            <a href="{AUTH_URL}" target="_blank">
            🔐 Open Google OAuth
            </a>
            </p>
            '''
        )
    )


    print("等待 Google OAuth callback...")


    # 等待 Flask 收到 callback
    for _ in range(300):

        if oauth_result.get("callback_url"):
            break

        await asyncio.sleep(1)


    if not oauth_result.get("callback_url"):
        raise TimeoutError(
            "OAuth callback timeout."
        )


    if oauth_result.get("error"):
        raise RuntimeError(
            f"OAuth error: {oauth_result['error']}"
        )


    # 把 Google callback 交回 ADK
    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_response_uri
    ) = oauth_result["callback_url"]


    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .redirect_uri
    ) = REDIRECT_URI


    auth_content = types.Content(
        role="user",

        parts=[
            types.Part(
                function_response=
                types.FunctionResponse(

                    id=auth_call.id,

                    name="adk_request_credential",

                    response=(
                        auth_config_from_adk
                        .model_dump()
                    ),
                )
            )
        ],
    )


    # Resume 同一個 Session
    events_after_auth = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=auth_content,
    )


    # 必須把 events consume 完
    async for event in events_after_auth:
        pass


    print("✅ Google Calendar OAuth initialized")
```

```python
async def ask_agent(prompt: str):

    content = types.Content(
        role="user",
        parts=[
            types.Part(text=prompt)
        ],
    )

    events = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=content,
    )

    final_text = None

    async for event in events:

        if not event.content:
            continue

        for part in event.content.parts or []:

            if part.text:
                final_text = part.text

            # OAuth 不應該再次出現
            if (
                part.function_call
                and
                part.function_call.name
                == "adk_request_credential"
            ):
                raise RuntimeError(
                    "OAuth credential is missing or expired. "
                    "Run await initialize_oauth() again."
                )


    if final_text:
        print(final_text)
```

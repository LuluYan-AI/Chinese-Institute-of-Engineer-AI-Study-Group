### Gemini Enterprise

#### 1. 建立 GCP 專案與啟用 Gemini Enterprise

影片假設你已完成：

1. 建立 Google Cloud Project。
2. 綁定 Billing Account。
3. 在 Cloud Console 搜尋 **Gemini Enterprise**。
4. 啟用試用方案或相應訂閱。
5. 啟用產品要求的 API。

這裡的「Application」不是手機或 Web 程式，而是一個 Gemini Enterprise 的邏輯工作區。它包含：

* 使用者登入方式
* 可搜尋的資料來源
* 可使用的 Agent
* 模型與功能開關
* Web App 的入口
* 權限、稽核與安全設定

因此，一個 Application 可以理解為：

```text
Gemini Enterprise App
├── Identity provider
├── Data stores / connectors
├── Core Assistant
├── Agent Designer agents
├── Registered ADK / A2A agents
└── Web app configuration
```

目前 Gemini Enterprise 被定位為企業搜尋、AI 助理與 Agent 平台，可以將回答建立在組織內部資料之上。([Google Cloud Documentation][2])


- https://console.cloud.google.com/
- https://console.cloud.google.com/gemini-enterprise

#### 2. 建立 Gemini Enterprise Application

影片中只填了兩個主要欄位：

* Display name
* Region

Region 不只是介面設定，它通常會影響：

* 資源部署位置
* 資料駐留
* 支援的功能
* 法規與企業政策
* 後續某些 Data Store 的可用區域

正式環境應在建立前決定區域，因為部分雲端資源建立後不容易直接更換 region。

建立完成後，Cloud Console 會產生一個管理端 Application，以及對應的終端使用者 Web App。

#### 3. 設定 Workforce Identity

影片選擇 **Google Identity**，也就是讓使用者以組織的 Google 帳號登入。

這一步的核心目的不是單純登入，而是建立：

```text
使用者身分 → 原始資料權限 → Gemini Enterprise 查詢權限
```

例如 Simon 在 Google Drive 只能看 A、B、C 三份文件，那麼透過 Gemini Enterprise 搜尋時，也不應因為使用 AI 而看到其他人的私有文件。

Gemini Enterprise 支援 Google Identity，也支援透過 Workforce Identity Federation 整合 Microsoft Entra ID、Okta、OIDC 或 SAML 身分系統。([Google Cloud Documentation][3])

使用 Google Drive Connector 時，Google Identity 通常是最簡單的配置。官方文件也指出，Google Drive 資料與使用者權限配置存在直接關係。([Google Cloud Documentation][4])

#### 4. 建立 Google Drive Data Store

影片路徑大致是：

```text
Gemini Enterprise App
→ Connected data sources / Data stores
→ Add data source
→ Google Drive
```

目前文件中的介面名稱可能顯示為：

```text
Gemini Enterprise
→ Data stores
→ Create data store
→ Google Drive
```

Google 把兩個概念分開：

* **Connector**：負責連接外部系統。
* **Data Store**：Gemini Enterprise 中代表該資料來源的邏輯資源。

因此，實際關係是：

```text
Google Drive
    ↓ connector
Google Drive Data Store
    ↓ attached to
Gemini Enterprise Application
    ↓ used by
Core Assistant / Agent Designer Agent
```

建立時通常要指定資料範圍，例如：

* 全部 Workspace Drive
* 特定資料夾
* Include 指定資料夾
* Exclude 指定資料夾

Google Drive Connector 可使用 federation，也就是查詢時直接向 Drive 取得資料，而不是一定將整份內容複製到另一個搜尋索引。官方文件指出，這種方式不需要另外管理資料副本，但搜尋範圍仍受網域擁有權和使用者權限限制。([Google Cloud Documentation][5])

### 容易忽略的設定

使用 Google Drive Connector 前，可能還需要：

* 啟用 Google Workspace smart features。
* 確認文件屬於正確的 Workspace domain。
* 將文件放在組織擁有的 Shared Drive。
* 確認使用者具備文件權限。
* 確認 Data Store 已連接到正確的 Application。

如果文件是另一個網域擁有，即使使用者可以在 Drive 中看到，也不一定會被預設的 federated search 找到。新版產品提供 cross-domain documents 選項，但官方特別提醒，外部文件會增加 prompt injection 和搜尋結果操控風險。([Google Cloud Documentation][5])

#### 5. 在 Gemini Enterprise Web App 測試 Connector

建立 Data Store 後，影片進入終端使用者 Web App，確認 Google Drive Connector 已開啟，再輸入：

```text
What is the title of the last Drive file I created?
```

這個流程大致是：

```text
User question
    ↓
Core Assistant 判斷需要企業資料
    ↓
呼叫 Google Drive Data Store
    ↓
使用登入者身分查詢 Drive
    ↓
取得符合權限的文件或 metadata
    ↓
將檢索結果交給模型
    ↓
產生 grounded answer
```

「last Drive file」這類問題通常主要依賴：

* 檔案 metadata
* 建立時間
* 擁有者
* 使用者可見範圍

不一定需要讀取完整文件內容。

這一步應先測試成功，再建立自訂 Agent。否則後續 Agent 查不到資料時，很難判斷是 Connector、權限還是 Agent 流程設定有問題。

#### 6. 啟用 Agent Designer

Agent Designer 預設可能未開啟。管理員需要到 Application 的 Web App 功能設定中，啟用：

```text
Enable Agent Designer
```

目前官方描述的路徑是 Application 的 **End user features** 或 **Manage web app features** 頁面。啟用後，使用者才可以在 Web App 中建立和管理自訂 Agent。([Google Cloud Documentation][6])

啟用後必須重新整理 Web App，因為左側導覽列和前端 feature flags 通常是在載入時取得。

重新整理後，側邊欄會出現類似：

```text
New agent
```

或 Agent Designer 入口。

## 7. 使用自然語言產生 Agent

影片輸入類似：

```text
Create a pirate-themed assistant for searching enterprise data.
```

Agent Designer 不是只把這句話當成普通 system prompt。它會根據描述產生一個初始 Agent 配置，可能包括：

* Agent 名稱
* Agent description
* Instructions
* 模型
* 可用工具
* Data Stores
* Workflow nodes
* 回答風格
* 預覽用的建議問題

可以把它理解成由模型產生一份視覺化 Agent specification：

```yaml
agent:
  name: Parrot Data Searcher
  model: Gemini
  instructions:
    - Search enterprise data when needed
    - Answer in a pirate style
  tools:
    - Google Drive data store
    - Enterprise search
```

這段 YAML 只是概念示意，不是 Agent Designer 實際匯出的格式。

## 8. Flow 分頁實際控制什麼

影片提到進入 **Flow** 分頁查看和修改設定。

目前 Agent Designer 已不只是單一 prompt editor，而是可以建立 single-step 和 multi-step Agent 的視覺化流程。官方列出的能力包括：

* 用自然語言建立 Agent
* 在 visual flow canvas 編輯流程
* 建立多步驟 Agent
* 使用 subagents
* 連接 Gmail、Drive、Jira 等資料或工具
* 排程執行 Agent ([Google Cloud Documentation][7])

最簡單的 Agent 流程可能只有：

```text
Start
  ↓
LLM Agent
  ├── Google Drive search
  └── Enterprise search
  ↓
Response
```

較複雜的流程可能是：

```text
User request
  ↓
分類問題
  ├── Drive 文件問題 → Drive Search Agent
  ├── Email 問題 → Gmail Agent
  └── Calendar 問題 → Calendar Agent
  ↓
整合結果
  ↓
產生最終答案
```

在 Flow 中最值得檢查的是以下四項。

### Instructions

這是 Agent 的核心行為規則。例如：

```text
你是公司內部文件研究助理。

回答前先搜尋已連接的企業資料。
只根據找到的資料回答。
找不到資料時，明確說明找不到。
回答時列出使用的文件名稱。
不要把未確認的推測寫成事實。
```

影片中的 pirate theme 只是風格要求：

```text
Respond like a pirate.
```

風格不會自動帶來工具能力。真正讓 Agent 可以搜尋 Drive 的，是 Flow 中連接的 Data Store 或 search tool。

### Model

模型決定：

* 推理能力
* 延遲
* 成本或配額
* 工具選擇能力
* 長上下文能力

可選模型會受到管理員的 **Model availability** 設定限制，不是每位 Agent 建立者都能任意選擇。([Google Cloud Documentation][8])

### Tools 與 Data Sources

需要確認 Google Drive Data Store 確實被加入 Agent。

「Application 已經有 Drive Connector」不一定等同「每個 Agent 都必定使用 Drive」。Agent 的 Flow 必須具有對應工具或資料來源，而且 instructions 應明確告訴模型何時使用它。

### 流程連線

多步驟 Agent 要確認：

* 每個 node 的輸入來源
* 每個 node 的輸出流向
* 是否存在無法到達的 node
* 工具輸出是否有送回 LLM
* 是否可能形成不必要的循環
* 錯誤時是否有 fallback

## 9. Create 與開始聊天

按下 **Create** 後，Agent 配置會被儲存，並可在 Gemini Enterprise Web App 中執行。

使用者提問時，流程通常是：

```text
1. 載入 Agent instructions
2. 載入可使用的工具定義
3. 模型判斷是否需要工具
4. 呼叫 Drive 或 Enterprise Search
5. 收到搜尋結果
6. 根據結果生成回答
7. 套用 Agent 的語氣與格式
```

所以影片中的 pirate 回答包含兩個獨立層次：

```text
內容來源：Google Drive / Enterprise Search
表達方式：Pirate instructions
```

這一點很重要。Grounding 負責「答案根據什麼」，persona 負責「答案怎麼說」。

## 10. Library 與 Pin

建立後，Agent 會出現在使用者的 Library。

Pin 的效果主要是：

* 放在側邊欄或常用區域
* 快速啟動 Agent
* 不必每次從 Agent Library 搜尋

它不等於部署成公開服務，也不代表 Agent 已分享給整個組織。

如果要讓其他使用者使用，還涉及：

* Enable agent sharing
* 是否需要管理員核准
* 分享給個人或群組
* Google Identity 與 WIF 的群組限制
* 組織的 Agent 治理政策

這些由 Gemini Enterprise 管理員控制。([Google Cloud Documentation][8])

## 11. 影片範例真正建立了什麼

從軟體架構角度，影片建立的不是一個獨立程式，而是一份平台代管的 Agent 配置：

```text
Agent configuration
├── Prompt / instructions
├── Model reference
├── Tool bindings
├── Data Store bindings
├── Visual workflow
├── Access policy
└── UI metadata
```

執行環境、OAuth、工具呼叫、聊天 UI、模型服務和資料存取，主要由 Gemini Enterprise 管理。

這正是它能做到 no-code 的原因。

## 12. 它與 ADK 的對應關係

可以用下表理解：

| Agent Designer       | ADK                              |
| -------------------- | -------------------------------- |
| 視覺化 Flow             | `Agent`、`SequentialAgent` 等類別    |
| Instructions         | `instruction`                    |
| Model selector       | `model`                          |
| Connected Data Store | Tool、Search、Connector            |
| Subagent node        | `sub_agents`                     |
| Web App preview      | ADK Web 或自建前端                    |
| Library              | Agent Registry／Gemini Enterprise |
| 平台代管                 | 自己控制程式與部署                        |

概念上，影片中的 Agent 可類比成：

```python
from google.adk.agents import Agent

root_agent = Agent(
    name="parrot_data_searcher",
    model="gemini-model",
    instruction="""
    Search connected enterprise data before answering.
    Base answers on retrieved information.
    Respond using a pirate style.
    """,
    tools=[
        # Enterprise search or Drive-related tool
    ],
)
```

差別是 Agent Designer 幫你處理了 UI、流程配置、工具綁定和平台部署，不需要直接寫 Python。

## 13. 建議的實際測試順序

不要直接從 pirate Agent 開始。較可靠的測試順序是：

```text
1. Core Assistant 是否能找到已知 Drive 文件
2. 權限不同的帳號是否得到不同結果
3. Agent Designer Agent 是否能呼叫相同 Data Store
4. 要求 Agent 回傳文件標題與來源
5. 測試找不到文件時的行為
6. 測試同名文件和模糊問題
7. 最後才加入 persona 或複雜工作流程
```

第一版 Instructions 可以使用：

```text
你是企業文件搜尋助理。

處理問題時：
1. 優先搜尋已連接的 Google Drive 資料。
2. 僅根據檢索到的內容回答。
3. 回答中列出相關文件名稱。
4. 若資料不足，明確說明缺少哪些資料。
5. 不要根據一般知識補造企業內部資訊。
6. 除非使用者要求，否則保持簡潔和專業。
```

這比 pirate persona 更適合驗證 Agent 是否真的使用了企業資料。

## 1. Google Cloud 帳戶、Project 與 Billing 是三件不同的事

「有 Google Cloud 帳戶」只代表你可以登入 Cloud Console。真正的資源都必須建立在一個 Google Cloud Project 裡。

一個 Project 會提供：

```text
Project ID
Project number
IAM 權限邊界
API 啟用狀態
Billing 關聯
Gemini Enterprise Apps
Data Stores
稽核紀錄
配額與費用
```

開始前至少需要：

1. 一個 Google Cloud Project。
2. 該 Project 已綁定 Billing Account。
3. 你的帳號具有啟用 API 和建立 Gemini Enterprise 資源的權限。

Google 目前的 quickstart 要求確認 Project 已啟用 billing，且操作者具備 Gemini Enterprise Admin。啟用 API 則需要 `serviceusage.services.enable` 權限，通常由 Project Owner 或 Service Usage Admin 提供。([Google Cloud Documentation][1])

常見 IAM 角色如下：

| 工作                   | 常見角色                    |
| -------------------- | ----------------------- |
| 建立 Project           | Project Creator         |
| 啟用 API               | Service Usage Admin     |
| 管理 Gemini Enterprise | Gemini Enterprise Admin |
| 建立或修改 Data Store     | Discovery Engine Editor |
| 一般使用 Web App         | Gemini Enterprise User  |

其中 Gemini Enterprise Admin 對應的角色 ID 通常是：

```text
roles/discoveryengine.agentspaceAdmin
```

這個名稱仍保留了舊產品名稱 `agentspace`，所以看到 Agentspace 不代表你進錯產品。

* Gemini Enterprise Agent Platform

這幾個產品名稱很接近，但用途不同。

影片使用的是：

它是提供企業員工使用的搜尋、聊天和 Agent Web App。

## 4. 啟用的 Gemini Enterprise API 是什麼

影片畫面說的是：

> activate the required API

Gemini Enterprise 核心使用的 API 是：

```text
Gemini Enterprise (Discovery Engine) API
```

其 service name 是：

```text
discoveryengine.googleapis.com
```

這也是為什麼許多 IAM role、REST resource 和舊文件仍包含 `discoveryengine`。

目前 Google 列出的必要 API 包括：

```text
Gemini Enterprise (Discovery Engine) API
Cloud Storage API
Identity and Access Management API
Vertex AI API
```

不同 onboarding 流程可能只在第一個畫面明確要求啟用 Discovery Engine API，其他 API 可能由精靈自動處理，或在後續功能使用時要求啟用。Google 的安全文件將 Vertex AI、Discovery Engine、Cloud Storage 和 IAM 都列為 Gemini Enterprise 所需 API；目前 quickstart 則明確要求啟用 Discovery Engine、Cloud Storage 和 IAM。([Google Cloud Documentation][3])

你也可以手動確認：

```text
Google Cloud Console
→ APIs & Services
→ Enabled APIs & services
```

搜尋：

```text
Gemini Enterprise
Discovery Engine
Vertex AI
Cloud Storage
Identity and Access Management
```

也可以使用 `gcloud`：

```bash
gcloud services enable \
  discoveryengine.googleapis.com \
  storage.googleapis.com \
  iam.googleapis.com \
  aiplatform.googleapis.com \
  --project=YOUR_PROJECT_ID
```

這段命令只是等價的 CLI 做法。使用 Cloud Console onboarding 時，通常不需要手動執行。

### 啟用 API 的意義

啟用 API 不等於建立了一個 Agent，也不等於產生 API key。

它只是允許這個 Project 建立和呼叫相關服務：

```text
API disabled
→ Project 無法建立 Gemini Enterprise 資源

API enabled
→ Project 可以建立 App、Data Store、Engine 等資源
```

Gemini Enterprise Web App 一般也不是靠你手動建立 API key 來運作。登入者身份、IAM、OAuth 和 Workspace identity 才是主要驗證機制。

## 5. 建立 Gemini Enterprise App

API 啟用後，進入：

```text
Gemini Enterprise
→ Apps
→ Create app
```

現在的建立頁面通常會要求：

```text
App name
App ID
Location
Organization/company display name
```

影片只提到 display name 和 region，是簡化過的說法。Google 目前的文件顯示，建立 App 時可以設定 App name、不可變更的 App ID、multi-region，以及對外顯示的組織名稱。([Google Cloud Documentation][4])

### App name

例如：

```text
Simon AI Workspace
```

這是介面中顯示的名稱，通常可以讓人閱讀。

### App ID

例如系統產生：

```text
simon-ai-workspace_174238
```

App ID 是 API 與資源路徑使用的穩定識別值。

建立後通常不能修改，所以正式環境最好使用有意義的命名，例如：

```text
corp-search-prod
corp-search-dev
ai-assistant-tw
```

不要把 production 和測試環境混在同一 App。

### App 在架構上的含義

App 不是 Agent 本身。它是所有終端使用者功能的容器：

```text
Gemini Enterprise App
├── Web App
├── Assistant
├── Agent Library
├── Agent Designer
├── Identity provider
├── Connected data stores
├── Feature toggles
├── Model settings
└── Access policy
```

因此，同一個 Project 可以有多個 App，例如：

```text
Project: enterprise-ai-prod
├── HR Assistant App
├── Engineering Knowledge App
└── Sales Intelligence App
```

每個 App 可以掛接不同資料來源和開放給不同使用者。

## 6. Region 或 Location 要怎麼選

影片說「choose the region」，但目前 Gemini Enterprise App 主要使用的是 multi-region：

```text
global
us
eu
```

Google 建議，沒有資料駐留需求時使用預設的 `global`；如果公司有法規、合約或資料駐留限制，才考慮 `us` 或 `eu`。([Google Cloud Documentation][4])

三者的概念是：

| Location | 適用情況                             |
| -------- | -------------------------------- |
| `global` | 教學、測試、沒有特定資料駐留要求                 |
| `us`     | 要求資料靜態儲存在美國 multi-region         |
| `eu`     | GDPR 或政策要求資料靜態儲存在歐盟 multi-region |

Location 是重要架構決策，因為它會影響：

* App resource 的位置
* Data Store 可使用的位置
* 部分 connector 的支援情況
* CMEK 設定
* Agent Gateway 或 Agent Registry 相容性
* 資料駐留與合規

通常 App 和所連接的 Data Store 應選擇相容的位置。

教學環境可以選：

```text
global
```

正式企業環境則應先詢問資安、法務或 Cloud 管理團隊。

## 7. 建立 App 後，為什麼要設定 Workforce Identity

App 建立完成後，下一步不是立刻接 Drive，而是設定「誰在使用」。

影片選：

```text
Google Identity
```

這表示 Gemini Enterprise Web App 使用者透過 Google Workspace 或 Cloud Identity 帳號登入。

這一步建立的邏輯是：

```text
Google Workspace user
    ↓ authentication
Gemini Enterprise user identity
    ↓ authorization
App access
    ↓ identity-aware connector query
Google Drive / Gmail / Calendar permissions
```

這裡最重要的是：使用者身份會向下傳遞到資料存取層。

例如：

```text
Alice 可讀 HR-policy.pdf
Bob 不可讀 HR-policy.pdf
```

即使 Alice 和 Bob 使用同一個 Gemini Enterprise App，搜尋結果也應不同：

```text
Alice query → 可以取得 HR-policy.pdf
Bob query   → 不應取得 HR-policy.pdf
```

這叫做 data source access control，而不只是 Web App 登入。

## 8. Google Identity 和 Workforce Identity Federation 的差異

目前主要有兩條路。

### Google Identity

適合：

* 公司已使用 Google Workspace
* 使用者有受管理的 Workspace 帳號
* 要連接 Gmail、Drive、Calendar
* 想快速完成教學

登入身份通常是：

```text
user@company.com
```

並由 Google Workspace 或 Cloud Identity 管理。

### Workforce Identity Federation

適合公司使用：

* Microsoft Entra ID
* Okta
* Ping Identity
* 自建 OIDC
* SAML identity provider

流程大致是：

```text
External IdP
→ Workforce Identity Pool
→ Workforce Identity Provider
→ Gemini Enterprise
```

Workforce Identity Federation 讓外部 IdP 的員工帳號可以登入 Google Cloud 服務，而不必替每個人建立傳統 Google Cloud 使用者身份。([Google Cloud][5])

影片為了簡化，選擇 Google Identity 是合理的。對 Google Workspace Connector 而言，這也是最直接的路徑。

目前介面大致是：

```text
Gemini Enterprise
→ Settings
→ Authentication
→ Add identity provider
```

然後針對 App 所在位置配置 identity provider。([Google Cloud Documentation][6])

## 9. Google Identity 並不等於個人 Gmail

這是一個很容易踩到的問題。

影片說 Google Identity，通常是指組織管理的：

```text
Google Workspace account
或
Cloud Identity account
```

不等於一般消費者帳號：

```text
someone@gmail.com
```

特別是 Google Drive Connector，官方目前要求 Cloud Console 登入帳號必須與要連接的 Google Drive instance 屬於相同的 Google Workspace customer；一般 `@gmail.com` 消費者帳號沒有 Workspace customer ID，因此不受支援。([Google Cloud Documentation][7])

所以做這個教學最好使用：

```text
simon@your-company-domain.com
```

而不是：

```text
simon123@gmail.com
```

即使個人 Gmail 可以建立 GCP Project，也不代表它能完成 Workspace enterprise connector 的完整流程。

## 10. Connected Data Source 實際建立的是什麼

身份設定完成後，影片進入：

```text
Connected data sources
→ Add data source
```

目前的介面更常稱為：

```text
Data stores
→ Create data store
```

Google 文件中的術語關係是：

```text
Data source
    ↓ accessed through
Connector
    ↓ represented inside Gemini Enterprise as
Data Store
    ↓ attached to
App
```

例如：

```text
Google Drive
→ Google Drive connector
→ drive-corporate-docs Data Store
→ Simon AI Workspace App
```

Connector 和 Data Store 有時在介面上會混用，但技術上不是完全相同：

* Connector 定義如何連到來源系統。
* Data Store 是 Gemini Enterprise 中可被 App 使用的資料資源。
* App 決定終端使用者可以搜尋哪些 Data Stores。

Google 將 connectors 定義為把 Google 或第三方資料來源帶入 Gemini Enterprise，並以獨立 Data Stores 管理。([Google Cloud Documentation][8])

## 11. 建立 Google Drive Data Store 的前置條件

對影片使用的 Google Drive connector，開始前應確認：

```text
1. 使用 Google Workspace 帳號
2. Cloud Console 帳號與 Drive Workspace 相符
3. 已設定 Identity Provider
4. 具有建立 Data Store 的 IAM 權限
5. App 與 Data Store 的 location 相容
6. Workspace 管理員沒有封鎖所需功能
```

官方特別要求：

* 使用 Cloud Console 的帳號必須與要連接的 Drive instance 相同。
* 必須是 Google Workspace，而不是一般消費者 Gmail。
* 為了正確執行 access control，應先設定 identity provider。([Google Cloud Documentation][7])

## 12. 建立 Drive Connector 時會填哪些東西

介面大致是：

```text
Gemini Enterprise
→ Data stores
→ Create data store
→ Google Drive
```

接下來通常需要設定：

### 要搜尋的資料範圍

可能包含：

```text
All supported Drive content
Specific folders
Shared drives
Include filters
Exclude filters
```

影片中說：

> Choose what you would like to share

這並不是把檔案權限改成公開，而是定義 Connector 的搜尋範圍。

例如：

```text
Connector scope:
  Include: Shared Drive / AI Study Group
  Exclude: Shared Drive / Finance Confidential
```

實際查詢時仍會再套用終端使用者原本的 Drive 權限。

也就是兩層過濾：

```text
Connector scope
AND
User's Drive permissions
```

某份文件要出現在結果中，通常必須同時滿足：

```text
文件在 Connector 搜尋範圍內
且
登入使用者本來就有權讀取
```

### Data connector name

例如：

```text
company-drive
engineering-shared-drive
workspace-documents-prod
```

這只是管理資源名稱，不是 Workspace 資料夾名稱。

命名最好帶環境：

```text
drive-rnd-dev
drive-rnd-prod
```

### Multi-region

通常應和 App 相同或相容：

```text
App: global
Data Store: global
```

或：

```text
App: eu
Data Store: eu
```

### Encryption

在 `us` 或 `eu` 等配置下，部分 Connector 會讓你選：

```text
Google-managed encryption key
Cloud KMS customer-managed key
```

教學通常選 Google-managed key。企業正式環境若有 CMEK 政策，才使用 Cloud KMS。

## 13. Drive 資料是複製進 Gemini Enterprise 嗎

影片的 Drive 連接通常是 federated retrieval，也就是查詢時從 Drive 取得資料，而不是先將所有 Drive 文件完整複製進 Gemini Enterprise index。

概念上：

```text
User asks question
→ Gemini Enterprise interprets query
→ Connector queries Google Drive
→ Drive returns authorized results
→ Gemini model synthesizes answer
```

官方 Drive 文件指出，這類連接可以直接從 Drive 擷取資訊，不需要你管理另一份資料儲存副本。([Google Cloud Documentation][7])

不過應注意：

* Metadata、設定和必要的服務資料仍可能由 Google Cloud 管理。
* 不同 connector 可能使用 federation 或 ingestion。
* 第三方 connector 未必與 Google Drive 採用完全相同的資料流。

因此不能把所有 Connected Data Sources 都理解成「即時 API 查詢」，也不能全部理解成「複製並建立向量索引」。要看 connector 類型。

## 14. 建立 Data Store 後還要掛到 App

建立 Data Store 和讓 App 使用 Data Store 是兩個動作。

可能的流程有兩種：

```text
先建立 App
→ 在 App 裡新增 Connected Data Source
→ 建立並自動掛載 Data Store
```

或：

```text
先建立 Data Store
→ 開啟 App
→ Connect data store
```

最終需要形成：

```text
App
└── Connected data stores
    └── company-drive
```

如果只建立 Data Store，卻沒有連到目前的 App，Web App 不一定能搜尋它。

可以在 App 的資料來源頁確認：

```text
Data store status: Active
Connected to app: Yes
Connector enabled in web app: Yes
```

## 15. 建議你實際操作時的完整檢查清單

```text
A. Cloud 基礎
[ ] 登入正確的 Google Workspace 帳號
[ ] 選到正確的 GCP Project
[ ] Billing 已啟用
[ ] Project ID 記錄下來

B. IAM
[ ] 有 Service Usage Admin 或等價權限
[ ] 有 Gemini Enterprise Admin
[ ] 有 Discovery Engine Editor
[ ] 測試使用者有 Gemini Enterprise User

C. APIs
[ ] discoveryengine.googleapis.com
[ ] storage.googleapis.com
[ ] iam.googleapis.com
[ ] aiplatform.googleapis.com

D. Gemini Enterprise App
[ ] App name 正確
[ ] App ID 命名合理
[ ] Location 已決定
[ ] 公司顯示名稱正確

E. Identity
[ ] Google Identity 已設定
[ ] 使用的是 Workspace 帳號而非個人 Gmail
[ ] 測試使用者能登入 Web App

F. Google Drive
[ ] 使用相同 Workspace customer
[ ] Drive Data Store 已建立
[ ] Data Store 狀態為 Active
[ ] 搜尋範圍正確
[ ] Data Store 已連接到 App
```

最重要的技術關係可以濃縮成：

```text
Google Cloud Project
├── Billing
├── Enabled APIs
├── IAM roles
└── Gemini Enterprise App
    ├── Identity provider
    ├── Web App
    └── Connected Data Stores
        └── Google Drive Connector
```

所以影片前一分鐘並不是單純「啟用一個 Gemini API」。它是在完成四個不同層級的工作：

```text
產品授權：Gemini Enterprise trial/subscription
服務啟用：Discovery Engine 等 Google Cloud APIs
應用資源：Gemini Enterprise App
企業整合：Identity Provider + Data Store Connector
```

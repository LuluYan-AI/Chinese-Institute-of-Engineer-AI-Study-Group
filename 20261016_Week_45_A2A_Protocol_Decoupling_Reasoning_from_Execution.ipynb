### A2A 1.0：跨語言 AI Agent 協作——整合 Python LangGraph 與 Go

https://www.youtube.com/watch?v=NsJ7UjRCnZU

#### (1)  問題背景：AI Agent 的框架孤島（Framework Silo）

在企業級 AI 系統中，不同團隊可能使用不同程式語言與開發框架。例如：

- Python + LangGraph：負責推理（Reasoning）、工作流程編排（Orchestration）與任務分派。
- Go + Execution Service：負責接收任務、呼叫 AI 模型及產生執行結果。

當 Python Agent 需要呼叫 Go 服務時，若缺乏統一的通訊介面，開發者可能需要建立大量客製化 API，增加系統整合與維護成本。

這類問題稱為框架孤島（Framework Silo）。

解決方式是透過標準化通訊協定，使不同語言與框架的 AI Agent 能夠互相協作，而不必重新開發既有服務。



#### (2) A2A Protocol：跨框架 Agent 通訊協定

A2A（Agent-to-Agent Protocol） 是一種讓獨立 AI Agent 跨系統交換訊息、委派任務及回傳結果的標準化協定。

A2A 的核心概念包括：

&#x20;    Text                Markdown&#x20;

| 核心概念                        | 說明                        |
| --------------------------- | ------------------------- |
| 互通性（Interoperability）       | 不同語言及框架的 Agent 可以互相通訊     |
| 任務委派（Task Delegation）       | 一個 Agent 可以將工作交給另一個 Agent |
| 標準化訊息（Standardized Message） | 使用共同的資料格式交換任務資訊           |
| 解耦（Decoupling）              | 推理系統與執行系統可以獨立開發、部署與維護     |

&#x20;     CSV          Excel&#x20;

本案例使用 HTTP POST 與 JSON，示範 Python LangGraph 與 Go 服務之間的任務交換。

#### (3) Python：使用 LangGraph 建立任務編排器

### 3.1 定義 AgentState

LangGraph 使用狀態（State）管理工作流程中需要共享的資料。

```
from typing import TypedDictclass AgentState(TypedDict):    objective: str    a2a_payload: str    final_result: str
```

其中：

- `objective`：任務目標（Task Objective）。
- `a2a_payload`：準備傳送的 JSON 訊息。
- `final_result`：遠端服務回傳的結果。

### 3.2 建立 Reasoning Node

推理節點（Reasoning Node）負責將任務轉換成結構化訊息。

```
import jsondef reasoning_node(state: AgentState):    envelope = {        "a2a_version": "1.0",        "sender_id": "langgraph-orchestrator",        "task": state["objective"],        "context": {            "urgency": "high",            "auth_token": "bearer_abc123"        }    }    return {"a2a_payload": json.dumps(envelope)}
```

程式將 `objective` 轉換為 JSON Payload，包含發送者、任務及相關背景資訊。

其中 `auth_token` 僅為教學示意值，並未實作真正的身分驗證（Authentication），不應作為正式系統的安全機制。

### 3.3 建立任務分派節點

使用 HTTP POST 將 JSON 訊息傳送至 Go 服務。

```
import urllib.requestdef a2a_dispatch_node(state: AgentState):    req = urllib.request.Request(        "http://localhost:18080/execute",        data=state["a2a_payload"].encode("utf-8"),        headers={"Content-Type": "application/json"}    )    with urllib.request.urlopen(req) as response:        result = json.loads(response.read().decode())    return {"final_result": result["final_result"]}
```

此節點完成三項工作：發送 HTTP Request、接收 JSON Response，以及更新 `AgentState`。

### 3.4 建立 LangGraph Workflow

```
from langgraph.graph import StateGraph, ENDworkflow = StateGraph(AgentState)workflow.add_node("reason", reasoning_node)workflow.add_node("delegate_via_a2a", a2a_dispatch_node)workflow.set_entry_point("reason")workflow.add_edge("reason", "delegate_via_a2a")workflow.add_edge("delegate_via_a2a", END)langgraph_app = workflow.compile()
```

工作流程包含兩個節點：

`reason` → `delegate_via_a2a` → `END`

LangGraph 負責控制任務執行順序，而 HTTP 通訊負責將工作交給外部服務。



#### (4) Go：建立遠端任務執行服務

Go 服務負責接收 Python 傳來的 JSON Payload，解析任務內容，再呼叫 Gemini 模型產生執行計畫。

### 4.1 定義 JSON 資料結構

Go 使用 `struct` 定義接收資料的格式。

```
type A2APayload struct {
    A2AVersion string            `json:"a2a_version"`
    SenderID   string            `json:"sender_id"`
    Task       string            `json:"task"`
    Context    map[string]string `json:"context"`
}
```

此結構必須與 Python 發送的 JSON 欄位相對應，才能正確完成資料反序列化（Deserialization）。

### 4.2 接收與解析任務

以下擷取原始 Go 服務中接收訊息的核心程式碼：

```
func handleA2A(w http.ResponseWriter, r *http.Request) {
    var payload A2APayload

    if err := json.NewDecoder(r.Body).Decode(&payload); err != nil {
        http.Error(w, err.Error(), http.StatusBadRequest)
        return
    }

    // 接續呼叫 Gemini API 處理 payload.Task
}
```

此段展示 HTTP Handler 如何解析 Python 發送的資料。完整程式另包含 Gemini API 呼叫、模型輸出解析及 JSON 回應。

### 4.3 使用 Gemini 3.1 Pro 產生執行計畫

原始程式使用以下模型 API：

```
apikey := os.Getenv("GEMINI_API_KEY")

url := "https://generativelanguage.googleapis.com/v1beta/models/gemini-3.1-pro-preview:generateContent?key=" + apikey
```

並以 System Instruction 指定模型角色：

```
reqBody := map[string]interface{}{
    "system_instruction": map[string]interface{}{
        "parts": []map[string]interface{}{
            {
                "text": "You are a platform execution agent. Parse the task and return a concise, step-by-step execution plan.",
            },
        },
    },
    "contents": []map[string]interface{}{
        {
            "parts": []map[string]interface{}{
                {
                    "text": payload.Task,
                },
            },
        },
    },
}
```

Gemini 會依據收到的任務內容生成逐步執行計畫（Step-by-Step Execution Plan）。

須注意：此案例雖然稱為 Go ADK Execution Service，但提供的 Go 程式實際使用標準 HTTP 與 Gemini REST API，並未展示 Google ADK SDK 的 Agent 建構方式。

此外，模型回傳的是文字形式的計畫，並不代表資料庫遷移等外部操作已真正執行。

## 7. 總結

本案例展示 AI Agent 系統的重要架構原則：推理與執行分離（Decoupling Reasoning from Execution）。

Python LangGraph 負責任務編排，Go 服務負責遠端任務處理，兩者透過 HTTP 與 JSON 完成跨語言協作。

其主要價值在於降低系統耦合（System Coupling）、提升模組化（Modularity），並支援異質 AI 服務整合（Heterogeneous AI Service Integration）。

不過，這份程式的定位是跨框架通訊的概念驗證（Proof of Concept），而不是完整符合 A2A 1.0 規範的正式實作。

### 延伸閱讀：LangGraph 多代理工作流程（Multi-Agent Workflows）

參考文章： [LangChain Blog — LangGraph: Multi-Agent Workflows](https://www.langchain.com/blog/langgraph-multi-agent-workflows?utm_source=chatgpt.com)

LangChain 官方文章介紹如何使用 LangGraph 建立多代理系統（Multi-Agent System），讓不同 AI Agent 分工合作，共同完成複雜任務。

文章介紹三種主要架構：

1. 多代理協作（Multi-Agent Collaboration）：多個 Agent 共享工作狀態與訊息，協同完成任務。
2. 代理監督者（Agent Supervisor）：由 Supervisor Agent 負責任務分派與協調，各 Agent 獨立處理專屬工作。
3. 階層式代理團隊（Hierarchical Agent Teams）：透過多層 Agent 結構，將複雜任務分解為不同子任務。

與 A2A Protocol 的關係：

LangGraph 主要處理工作流程編排（Workflow Orchestration）及 Agent 之間的控制流程；A2A 則著重跨服務、跨框架的通訊與互通性（Interoperability）。

兩者可互補使用：LangGraph 負責協調任務，A2A 負責不同系統之間的 Agent 通訊。

### AI Agent 架構選擇：RAG、MCP、Sub-agents 與 A2A

影片來源： [Google Cloud Tech — Stop using MCP for everything! (RAG vs. A2A vs. Sub-agents)](https://www.youtube.com/watch?v=ks4M8b9Ul6E\&utm_source=chatgpt.com)

## 1. 核心概念：不同 AI 技術解決不同問題

開發 AI 應用程式時，常見的技術包括：

- RAG（Retrieval-Augmented Generation，檢索增強生成）：讓 AI 從外部知識庫取得相關資訊。
- MCP（Model Context Protocol，模型上下文協定）：標準化 AI 與外部資料、工具及系統的連接方式。
- Sub-agents（子代理）：將複雜任務拆分給不同專業 Agent。
- A2A（Agent-to-Agent Protocol，代理間通訊協定）：讓獨立的 AI Agent 跨系統協作。

這些技術並非互相取代，而是解決不同層次的系統設計需求。

## 2. RAG：從外部知識庫取得資訊

應用案例：Technician Helper Bot（技術人員維修助手）

維修人員需要查詢 20 年前洗衣機的錯誤代碼，AI 必須從大量維修手冊中找出正確內容。

適合使用 RAG（檢索增強生成）。

RAG 的基本流程：

使用者問題（User Query）

知識檢索（Retrieval）

從維修手冊搜尋相關內容

LLM 生成回答（Generation）

結合問題與檢索內容產生答案

Google Cloud 相關技術包括：

- Vertex AI Search / RAG Engine：建立知識檢索能力。
- ADK（Agent Development Kit，代理開發套件）：建立 AI 應用。
- Cloud Run / Agent Engine：部署及執行 AI 服務。

適用情境： 文件搜尋、企業知識問答、技術手冊查詢。

## 3. MCP：標準化資料存取與工具操作

MCP 在影片中分成兩種主要應用。

### 3.1 MCP 作為資料查詢工具（Federated Lookup）

應用案例：Executive Insights Dashboard（企業高階主管分析儀表板）

主管希望比較不同區域的營收與行銷支出，但資料分散於訂單及行銷系統。

透過 MCP，每個業務團隊可以維護自己的 MCP Server，提供受控的資料存取介面。

AI Agent

分析使用者問題並選擇工具

Order MCP Server

訂單與營收資料

Campaign MCP Server

行銷活動與支出資料

其優點在於資料擁有者可以自行管理存取權限與業務邏輯，不必直接開放完整資料庫。

### 3.2 MCP 作為動作執行工具（Action Tool）

應用案例：Supply Bot（零件訂購助手）

維修人員提出：

> 訂購 XYZ 型號的壓縮機，並將費用記錄至工作單 123。

AI Agent 透過 MCP 呼叫後端訂購工具，完成零件訂購。

此時 MCP 不只是讀取資訊，也提供 工具呼叫（Tool Calling） 能力。

因此，MCP 的兩種用途可以區分為：

| 類型                | 功能         | 案例        |
| ----------------- | ---------- | --------- |
| Lookup Tool（查詢工具） | 讀取外部資料     | 查詢訂單、營收   |
| Action Tool（動作工具） | 呼叫外部系統執行操作 | 訂購零件、更新工單 |

&#x20;     CSV          Excel&#x20;

## 4. Sub-agents：應用程式內部的多代理分工

應用案例：Triage & Dispatch System（案件分類與派工系統）

當客戶回報「冰箱漏水並發出異常聲音」時，系統需要完成故障分析、排程、派工與通知。

這類多步驟任務適合採用 Sub-agents（子代理） 架構。

Manager / Root Agent

負責協調與任務分派

Diagnosis Agent

故障診斷

Scheduling Agent

排程安排

Notification Agent

通知派工

這種架構以管理代理（Manager Agent）負責協調，多個專業代理（Specialist Agents）各自處理不同任務，並透過共享狀態（Shared State）合作。

如果所有 Agent 都屬於同一個應用程式，通常可以直接使用 ADK 或 LangGraph 的內部協作機制，不一定需要 A2A。

## 5. A2A：跨組織、跨系統的 Agent 協作

應用案例：Warranty Verifier（產品保固驗證系統）

維修公司需要向製造商確認某個產品是否仍在保固期內，但製造商不希望開放完整的內部資料庫。

此時可使用 A2A（Agent-to-Agent Protocol）。

Fix It Fast Agent

提出保固驗證請求

A2A Protocol

任務與結果交換

Manufacturer Agent

使用內部資料與規則判斷保固

製造商 Agent 只需要回傳驗證結果，不必將原始資料或內部判斷邏輯暴露給維修公司。

A2A 也能支援長時間任務（Long-running Tasks）；影片提及可搭配 Pub/Sub 與人工審核流程處理複雜保固申請。

重要差異： Sub-agents 偏重單一應用中的內部分工；A2A 偏重獨立 Agent 服務之間的互通性。跨公司是 A2A 的典型應用，但不是必要條件。

## 8. 總結

這四種技術分別處理 AI 系統中的不同需求：

- RAG 解決知識取得問題。
- MCP 解決資料與工具連接問題。
- Sub-agents 解決複雜任務的內部分工問題。
- A2A 解決獨立 Agent 之間的通訊與協作問題。

在實際的企業級 AI 系統中，這些技術可以組合使用。例如，主 Agent 透過 Sub-agents 分配工作，各子代理利用 RAG 查詢知識、透過 MCP 操作業務工具，並透過 A2A 與外部公司的 Agent 協作。

因此，AI Agent 架構設計的重點不是選擇單一技術，而是根據知識來源、工具存取、任務分工及系統邊界，選擇適當的技術組合。

### Google Cloud：Building AI Workflows — AI 推理與執行解耦架構

## 1. 核心概念：推理與執行分離（Reasoning–Execution Decoupling）

在企業級 AI 系統中，大型語言模型（Large Language Model, LLM）主要負責理解問題、規劃任務與選擇工具；外部服務則負責實際執行操作。

透過解耦（Decoupling），兩個部分可以獨立開發、部署與擴充。



| 系統層                      | 主要職責               |
| ------------------------ | ------------------ |
| 推理層（Reasoning Layer）     | 分析問題、規劃步驟、決定行動     |
| 編排層（Orchestration Layer） | 管理任務順序、狀態與代理協作     |
| 執行層（Execution Layer）     | 呼叫 API、存取資料及執行業務操作 |



## 2. AI Workflow 系統架構

User Request（使用者請求）

Gemini / LLM

理解任務與產生執行計畫

此圖是概念性架構；實際系統可能將推理與編排整合在同一個 Agent 服務中。

## 3. Google Cloud 相關技術

- ADK（Agent Development Kit）：建立可使用工具、規劃任務及協調工作流程的 AI Agent。
- Cloud Run：部署可獨立擴充的 API 與 Agent 服務。
- MCP（Model Context Protocol）：讓 Agent 透過標準介面存取外部工具。
- A2A（Agent-to-Agent Protocol）：支援不同 Agent 服務之間的任務委派。
- Agent Runtime / Agent Executor：提供 Agent 執行環境；部分架構支援長時間任務、執行恢復與分散式部署。




## 4. 生產環境設計原則

從概念驗證（Proof of Concept）進入正式環境（Production）時，需要考慮：



| 設計原則                  | 說明            |
| --------------------- | ------------- |
| 模組化（Modularity）       | 各服務具有明確職責     |
| 可擴充性（Scalability）     | 推理與執行服務可獨立擴充  |
| 容錯能力（Fault Tolerance） | 處理服務失敗、逾時與重試  |
| 安全性（Security）         | 驗證身分並限制工具存取權限 |
| 可觀測性（Observability）   | 記錄任務執行、錯誤與效能  |



這些原則有助於提高 AI Workflow 的可靠性（Reliability）及可維護性（Maintainability）。

```python
%%writefile /content/a2a_demo/main.go
package main

import (
    "bytes"
    "encoding/json"
    "fmt"
    "io"
    "log"
    "net/http"
    "os"
    "strings"
    "time"
)

// ----------------------------------------------------
// 1. JSON Payload Schema
// ----------------------------------------------------

type A2APayload struct {
    A2AVersion string            `json:"a2a_version"`
    SenderID   string            `json:"sender_id"`
    Task       string            `json:"task"`
    Context    map[string]string `json:"context"`
}

type ExecutionResult struct {
    FinalResult string `json:"final_result"`
}

// ----------------------------------------------------
// 2. Gemini Response Schema
// ----------------------------------------------------

type GeminiResponse struct {
    Candidates []struct {
        Content struct {
            Parts []struct {
                Text string `json:"text"`
            } `json:"parts"`
        } `json:"content"`
    } `json:"candidates"`
}

// ----------------------------------------------------
// 3. Health Check
// ----------------------------------------------------

func healthHandler(w http.ResponseWriter, r *http.Request) {
    w.Header().Set("Content-Type", "application/json")
    json.NewEncoder(w).Encode(map[string]string{
        "status": "ok",
        "service": "Go Execution Agent",
    })
}

// ----------------------------------------------------
// 4. Gemini API
// ----------------------------------------------------

func callGemini(task string) (string, error) {

    apiKey := os.Getenv("GEMINI_API_KEY")
    if apiKey == "" {
        return "", fmt.Errorf("GEMINI_API_KEY not configured")
    }

    model := os.Getenv("GEMINI_MODEL")
    if model == "" {
        model = "gemini-2.5-flash"
    }

    scheme := "https" + "://"
    url := fmt.Sprintf(
        scheme+"[generativelanguage.googleapis.com/v1beta/models/%s:generateContent](https://generativelanguage.googleapis.com/v1beta/models/%s:generateContent)",
        model,
    )

    requestBody := map[string]interface{}{
        "systemInstruction": map[string]interface{}{
            "parts": []map[string]string{
                {
                    "text": "You are an AI execution planning agent. " +
                        "Analyze the task and return a clear, step-by-step " +
                        "execution plan. Respond in Traditional Chinese. " +
                        "Do not claim that real-world actions were executed.",
                },
            },
        },
        "contents": []map[string]interface{}{
            {
                "parts": []map[string]string{
                    {"text": task},
                },
            },
        },
    }

    jsonData, err := json.Marshal(requestBody)
    if err != nil {
        return "", err
    }

    req, err := http.NewRequest(
        "POST",
        url,
        bytes.NewBuffer(jsonData),
    )
    if err != nil {
        return "", err
    }

    req.Header.Set("Content-Type", "application/json")
    req.Header.Set("x-goog-api-key", apiKey)

    client := &http.Client{
        Timeout: 120 * time.Second,
    }

    resp, err := client.Do(req)
    if err != nil {
        return "", err
    }
    defer resp.Body.Close()

    body, err := io.ReadAll(io.LimitReader(resp.Body, 4<<20))
    if err != nil {
        return "", err
    }

    if resp.StatusCode != http.StatusOK {
        return "", fmt.Errorf(
            "Gemini API returned HTTP %d: %s",
            resp.StatusCode,
            string(body),
        )
    }

    var result GeminiResponse

    if err := json.Unmarshal(body, &result); err != nil {
        return "", err
    }

    if len(result.Candidates) == 0 {
        return "", fmt.Errorf("no Gemini candidates returned")
    }

    var texts []string

    for _, part := range result.Candidates[0].Content.Parts {
        if part.Text != "" {
            texts = append(texts, part.Text)
        }
    }

    finalText := strings.Join(texts, "\n")

    if finalText == "" {
        return "", fmt.Errorf("Gemini returned no text")
    }

    return finalText, nil
}

// ----------------------------------------------------
// 5. A2A-style Task Handler
// ----------------------------------------------------

func handleA2A(w http.ResponseWriter, r *http.Request) {

    if r.Method != http.MethodPost {
        http.Error(w, "POST required", http.StatusMethodNotAllowed)
        return
    }

    var payload A2APayload

    r.Body = http.MaxBytesReader(w, r.Body, 1<<20)

    if err := json.NewDecoder(r.Body).Decode(&payload); err != nil {
        http.Error(w, "Invalid JSON payload", http.StatusBadRequest)
        return
    }

    if strings.TrimSpace(payload.Task) == "" {
        http.Error(w, "Task is empty", http.StatusBadRequest)
        return
    }

    log.Printf(
        "[Agent B] Received from=%s task=%q",
        payload.SenderID,
        payload.Task,
    )

    log.Println("[Agent B] Calling Gemini API...")

    result, err := callGemini(payload.Task)

    if err != nil {
        log.Printf("[Agent B] Error: %v", err)
        http.Error(
            w,
            "Execution failed; see server log",
            http.StatusBadGateway,
        )
        return
    }

    w.Header().Set("Content-Type", "application/json")

    json.NewEncoder(w).Encode(ExecutionResult{
        FinalResult: result,
    })

    log.Println("[Agent B] Response sent to Agent A")
}

// ----------------------------------------------------
// 6. Main Server
// ----------------------------------------------------

func main() {

    mux := http.NewServeMux()

    mux.HandleFunc("/health", healthHandler)
    mux.HandleFunc("/execute", handleA2A)

    log.Println("===================================")
    log.Println("Go Execution Agent Started")
    log.Println("Listening on 127.0.0.1:18080")
    log.Println("===================================")

    // Listen on port 18080 to avoid conflict with Colab's internal port 8080
    log.Fatal(http.ListenAndServe("127.0.0.1:18080", mux))
}
```

### Part III：啟動 Go Agent 背景服務

#### Cell 5：以 Background Process 啟動 Go Server
這個 Cell 是整個實驗最關鍵的部分。
使用 Python 的 `subprocess.Popen()` 啟動 Go Server。`Popen()` 不會等待 Server 結束，而是讓 Server 以另一個作業系統 Process 持續在背景執行，並將輸出導向 `go_agent.log`。

> **Colab 環境關鍵設定**：由於 Colab 預設配置了內部網路 Proxy，直接請求 `127.0.0.1` 會被 Proxy 攔截導致連線失敗。因此必須先設定 `os.environ["NO_PROXY"] = "127.0.0.1,localhost"`，讓本機端對端通訊直連。

```text
+-------------------------------------------------------------------------+
|                   Colab Runtime (Linux OS Environment)                  |
|                                                                         |
|  [Python Notebook Process] (設定 NO_PROXY="127.0.0.1,localhost")        |
|       |                                                                 |
|       | 1. subprocess.Popen(["/content/a2a_demo/go_agent"])             |
|       +---------------------------------------+                         |
|       |                                       v                         |
|       |                        [Go Server Process (PID: xxxx)] (背景常駐)|
|       |                         ├── 持續監聽 127.0.0.1:18080             |
|       |                         └── 輸出寫入 ──> go_agent.log            |
|       |                                       ^                         |
|       | 2. GET /health (繞過 Proxy 本機直連)    |                         |
|       +---------------------------------------+                         |
+-------------------------------------------------------------------------+
```

**[Cell 5 執行程式碼]**

````python
import subprocess
import os
import time
import requests

# 關鍵修復：設定 NO_PROXY，避免 Colab 內部 Proxy 攔截 127.0.0.1 的本機連線
os.environ["NO_PROXY"] = "127.0.0.1,localhost"
os.environ["no_proxy"] = "127.0.0.1,localhost"

SERVER_URL = "http://127.0.0.1:18080"
LOG_PATH = PROJECT_DIR / "go_agent.log"

# 先清理可能已經在背景執行的舊 go_agent 行程
subprocess.run(["pkill", "-f", "go_agent"], check=False)
time.sleep(0.5)

log_handle = open(LOG_PATH, "w")
go_process = subprocess.Popen(
    [str(PROJECT_DIR / "go_agent")],
    cwd=str(PROJECT_DIR),
    stdout=log_handle,
    stderr=subprocess.STDOUT,
    env=os.environ.copy(),
)
log_handle.close()
print("Started Go process PID:", go_process.pid)

# Wait until HTTP health check succeeds
ready = False
for _ in range(30):
    if go_process.poll() is not None:
        break
    try:
        response = requests.get(
            SERVER_URL + "/health",
            timeout=1
        )
        if response.ok:
            ready = True
            break
    except requests.RequestException:
        pass
    time.sleep(0.5)

if not ready:
    print("Process poll status:", go_process.poll())
    print(LOG_PATH.read_text()[-3000:])
    raise RuntimeError("Go Server did not start correctly")

print("Go Server is ready.")
print(response.json())
````

### Part IV：建立 Python LangGraph Orchestrator（前端 Agent）

#### Cell 7：定義 AgentState 與兩個 LangGraph Nodes
原始 Python 程式使用 `AgentState` 管理三項資訊，並由兩個 Node 負責處理：
1. **`reasoning_node`**：建立結構化任務訊息（封裝 JSON Payload）。
2. **`a2a_dispatch_node`**：將訊息透過 HTTP POST 送至 Go Agent，並取得回應。

```text
+-------------------------------------------------------------------------+
|                        AgentState (狀態結構定義)                         |
+------------------+------------------------------------------------------+
| 狀態欄位         | 用途                                                 |
+------------------+------------------------------------------------------+
| objective        | 使用者設定的任務 (輸入字串)                          |
| a2a_payload      | 準備發送至 Go Server 的結構化 JSON 字串               |
| final_result     | Go Agent 呼叫 Gemini 後回傳的最終文字結果            |
+------------------+------------------------------------------------------+
```

**[Cell 7 執行程式碼]**
```python
import json
import requests
from typing import TypedDict

# ----------------------------------------------------
# 1. Agent State
# ----------------------------------------------------
class AgentState(TypedDict):
    objective: str
    a2a_payload: str
    final_result: str

# ----------------------------------------------------
# 2. Reasoning Node
# ----------------------------------------------------
def reasoning_node(state: AgentState):
    print("[Agent A] Preparing task payload...")
    envelope = {
        "a2a_version": "1.0",
        "sender_id": "langgraph-orchestrator",
        "task": state["objective"],
        "context": {
            "urgency": "high",
            "source": "Google Colab"
        }
    }
    return {
        "a2a_payload": json.dumps(
            envelope,
            ensure_ascii=False
        )
    }

# ----------------------------------------------------
# 3. Dispatch Node
# ----------------------------------------------------
def a2a_dispatch_node(state: AgentState):
    print("[Agent A] Sending task to Go Agent...")
    payload = json.loads(state["a2a_payload"])
    execute_url = f"http://{'127.0.0.1:18080'}/execute"
    
    response = requests.post(
        execute_url,
        json=payload,
        timeout=150

    response.raise_for_status()
    result = response.json()

    return {
        "final_result": result["final_result"]
    }
```

````python
from langgraph.graph import StateGraph, END

workflow = StateGraph(AgentState)

# Add Nodes
workflow.add_node(
    "reason",
    reasoning_node
)
workflow.add_node(
    "delegate_via_a2a",
    a2a_dispatch_node
)

# Define Entry Point
workflow.set_entry_point("reason")

# Define Edges
workflow.add_edge(
    "reason",
    "delegate_via_a2a"
)
workflow.add_edge(
    "delegate_via_a2a",
    END
)

# Compile Graph
langgraph_app = workflow.compile()

print("LangGraph Orchestrator is ready.")
````

Go Server 負責接收 Agent A 的 JSON Payload，透過環境變數中的 Vertex AI 專案設定與 Bearer Token 向 Google Cloud Vertex AI 發送請求，再將結果回傳。

```text
                               │      ├── 附加 Header: Authorization: Bearer <TOKEN>

                               └── 3. 封裝並回傳 ExecutionResult (JSON)
+-------------------------------------------------------------------------+
```

**[Cell 3 執行程式碼]**
```python
%%writefile /content/a2a_demo/main.go
package main

import (
    "bytes"
    "encoding/json"
    "fmt"
    "io"
    "log"
    "net/http"
    "os"
    "strings"
    "time"
)

// ----------------------------------------------------
// 1. JSON Payload Schema
// ----------------------------------------------------

type A2APayload struct {
    A2AVersion string            `json:"a2a_version"`
    SenderID   string            `json:"sender_id"`
    Task       string            `json:"task"`
    Context    map[string]string `json:"context"`
}

type ExecutionResult struct {
    FinalResult string `json:"final_result"`
}

// ----------------------------------------------------
// 2. Gemini Response Schema
// ----------------------------------------------------

type GeminiResponse struct {
    Candidates []struct {
        Content struct {
            Parts []struct {
                Text string `json:"text"`
            } `json:"parts"`
        } `json:"content"`
    } `json:"candidates"`
}

// ----------------------------------------------------
// 3. Health Check
// ----------------------------------------------------

func healthHandler(w http.ResponseWriter, r *http.Request) {
    w.Header().Set("Content-Type", "application/json")
    json.NewEncoder(w).Encode(map[string]string{
        "status":  "ok",
        "service": "Go Execution Agent (Vertex AI)",
    })
}

// ----------------------------------------------------
// 4. Vertex AI Gemini API
// ----------------------------------------------------

func callGemini(task string) (string, error) {
    token := strings.TrimSpace(os.Getenv("VERTEX_ACCESS_TOKEN"))
    project := strings.TrimSpace(os.Getenv("GCP_PROJECT_ID"))
    location := strings.TrimSpace(os.Getenv("GCP_LOCATION"))
    model := strings.TrimSpace(os.Getenv("GEMINI_MODEL"))

    if token == "" || project == "" {
        return "", fmt.Errorf("Vertex AI credentials (VERTEX_ACCESS_TOKEN / GCP_PROJECT_ID) not configured")
    }
    if location == "" {
        location = "us-central1"
    }
    if model == "" {
        model = "gemini-2.5-flash"
    }

    // 組合 Vertex AI REST API 網址 (拆開字串避免複製時被轉成 Markdown 超連結)
    domain := "aiplatform" + "." + "googleapis" + "." + "com"
    url := fmt.Sprintf(
        "https"+"://%s-%s/v1/projects/%s/locations/%s/publishers/google/models/%s:generateContent",
        location, domain, project, location, model,
    )

    requestBody := map[string]interface{}{
        "systemInstruction": map[string]interface{}{
            "role": "system",
            "parts": []map[string]string{
                {
                    "text": "You are an AI execution planning agent. " +
                        "Analyze the task and return a clear, step-by-step " +
                        "execution plan. Respond in Traditional Chinese. " +
                        "Do not claim that real-world actions were executed.",
                },
            },
        },
        "contents": []map[string]interface{}{
            {
                "role": "user",
                "parts": []map[string]string{
                    {"text": task},
                },
            },
        },
    }

    jsonData, err := json.Marshal(requestBody)
    if err != nil {
        return "", err
    }

    req, err := http.NewRequest("POST", url, bytes.NewBuffer(jsonData))
    if err != nil {
        return "", err
    }

    req.Header.Set("Content-Type", "application/json")
    req.Header.Set("Authorization", "Bearer "+token)

    client := &http.Client{
        Timeout: 120 * time.Second,
    }

    resp, err := client.Do(req)
    if err != nil {
        return "", err
    }
    defer resp.Body.Close()

    body, err := io.ReadAll(io.LimitReader(resp.Body, 4<<20))
    if err != nil {
        return "", err
    }

    if resp.StatusCode != http.StatusOK {
        return "", fmt.Errorf(
            "Vertex AI returned HTTP %d: %s",
            resp.StatusCode,
            string(body),
        )
    }

    var result GeminiResponse
    if err := json.Unmarshal(body, &result); err != nil {
        return "", err
    }

    if len(result.Candidates) == 0 {
        return "", fmt.Errorf("no Gemini candidates returned")
    }

    var texts []string
    for _, part := range result.Candidates[0].Content.Parts {
        if part.Text != "" {
            texts = append(texts, part.Text)
        }
    }

    finalText := strings.Join(texts, "\n")
    if finalText == "" {
        return "", fmt.Errorf("Gemini returned no text")
    }

    return finalText, nil
}

// ----------------------------------------------------
// 5. A2A-style Task Handler
// ----------------------------------------------------

func handleA2A(w http.ResponseWriter, r *http.Request) {
    if r.Method != http.MethodPost {
        http.Error(w, "POST required", http.StatusMethodNotAllowed)
        return
    }

    var payload A2APayload
    r.Body = http.MaxBytesReader(w, r.Body, 1<<20)

    if err := json.NewDecoder(r.Body).Decode(&payload); err != nil {
        http.Error(w, "Invalid JSON payload", http.StatusBadRequest)
        return
    }

    if strings.TrimSpace(payload.Task) == "" {
        http.Error(w, "Task is empty", http.StatusBadRequest)
        return
```

````python
%%writefile /content/a2a_demo/main.go
package main

import (
    "bytes"
    "encoding/json"
    "fmt"
    "io"
    "log"
    "net/http"
    "os"
    "strings"
    "time"
)

// ----------------------------------------------------
// 1. JSON Payload Schema
// ----------------------------------------------------

type A2APayload struct {
    A2AVersion string            `json:"a2a_version"`
    SenderID   string            `json:"sender_id"`
    Task       string            `json:"task"`
    Context    map[string]string `json:"context"`
}

type ExecutionResult struct {
    FinalResult string `json:"final_result"`
}

// ----------------------------------------------------
// 2. Gemini Response Schema
// ----------------------------------------------------

type GeminiResponse struct {
    Candidates []struct {
        Content struct {
            Parts []struct {
                Text string `json:"text"`
            } `json:"parts"`
        } `json:"content"`
    } `json:"candidates"`
}

// ----------------------------------------------------
// 3. Health Check
// ----------------------------------------------------

func healthHandler(w http.ResponseWriter, r *http.Request) {
    w.Header().Set("Content-Type", "application/json")
    json.NewEncoder(w).Encode(map[string]string{
        "status":  "ok",
        "service": "Go Execution Agent (Vertex AI)",
    })
}

// ----------------------------------------------------
// 4. Vertex AI Gemini API
// ----------------------------------------------------

func callGemini(task string) (string, error) {
    token := strings.TrimSpace(os.Getenv("VERTEX_ACCESS_TOKEN"))
    project := strings.TrimSpace(os.Getenv("GCP_PROJECT_ID"))
    location := strings.TrimSpace(os.Getenv("GCP_LOCATION"))
    model := strings.TrimSpace(os.Getenv("GEMINI_MODEL"))

    if token == "" || project == "" {
        return "", fmt.Errorf("Vertex AI credentials (VERTEX_ACCESS_TOKEN / GCP_PROJECT_ID) not configured")
    }
    if location == "" {
        location = "us-central1"
    }
    if model == "" {
        model = "gemini-2.5-flash"
    }

    // 組合 Vertex AI REST API 網址 (拆開字串避免複製時被轉成 Markdown 超連結)
    domain := "aiplatform" + "." + "googleapis" + "." + "com"
    url := fmt.Sprintf(
        "https"+"://%s-%s/v1/projects/%s/locations/%s/publishers/google/models/%s:generateContent",
        location, domain, project, location, model,
    )

    requestBody := map[string]interface{}{
        "systemInstruction": map[string]interface{}{
            "role": "system",
            "parts": []map[string]string{
                {
                    "text": "You are an AI execution planning agent. " +
                        "Analyze the task and return a clear, step-by-step " +
                        "execution plan. Respond in Traditional Chinese. " +
                        "Do not claim that real-world actions were executed.",
                },
            },
        },
        "contents": []map[string]interface{}{
            {
                "role": "user",
                "parts": []map[string]string{
                    {"text": task},
                },
            },
        },
    }

    jsonData, err := json.Marshal(requestBody)
    if err != nil {
        return "", err
    }

    req, err := http.NewRequest("POST", url, bytes.NewBuffer(jsonData))
    if err != nil {
        return "", err
    }

    req.Header.Set("Content-Type", "application/json")
    req.Header.Set("Authorization", "Bearer "+token)

    client := &http.Client{
        Timeout: 120 * time.Second,
    }

    resp, err := client.Do(req)
    if err != nil {
        return "", err
    }
    defer resp.Body.Close()

    body, err := io.ReadAll(io.LimitReader(resp.Body, 4<<20))
    if err != nil {
        return "", err
    }

    if resp.StatusCode != http.StatusOK {
        return "", fmt.Errorf(
            "Vertex AI returned HTTP %d: %s",
            resp.StatusCode,
            string(body),
        )
    }

    var result GeminiResponse
    if err := json.Unmarshal(body, &result); err != nil {
        return "", err
    }

    if len(result.Candidates) == 0 {
        return "", fmt.Errorf("no Gemini candidates returned")
    }

    var texts []string
    for _, part := range result.Candidates[0].Content.Parts {
        if part.Text != "" {
            texts = append(texts, part.Text)
        }
    }

    finalText := strings.Join(texts, "\n")
    if finalText == "" {
        return "", fmt.Errorf("Gemini returned no text")
    }

    return finalText, nil
}

// ----------------------------------------------------
// 5. A2A-style Task Handler
// ----------------------------------------------------

func handleA2A(w http.ResponseWriter, r *http.Request) {
    if r.Method != http.MethodPost {
        http.Error(w, "POST required", http.StatusMethodNotAllowed)
        return
    }

    var payload A2APayload
    r.Body = http.MaxBytesReader(w, r.Body, 1<<20)

    if err := json.NewDecoder(r.Body).Decode(&payload); err != nil {
        http.Error(w, "Invalid JSON payload", http.StatusBadRequest)
        return
    }

    if strings.TrimSpace(payload.Task) == "" {
        http.Error(w, "Task is empty", http.StatusBadRequest)
        return
    }

    log.Printf("[Agent B] Received from=%s task=%q", payload.SenderID, payload.Task)
    log.Println("[Agent B] Calling Vertex AI Gemini API...")

    result, err := callGemini(payload.Task)
    if err != nil {
        log.Printf("[Agent B] Error: %v", err)
        http.Error(w, fmt.Sprintf("Execution failed: %v", err), http.StatusBadGateway)
        return
    }

    w.Header().Set("Content-Type", "application/json")
    json.NewEncoder(w).Encode(ExecutionResult{
        FinalResult: result,
    })

    log.Println("[Agent B] Response sent to Agent A")
}

// ----------------------------------------------------
// 6. Main Server
// ----------------------------------------------------

func main() {
    mux := http.NewServeMux()
    mux.HandleFunc("/health", healthHandler)
    mux.HandleFunc("/execute", handleA2A)

    log.Println("===================================")
    log.Println("Go Execution Agent Started (Vertex AI)")
    log.Println("Listening on 127.0.0.1:18080")
    log.Println("===================================")

    log.Fatal(http.ListenAndServe("127.0.0.1:18080", mux))
}
````

#### 這個 Agent 具備什麼能力？該對它說什麼？
雖然後端是用 Go 語言編寫的，但 Go Server 在這裡的角色是**「高效能的 API 閘道與任務執行器」**，負責管理連線與封裝請求；真正的大腦則是背後的 **Vertex AI Gemini 模型**，並被設定為**「AI 執行規劃代理（Execution Planning Agent）」**。

因此，這個 Agent 最擅長將**「一句話的高階目標」**拆解為**「結構化的 Step-by-Step 執行計畫、風險評估與驗證步驟」**。非常適合對它提出以下四大類任務：

4. **技術社群與大型活動籌辦**：例如規劃一場 100 人的 AI 開發者黑客松或雙週技術讀書會的完整執行 SOP。

### Part VII：觀察背景 Agent 與管理服務

#### Cell 11：查看 Go Agent 的背景日誌
Go Server 的輸出被寫入 `/content/a2a_demo/go_agent.log`，因此即使 Server 在背景執行，仍可以隨時檢查任務紀錄。這些日誌可以用於教學說明：Go Server 並非在 Python Cell 結束後消失，而是一直在背景等待任務。

```text
+-------------------------------------------------------------------------+
|                 獨立執行生命週期 (Independent Lifecycle)                |
+-------------------------------------------------------------------------+
  [前景 Python Cell] ──(讀取最後 30 行)──> [/content/a2a_demo/go_agent.log]
                                                        ^
                                                        | (持續寫入日誌)
  [前景 Python Cell] ──(terminate / kill)─> [背景 Go Process (Port 18080)]
+-------------------------------------------------------------------------+
```

#### Cell 12：停止 Go Server
結束實驗時，可以停止背景 Go Process，釋放 Port `18080`。
停止後，Agent A 將無法繼續透過 `127.0.0.1:18080` 呼叫 Go Server，除非重新執行 **Cell 5**。這也展示了跨服務架構中的一個重要概念：前端 Agent 與後端 Agent 具有獨立的執行生命週期（Independent Execution Lifecycle）。

```python
import subprocess

    try:

        go_process.wait()
    print("Go Agent stopped.")
```

# Step 7：建立「只執行一次」的 OAuth 初始化函式

這個函式的工作是：

```text
觸發 Calendar Tool
↓
ADK 發出 OAuth Request
↓
User Google Login / Consent
↓
Flask 收 callback
↓
把 callback 交回 ADK
↓
ADK Token Exchange
↓
Credential 留在目前 Session
```

```python
import asyncio

from IPython.display import display, HTML


async def initialize_oauth():

    oauth_result.clear()

    # 用一個 harmless read request 觸發 Calendar OAuth
    content = types.Content(
        role="user",
        parts=[
            types.Part(
                text="請查看我今天的 Calendar 行程。"
            )
        ],
    )

    events = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=content,
    )

    auth_call = None
    auth_config_from_adk = None

    async for event in events:

        if not event.content:
            continue

        for part in event.content.parts or []:

            if (
                part.function_call
                and
                part.function_call.name
                == "adk_request_credential"
            ):

                auth_call = part.function_call

                raw_config = (
                    auth_call.args.get("authConfig")
                    or
                    auth_call.args.get("auth_config")
                )

                auth_config_from_adk = (
                    AuthConfig.model_validate(
                        raw_config
                    )
                )

                break

        if auth_call:
            break


    if not auth_call:
        print(
            "✅ OAuth credential already available."
        )
        return


    AUTH_URL = (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_uri
    )


    print("請打開以下網址完成 Google OAuth：\n")
    print(AUTH_URL)

    display(
        HTML(
            f'''
            <p>
            <a href="{AUTH_URL}" target="_blank">
            🔐 Open Google OAuth
            </a>
            </p>
            '''
        )
    )


    print("等待 Google OAuth callback...")


    # 等待 Flask 收到 callback
    for _ in range(300):

        if oauth_result.get("callback_url"):
            break

        await asyncio.sleep(1)


    if not oauth_result.get("callback_url"):
        raise TimeoutError(
            "OAuth callback timeout."
        )


    if oauth_result.get("error"):
        raise RuntimeError(
            f"OAuth error: {oauth_result['error']}"
        )


    # 把 Google callback 交回 ADK
    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .auth_response_uri
    ) = oauth_result["callback_url"]


    (
        auth_config_from_adk
        .exchanged_auth_credential
        .oauth2
        .redirect_uri
    ) = REDIRECT_URI


    auth_content = types.Content(
        role="user",

        parts=[
            types.Part(
                function_response=
                types.FunctionResponse(

                    id=auth_call.id,

                    name="adk_request_credential",

                    response=(
                        auth_config_from_adk
                        .model_dump()
                    ),
                )
            )
        ],
    )


    # Resume 同一個 Session
    events_after_auth = runner.run_async(
        session_id=session.id,
        user_id=USER_ID,
        new_message=auth_content,
    )


    # 必須把 events consume 完
    async for event in events_after_auth:
        pass


    print("✅ Google Calendar OAuth initialized")
```

### Appendix：Cell 3 Go Execution Server (`main.go`) 程式碼原理解析

對於熟悉 Python 但較少接觸 Go 語言的開發者來說，可以把這支 `main.go` 想像成一個**「內建型別檢查的輕量級 FastAPI / Flask 伺服器」**。它完全只使用 Go 語言內建的標準函式庫（Standard Library），不需要 `pip install` 任何外部套件，就能編譯成一個獨立的高效能執行檔。

整支程式由上而下共分為 **6 個核心區塊**：

```text
+-------------------------------------------------------------------------+
|                      main.go 程式結構與運作流程                         |
+-------------------------------------------------------------------------+
  [區塊 1 & 2] 資料結構定義 (Structs)
   ├── A2APayload / ExecutionResult  (定義 Python 與 Go 溝通的 JSON 格式)
   └── GeminiResponse                (定義拆解 Vertex AI 回應的 JSON 格式)
          |
          v
  [區塊 6] 程式進入點 main()
   └── 啟動 HTTP Server 監聽 127.0.0.1:18080，並註冊兩條路由：
          |
          +──> 路由 A: GET /health   ──> [區塊 3] healthHandler()
          |                              回傳 {"status": "ok"} 供存活檢查
          |
          └──> 路由 B: POST /execute ──> [區塊 5] handleA2A()
                                         1. 接收並檢查 Python 傳來的 JSON
                                         2. 呼叫 [區塊 4] callGemini(task)
                                            └── 帶上 Bearer Token 呼叫 Vertex AI
                                         3. 將 Gemini 生成的計畫打包回傳給 Python
+-------------------------------------------------------------------------+
```

---

#### 區塊 1：套件引入與 A2A 通訊資料結構（對應 Python 的 `import` 與 `Pydantic`）

```go
package main

import (
    "bytes"
    "encoding/json"
    "fmt"
    "io"
    "log"
    "net/http"
    "os"
    "strings"
    "time"
)

type A2APayload struct {
    A2AVersion string            `json:"a2a_version"`
    SenderID   string            `json:"sender_id"`
    Task       string            `json:"task"`
    Context    map[string]string `json:"context"`
}

type ExecutionResult struct {
    FinalResult string `json:"final_result"`
}
```
* **工作原理**：
  * **`import`**：引入 Go 內建的網路（`net/http`）、JSON 處理（`encoding/json`）與系統環境變數（`os`）工具。
  * **`type ... struct`**：Go 是強型別語言，不能像 Python 一樣隨意拿一個 `dict` 就直接塞資料。這裡的 `struct` 就像 Python 的 `TypedDict` 或 `Pydantic BaseModel`。
  * **`` `json:"..."` `` 標籤**：告訴 Go 在把資料轉成 JSON 或從 JSON 讀取時，對應的欄位名稱是什麼。`A2APayload` 是準備**收進來**的包裹格式，而 `ExecutionResult` 則是最後要**回傳給 Python** 的包裹格式。

```go
func callGemini(task string) (string, error) {
    token := strings.TrimSpace(os.Getenv("VERTEX_ACCESS_TOKEN"))
    project := strings.TrimSpace(os.Getenv("GCP_PROJECT_ID"))
    location := strings.TrimSpace(os.Getenv("GCP_LOCATION"))
    model := strings.TrimSpace(os.Getenv("GEMINI_MODEL"))

    if token == "" || project == "" {
        return "", fmt.Errorf("Vertex AI credentials not configured")
    }
    if location == "" {
        location = "us-central1"
    }
    if model == "" {
        model = "gemini-2.5-flash"
    }

    domain := "aiplatform" + "." + "googleapis" + "." + "com"
    url := fmt.Sprintf(
        "https"+"://%s-%s/v1/projects/%s/locations/%s/publishers/google/models/%s:generateContent",
        location, domain, project, location, model,
    )

    requestBody := map[string]interface{}{
        "systemInstruction": map[string]interface{}{
            "role": "system",
            "parts": []map[string]string{
                {
                    "text": "You are an AI execution planning agent. " +
                        "Analyze the task and return a clear, step-by-step " +
                        "execution plan. Respond in Traditional Chinese. " +
                        "Do not claim that real-world actions were executed.",
                },
            },
        },
        "contents": []map[string]interface{}{
            {
                "role": "user",
                "parts": []map[string]string{
                    {"text": task},
                },
            },
        },
    }

    jsonData, _ := json.Marshal(requestBody)
    req, _ := http.NewRequest("POST", url, bytes.NewBuffer(jsonData))
    req.Header.Set("Content-Type", "application/json")
    req.Header.Set("Authorization", "Bearer "+token)

    client := &http.Client{Timeout: 120 * time.Second}
    resp, err := client.Do(req)
    // ... (檢查 HTTP 狀態碼並用 GeminiResponse 解析回傳文字)
```
* **工作原理**：這是 Agent B 的「大腦連線器」，等價於在 Python 裡用 `requests.post()` 呼叫 Vertex AI：
  1. **讀取環境變數 (`os.Getenv`)**：把我們在 Cell 2 準備好的專案 ID、模型名稱與 `VERTEX_ACCESS_TOKEN` 讀進來。
  2. **組裝 Prompt (`requestBody`)**：在這裡加上了關鍵的 **`systemInstruction`（系統指令）**，規定 Gemini 扮演「AI 執行規劃代理」並使用繁體中文回答；接著把使用者傳來的 `task` 放進 `contents`。
  3. **發送請求與授權 (`Authorization: Bearer`)**：把組裝好的資料轉成 JSON，帶上 GCP 的通行證（Bearer Token），設好 120 秒 Timeout 後發送給 Google Cloud Vertex AI，最後把生成的文字抽出來回傳。

#### 區塊 5：A2A 任務接收與派發處理器（`handleA2A`）

```go
func handleA2A(w http.ResponseWriter, r *http.Request) {
    if r.Method != http.MethodPost {
        http.Error(w, "POST required", http.StatusMethodNotAllowed)
        return
    }

    var payload A2APayload
    r.Body = http.MaxBytesReader(w, r.Body, 1<<20) // 限制最大 1MB

    if err := json.NewDecoder(r.Body).Decode(&payload); err != nil {
        http.Error(w, "Invalid JSON payload", http.StatusBadRequest)
        return
    }

    log.Printf("[Agent B] Received from=%s task=%q", payload.SenderID, payload.Task)
    log.Println("[Agent B] Calling Vertex AI Gemini API...")

    result, err := callGemini(payload.Task)
    if err != nil {
        log.Printf("[Agent B] Error: %v", err)
        http.Error(w, fmt.Sprintf("Execution failed: %v", err), http.StatusBadGateway)
        return
    }

    w.Header().Set("Content-Type", "application/json")
    json.NewEncoder(w).Encode(ExecutionResult{
        FinalResult: result,
    })
    log.Println("[Agent B] Response sent to Agent A")
}
```
* **工作原理**：這是負責接待 Python LangGraph 的主櫃檯（處理 `/execute` 請求）：
  1. **安全檢查**：確認對方是用 `POST` 方法，且傳來的資料不超過 1MB，並把 JSON 解碼放進 `payload` 變數裡。
  2. **寫入日誌 (`log.Printf`)**：在背景把「收到來自誰的什麼任務」寫進 `go_agent.log`（也就是我們在 Cell 11 看到的紀錄）。
  3. **呼叫模型與錯誤轉發**：呼叫上一個區塊的 `callGemini(payload.Task)`。如果發生錯誤，就回傳 `HTTP 502 (Bad Gateway)` 並附上錯誤原因；如果成功，就把結果打包成 `ExecutionResult` 的 JSON 格式傳回給 Python。

### Appendix II：Cell 7 & Cell 8 Python LangGraph 程式碼原理解析

如果說後端的 Go Server 是負責衝鋒陷陣的「執行官」，那麼前端的 **LangGraph（Agent A）** 就是負責控管流程與傳遞工作記憶的**「專案經理（Orchestrator）」**。

LangGraph 的核心精神是**「狀態機（State Machine）」**：我們先定義好一張共用的**「工作表單（`AgentState`）」**，接著讓這張表單依序流過不同的**「處理站點（Node）」**。每個站點只負責讀取表單上的資料、完成自己的工作，再把新結果寫回表單上，直到抵達終點（`END`）。

整段 Python 程式碼（橫跨 Cell 7 與 Cell 8）共分為 **4 個核心區塊**：

```text
+-------------------------------------------------------------------------+
|                 LangGraph 狀態流轉與工作表單 (AgentState)               |
+-------------------------------------------------------------------------+
  [ 呼叫 invoke() 傳入初始表單 ]
   ├── objective:    "請規劃 PostgreSQL 遷移..."
   ├── a2a_payload:  "" (空)
   └── final_result: "" (空)
          |
          v
  +---------------------------------------------------------------------+
  | [區塊 2] 站點 1：reasoning_node (任務封裝站)                        |
  | 讀取 objective ──> 包裝成 A2A 信封格式 ──> 寫入 a2a_payload 欄位    |
  +---------------------------------------------------------------------+
          |
          | (表單更新：a2a_payload 已填寫)
          v
  +---------------------------------------------------------------------+
  | [區塊 3] 站點 2：a2a_dispatch_node (跨語言派發站)                   |
  | 讀取 a2a_payload ──> HTTP POST 給 Go Server (18080)                 |
  | ──> 取得 Gemini 分析結果 ──> 寫入 final_result 欄位                 |
  +---------------------------------------------------------------------+
          |
          | (表單完成：final_result 已填寫)
          v
  [ 抵達 END：回傳完整的最終表單 final_state ]
+-------------------------------------------------------------------------+
```

#### 區塊 1：定義共用工作表單（`AgentState`）

```python
import json
import requests
from typing import TypedDict

# ----------------------------------------------------
# 1. Agent State
# ----------------------------------------------------
class AgentState(TypedDict):
    objective: str
    a2a_payload: str
    final_result: str
```
* **工作原理**：
  * **`TypedDict`**：在 LangGraph 中，所有節點（Node）之間不會直接把變數傳來傳去，而是共用同一個「狀態字典（State Dictionary）」。
  * 這裡的 `AgentState` 定義了這張共用表單上的三個欄位：
    1. **`objective`**：最初接收到的使用者任務字串。
    2. **`a2a_payload`**：準備跨語言傳送給 Go Server 的 JSON 字串（由第一個 Node 負責填寫）。
    3. **`final_result`**：Go Server 呼叫完 Gemini 後傳回來的最終執行計畫（由第二個 Node 負責填寫）。

#### 區塊 2：第一站 — 任務封裝節點（`reasoning_node`）

```python
# ----------------------------------------------------
# 2. Reasoning Node
# ----------------------------------------------------
def reasoning_node(state: AgentState):
    print("[Agent A] Preparing task payload...")
    envelope = {
        "a2a_version": "1.0",
        "sender_id": "langgraph-orchestrator",
        "task": state["objective"],
        "context": {
            "urgency": "high",
            "source": "Google Colab"
        }
    }
    return {
        "a2a_payload": json.dumps(
            envelope,
            ensure_ascii=False
        )
    }
```
* **工作原理**：
  * **輸入參數 (`state: AgentState`)**：當流程走到這一站時，LangGraph 會自動把當前的 `state` 表單交給這個函式。
  * **組裝信封 (`envelope`)**：它從表單中讀出 `state["objective"]`，並加上版本號（`a2a_version`）、發送者身分（`sender_id`）與上下文資訊（`context`），組裝成符合 A2A 風格的標準字典（正好對應 Go 裡面的 `A2APayload` 結構）。
  * **局部更新狀態 (`return {"a2a_payload": ...}`)**：注意它不需要把整個 `state` 回傳，**只需要回傳「想要更新的欄位」即可**。LangGraph 收到回傳的字典後，會自動幫我們把表單上的 `a2a_payload` 欄位更新好（`ensure_ascii=False` 則是確保中文字不會被轉成 `\uXXXX` 亂碼）。

#### 區塊 3：第二站 — 跨語言通訊派發節點（`a2a_dispatch_node`）

```python
# ----------------------------------------------------
# 3. Dispatch Node
# ----------------------------------------------------
def a2a_dispatch_node(state: AgentState):
    print("[Agent A] Sending task to Go Agent...")
    payload = json.loads(state["a2a_payload"])
    execute_url = f"http://{'127.0.0.1:18080'}/execute"
    
    response = requests.post(
        execute_url,
        json=payload,
        timeout=150
    )
    response.raise_for_status()
    result = response.json()
    print("[Agent A] Response received!")
    return {
        "final_result": result["final_result"]
    }
```
* **工作原理**：
  * **承接上一站的成果**：這一站直接從 `state["a2a_payload"]` 取出上一站打包好的 JSON 字串。
  * **發送 HTTP POST (`requests.post`)**：透過本機網路（`127.0.0.1:18080/execute`）把任務丟給背景的 Go Server，並設定 `timeout=150` 秒等待 Go 與 Gemini 完成推理。
  * **錯誤檢查 (`raise_for_status`)**：如果 Go Server 回傳 200 OK 以外的狀態碼（例如 502 Bad Gateway），這行會立刻拋出例外並中斷流程。
  * **寫回最終結果**：把 Go Server 回傳的 `result["final_result"]` 寫回共用表單的 `final_result` 欄位中。

#### 區塊 4：繪製工作流程圖並編譯（`StateGraph` & `compile`）

```python
from langgraph.graph import StateGraph, END

workflow = StateGraph(AgentState)

# 1. 註冊兩個工作站點 (Nodes)
workflow.add_node("reason", reasoning_node)
workflow.add_node("delegate_via_a2a", a2a_dispatch_node)

# 2. 設定起點 (Entry Point)
workflow.set_entry_point("reason")

# 3. 連接站點之間的移動路線 (Edges)
workflow.add_edge("reason", "delegate_via_a2a")
workflow.add_edge("delegate_via_a2a", END)

# 4. 編譯成可執行的應用程式
langgraph_app = workflow.compile()
```
* **工作原理**：前面三個區塊只是定義了「表單格式」與「個別工人的工作內容」，這個區塊則是把整條生產線組裝起來：
  1. **`StateGraph(AgentState)`**：宣告建立一條以 `AgentState` 為共用表單的工作流。
  2. **`add_node`**：把剛剛寫好的兩個 Python 函式掛上名字（`"reason"` 與 `"delegate_via_a2a"`）放進圖裡。
  3. **`set_entry_point` & `add_edge`**：畫出箭頭方向：`起點 ──> "reason" ──> "delegate_via_a2a" ──> END（終點）`。未來若想加入「自我檢查」或「失敗重試」，只需要在這裡把 `add_edge` 改成條件分支（`add_conditional_edges`）即可，完全不用改動原本的架構。
  4. **`workflow.compile()`**：把畫好的設計圖「編譯」成一個真正可以執行的物件 `langgraph_app`。之後無論是在 Cell 9 單次測試，或是在 Cell 10 按下互動按鈕，都只需要呼叫 `langgraph_app.invoke({...})` 就能一鍵啟動整條跨語言流水線。

#### 二、正式 Google A2A 必須實作的 5 大核心要素（差異最大之處）

##### 1. Agent Card（代理名片與自動探索機制）— *從「寫死網址」到「自動認識」*
* **目前 Demo**：Python 程式碼裡直接寫死了 `127.0.0.1:18080/execute`，而且我們心裡早就知道 Go Server 是做「執行計畫」的。
* **正式 A2A**：Go Server 必須公開一個標準端點 **`GET /.well-known/agent.json`**（稱為 **Agent Card**）。這張 JSON 名片會記載：
  * 這個 Agent 叫什麼名字、版本、Endpoint 在哪裡。
  * **`skills`（技能清單）**：它擅長做什麼（例如：「雲端架構遷移規劃」、「資料庫效能調校」）、接受什麼輸入與輸出格式。
  * **`authentication`（認證方式）**：需要什麼樣的 OAuth2 或 API Key 才能呼叫它。
  * **價值**：Agent A 只要讀取各個 Agent 的名片，就能由 LLM **動態決定**今天這個任務該發包給 Agent B 還是 Agent C，完全不用改寫程式碼。

##### 2. 標準化 JSON-RPC 2.0 通訊與多模態資料（`Message` / `Part` / `Artifact`）
* **目前 Demo**：我們自己發明了 `objective`、`a2a_payload`、`final_result` 這些欄位名稱，只能傳遞單一純文字字串。如果換成別人的 Agent 就完全看不懂。
* **正式 A2A**：統一採用 **JSON-RPC 2.0** 規格（例如方法名稱固定為 `tasks/send`、`tasks/get`、`tasks/cancel`），並規定了標準資料載體：
  * **`Part`（內容片段）**：溝通內容不再只是字串，而是由 `TextPart`（文字）、`FilePart`（檔案/圖片/PDF）、`DataPart`（結構化 JSON 表單）組成的陣列。
  * **`Artifact`（正式產出物）**：Agent B 完成任務後，不會只把長篇大論塞在同一個字串裡，而是會將產出的架構圖、SQL 腳本、遷移計畫書分別打包成獨立的 `Artifact` 交付給 Agent A。

##### 3. Task 任務生命週期管理（Stateful Task Lifecycle）— *差異最關鍵的核心*
* **目前 Demo**：採用「同步阻塞（Synchronous Blocking）」。Python 發出 POST 後就傻傻等在那裡（設定 `timeout=150` 秒），Go 算完才一口氣回傳。如果一個任務要跑 10 分鐘，HTTP 連線早就斷了。
* **正式 A2A**：每個任務都有獨一無二的 `Task ID`，並具備明確的**狀態機（State Machine）**：
  * `submitted`（已收件）➔ `working`（處理中）➔ `input-required`（**需要補充資訊！**）➔ `completed`（完成）/ `failed`（失敗）/ `canceled`（取消）。
  * **為什麼 `input-required` 超級重要？**
    在正式 A2A 中，當 Agent A 叫 Agent B 規劃「PostgreSQL 遷移」時，Agent B 可以回傳 `input-required` 並反問 Agent A：*「請問你們的資料庫容量多大？允許停機幾分鐘？」*，讓兩個 Agent 進行**多輪交涉（Multi-turn negotiation）**，而不是像現在這樣一次定生死。

##### 4. 非同步串流與推播（SSE Streaming & Push Notifications）
* **目前 Demo**：使用者按下按鈕後，畫面會卡住好幾秒，直到整篇報告全部生成完畢才一次跳出來。
* **正式 A2A**：
  * 支援 **`tasks/sendSubscribe`（Server-Sent Events, SSE）**：Go Server 可以一邊思考、一邊把目前的執行進度（`TaskStatusUpdateEvent`）與正在生成的字句（`TaskArtifactUpdateEvent`）即時串流推給 Python 前端。
  * 針對耗時數小時的任務，甚至支援註冊 Webhook（Push Notification），等 Go 背景做完再主動回撥通知 Python。

##### 5. 真正的「雙向獨立大腦」（Dual Autonomous Reasoning）
* **目前 Demo**：嚴格來說，我們的 Agent A（LangGraph）裡的 `reasoning_node` 只是用 Python 字串拼接把任務包成 JSON，並沒有呼叫 LLM；只有後端的 Go Server 呼叫了 Gemini。
* **正式 A2A**：
  * **Agent A 自己也有一顆 Gemini 大腦**：負責接見人類、分析需求、閱讀各個外部 Agent 的 Agent Card、拆解子任務、驗收 Agent B 交回來的報告，如果不滿意還會退件重做。
  * **Agent B 也有自己的大腦與專屬工具（Tools / MCP）**：不光只是生成文字計畫，還能透過 MCP（Model Context Protocol）去查詢真實的 GCP 帳單或資料庫狀態，再透過 A2A 協定回報給 Agent A。

#### 一、Python 與 Go 各有哪些現成套件（Library）可以直接用？

##### 1. Python 生態系（最完整、開箱即用）
在 Python 端，Google 主推兩大核心套件，能無縫對接你熟悉的 LangGraph：
* **`google-adk`（Google Agent Development Kit）**：
  * Google 官方推出的 Agent 開發框架（`pip install google-adk`）。
  * **內建原生 A2A 支援**：只要用幾行程式碼把寫好的 Agent 包裝起來，ADK 就會**自動幫你生成 `/.well-known/agent.json`（Agent Card）**，並自動啟動符合 A2A 標準的 Server；當它要呼叫別人的 Agent 時，也有內建的 `RemoteA2aAgent` Client 可以直接把遠端 Go Agent 當成自己的一個 Sub-agent 來呼叫。
* **`a2a-sdk`（A2A Protocol 官方 Python SDK）**：
  * 如果你不想換框架，想繼續用原本的 **LangGraph** 或 **CrewAI**，可以直接安裝官方協定套件（`pip install a2a-sdk`）。
  * 它提供了現成的 **`A2AServer`**、**`A2AClient`**、**`TaskManager`** 以及所有標準的 **Pydantic Models**（如 `AgentCard`, `Task`, `Message`, `TextPart`, `Artifact`）。你只需要把 `langgraph_app.invoke()` 塞進它的 `TaskManager` 處理函式中，底層的 JSON-RPC 與 SSE 串流全部由 SDK 自動搞定。

##### 2. Go 語言生態系（輕量、高併發首選）
在 Go 語言端，不需要像 Cell 3 手刻 `type A2APayload struct`，可以直接引入以下資源：
* **Google ADK for Go / 官方 A2A Go 套件（與 Reference Structs）**：
  * Google 在 A2A 官方開源專案（`github.com/google/A2A` 及相關 Go SDK 模組）中，提供了定義完整的 **Go Structs（`AgentCard`, `Task`, `TaskStatus`, `Artifact`）** 以及標準的 **JSON-RPC 2.0 / SSE Server Handler**。
  * **Google GenAI Go SDK (`google.golang.org/genai`)**：
    在我們的 Demo 中，為了避免在 Colab 初始化複雜的 `go.mod`，我們用 HTTP 直接打 Vertex AI 網址。但在正式 Go 專案中，可以直接 `go get google.golang.org/genai`，跟 Python 一樣用短短幾行 `client.Models.GenerateContent(...)` 就能自動處理好 Vertex AI 的認證與模型呼叫，連 Token 都不用自己從環境變數傳。

#### 三、總結：真實開發時的體感差異

如果用一句話來總結使用官方 Library 後的開發體驗：
* **你不需要再寫任何關於「怎麼收發 HTTP、怎麼解析 JSON-RPC、怎麼建立 SSE 串流」的程式碼**。
* **你只需要專注寫 3 件事**：
  1. **填好名片（Agent Card）**：宣告你的 Agent 會什麼技能。
  2. **寫好任務處理函式（Task Handler）**：當收到對方的 `Task` 時，呼叫 Gemini 或你的業務邏輯（例如查詢資料庫、跑 LangGraph）。
  3. **回報狀態（Yield Status / Artifact）**：在函式裡呼叫 SDK 提供的 `update_status("working")` 或 `add_artifact(result)`，剩下的跨語言網路傳輸就全部自動完成了！

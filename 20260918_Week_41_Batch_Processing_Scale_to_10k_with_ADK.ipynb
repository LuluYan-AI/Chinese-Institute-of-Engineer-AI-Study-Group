### 1. 為什麼需要 Batch Processing？

## 1.1 傳統同步處理方式

建立 AI Agent 時，最直覺的做法通常是讓 Agent 逐筆讀取資料、呼叫模型、等待結果，再處理下一筆。

例如：

```python
for pdf in pdf_files:
    text = read_pdf(pdf)
    result = model.generate_content(text)
    save_result(result)
```

流程如下：

```text
PDF 1 → Gemini → 等待結果
PDF 2 → Gemini → 等待結果
PDF 3 → Gemini → 等待結果
...
PDF 10000 → Gemini → 等待結果
```

這種模式稱為同步處理（Synchronous Processing）。

它適合少量、需要即時回應的工作，但當資料規模變大時，會開始出現架構上的問題。

## 1.2 同步處理在大量資料下的問題

如果要處理數千或數萬筆資料，例如 10,000 份 PDF，逐筆同步呼叫模型會帶來幾個問題：

* Python 程式必須長時間持續執行。
* API request 容易受到速率限制（Rate Limit）。
* Colab、Notebook 或 Client connection 可能中斷。
* 使用者其實不需要立即取得每一筆結果。
* 大量標準 API 呼叫的成本較高。

因此問題不只是：

```text
如何讓迴圈跑得更快？
```

真正應該改變的是：

```text
不要讓 Agent 自己執行這 10,000 次模型呼叫。
```

## 1.3 從 Worker 思維轉向 Orchestrator 思維

傳統架構中，Agent 同時負責：

```text
讀取資料
↓
呼叫模型
↓
等待結果
↓
處理下一筆資料
```

也就是把 Agent 當成 Worker。

Batch Processing 則改成讓 Agent 負責：

```text
準備資料
↓
建立大量 requests
↓
提交 Batch Job
↓
結束
```

真正的大量模型推論則交給 Gemini 的後端基礎設施執行。

這種模式稱為 Agent as Orchestrator。

---



### 2. Batch Processing 的核心架構

## 2.1 什麼是批次處理（Batch Processing）？

批次處理（Batch Processing）是把大量彼此獨立的 request 集中提交，讓後端系統統一排程與執行。

概念上：

```text
Agent
  │
  ├── 收集資料
  ├── 建立 prompts
  ├── 建立 requests
  │
  ▼
Batch Input
  │
  ▼
Gemini Batch API
  │
  ▼
Google Backend
  │
  ├── Request 1
  ├── Request 2
  ├── Request 3
  └── ...
  │
  ▼
Results
```

Agent 不需要等待每一筆 request 完成。

它只需要：

```text
Prepare → Submit → Exit
```

## 2.2 什麼是非同步處理（Asynchronous Processing）？

非同步處理（Asynchronous Processing）表示：

```text
提交工作
↓
取得 Job ID
↓
程式可以繼續執行或直接結束
```

而模型推論會在遠端後端繼續執行。

這和一般同步 API 很不同。

同步模式：

```text
request
↓
等待
↓
response
```

Batch 模式：

```text
submit
↓
job_id
```

接著由 Google backend 處理：

```text
Pending
↓
Running
↓
Succeeded
```

因此本地端程式的生命週期，不再和模型執行的生命週期綁在一起。

## 2.3 Agent as Orchestrator

編排器（Orchestrator）的工作不是自己完成所有推論，而是負責決定：

```text
What should be processed?
What prompt should be used?
How should requests be structured?
Where should the job be submitted?
```

例如使用者要求：

```text
分析 Cloud Storage 裡的所有 customer feedback。
```

Agent 可以：

```text
1. 找出所有資料
2. 建立 prompts
3. 建立 Batch Requests
4. 上傳 Batch Input
5. 建立 Batch Job
6. 回傳 Job ID
```

Agent 負責控制流程，真正的大量模型推論則由後端執行。

## 2.4 ADK 與 Gemini Batch API 的角色分工

Google Agent Development Kit（ADK）本身不是 Batch inference engine。

比較準確的架構是：

```text
ADK Agent
   ↓
Tool
   ↓
Gemini Batch API
```

角色可以理解為：

```text
ADK
=
Orchestration Layer

Gemini Batch API
=
Execution Layer
```

ADK 負責理解任務、準備資料與呼叫工具。

Gemini Batch API 負責真正的大規模模型推論。

### 3. Gemini Batch Processing 的實作流程

## 3.1 準備大量請求資料

首先需要把每一筆資料轉換成一個獨立的模型 request。

例如：

```python
requests = []

for item in items:
    requests.append({
        "key": item["id"],
        "request": {
            "contents": [
                {
                    "parts": [
                        {"text": item["text"]}
                    ]
                }
            ]
        }
    })
```

此時尚未呼叫 Gemini。

這個步驟只是先建立：

```text
Request 1
Request 2
Request 3
...
Request N
```

## 3.2 使用 JSONL 組織 Batch Requests

大型 Batch 通常使用 JSON Lines（JSONL）格式。

JSONL 的特色是：

```text
一行 = 一個 JSON object
```

例如：

```json
{"key": "request-1", "request": {...}}
{"key": "request-2", "request": {...}}
{"key": "request-3", "request": {...}}
```

可以用 Python 建立：

```python
import json

with open("requests.jsonl", "w") as f:
    for request in requests:
        f.write(json.dumps(request) + "\n")
```

`key` 很重要，因為它可以用來對應 input 與 output。

例如：

```text
pdf-123 → result A
pdf-456 → result B
```

當處理數千或數萬筆資料時，這可以確保每個結果都能對應回原始資料。

## 3.3 上傳 Batch Input

建立 JSONL 後，可以將檔案上傳到 Google backend。

例如：

```python
from google import genai
from google.genai import types

client = genai.Client()

uploaded_file = client.files.upload(
    file="requests.jsonl",
    config=types.UploadFileConfig(
        display_name="batch-requests",
        mime_type="jsonl"
    )
)
```

此時只是把資料從本地環境上傳：

```text
Local Machine
      ↓
Google Backend
```

模型推論尚未開始。

## 3.4 建立 Batch Job

接下來建立 Batch Job：

```python
batch_job = client.batches.create(
    model="...",
    src=uploaded_file.name,
    config={
        "display_name": "my-batch-job"
    }
)
```

這個操作的意思不是：

```text
現在立即處理全部 request 並等待結果。
```

而是：

```text
建立一個後端工作，交給系統排程。
```

可以把 `batches.create()` 理解為：

```text
建立工作單
```

而不是：

```text
執行完工作並回傳結果
```

## 3.5 查詢 Job 狀態與取得結果

建立 Batch Job 後，系統會回傳對應的 Job ID。

例如：

```text
batches/abc123
```

之後可以透過這個 ID 查詢工作狀態。

典型流程如下：

```text
Submitted
↓
Pending
↓
Running
↓
Succeeded
```

當工作完成後，再從 Batch Job 取得結果並進行後續處理。

### 4. Batch Job 的執行模型

## 4.1 Request 與 Job 的差異

一般模型 API 通常是 Request-based。

例如：

```python
response = model.generate_content(prompt)
```

流程是：

```text
Request
↓
Processing
↓
Response
```

呼叫者通常需要等待結果。

Batch API 則是 Job-based。

```python
job = client.batches.create(...)
```

流程是：

```text
Submit
↓
Job ID
```

接著由後端獨立處理：

```text
Job ID
↓
Pending
↓
Running
↓
Succeeded
```

這種架構稱為 Job-based Architecture。

## 4.2 Batch Job 的生命週期

Batch Job 不一定會立即開始執行。

常見狀態可能是：

```text
Submitted
↓
Pending
↓
Running
↓
Succeeded
```

`Pending` 代表工作已成功送入系統，但仍在等待後端排程。

它不代表錯誤。

Batch API 的設計重點不是最低延遲，而是：

$$
\text{High Throughput}
+
\text{Lower Cost}
$$

代價則是：

$$
\text{Higher Latency}
$$

因此 Batch 更適合非即時工作。

## 4.3 為什麼程式結束後工作仍會繼續？

建立 Batch Job 後，真正的工作已經存在 Google backend。

例如：

```text
batches/abc123
```

此時即使：

```text
關閉 Browser
關閉 Laptop
Colab Disconnect
Python Process Terminate
```

都不會影響已提交的 Batch Job。

因為模型推論並不是在本地 Python process 中執行，而是在 Google-managed infrastructure 中執行。

這也是非同步 Batch Processing 的核心優勢。

## 4.4 Pending、Running 與 Succeeded 的意義

可以把主要狀態理解為：

```text
Pending
=
等待排程

Running
=
正在執行

Succeeded
=
工作完成
```

Batch Processing 不保證立即執行。

因此影片中提到少量資料可能在數分鐘內完成，只能視為實際觀察結果，而不是固定完成時間。

大型工作則可能需要更長時間。

### 5. 什麼情況適合使用 Batch Processing？

## 5.1 適合的工作負載

Batch Processing 最適合：

> 大量、彼此獨立、不需要立即取得結果的工作。

例如：

```text
10,000 PDFs → Summary

100,000 Reviews → Sentiment Classification

50,000 Products → Categorization

20,000 Images → Description

Dataset → Synthetic Labels

Prompt Dataset → Model Evaluation
```

這些任務具有共同結構：

$$
x_1 \rightarrow f(x_1)
$$

$$
x_2 \rightarrow f(x_2)
$$

$$
x_3 \rightarrow f(x_3)
$$

每一筆 request 都可以獨立執行。

也就是：

$$
f(x_2)
$$

不需要等待：

$$
f(x_1)
$$

的結果。

這類問題非常適合 Batch Processing。

## 5.2 不適合的 Agent Workflow

如果 Agent 的下一步取決於前一步的結果，就不適合直接轉成 Batch。

例如：

```text
LLM
↓
Search
↓
根據搜尋結果決定下一步
↓
Database
↓
根據 Database 結果繼續推理
↓
LLM
```

這種流程具有序列相依性（Sequential Dependency）。

可以表示為：

$$
x_{t+1}
=
f(x_t, y_t)
$$

也就是下一步 request 必須等前一步 output 出現後，才能決定內容。

因此 Batch Processing 最適合大量獨立任務，而不是需要連續推理與工具互動的 Agent loop。

## 5.3 成本、延遲與吞吐量的取捨

Batch Processing 的核心取捨可以理解為：

```text
降低成本
+
提高大量任務處理能力
+
不需要保持 Client Connection
```

交換：

```text
無法保證即時完成
```

因此它不是「更快的同步 API」，而是一種不同的執行模式。

另外，「Batch 成本降低 50%」並不是：

```text
Token 數量減少 50%
```

而是：

```text
相同的 Token
使用 Batch Processing 的計價較低
```

可以概念化為：

$$
\text{Standard Cost}
=
C
$$

$$
\text{Batch Cost}
\approx
0.5C
$$

因此 Batch 特別適合大量、離線（Offline）、不要求即時回應的模型推論。

---

整體架構可以濃縮成：

```text
ADK Agent
   ↓
Collect Inputs
   ↓
Build Batch Requests
   ↓
JSONL
   ↓
Upload
   ↓
Gemini Batch API
   ↓
Batch Job
   ↓
Pending → Running → Succeeded
   ↓
Results
```

最重要的架構概念是：

> Batch Processing 並不是讓 Agent 更快執行大量 API calls，而是讓 Agent 從執行工作的 Worker，轉變為負責準備與提交工作的 Orchestrator，再由 Gemini Batch API 在後端非同步完成真正的大規模推論。

Agent Platform pricing：https://cloud.google.com/products/gemini-enterprise-agent-platform/pricing

除了前面介紹的 Batch Processing 基本架構之外，官方文件還補充了一些在實際 production 環境中很重要的限制與操作細節。

## 1. Batch Job 有明確的規模限制

在 Gemini Enterprise Agent Platform 的 Batch Inference 中，單一 Batch Job 最多可包含約 **200,000 個 requests**。

如果使用 Cloud Storage 作為輸入來源，單一輸入檔案也有大小限制。單一輸入檔案大小上限為 1 GB。這裡的「1 GB」指的是你實際放到 Cloud Storage、拿去當 Batch input 的那個檔案本身的檔案大小。如果你的輸入是 JSONL 那麼判斷方式就是看這個 .jsonl 檔案實際佔多少 bytes。

這表示 Batch Processing 雖然適合大量資料，但仍然需要在設計 pipeline 時考慮：

```text
資料總量
↓
是否需要切成多個 Batch Jobs
↓
每個 Job 的 request 數量與檔案大小
```

因此，大型資料集通常不會無限制地塞進單一 Job，而是需要適當分批。

---

## 2. Batch Job 可能會長時間等待排程

Batch Job 提交後，不一定會立刻執行。

工作可能先停留在 queue 中，等待後端資源。

官方文件指出，Job 最長可能在 queue 中等待一段較長時間；真正開始執行後，多數工作通常會在一定時間範圍內完成。

因此 Batch Processing 的設計前提應該是：

```text
允許延遲
+
不要求即時結果
```

如果應用程式要求秒級或分鐘級回應，就不適合使用 Batch。

---

## 3. 後端會處理驗證、平行執行與重試

Batch API 不只是單純把很多 request 放進 queue。

Google 的後端也會負責部分批次執行工作，例如：

* 請求格式驗證
* 平行處理
* 失敗重試

這代表開發者不需要自己建立完整的 retry loop 來管理大量模型請求。

概念上可以理解為：

```text
Application
↓
Submit Batch Job
↓
Managed Batch Infrastructure
    ├── Validation
    ├── Parallel Execution
    └── Retry
```

這也是 Batch Processing 相較於自己撰寫大量 concurrent API calls 的重要優勢。

---

## 4. Batch Job 可以取消

已提交的 Batch Job 可以在完成前取消。

這在大量資料處理中特別重要，例如：

```text
發現輸入資料錯誤
↓
取消 Batch Job
↓
修正資料
↓
重新提交
```

取消後，通常只會針對已經完成處理的 request 計算費用。

因此在 production 系統中，保留 Job ID 不只是為了查詢狀態，也方便後續進行取消或管理。

---

## 5. 並不是所有 Gemini 功能都支援 Batch

Batch Processing 並不等同於一般 Gemini API 的完整功能集合。

部分功能可能無法在 Batch 模式中使用，例如某些：

```text
Provisioned Throughput
Explicit Caching
RAG
```

因此如果原本的 Agent workflow 依賴這些能力，就不能直接假設可以完整搬到 Batch API。

在設計系統前，應先確認：

```text
目前使用的 Gemini 功能
↓
是否支援 Batch
↓
再決定是否改成 Batch Architecture
```

---

## 6. Batch Discount 與 Cache Discount 不一定可以疊加

Batch API 的成本通常比標準同步 inference 低，但還需要注意 caching 的計價規則。

某些 Gemini 模型支援隱式快取（Implicit Caching），可以降低重複 prompt 的成本。

但是：

```text
Batch Discount
+
Caching Discount
```

不一定可以同時疊加。

因此在做成本估算時，不應直接把所有折扣相乘。

更合理的做法是分別比較：

```text
Standard Inference
Batch Inference
Cached Inference
```

再選擇實際成本最低且符合 latency 需求的方案。

---

## 7. Production 成本不只有 Token

在 Agent Platform 中，模型 token 通常只是整體成本的一部分。

實際 production system 還可能包含：

```text
Compute
Memory
Storage
Sessions
Memory Bank
```

因此：

```text
Total Cost
≠
LLM Token Cost
```

更完整的理解應該是：

```text
Total Cost
=
Model Inference
+
Agent Runtime
+
Storage
+
State Management
+
Other Infrastructure
```

所以「Batch 可以降低模型 inference 成本」並不代表整個 Agent 系統的總成本也一定降低相同比例。

---

## 8. 完成結果可以用 Polling 或 Webhook 取得

Batch Job 提交後，常見有兩種方式取得完成狀態。

第一種是輪詢（Polling）：

```text
Application
↓
Check Job Status
↓
Not finished
↓
Wait
↓
Check Again
```

第二種是 Webhook Callback：

```text
Batch Job
↓
Completed
↓
Webhook Notification
↓
Application retrieves results
```

Polling 比較簡單，適合測試或小型系統。

Webhook 則更適合 production，因為應用程式不需要持續查詢 Job 狀態。

因此實務上常見的架構會是：

```text
Submit Batch Job
↓
Save Job ID
↓
End Current Session
↓
Webhook Receives Completion Event
↓
Retrieve Results
↓
Continue Workflow
```

這讓 Batch Processing 更容易整合進完整的非同步 Agent workflow。

```python
import os
import json
import urllib.request
from google import genai
from google.genai import types

# 1. Initialize the Gemini Client
api_key = os.environ.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

# 2. Fetch live data from the web to create the JSONL dataset
dataset_path = "unprocessed_batch_01.jsonl"
print("Fetching live headlines from Hacker News...")

# Grab the top 5 story IDs from the public Hacker News API
top_stories_url = "https://hacker-news.firebaseio.com/v0/topstories.json"
story_ids = json.loads(urllib.request.urlopen(top_stories_url).read())[:5]

# Format the live data into Gemini's JSONL Batch structure
with open(dataset_path, "w") as f:
    for sid in story_ids:
        story_url = f"https://hacker-news.firebaseio.com/v0/item/{sid}.json"
        story = json.loads(urllib.request.urlopen(story_url).read())
        prompt = f"Extract companies, products, and technologies from this headline: {story.get('title', '')}"
        f.write(json.dumps({"request": {"contents": [{"parts": [{"text": prompt}]}]}}) + "\n")

# 3. Upload your JSONL dataset to the Gemini API
uploaded_file = client.files.upload(
    file=dataset_path,
    config={'mime_type': 'application/jsonl'}
)
print(f"Uploaded file: {uploaded_file.uri}")

# 4. Submit the Asynchronous Batch Job at 50% token cost
batch_job = client.batches.create(
    model="gemini-3.1-pro-preview",
    src=uploaded_file.name,
    config=types.CreateBatchJobConfig(
        display_name="massive_document_extractor",
    )
)

print(f"Batch job {batch_job.name} submitted successfully.")
print(f"Current State: {batch_job.state}")

with open(".latest_batch_job.txt", "w") as cached_file:
    cached_file.write(batch_job.name)
```

```python
import os
import sys
from google import genai

# 1. Initialize Client
api_key = os.environ.get("GEMINI_API_KEY")
if not api_key:
    print("Error: Please export your GEMINI_API_KEY before running this script.")
    sys.exit(1)

client = genai.Client(api_key=api_key)

# 2. Get the Batch Job ID to poll
job_name = None
if len(sys.argv) > 1:
    job_name = sys.argv[1]
elif os.path.exists(".latest_batch_job.txt"):
    with open(".latest_batch_job.txt", "r") as cached_file:
        job_name = cached_file.read().strip()
    print(f"Auto-detected latest job ID from cache: {job_name}")

if not job_name:
    job_name = input("Enter the Batch Job name (e.g. batches/12345ABCD): ").strip()

if not job_name.startswith("batches/"):
    job_name = f"batches/{job_name}"

# 3. Request Status from Backend
try:
    print(f"\nPolling Google Cloud for status of: {job_name}...")
    job = client.batches.get(name=job_name)
    
    print("=" * 50)
    print(f"Job Name    : {job.name}")
    print(f"State       : {job.state}")
    
    if job.state == "JOB_STATE_SUCCEEDED":
        uri = getattr(job, "output_uri", "Available via Google Cloud Console")
        print(f"Output URI  : {uri}")
        print("\nSuccess! You can now download and parse the JSON payload.")
    elif job.state == "JOB_STATE_FAILED":
        print("\nThe job failed on the backend.")
        print("Error details:", getattr(job, "error", "Unknown API Error"))
        
    print("=" * 50)
except Exception as e:
    print(f"\nFailed to fetch job status. Ensure the ID is correct.")
    print(f"Error: {e}")
```

# Batch Processing JSONL

對 Batch Processing 來說，通常更好的做法是：**JSONL 裡只放檔案的 reference / URI，讓 Gemini 從已上傳的檔案讀取內容。**

Google 官方文件明確說明，使用多模態輸入（Multimodal Input）時，可以在 JSONL 裡引用其他已上傳的檔案。([Google AI for Developers][1])

概念上會像這樣：

```text
圖片 / PDF / 影片
        ↓
先上傳到 Google File API / 可支援的儲存位置
        ↓
取得 File URI
        ↓
JSONL 裡只放 URI
        ↓
Gemini Batch API 讀取該檔案
```

例如概念性的 JSONL request：

```json
{
  "key": "request-1",
  "request": {
    "contents": [
      {
        "parts": [
          {
            "file_data": {
              "mime_type": "application/pdf",
              "file_uri": "..."
            }
          },
          {
            "text": "請摘要這份 PDF"
          }
        ]
      }
    ]
  }
}
```

所以 JSONL 本身不需要長成：

```text
JSONL
└── 整份 PDF 的 Base64
```

而可以是：

```text
JSONL
└── PDF URI
```

這差異非常重要。

假設原始 PDF 是 30 MB：

如果用 Base64 直接塞：

```text
30 MB PDF
↓
Base64
↓
約 40 MB
↓
寫進 JSONL
```

但如果使用 File URI：

```text
30 MB PDF → File API / Storage

JSONL：
"file_uri": "files/abc123"
```

那 JSONL 裡可能只有幾百 bytes 的 reference，原本 30 MB 的 PDF **不會直接算進 JSONL 本身的檔案大小**。

因此，大量圖片、PDF、audio、video 這類資料時，實務上通常應該採用：

```text
大型內容
→ 外部儲存 / File API

JSONL
→ instruction + metadata + file reference
```

而不是：

```text
大型內容
→ Base64
→ 全部塞進 JSONL
```

這也修正了我們前面談「什麼東西容易把 JSONL 撐大」時的一個重要細節：**只要採用檔案 reference 的設計，大型 PDF、圖片本身通常就不是 JSONL 大小的主要來源；真正會把 JSONL 撐大的，反而比較可能是大量 inline text、重複的 system instruction 或直接內嵌的資料。** ([Google AI for Developers][1])

另外順便更正前面的一個數字：我們剛才討論的 Enterprise Agent Platform 文件有 **1 GB** 限制，但目前 Gemini API 的 Batch API + File API 文件寫的是 **2 GB input file limit**。這是不同產品/API 路徑的限制，教材裡最好不要把兩者混成同一個數字。([Google AI for Developers][2])

# OpenAI Batch Process

OpenAI 也有對應的 **Batch API**，用途和你前面看的 Gemini Batch Processing 很接近：把大量 request 先整理成檔案，上傳後建立 batch job，由 OpenAI 非同步處理，完成後再取得輸出檔案。官方說明的目標完成時間是 **24 小時內**，而且 Batch API 相較同步 API 有 **50% 的價格折扣**。([OpenAI Help Center][1])

你可以看這幾個官方位置：

* OpenAI Batch API FAQ：說明整體運作方式、24 小時 completion window、50% 折扣、狀態、取消行為等。
  [OpenAI Batch API FAQ](https://help.openai.com/en/articles/9197833-batch-api-faq?utm_source=chatgpt.com)
* OpenAI Files API：Batch 的輸入檔要先透過 Files API 上傳，`purpose` 設成 `batch`。官方目前規定 Batch API 的 `.jsonl` input file 上限是 **200 MB**。([OpenAI Developers][2])
  [OpenAI Files API Reference](https://developers.openai.com/api/reference/typescript/resources/files/methods/create?utm_source=chatgpt.com)
* Batch API 本身的 API Reference 可以從 OpenAI Developers 的 Batches resource 進去，例如查詢 batch、狀態、output file 等。([OpenAI Developers][3])

OpenAI 的流程大致是：

```text
建立 JSONL
↓
Files API 上傳
purpose = "batch"
↓
取得 input_file_id
↓
建立 Batch
↓
OpenAI 非同步處理
↓
取得 output_file_id
↓
下載結果
```

和 Gemini 最大的概念差異之一，是 OpenAI 官方 Batch API 明確要求 batch input 為 **`.jsonl`**，而且目前 Batch 專用輸入檔上限是 **200 MB**；一般 Files API 雖然單檔可到 512 MB，但 Batch input 本身仍受 200 MB 限制。([OpenAI Developers][2])

另外，OpenAI 的 Batch API 也支援圖片輸入，但不支援 streaming；結果是寫到 output file，而不是一邊產生一邊串流回傳。([OpenAI Help Center][1])

如果你要拿它和 Gemini 做教材對照，最值得對比的是這幾個欄位：

```text
Gemini Batch API
vs.
OpenAI Batch API

Input format
Input file size limit
Completion window
Pricing discount
Job status model
Multimodal support
Result retrieval
```

這樣會很容易看出兩家的 Batch 架構其實非常相似，只是檔案限制、API 介面和支援能力不同。

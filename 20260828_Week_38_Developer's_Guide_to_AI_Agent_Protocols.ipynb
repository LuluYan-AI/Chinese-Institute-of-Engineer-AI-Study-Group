### AI Agent Protocols

## Agent Protocols 名稱對照

| 縮寫        | 英文全名                                | 繁體中文譯名         | 簡體中文譯名         | 核心用途                   |
| --------- | ----------------------------------- | -------------- | -------------- | ---------------------- |
| **MCP**   | **Model Context Protocol**          | **模型上下文協定**    | **模型上下文协议**    | Agent 與工具、資料來源之間的標準介面  |
| **A2A**   | **Agent2Agent Protocol**            | **代理間通訊協定**    | **智能体间通信协议**   | 不同 Agent 之間的發現、通訊與任務協作 |
| **UCP**   | **Universal Commerce Protocol**     | **通用商務協定**     | **通用商务协议**     | 標準化商品、購物、結帳與商務交易流程     |
| **AP2**   | **Agent Payments Protocol**         | **代理支付協定**     | **智能体支付协议**    | Agent 支付授權、權限約束與交易稽核   |
| **A2UI**  | **Agent-to-User Interface**         | **代理對使用者介面**   | **智能体对用户界面**   | Agent 以宣告式資料描述互動式使用者介面 |
| **AG-UI** | **Agent–User Interaction Protocol** | **代理－使用者互動協定** | **智能体－用户交互协议** | Agent 與前端之間的即時事件與串流互動  |

## 1. MCP — Model Context Protocol

**繁體中文：模型上下文協定**  
**簡體中文：模型上下文协议**

MCP 解決的是：

> **Agent 如何用標準方式連接外部工具與資料來源。**

例如：

```text
Agent
  ↓
MCP
  ↓
Database
API
File System
Search Service
External Tools
```

過去每一個外部服務可能都需要另外撰寫整合程式；MCP 的目的則是建立共同的工具與資料存取介面。

最簡單可以記成：

```text
MCP
=
Agent ↔ Tools / Data
```

---

## 2. A2A — Agent2Agent Protocol

**繁體中文：代理間通訊協定**  
**簡體中文：智能体间通信协议**

A2A 解決的是：

> **一個 Agent 如何發現另一個 Agent，並與它交換任務與結果。**

例如：

```text
Main Agent
    │
    ├──→ Research Agent
    │
    ├──→ Pricing Agent
    │
    └──→ Logistics Agent
```

A2A 的重點不是工具呼叫，而是讓不同 Agent，即使由不同團隊、不同 framework 或不同 Model 建立，仍然可以透過共同協定互相協作。

簡化來說：

```text
A2A
=
Agent ↔ Agent
```

---

## 3. UCP — Universal Commerce Protocol

**繁體中文：通用商務協定**  
**簡體中文：通用商务协议**

UCP 解決的是：

> **Agent 如何用標準方式進行商務交易。**

例如：

```text
Discover Product
      ↓
Select Product
      ↓
Create Cart
      ↓
Checkout
      ↓
Create Order
```

如果每個商家都有完全不同的商品、購物車與結帳 API，Agent 就需要大量客製整合。

UCP 的目標是將這些商務流程標準化。

因此：

```text
UCP
=
Agent ↔ Commerce
```

---

## 4. AP2 — Agent Payments Protocol

**繁體中文：代理支付協定**  
**簡體中文：智能体支付协议**

AP2 關注的不是「買什麼」，而是：

> **Agent 是否被授權支付，以及這項授權可以在什麼條件下使用。**

例如：

```text
Purchase Intent
      ↓
Authorization
      ↓
Payment Mandate
      ↓
Payment
      ↓
Audit Record
```

它處理的重點包括：

* Payment authorization
* Spending limits
* User intent
* Mandates
* Audit trail
* Transaction accountability

因此 UCP 與 AP2 要區分：

```text
UCP
→ 如何完成商務交易

AP2
→ 如何授權與管理付款
```

可以簡化成：

```text
AP2
=
Payment Authorization
+
Audit
```

---

## 5. A2UI — Agent-to-User Interface

**繁體中文：代理對使用者介面**  
**簡體中文：智能体对用户界面**

A2UI 解決的是：

> **Agent 如何告訴前端「應該呈現什麼 UI」。**

例如 Agent 不只回答：

```text
目前有 5 個供應商。
```

而是產生一個結構化 UI 描述：

```text
Supplier Comparison
├── Supplier A
├── Supplier B
├── Price
├── Quantity
├── Checkbox
└── Order Button
```

前端再根據這些 declarative components 渲染真正的 UI。

因此可以理解成：

```text
Agent
  ↓
A2UI Description
  ↓
UI Renderer
  ↓
Interactive Interface
```

簡化來說：

```text
A2UI
=
What UI should be rendered?
```

---

## 6. AG-UI — Agent–User Interaction Protocol

**繁體中文：代理－使用者互動協定**  
**簡體中文：智能体－用户交互协议**

AG-UI 主要解決：

> **Agent 執行中的狀態與事件，如何即時傳送給前端。**

Agent 的工作通常不是一次 Request、一次 Response，而是一連串事件：

```text
Run Started
    ↓
Text Streaming
    ↓
Tool Call
    ↓
Tool Result
    ↓
More Text
    ↓
Run Finished
```

AG-UI 將這些 runtime events 標準化，使不同 Agent backend 可以用一致方式與前端互動。

所以：

```text
AG-UI
=
How Agent events are streamed
to the frontend
```

---

## 六個 Protocol 的快速辨識方式

最簡單可以記成：

```text
MCP
→ 工具與資料
→ Tools / Data

A2A
→ Agent 之間
→ Agent / Agent

UCP
→ 商務交易
→ Commerce

AP2
→ 支付授權
→ Payment Authorization

A2UI
→ 畫面內容
→ Rich UI

AG-UI
→ 即時互動
→ Streaming Events
```

其中三組最容易混淆，可以直接這樣區分：

| 容易混淆              | 關鍵差異                                       |
| ----------------- | ------------------------------------------ |
| **MCP vs A2A**    | MCP 是 Agent 使用工具；A2A 是 Agent 與另一個 Agent 協作 |
| **UCP vs AP2**    | UCP 管商務交易流程；AP2 管付款授權與稽核                   |
| **A2UI vs AG-UI** | A2UI 描述「畫面要長什麼樣」；AG-UI 處理「執行事件如何即時傳到前端」    |

最後可以濃縮成一句：

> **MCP 管工具與資料，A2A 管 Agent 協作，UCP 管商務交易，AP2 管支付授權，A2UI 管介面描述，AG-UI 管即時前端互動。**



##### A2A v.s. MCP

> **A2A 的早期設計確實大量借鑑了 MCP 已經驗證過的「協定化互通」思路與部分技術風格，但 A2A 並不是把 MCP 直接延伸成 Agent-to-Agent。**

兩者最初就被定位成互補關係：MCP 處理 **Agent ↔ Tools / Data**，A2A 處理 **Agent ↔ Agent**。A2A 官方文件一直明確強調這一點。([GitHub][1])

你可以把「借鑑 MCP」理解成三個層面。

第一，A2A 延續了 MCP 已經證明有效的概念：**用一個公開協定，把原本各家框架私有的 integration boundary 標準化。**

```text
MCP
Agent ↔ Tool / Resource

A2A
Agent ↔ Agent
```

也就是先定義共同語言、能力描述與互通方式，而不是要求大家使用同一個 Agent framework。

第二，A2A 也沿用了很多 Web-native 的設計思想，例如：

* JSON / HTTP
* capability discovery
* structured messages
* streaming
* extensibility
* framework independence

不過 A2A 因為要處理 Agent-to-Agent，所以又新增 MCP 不需要那麼強調的概念：

```text
Agent Card
Task
Task Lifecycle
contextId
Artifacts
Push Notifications
Long-running collaboration
```

也就是它必須處理「一個遠端 Agent 接下一個任務後，可能跑很久，而且過程中狀態會變化」這種情況。([A2A Protocol][2])

第三，也就是你剛才問的 Stateless，兩邊現在反而越來越接近同一套 Web architecture philosophy。

新版 MCP：

```text
Stateless Protocol
        +
Explicit Application State
```

A2A v1.0：

```text
Stateless Web Transport
        +
Explicit Task / Context State
```

A2A v1.0 官方甚至直接強調：

> stateless、layered architecture，以及 infrastructure-friendly communication patterns。

目的也是讓既有的 load balancer、gateway、security、observability infrastructure 可以直接使用。([A2A Protocol][3])

所以如果畫成演化關係，可以這樣理解：

```text
MCP
先證明 Protocol-based interoperability
是可行的
      ↓
Agent ↔ Tool
      ↓
相同思想往 Agent collaboration 延伸
      ↓
A2A
Agent ↔ Agent
      ↓
加入 Task / Context / Lifecycle
      ↓
A2A v1.0
進一步 Web-native / Stateless
```

但我會避免直接寫：

> 「A2A 是建立在 MCP 上。」

#### 3. UCP — Universal Commerce Protocol

https://ucp.dev/

UCP 解決的是：

> **Agent 如何用標準方式進行商務交易。**

例如：

```text
Discover Product
      ↓
Select Product
      ↓
Create Cart
      ↓
Checkout
      ↓
Create Order
```

如果每個商家都有完全不同的商品、購物車與結帳 API，Agent 就需要大量客製整合。

UCP 的目標是將這些商務流程標準化。

因此：

```text
UCP
=
Agent ↔ Commerce
```

UCP 目前可以評價為：**規格成熟度已經進入「可正式實作」階段，產業支持度很高，但真正大規模 production adoption 還在快速形成中。**

截至 2026 年 8 月，UCP 已經不只是 draft。官方已發布多個正式版本，包括 `2026-01-11`、`2026-01-23`、`2026-04-08`，並採用明確的 date-based versioning、release branch 與 compatibility negotiation 機制。規格也已涵蓋 discovery、checkout、cart、order，以及 REST、MCP 等 transport binding。([Universal Commerce Protocol][1])

所以在技術成熟度上，我會給它：

| 面向                             | 評價  |
| ------------------------------ | --- |
| 規格完整度                          | 高   |
| Versioning / Governance        | 高   |
| SDK / Reference implementation | 已具備 |
| Production 可實作性                | 中高  |
| 跨產業實際 deployment               | 成長中 |

產業接受度則相當強。UCP 最初由 Google、Shopify 與 Etsy、Wayfair、Target、Walmart 等共同推動；目前 Stripe 已加入 Governing Council，Technical Council 也納入 Amazon、Meta、Microsoft、Salesforce 等公司。2026 年 7 月還成立 Food Technical Council，成員包括 Block（Square）、DoorDash、Google、Toast、Uber Eats。([Universal Commerce Protocol][1])

| Food                       | DoorDash、Uber Eats、Toast                                     |

([Universal Commerce Protocol][2])

這代表 UCP 最大的優勢是：**它不是只有一家平台自己推的規格，而是已經形成跨零售、支付、旅遊、餐飲與企業軟體的治理聯盟。**

不過要區分：

```text
Endorsement / Governance Participation
                ≠
Full Production Deployment
```

很多企業目前屬於支持、共同制定、測試或逐步整合階段，不能把官方 logo 名單直接解讀成「全部都已全面上線 UCP」。

官方 roadmap 正在制定旅館訂房、餐廳菜單、餐點客製、外送與 tipping 等 capability。([Universal Commerce Protocol][3])

因此一句話總結：

> **UCP 目前屬於「規格已成熟、產業聯盟很強、實際部署正在擴張」的階段；在 Agentic Commerce 協定中，它已經是最值得關注的主流候選標準之一，但還不能說已經成為唯一的事實標準。**

##### ACP ( OpenAI)

ACP — Agentic Commerce Protocol

繁體中文可以翻成：代理式商務協定
簡體中文可以翻成：智能体商务协议

OpenAI 與 Stripe 共同開發 ACP，目的是讓 AI Agent、使用者與商家之間可以用標準方式完成商品發現、結帳、訂單處理與付款流程。OpenAI 的 Instant Checkout 就是以 ACP 為核心。([OpenAI][1])

可以把它簡化理解成：

```text
ACP
=
Agent ↔ Merchant ↔ Checkout / Order / Payment Flow
```

它和 Google **UCP — Universal Commerce Protocol** 很接近，兩者都在處理 Agentic Commerce 的標準化，但不是同一個 protocol。

目前可以先這樣對照：

| Protocol | 英文全名                        | 繁體中文    | 主要用途                                 |
| -------- | --------------------------- | ------- | ------------------------------------ |
| **ACP**  | Agentic Commerce Protocol   | 代理式商務協定 | OpenAI / Stripe 推動的 Agent 商務、結帳與訂單流程 |
| **UCP**  | Universal Commerce Protocol | 通用商務協定  | Google 生態推動的標準化 Agent Commerce 流程    |
| **AP2**  | Agent Payments Protocol     | 代理支付協定  | 支付授權、Mandate、稽核與付款責任                 |

尤其要注意：

```text
ACP / UCP
→ Commerce workflow

AP2
→ Payment authorization
```

另外，OpenAI 在 2026 年 3 月已進一步擴展 ACP，不只支援 checkout，也支援 **product discovery**，也就是讓 ChatGPT 能取得更完整、即時的商品資訊來做搜尋與比較。([OpenAI][2])

##### 為什麼 Amazon 可以支持 UCP，但是反對 OpenAI 的 ACP

因為 Amazon 支持 UCP、但對 OpenAI 的 ACP 較保留，背後其實不是「技術協定好不好」這麼單純，而是商業控制權與平台位置不同。

UCP 對 Amazon 比較容易接受，因為它是較中性的開放商務標準，而且 Amazon 已經進入 UCP 的 Technical Council。UCP 的設計目標是讓不同商家、支付商、平台與 Agent 都能用共同方式做商品發現、結帳與訂單流程；它不是把交易鎖在某一個 AI 入口裡。Amazon 因此可以參與規格制定，同時保留自己的 merchant relationship、checkout、payment 與 customer experience。([Universal Commerce Protocol][1])

ACP 的情況不同。OpenAI 的 ACP 是直接服務於 ChatGPT 內的 product discovery 與 Instant Checkout。也就是使用者可以：

```text
ChatGPT
   ↓
找商品
   ↓
比較商品
   ↓
直接 Buy
   ↓
在 ChatGPT 內完成 checkout
```

OpenAI 官方也明確表示，ACP 正在擴展到 product discovery，讓更多商品資訊直接進入 ChatGPT 的購物體驗。([OpenAI][2])

這對 Amazon 的戰略意義就不一樣了。

Amazon 最有價值的不只是「賣東西」，而是掌握整個 shopping funnel：

```text
Discovery
   ↓
Search
   ↓
Recommendation
   ↓
Product Page
   ↓
Checkout
   ↓
Payment
   ↓
Fulfillment
   ↓
Customer Relationship
```

如果 ChatGPT 透過 ACP 成為使用者主要的 shopping interface，Amazon 就可能被壓縮成後端商品與履約供應商：

```text
User
 ↓
ChatGPT
 ↓
ACP
 ↓
Amazon
```

Amazon 最不希望失去的，正是最前面的 discovery、recommendation 與 customer relationship。

所以兩者最核心的差別可以濃縮成：

|                       | UCP               | OpenAI ACP                         |
| --------------------- | ----------------- | ---------------------------------- |
| 性質                    | 跨平台商務標準           | ChatGPT-centered commerce protocol |
| Amazon 的角色            | 可以參與治理與制定         | 比較像接入 OpenAI 的 commerce channel    |
| Shopping入口            | 可由不同 Agent / 平台掌握 | ChatGPT 是重要入口                      |
| Customer relationship | 商家較容易保留           | 有部分可能被 ChatGPT 中介                  |
| Amazon 戰略風險           | 較低                | 較高                                 |

而且 Amazon 2026 年的新 Agent Policy 已經明確表示，AI Agent 存取 Amazon Services 必須辨識自身身分、遵守 Amazon 的 Agent Policy，而且 Amazon 可以限制其存取。這顯示 Amazon 並不是反對 Agentic Commerce，而是希望 **Agent 如何進入 Amazon 生態，仍由 Amazon 自己控制**。([Amazon Seller Central][3])

所以比較準確的說法不是：

> Amazon 支持 Google、反對 OpenAI。

而是：

> **Amazon 支持不會削弱自己平台控制權的開放協定，但對可能把購物入口、商品發現與顧客關係轉移到第三方 AI 平台的協定會比較防備。**

這也解釋了為什麼 Amazon 可以一方面加入 UCP、支援 AP2 等開放 protocol，另一方面又對 ChatGPT 直接成為 shopping intermediary 保持警戒。

#### 4. AP2 — Agent Payments Protocol

AP2 關注的不是「買什麼」，而是：

> **Agent 是否被授權支付，以及這項授權可以在什麼條件下使用。**

例如：

```text
Purchase Intent
      ↓
Authorization
      ↓
Payment Mandate
      ↓
Payment
      ↓
Audit Record
```

它處理的重點包括：

* Payment authorization
* Spending limits
* User intent
* Mandates
* Audit trail
* Transaction accountability

因此 UCP 與 AP2 要區分：

```text
UCP
→ 如何完成商務交易

AP2
→ 如何授權與管理付款
```

可以簡化成：

```text
AP2
=
Payment Authorization
+
Audit
```

##### Cloudflare Wallets

## Cloudflare Wallets：從 Agent 身分、支付到可程式化錢包

Cloudflare Wallets 可以理解為一套讓 AI Agent 擁有「可被控制的支付能力」的基礎設施。它的目的不是單純提供一個一般使用者錢包，而是讓 Agent 可以在明確的權限、額度與規則下，代表使用者或組織進行付款。

Cloudflare 在推出 Wallets 之前，已經逐步建立幾個相關基礎：

```text
Web Bot Auth
→ 驗證 Agent 身分

x402
→ 支援機器對機器付款

Monetization Gateway
→ 讓 API、資料與服務可以向 Agent 收費

Wallets
→ 讓 Agent 擁有可程式化的資金與支付能力
```

因此 Wallets 可以看成 Cloudflare 前面這些工作的整合延伸：前面先處理「Agent 是誰」、「Agent 如何付款」以及「服務如何向 Agent 收費」，最後再補上 Agent 自己可以持有與使用支付能力的這一層。

Wallets 的典型用途包括：

* 為 Agent 建立專屬 Wallet
* 設定可用額度
* 限制可交易的對象
* 控制單筆或總支出
* 讓 Agent 自動支付 API、資料或其他服務

整體概念可以簡化為：

```text
User / Organization
        ↓
Cloudflare Wallets
        ↓
AI Agent
        ↓
x402 / Payment Rail
        ↓
API / Data / Merchant
```

### 與 AP2 的差異

Cloudflare Wallets 與 AP2 並不是同一類產品。

Cloudflare Wallets 比較偏向：

> **提供 Agent 實際持有與使用支付能力的工具。**

AP2（Agent Payments Protocol）則比較偏向：

> **定義 Agent 的付款授權、使用者意圖與交易責任如何被證明與記錄。**

因此可以簡單區分為：

```text
Cloudflare Wallets
→ Agent 如何實際付款

AP2
→ Agent 為什麼有權付款
```

兩者可以互補，而不一定互相取代：

```text
User Authorization
      ↓
AP2
      ↓
Agent
      ↓
Cloudflare Wallet
      ↓
Payment
```

所以 Cloudflare Wallets 比較接近 Agent 的「支付工具與資金執行層」，而 AP2 則偏向「授權與信任協定層」。


**Cloudflare Wallets 是一個新推出、仍在早期 rollout 的可程式化錢包產品；AP2 則是一個已獲得廣泛產業參與的開放支付協定**。兩者不在同一層，也不能直接拿「誰取代誰」來比較。

Cloudflare Wallets 的定位是讓 AI Agent 有一個穩定身分，以及可以在受限額度下自主支付的 wallet。官方目前強調的支付路徑主要是 **x402**，而 Cloudflare Agents SDK 也支援 **x402 與 MPP（Machine Payments Protocol）**。Wallets 初期預計支援 stablecoin 的持有、收款與付款，並提供 Account Wallet / Virtual Wallet、allowance、allow list、單筆上限等控制。([Cloudflare Docs][1])

AP2，也就是 **Agent Payments Protocol**，則是 Google 推動的 payment-agnostic protocol。它不負責「錢包裡的錢怎麼存」，而是處理 Agent 交易中的 **授權、真實意圖、責任歸屬與可稽核證據**。AP2 使用 mandates 與 Verifiable Digital Credentials，讓商家、支付網路與發卡方可以知道：這筆交易是否真的代表使用者意圖、Agent 被授權到什麼程度，以及出了問題後如何追溯。([Google Cloud][2])

所以兩者若放在同一筆未來交易中，可以是：

```text
Human / Organization
        ↓
Cloudflare Account Wallet
        ↓
Virtual Wallet for Agent
        ↓
Agent
        ↓
AP2 authorization / mandates
        ↓
Merchant / Payment Provider
        ↓
Payment rail
```

或者，如果底層是機器對 API 的小額支付：

```text
Agent
  ↓
Cloudflare Wallet
  ↓
x402 / MPP
  ↓
Paid API / MCP Tool / Content
```

這也反映出一個重要差別：**Cloudflare Wallets 目前更偏向「Agent 原生 wallet + machine payment」；AP2 更偏向「Agent commerce 的 trust / authorization layer」。**

Google 在 2025 年 9 月推出 AP2 時，公開表示有超過 60 家組織共同參與；到 2025 年 10 月與 PayPal 的合作公告時，已經表述為 **超過 100 個 industry partners**。([Google Cloud][2])

### Cloudflare Wallets 的接受程度則要更保守看

Cloudflare Wallets 是 **2026 年 8 月 4 日才正式宣布**的新產品。截至 2026 年 8 月 19 日官方文件仍明確寫著，目前真正可用的功能主要是：

```text
Reserve a Cloudflare Wallet handle
```

而：

```text
Send funds
Receive funds
Hold funds
Create usable Virtual Wallets
```

仍屬於即將提供的能力。官方甚至特別說，現在 reserved handle **還不能收、付或持有資金**。([Cloudflare Docs][1])

所以目前不能說：

> 「Cloudflare Wallets 已經被多少萬家 merchant 接受。」

這樣的數字目前沒有公開依據。

目前比較準確的說法是：

> **Cloudflare Wallets 有很強的潛在 distribution，但實際 wallet acceptance network 尚未形成。**

它背後最大的優勢並不是已經有一張 merchant acceptance list，而是 Cloudflare 本身位於大量網站、API、Workers、MCP server 與網路流量前方，因此未來可以把 Wallets、Monetization Gateway、x402、Agent SDK 等能力整合在同一個 infrastructure layer。Cloudflare 已公開表示 Monetization Gateway 將透過 x402 讓 API、資料與內容接受 machine payments。([Cloudflare Blog][4])

### 但 Cloudflare 整體在 Agentic Commerce 的支付合作其實很廣

這裡要區分：

```text
Cloudflare Wallets adoption
```

和：

```text
Cloudflare agentic-commerce ecosystem partnerships
```

Cloudflare 已經與 **Visa、Mastercard、American Express** 合作 agentic commerce identity / verification，例如 Visa Trusted Agent Protocol、Mastercard Agent Pay，以及 Web Bot Auth。Cloudflare 也在自己的 agentic commerce solution page 明確把 **AP2、x402、Visa Trusted Agent Protocol、Mastercard Agent Pay** 列為可連接的 emerging agentic-payment standards。([Cloudflare][5])

這點很重要，因為它代表 Cloudflare 並沒有把自己鎖死在單一支付協定。

比較像：

```text
                    Cloudflare
                        │
          ┌─────────────┼─────────────┐
          ↓             ↓             ↓
        x402           AP2       Card-network
                                  protocols
                                   │
                         Visa / Mastercard / Amex
```

因此 Cloudflare 的策略目前比較偏向：

> **做 agentic commerce 的 infrastructure / identity / wallet / protocol-neutral execution layer。**

而不是：

> **推出一個 proprietary protocol，要求整個產業只能使用它。**

#### 5. A2UI — Agent-to-User Interface

A2UI 解決的是：

> **Agent 如何告訴前端「應該呈現什麼 UI」。**

例如 Agent 不只回答：

```text
目前有 5 個供應商。
```

而是產生一個結構化 UI 描述：

```text
Supplier Comparison
├── Supplier A
├── Supplier B
├── Price
├── Quantity
├── Checkbox
└── Order Button
```

前端再根據這些 declarative components 渲染真正的 UI。

因此可以理解成：

```text
Agent
  ↓
A2UI Description
  ↓
UI Renderer
  ↓
Interactive Interface
```

簡化來說：

```text
A2UI
=
What UI should be rendered?
```

A2UI 目前可以說是：**技術上已經相當可用，但產業採用仍在早期。**

它目前仍屬 **Early Stage Public Preview**，v0.9 已經相對穩定，v1.0 進入候選階段，因此規格本身已經不只是實驗性概念，但還不到「成熟產業標準」的程度。

它的核心用途是讓 Agent 產生宣告式 UI，而不是直接輸出 HTML / JavaScript：

```text
Agent
  ↓
A2UI JSON
  ↓
Client Renderer
  ↓
Native / Interactive UI
```

目前主要支持力量來自 Google 生態，包括 Flutter GenUI、Angular / Lit renderer，以及與 A2A、AG-UI、MCP 等協定的整合方向。這代表它的 interoperability 設計已經逐漸成形。

但目前最大的限制是：

> **Google 以外的大型平台採用還不夠廣。**

因此可以這樣評估：

| 面向                       | 目前狀態      |
| ------------------------ | --------- |
| 技術成熟度                    | 中高        |
| Reference implementation | 已具備       |
| Google 生態支持              | 很強        |
| 跨協定整合                    | 不錯        |
| 跨公司採用                    | 仍早期       |
| Production maturity      | 中等        |
| 成為產業標準的可能性               | 有潛力，但尚未確定 |

一句話總結：

> **A2UI 已經可以實際使用，但目前更像「正在形成中的標準」，還不是像 MCP 那樣已經有明顯跨產業 network effect 的協定。**


#### 6. AG-UI — Agent–User Interaction Protocol


AG-UI 主要解決：

> **Agent 執行中的狀態與事件，如何即時傳送給前端。**

Agent 的工作通常不是一次 Request、一次 Response，而是一連串事件：

```text
Run Started
    ↓
Text Streaming
    ↓
Tool Call
    ↓
Tool Result
    ↓
More Text
    ↓
Run Finished
```

AG-UI 將這些 runtime events 標準化，使不同 Agent backend 可以用一致方式與前端互動。

所以：

```text
AG-UI
=
How Agent events are streamed
to the frontend
```

AG-UI 主要是用來處理：

> **Agent 後端和前端介面之間的即時互動。**

它不是拿來連接工具，也不是拿來做 Agent-to-Agent，而是讓使用者可以在前端「看見 Agent 正在做什麼、收到即時結果，並且可以中途互動」。

主要可以做這幾類事情：

| 功能                | AG-UI 能做什麼                        |
| ----------------- | --------------------------------- |
| Streaming Text    | 一邊生成、一邊把文字送到前端                    |
| Tool Call Events  | 顯示 Agent 正在呼叫哪個 Tool              |
| Tool Results      | 把工具執行結果即時回傳前端                     |
| Run Lifecycle     | 顯示 Started、Running、Finished、Error |
| Shared State      | 同步 Agent 與前端的狀態                   |
| Human-in-the-loop | 執行到某一步時要求使用者確認或輸入                 |
| Generative UI     | 傳送與動態 UI 有關的事件                    |

例如一個 Research Agent：

```text
User
  ↓
開始研究
  ↓
AG-UI
  ↓
Frontend 顯示：
"正在搜尋資料..."
  ↓
Tool Call：Web Search
  ↓
搜尋結果回傳
  ↓
"正在整理來源..."
  ↓
Streaming Answer
  ↓
完成
```

如果是 Coding Agent：

```text
User
  ↓
"幫我修這個 bug"
  ↓
Agent
  ↓
AG-UI Events
  ↓
Frontend

Reading file...
Running tests...
Editing code...
Running tests again...
✓ Tests passed
```

所以使用者不需要等到 Agent 全部完成之後才看到結果。

AG-UI 也很適合 Human-in-the-loop，例如：

```text
Agent:
準備執行付款 $500
        ↓
AG-UI
        ↓
Frontend 顯示 Approval UI
        ↓
[Approve] [Reject]
        ↓
使用者選擇
        ↓
Agent 繼續執行
```

它最核心的價值，可以濃縮成：

```text
Traditional API

Frontend
   ↓
Request
   ↓
Agent
   ↓
Final Response
```

變成：

```text
AG-UI

Frontend
   ↕
Real-time Events
   ↕
Agent
```

所以一句話來說：

> **AG-UI 是讓 Agent 從「黑箱式等待結果」，變成「可即時觀看、可互動、可控制的前端體驗」。**

而它和 A2UI 最大差別是：

```text
A2UI
→ 定義「畫面要呈現什麼」

AG-UI
→ 定義「Agent 執行事件怎麼即時送到前端」
```

兩者其實很適合一起使用。


#### 六個 Protocol 的快速辨識方式

最簡單可以記成：

```text
MCP
→ 工具與資料
→ Tools / Data

A2A
→ Agent 之間
→ Agent / Agent

UCP
→ 商務交易
→ Commerce

AP2
→ 支付授權
→ Payment Authorization

A2UI
→ 畫面內容
→ Rich UI

AG-UI
→ 即時互動
→ Streaming Events
```

其中三組最容易混淆，可以直接這樣區分：

| 容易混淆              | 關鍵差異                                       |
| ----------------- | ------------------------------------------ |
| **MCP vs A2A**    | MCP 是 Agent 使用工具；A2A 是 Agent 與另一個 Agent 協作 |
| **UCP vs AP2**    | UCP 管商務交易流程；AP2 管付款授權與稽核                   |
| **A2UI vs AG-UI** | A2UI 描述「畫面要長什麼樣」；AG-UI 處理「執行事件如何即時傳到前端」    |

最後可以濃縮成一句：

> **MCP 管工具與資料，A2A 管 Agent 協作，UCP 管商務交易，AP2 管支付授權，A2UI 管介面描述，AG-UI 管即時前端互動。**

# 真正執行時會發生什麼？

```text
先查 recipe
→ 再查 inventory
→ 再比較
→ 再回答
```

這正是 Agentic AI 和傳統 workflow programming 的分界之一。

### A2A code 解讀

這段 code 展示的是 A2A（Agent-to-Agent）架構：一邊把既有 ADK Agent 包裝成「可以被遠端呼叫的 Agent Service」，另一邊則示範另一個程式如何發現這個 Agent 的能力、送出任務，並接收結果。

如果上一段 MCP 可以理解成：

```text
Agent → Tool
```

那這一段 A2A 就是：

```text
Agent → Agent
```

兩者解決的是不同層級的問題。

先看整體架構：

```text
                    A2A Protocol
                         │
        ┌────────────────┴────────────────┐
        │                                 │
        ▼                                 ▼
 Client / Orchestrator              pricing_agent
        │                                 │
        │       discover / message        │
        ├────────────────────────────────>│
        │                                 │
        │          result / task          │
        │<────────────────────────────────┤
        │                                 │
```

更具體一點，這段 code 做了兩件事：

```text
Server Side
pricing_agent
    ↓
to_a2a(...)
    ↓
A2A Service
    ↓
port 8001


Client Side
ClientFactory
    ↓
連到 pricing-agent:8001
    ↓
取得 Agent Card
    ↓
送出 Message
    ↓
接收 Task / Message
```

最重要的觀念是：

> `pricing_agent` 原本只是一個 ADK Agent；`to_a2a()` 把它暴露成一個遵循 A2A protocol 的網路服務。

### A2A Agent Call

對，你現在真正想要的是：

> 本地 Agent 仍然是主控；遇到特定工作時，呼叫遠端 A2A Agent 去完成，拿到結果後，本地 Agent 繼續處理。

這和「把整個 conversation transfer 給 sub-agent」不是同一件事。

Google ADK 裡目前可以看到兩種不同控制模型：

```text
Chat / transfer 模式
    transfer_to_agent
    → 把控制權交給另一個 Agent

Task / delegation 模式
    request_task_...
    → 把一個工作交出去
    → 等結果回來
    → parent 繼續執行
```

ADK 現在的 task mode 文件也明確區分這兩種模式：chat mode 使用 `transfer_to_agent`；task mode 使用 `request_task`，由被呼叫的 Agent 完成後透過 `finish_task` 回傳結果。Task mode 也支援平行 delegation。([GitHub][1])

所以先回答你第二個問題：

「只能叫 sub-agent 嗎？」

不是。

`sub_agent` 是 ADK agent graph 裡的一個結構關係，不代表一定要做「控制權 transfer」。

同一個 child/remote agent，可以有不同的 interaction semantics：

```text
Parent
  │
  ├── transfer
  │      ↓
  │   Sub-agent 接管
  │
  └── task delegation
         ↓
      Sub-agent 做完
         ↓
      result 回 Parent
```

你真正要的是第二種。

而你剛才說：

> 「交代出去是不是就可能不回來？」

如果使用 `transfer_to_agent`，你的擔心是合理的。

`transfer_to_agent` 本質比較像：

```text
目前這個 Agent 不適合處理
→ 把 active conversation / execution routing
   轉交給另一個 Agent
```

ADK source code 的 transfer flow 也是把 target agent 找出來後，改變 execution routing，而不是把它當成普通 function call。([GitHub][2])

因此：

```text
Manager Agent
   │
   │ transfer_to_agent
   ▼
Pricing Agent
```

語意比較像：

> 「接下來你處理。」

而不是：

> 「幫我查一下價格，然後把結果還給我。」

這兩種非常不同。

甚至 ADK 目前就有相關 issue，指出 A2A remote sub-agent 若透過 `transfer_to_agent` 執行，有情況會讓 parent flow 結束，而 parent LLM 沒有重新拿到 remote agent 的結果繼續處理。該 issue 直接把問題描述成「handoff control」與「call-and-return」語意不一致。([GitHub][3])

所以你的直覺：

> 「如果我要的是交辦後回來，transfer 好像不太合適。」

是正確的。

---

如果你的 architecture 是：

```text
Kitchen Manager
       │
       │ 問價
       ▼
Remote Pricing Agent
       │
       │ $24/lb
       ▼
Kitchen Manager
       │
       │ 再結合庫存判斷
       ▼
User
```

這是一個典型的 call-and-return delegation。

你不希望：

```text
Kitchen Manager
       │
       │ transfer
       ▼
Pricing Agent
       │
       ▼
User
```

因為這樣 Kitchen Manager 後面的 business logic 就沒機會執行。

例如：

```text
1. 查 inventory
2. 查 market price
3. 綜合 inventory + price
4. 決定要不要補貨
```

第二步只是整體 reasoning 的一部分。

所以 Price Agent 應該是「service-like specialist agent」，不是新的 conversation owner。

概念上：

```python
pricing_agent = RemoteA2aAgent(
    name="pricing",
    agent_card="http://pricing-agent:8001/.well-known/agent-card.json",
)

kitchen_manager = Agent(
    model="gemini-3-flash-preview",
    name="kitchen_manager",
    instruction="""
You are the kitchen manager.

For wholesale pricing questions, request a task from the
pricing specialist and wait for the result.

After receiving the pricing result, continue your own reasoning.
Do not hand over the conversation to the pricing agent.

Combine pricing information with inventory information before
answering the user.
""",
    sub_agents=[
        pricing_agent,
    ],
)
```

但這裡我要特別提醒一點：

具體是否能直接把 `RemoteA2aAgent` 用成 task-mode child，會依你目前的 Google ADK 版本而定。ADK 對 remote A2A agent、`sub_agents` 與 task semantics 這塊截至 2026 年仍在快速演進，GitHub 上也還有相關功能與行為討論。([GitHub][4])

所以 architecture 上我會建議你明確要求：

```text
call-and-return
```

而不是：

```text
transfer
```

你可以把這兩種模式想成公司裡的兩句話。

`transfer_to_agent`：

```text
「這個客戶接下來交給你。」
```

控制權轉走。

而 task delegation：

```text
「幫我查一下今天 salmon 的批發價，
查完把結果給我。」
```

控制權還在 manager。

後者更符合你的 case。

---

至於「A2A 本身是不是不太做這種交辦？」

不是。

恰好相反，A2A protocol 本身非常適合這種 task-oriented delegation。

A2A 的核心資料模型裡本來就有：

```text
Message
Task
Task state
Artifact
```

也就是：

```text
Agent A
   │
   │ Message / Task
   ▼
Agent B
   │
   │ processing
   ▼
Task
   ├── status
   └── artifacts
```

所以 A2A 不只支援：

```text
Agent A → Agent B
```

它本來就可以表達：

> 「這是一個工作，你處理它，產生結果。」

只是 Google ADK 在自己的 agent orchestration layer 裡，又有自己的：

```text
transfer_to_agent
```

這個概念。

不要把：

```text
A2A Task
```

和：

```text
ADK transfer_to_agent
```

當成同一個東西。

這是最容易混淆的地方。

---

我會把整個概念整理成三層。

第一層是 A2A protocol：

```text
Agent A
   │
   │ A2A Message / Task
   ▼
Agent B
```

它只負責 agent-to-agent communication。

第二層是 ADK remote adapter：

```text
RemoteA2aAgent
```

它讓 ADK runtime 可以跟遠端 A2A Agent 溝通。

第三層才是 orchestration semantics：

```text
Transfer
or
Task delegation
```

也就是：

```text
                Remote Pricing Agent
                       ▲
                       │ A2A
                       │
              RemoteA2aAgent
                       ▲
                       │
          ┌────────────┴─────────────┐
          │                          │
     transfer                  task request
          │                          │
「你接手 conversation」      「幫我完成這件事」
          │                          │
 parent 不一定回來            result 回 parent
```

你的 kitchen manager case 明顯是右邊。

---

所以我會把你的 target architecture 定成：

```text
                         User
                          │
                          ▼
                   Kitchen Manager
                          │
              ┌───────────┴───────────┐
              │                       │
              ▼                       ▼
        Inventory Tool          Pricing Task
              │                       │
             MCP                      │ A2A
              │                       ▼
              ▼               Remote Price Agent
         Inventory DB                  │
                                      │ result
                                      ▼
                               Kitchen Manager
                                      │
                         combine price + stock
                                      │
                                      ▼
                                    User
```

Kitchen Manager 永遠是 orchestrator。

Price Agent 是 specialist worker。

這是比「sub-agent handoff」更符合你需求的 mental model。

一句話總結：

> `sub_agent` 只是 topology；`transfer_to_agent` 是 handoff；`task delegation` 才是你現在要的「交辦—回傳—繼續」。

而 A2A 很適合做「交辦」，只是你要避免把它用成 ADK 的 conversation transfer。

## A2UI Example Code

```python
# A2UI: Agent sends declarative JSON, renderer turns it into native UI.
# 18 component primitives — same building blocks, any frontend.

a2ui_messages = [
    # 1. Create a rendering surface
    {"beginRendering": {"surfaceId": "default", "root": "card"}},

    # 2. Send the component tree (flat list, ID references)
    {"surfaceUpdate": {"surfaceId": "default", "components": [
        {"id": "card", "component": {"Card": {"child": "col"}}},
        {"id": "col", "component": {"Column": {"children": {
            "explicitList": ["title", "price", "buy"]}}}},
        {"id": "title", "component": {"Text": {"usageHint": "h3",
            "text": {"path": "name"}}}},
        {"id": "price", "component": {"Text": {"text": {"path": "price"}}}},
        {"id": "buy", "component": {"Button": {"child": "btn-label",
            "action": {"name": "purchase",
            "context": [{"key": "item", "value": {"path": "name"}}]}}}},
        {"id": "btn-label", "component": {"Text": {
            "text": {"literalString": "Buy Now"}}}},
    ]}},

    # 3. Send the data (separate from structure)
    {"dataModelUpdate": {"surfaceId": "default", "contents": [
        {"key": "name", "valueString": "Fresh Atlantic Salmon"},
        {"key": "price", "valueString": "$24.00/lb"},
    ]}},
]
```

### A2UI code 解讀

這段 code 在展示 A2UI 的核心思想：Agent 不直接產生 HTML、React、SwiftUI 或 Flutter，而是輸出一份「宣告式 UI 描述」，由前端 renderer 轉成各平台原生介面。

如果前面兩個概念分別是：

```text
MCP  = Agent → Tool
A2A  = Agent → Agent
```

那這裡可以先記成：

```text
A2UI = Agent → UI
```

它解決的問題是：

> Agent 要怎麼把「我想呈現什麼介面」傳給前端，而不需要知道前端到底是 Web、iOS、Android 還是其他 UI framework。

整體架構可以想成：

```text
Agent
  │
  │ A2UI declarative messages
  ▼
Frontend Renderer
  │
  ├── React / Web
  ├── SwiftUI / iOS
  ├── Jetpack Compose / Android
  └── 其他 frontend
```

Agent 傳的不是：

```html
<div class="card">
  ...
</div>
```

也不是：

```jsx
<Card>
  <Text>{name}</Text>
</Card>
```

而是比較抽象的：

```text
我要一張 Card
裡面放一個 Column
Column 裡面有 title、price、buy button
title 綁定 data.name
price 綁定 data.price
button 被按下時觸發 purchase action
```

Renderer 再負責決定實際長什麼樣。

這是這段 code 最重要的觀念。

---

先看最外層：

```python
a2ui_messages = [
    ...
]
```

這不是一個 UI object，而是一連串 protocol messages。

這一點很重要。

它不是：

```python
ui = Card(...)
```

而比較像：

```text
Message 1：建立 UI surface
Message 2：更新 UI structure
Message 3：更新 UI data
```

所以 A2UI 是 message-driven。

這代表 UI 可以被逐步建立、更新，而不一定每次都重新傳完整頁面。

完整流程是：

```text
beginRendering
      ↓
surfaceUpdate
      ↓
dataModelUpdate
      ↓
Renderer
      ↓
畫出 UI
```

---

第一步：

```python
{"beginRendering": {
    "surfaceId": "default",
    "root": "card"
}}
```

這是在建立一個 rendering surface。

可以把 `surface` 理解成：

> 一塊由 A2UI 控制的 UI 區域。

例如可能是：

```text
Chat window
┌──────────────────────────┐
│ Assistant message        │
│                          │
│ ┌──────────────────────┐ │
│ │ A2UI Surface         │ │
│ │                      │ │
│ │ Product Card         │ │
│ │                      │ │
│ └──────────────────────┘ │
└──────────────────────────┘
```

其中：

```python
"surfaceId": "default"
```

就是這塊 UI surface 的識別碼。

`我要操作一個 surface ID 叫做 "default" 的 UI surface`

如果一個 Agent 同時管理多塊 UI，也可能有：

```text
product-panel
checkout-panel
recommendation-panel
```

之類的不同 surface。

接著：

```python
"root": "card"
```

告訴 renderer：

> 這棵 UI tree 的 root component 是 ID 為 `card` 的 component。

注意，它不是直接把 Card 塞在這裡，而只是引用：

```text
card
```

真正的 Card definition 稍後才送。

這也是整段程式另一個重要特色：

> UI component tree 是以 ID reference 建立的。

---

第二步：

```python
{"surfaceUpdate": {
    "surfaceId": "default",
    "components": [...]
}}
```

這是在告訴 renderer：

> `default` 這個 surface 裡有哪些 component，以及它們的關係。

裡面是 flat list：

```python
"components": [
    ...
]
```

注意不是典型 nested tree：

```python
Card(
    Column(
        Text(...),
        Text(...),
        Button(...)
    )
)
```

而是：

```text
card
col
title
price
buy
btn-label
```

全部攤平成一個 list。

然後透過 ID reference 建立 tree。

這是一種 normalized representation。

概念很像資料庫正規化或者 DOM node references：

```text
components
────────────────────────
id          type
────────────────────────
card        Card
col         Column
title       Text
price       Text
buy         Button
btn-label   Text
```

然後 component 彼此引用。

---

先看：

```python
{"id": "card",
 "component": {
     "Card": {
         "child": "col"
     }
 }}
```

意思是：

```text
card
  │
  └── child → col
```

所以：

```text
Card
└── Column
```

接著 `col`：

```python
{"id": "col",
 "component": {
     "Column": {
         "children": {
             "explicitList": [
                 "title",
                 "price",
                 "buy"
             ]
         }
     }
 }}
```

這代表：

```text
Column
├── title
├── price
└── buy
```

目前整棵 tree 就變成：

```text
card
└── col
    ├── title
    ├── price
    └── buy
```

Renderer 最後看到的不是 flat list，而會根據 references reconstructed 成 UI tree。

---

接著看 `title`：

```python
{"id": "title",
 "component": {
     "Text": {
         "usageHint": "h3",
         "text": {
             "path": "name"
         }
     }
 }}
```

這裡有兩個不同概念。

第一個：

```python
"usageHint": "h3"
```

它比較接近 semantic hint：

> 這段 Text 是三級標題性質。

它沒有直接說：

```css
font-size: 22px;
font-weight: 700;
```

因為 A2UI 希望保持 platform independence。

Web renderer 可以決定：

```html
<h3>
```

iOS renderer 可以選擇某個 title typography。

Android renderer也可以套用自己的 Material typography。

所以：

```text
usageHint = semantic intent
```

而不是：

```text
usageHint = exact visual implementation
```

這是 declarative cross-platform UI 很重要的設計。

---

第二個更重要：

```python
"text": {
    "path": "name"
}
```

注意這裡沒有寫：

```python
"text": "Fresh Atlantic Salmon"
```

而是：

```text
path → name
```

這表示：

> 這個 Text 的值要從 data model 的 `name` 欄位取得。

因此它其實是一個 data binding。

可以類比：

React：

```jsx
<Text>{data.name}</Text>
```

Vue：

```html
{{ name }}
```

SwiftUI：

```swift
Text(model.name)
```

A2UI 則用：

```json
{
  "text": {
    "path": "name"
  }
}
```

來表達。

所以 structure 與 data 是分離的。

---

下一個：

```python
{"id": "price",
 "component": {
     "Text": {
         "text": {
             "path": "price"
         }
     }
 }}
```

一樣：

```text
price Text
   │
   └── bind → data.price
```

所以目前：

```text
Card
└── Column
    ├── Text ← data.name
    ├── Text ← data.price
    └── Button
```

---

然後最值得看的就是 Button：

```python
{"id": "buy",
 "component": {
     "Button": {
         "child": "btn-label",
         "action": {
             "name": "purchase",
             "context": [
                 {
                     "key": "item",
                     "value": {
                         "path": "name"
                     }
                 }
             ]
         }
     }
 }}
```

Button 有兩部分：

```text
Visual content
+
Interaction
```

首先：

```python
"child": "btn-label"
```

意思是 Button 裡面顯示 `btn-label` component。

所以：

```text
Button
└── btn-label
```

後面：

```python
{"id": "btn-label",
 "component": {
     "Text": {
         "text": {
             "literalString": "Buy Now"
         }
     }
 }}
```

注意這次不是：

```python
"path": ...
```

而是：

```python
"literalString": "Buy Now"
```

差別非常重要：

```text
{"path": "name"}
```

代表：

```text
從 data model 讀取
```

而：

```text
{"literalString": "Buy Now"}
```

代表：

```text
固定字串
```

所以可以把兩者理解成：

```text
Dynamic binding:
path → name

Static value:
literalString → "Buy Now"
```

---

Button 更關鍵的是：

```python
"action": {
    "name": "purchase",
    ...
}
```

這代表按下 Button 時，不只是前端做某個 local callback。

它宣告的是一個 semantic action：

```text
purchase
```

概念上：

```text
User
  │
  │ click "Buy Now"
  ▼
Renderer
  │
  │ Action event:
  │ name = purchase
  ▼
Agent / Application
```

因此 UI 不只是一個 output channel。

它也變成 input channel。

流程變成：

```text
Agent
   │
   │ generate UI
   ▼
Renderer
   │
   ▼
User sees Button
   │
   │ click
   ▼
Action
   │
   ▼
Agent
```

所以 A2UI 可以支援：

> Agent → UI → User interaction → Agent

形成完整 interactive loop。

---

接著看：

```python
"context": [
    {
        "key": "item",
        "value": {
            "path": "name"
        }
    }
]
```

這是在說：

> 當 `purchase` action 被觸發時，附帶一些 context。

這裡會帶：

```text
item = data.name
```

目前 data.name 是：

```text
Fresh Atlantic Salmon
```

所以使用者按下 Buy Now，概念上可能產生：

```json
{
  "action": "purchase",
  "context": {
    "item": "Fresh Atlantic Salmon"
  }
}
```

這樣後端就知道：

> 使用者到底想買什麼？

否則如果畫面上有 20 個 Buy Now：

```text
Salmon      [Buy Now]
Tuna        [Buy Now]
Scallops    [Buy Now]
```

只有：

```text
action = purchase
```

是不夠的。

你還需要：

```text
purchase what?
```

所以 context 就負責提供 action 所需的 contextual data。

---

第三步：

```python
{"dataModelUpdate": {
    "surfaceId": "default",
    "contents": [
        {
            "key": "name",
            "valueString":
                "Fresh Atlantic Salmon"
        },
        {
            "key": "price",
            "valueString":
                "$24.00/lb"
        },
    ]
}}
```

這是在送真正的 data。

注意前面 `surfaceUpdate` 只說：

```text
title → data.name
price → data.price
```

它沒有真正提供資料。

現在才提供：

```text
name  = Fresh Atlantic Salmon
price = $24.00/lb
```

所以 renderer 執行 data binding：

```text
Text(path="name")
       ↓
"Fresh Atlantic Salmon"

Text(path="price")
       ↓
"$24.00/lb"
```

最後 UI 大概變成：

```text
┌────────────────────────────────┐
│ Fresh Atlantic Salmon          │
│                                │
│ $24.00/lb                      │
│                                │
│ ┌──────────────┐               │
│ │   Buy Now    │               │
│ └──────────────┘               │
└────────────────────────────────┘
```

這就是整段 JSON 最後表達的 UI。

---

真正值得學的地方是「structure / data separation」。

整段 code 有：

```text
surfaceUpdate
```

負責：

```text
UI structure
```

以及：

```text
dataModelUpdate
```

負責：

```text
UI data
```

可以畫成：

```text
        UI Structure
        surfaceUpdate
             │
             │
             ▼
        ┌──────────┐
        │ Renderer │
        └──────────┘
             ▲
             │
             │
          UI Data
      dataModelUpdate
```

這種設計有一個很實際的好處。

假設 salmon 價格從：

```text
$24.00/lb
```

變成：

```text
$21.50/lb
```

理論上 Agent 不需要重新傳整棵：

```text
Card
Column
Text
Text
Button
```

只需要送 data update：

```text
price = $21.50/lb
```

renderer 就更新：

```text
$24.00/lb
    ↓
$21.50/lb
```

這非常適合 dynamic agent applications。

例如：

```text
stock price dashboard
flight status
shopping cart
inventory
workflow status
approval UI
```

都可能持續變動。

---

再回來看註解：

```python
# 18 component primitives —
# same building blocks, any frontend.
```

這句話其實透露 A2UI 的 design philosophy。

它不是讓 Agent 隨便產生：

```text
任意 HTML
任意 JavaScript
任意 CSS
```

而是限定一組 component primitives。

例如概念上可能是：

```text
Text
Button
Card
Column
Row
Image
TextField
...
```

Agent只使用這些標準 building blocks。

因此：

```text
Agent output
     │
     ▼
A2UI primitives
     │
     ├──────────────┬───────────────┐
     ▼              ▼               ▼
 Web Renderer   iOS Renderer   Android Renderer
     │              │               │
     ▼              ▼               ▼
 React-ish UI    SwiftUI       Compose
```

同一份 declarative description，可以在不同 frontend 上有不同 native implementation。

這也是它相對於直接讓 Agent 產 HTML 的巨大差別。

---

例如 Agent 如果直接產：

```html
<button
  style="background:#007aff">
  Buy Now
</button>
```

就把自己跟 Web implementation 綁死了。

但如果 Agent 說：

```json
{
  "Button": {
    "action": {
      "name": "purchase"
    }
  }
}
```

那 iOS renderer 可以把它轉成：

```text
SwiftUI Button
```

Android 可以轉成：

```text
Material Button
```

Web 可以轉成：

```text
React Button
```

所以這裡真正的 abstraction boundary 是：

```text
Agent 決定：
WHAT UI should exist

Renderer 決定：
HOW that UI should look
```

這跟 declarative UI framework 的哲學非常接近。

---

現在把你目前三段 code 串起來，整個 Agent architecture 就開始完整了。

第一層，MCP：

```text
Agent
 ↓
Tool
```

解決外部能力：

```text
Database
Notion
API
filesystem
...
```

第二層，A2A：

```text
Agent
 ↓
Agent
```

解決 delegation / multi-agent communication。

第三層，A2UI：

```text
Agent
 ↓
UI
```

解決 agent-generated interactive interface。

所以一個完整系統可以變成：

```text
                          User
                           ▲
                           │
                        A2UI
                           │
                           ▼
                 Restaurant Manager
                      Agent
                           │
             ┌─────────────┴─────────────┐
             │ A2A                       │ A2A
             ▼                           ▼
       Kitchen Agent              Pricing Agent
             │                           │
        ┌────┴─────┐                    MCP
        │          │                     │
       MCP        MCP                    ▼
        │          │                Market API
        ▼          ▼
   Inventory    Notion
   Database     Recipes
```

例如使用者問：

```text
今晚想做 salmon，
幫我看看能不能做，而且價格可以的話幫我準備購買。
```

Restaurant Manager Agent 可能：

```text
1. A2A → Kitchen Agent
   「需要多少 salmon？」

2. Kitchen Agent
   MCP → Notion recipe
   MCP → inventory DB

3. A2A → Pricing Agent
   「今天 salmon 價格多少？」

4. Pricing Agent
   MCP → market price data

5. Restaurant Manager 判斷：
   價格合理。

6. A2UI → User
```

最後不是回一大段文字：

```text
Salmon costs $24/lb.
Would you like to buy it?
```

而是直接產生：

```text
┌───────────────────────────────┐
│ Fresh Atlantic Salmon         │
│                               │
│ $24.00/lb                     │
│                               │
│          [ Buy Now ]          │
└───────────────────────────────┘
```

使用者點：

```text
Buy Now
```

再產生：

```text
action = purchase
item = Fresh Atlantic Salmon
```

交回 Agent 處理下一步。

因此三者其實可以用一句非常好記的方式整理：

```text
MCP   = 給 Agent「手」
A2A   = 給 Agent「同事」
A2UI  = 給 Agent「介面」
```

再換成比較工程化的語言：

```text
MCP
Tool interoperability

A2A
Agent interoperability

A2UI
UI interoperability
```

而這段 A2UI code 最值得你掌握的不是 JSON syntax，而是四個 architecture concepts：

```text
1. Declarative UI
   Agent 描述「要什麼 UI」，
   不描述「怎麼畫」。

2. Component graph
   component 使用 ID + reference 組合，
   structure 可以被獨立更新。

3. Data binding
   UI structure 與 data model 分離。

4. Semantic actions
   UI interaction 不是只有 frontend callback，
   而可以回到 Agent 成為下一輪行為。
```

一旦把這四點理解了，這段看起來很繁瑣的 JSON，其實只是在表達一句非常簡單的事情：

```text
「請建立一張商品卡，
顯示商品名稱與價格，
並提供一個 Buy Now 按鈕；
按下時告訴我使用者要 purchase 哪一個商品。」
```

而真正有價值的地方，是這句描述可以脫離特定 frontend technology，由不同 renderer 用各自的 native UI 實現。


### A2UI 設計哲學

 A2UI 可以控制很多細節，但 A2UI 的設計哲學偏向「語意與結構優先」，不是讓 Agent 像寫 CSS 一樣控制每一個 pixel。

以目前 A2UI v1.0 的方向來說，Agent 主要描述：

```text
這是一張 Card
裡面有 Column
這段文字是標題
這個 Button 是 primary
這裡放圖片
```

Renderer 再決定實際顏色、字體、圓角、spacing 等視覺細節。官方 v1.0 特別強調把 branding / styling 從 protocol 解耦，讓 target framework 的 native theme 決定外觀。([GitHub][1])

所以像你問的：

```text
「這塊要紅色」
「這個按鈕圓角 12px」
「這裡畫一個 120×80 的藍色橢圓」
```

在 basic A2UI catalog 裡通常不是 Agent 直接傳這種 low-level style。

比較像：

```json
{
  "component": "Button",
  "variant": "primary"
}
```

Renderer 收到 `"primary"` 後：

```text
Web renderer
→ 藍底白字、8px radius

iOS renderer
→ 系統 accent color、native corner radius

Android renderer
→ Material primary button
```

也就是：

> Agent 傳「它是什麼」，Renderer 決定「它長什麼樣」。

這也是 `usageHint: "h3"` 之類欄位的概念：它是在說「這是 heading」，而不是直接說 `font-size: 20px; font-weight: 700`。([A2UI][2])

不過 A2UI 不是完全不能控制視覺。

有三個層次。

第一層是 semantic styling，最標準。例如：

```json
{
  "component": "Text",
  "variant": "h3"
}
```

或：

```json
{
  "component": "Button",
  "variant": "primary"
}
```

這種最符合 A2UI philosophy。

第二層是 theme。較早的 v0.9 basic catalog 曾經允許 surface 傳像：

```json
{
  "theme": {
    "primaryColor": "#FF0000"
  }
}
```

所以 Agent 可以影響整體品牌色。([GitHub][3])

但到了 v1.0，A2UI 有意把這類硬編碼 brand color 從核心 protocol/basic styling 中拿掉，傾向交給 renderer/native theme。([GitHub][1])

第三層最重要：Custom Catalog。

A2UI 並不限制你只能使用官方 basic components。Protocol 本身把 component 定義放在 Catalog 裡，而且 production application 通常可以定義自己的 catalog。([GitHub][1])

所以你完全可以自己設計一套：

```text
MyCompany Catalog

ProductCard
PriceBadge
StatusChip
RoundedPanel
WarningBanner
Chart
Gauge
Map
```

甚至定義：

```json
{
  "component": "StatusBadge",
  "status": "danger"
}
```

你的 renderer 可以決定：

```text
danger
→ red background
→ white text
→ pill shape
→ 16px radius
```

Agent還是不需要知道：

```text
#FF3B30
border-radius: 16px
padding: 4px 12px
```

這是一個很重要的 abstraction。

如果真的希望 Agent 能指定更細的形狀，也可以在 custom catalog 裡定義，例如：

```json
{
  "component": "Shape",
  "shape": "circle",
  "size": "large",
  "variant": "warning"
}
```

Renderer：

```text
shape="circle"
→ 畫圓形

variant="warning"
→ 用 application warning color
```

甚至你也可以把 schema 做得更 low-level：

```json
{
  "component": "Shape",
  "shape": "ellipse",
  "fill": "#FF0000",
  "width": 120,
  "height": 80
}
```

技術上可以，因為 custom catalog 可以定義自己的 component properties。

但這通常不是 A2UI 最推薦的方向。

原因是如果 Agent 開始輸出：

```text
width=117
height=43
color=#18A0FB
radius=13
margin-left=7
```

你就慢慢把 A2UI 變成：

```text
Agent-generated CSS
```

這會犧牲：

```text
跨平台一致性
native look & feel
accessibility
responsive design
design-system governance
```

所以比較好的 production 設計通常是：

```text
Agent：
"status": "warning"
"size": "large"
"variant": "primary"

Renderer / Design System：
warning = orange
large = 48px
primary = brand styling
```

而不是：

```text
Agent：
color = #FF9800
height = 48
borderRadius = 11
```

所以你可以把 A2UI 的控制粒度想成一條光譜：

```text
最推薦
──────────────────────────────────────→ 比較不推薦

semantic          design tokens       raw styling
────────          ─────────────       ───────────
"h3"              "danger"            "#FF0000"
"primary"         "large"             "12px"
"warning"         "rounded"           "width:137"
```

A2UI 最舒服的位置通常在左邊到中間。

因此，如果你問：

> 「Agent 可不可以說這一塊是圓形、紅色？」

答案是：

可以，尤其透過 custom catalog；但更好的 A2UI 設計通常不是讓 Agent直接指定「紅色 #FF0000、圓角 16px」，而是讓它說「這是一個 danger badge / circular status indicator」，由 renderer 和 design system 決定真正的形狀與顏色。([GitHub][1])

一句話來說：

```text
A2UI 可以很細，
但刻意鼓勵你不要太細。
```

它希望 Agent 控制的是 UI 的「意圖、結構、語意」，而不是像 Figma 或 CSS 一樣控制每一個視覺像素。

### AG-UI code 解讀

這段 code 在展示 AG-UI：把一個既有的 ADK Agent 包裝成「前端可以即時監看與互動的 streaming endpoint」。

如果沿用前面三個概念：

```text
MCP  = Agent → Tool
A2A  = Agent → Agent
A2UI = Agent → UI structure
```

那 AG-UI 可以先記成：

```text
AG-UI = Agent ↔ Frontend runtime
```

它主要解決的不是「畫什麼 UI」，而是：

> Agent 執行過程中的狀態、文字、tool call、結果，要怎麼即時、結構化地送到前端。

所以它更像一條 agent runtime event stream。

整體架構大概是：

```text
User
  │
  ▼
Frontend
  │
  │ HTTP / SSE
  ▼
AG-UI Endpoint
  │
  ▼
ADK Agent
  │
  ├── reasoning
  ├── tool calls
  └── response generation
```

而 Agent 執行過程中發生的事件，會一路 stream 回 frontend：

```text
ADK Agent
   │
   ├── RUN_STARTED
   ├── TOOL_CALL_START
   ├── TOOL_CALL_RESULT
   ├── TOOL_CALL_END
   ├── TEXT_MESSAGE_CONTENT
   └── RUN_FINISHED
   │
   ▼
Frontend
```

這是這段 code 最重要的地方。

也就是 server 可以在單一 HTTP connection 上，不斷往 client 推事件。

普通 HTTP：

AG-UI 就把這些過程變成 typed events。

這裡特別值得區分 AG-UI 與 A2UI。

名字非常像，但它們處理的是兩件不同的事情。

A2UI 比較關心：

```text
What UI should be rendered?
```

例如 Agent 說：

```text
Card
├── Text: Salmon
├── Text: $24/lb
└── Button: Buy Now
```

它是 declarative UI specification。

AG-UI 比較關心：

```text
What is the Agent doing right now?
```

例如：

```text
run started
tool started
tool returned
text delta
run finished
```

它是 agent runtime protocol。

所以可以粗略分成：

```text
A2UI
Agent → UI description

AG-UI
Agent ↔ UI runtime communication
```

兩者甚至可以互補。

例如：

```text
User
  │
  ▼
Frontend
  │
  │ AG-UI
  ▼
Kitchen Agent
  │
  ├── MCP → Inventory
  │
  ├── A2A → Pricing Agent
  │
  └── A2UI → Product Card
```

同時 AG-UI 仍然可以持續 stream：

```text
RUN_STARTED
TOOL_CALL_START
TOOL_CALL_RESULT
...
RUN_FINISHED
```

所以 frontend 一方面知道 Agent 正在做什麼，另一方面又可以收到 Agent 產生的 rich UI。

現在把你前面四個概念放在一起：

```text
MCP
Agent ↔ Tools

A2A
Agent ↔ Agents

A2UI
Agent → Declarative UI

AG-UI
Agent ↔ Frontend Runtime
```

更完整地畫：

```text
                           User
                            │
                            ▼
                     ┌────────────┐
                     │  Frontend  │
                     └─────┬──────┘
                           │
                        AG-UI
                           │
                           ▼
                ┌────────────────────┐
                │ Restaurant Manager │
                │       Agent        │
                └─────────┬──────────┘
                          │
             ┌────────────┼─────────────┐
             │            │             │
            A2A           MCP           A2UI
             │            │             │
             ▼            ▼             ▼
      Pricing Agent   Database       UI structure
             │
            MCP
             │
             ▼
        Market API
```

這裡每一層 responsibility 很清楚：

```text
AG-UI
「前端怎麼知道 Agent 正在做什麼？」

A2UI
「Agent 想讓前端呈現什麼？」

A2A
「Agent 怎麼把工作委派給另一個 Agent？」

MCP
「Agent 怎麼使用外部工具？」
```

如果要用前面那種好記的比喻，可以變成：

```text
MCP   = Agent 的手

A2A   = Agent 的同事

A2UI  = Agent 設計的介面

AG-UI = Agent 與前端之間的神經系統
```

其中最後一個「神經系統」的意思是：它不是畫面本身，而是把 execution state、tool activity、message streaming、run lifecycle 持續傳送到前端。

對已經學過一段時間 AI 的人來說，這段 code 最值得掌握的其實是三個 engineering concepts：

```text
1. Protocol adapter
   ADKAgent 把既有 ADK runtime
   轉成 AG-UI-compatible runtime。

2. Streaming execution model
   Agent 不再被視為單純的
   request → response function，
   而是一個會產生一連串 events 的 process。

3. Typed events
   Frontend 不需要從文字猜：
   「Agent 現在是不是在 call tool？」
   而是直接收到 TOOL_CALL_START、
   TOOL_CALL_RESULT 等結構化事件。
```

最後這一點尤其重要。

傳統 chatbot API 往往只有：

```json
{
  "text": "I checked inventory. You have 3 lbs..."
}
```

Frontend 其實不知道中間到底發生什麼。

AG-UI 的思路則是：

```text
不要只給我最終答案。

把 Agent execution 本身，
用標準化事件 stream 給我。
```

這樣前端才有能力做真正的 agent-native UX，例如 tool progress、approval flow、running status、live intermediate results、error states、multi-step workflow，甚至把 Agent 的 execution timeline 視覺化出來。

所以這段程式表面上只是「三行把 ADK Agent 掛到 FastAPI」，但 architecture 上真正做的是：

> 把一個 backend Agent runtime，轉成 frontend 可以即時理解、監看與互動的事件驅動服務。


### AG-UI 設計哲學

AG-UI 的主要 consumer 是「frontend / agentic frontend」，不是 telemetry backend。

AG-UI 的核心設計哲學，不是「把 Agent 的文字串流到前端」，而是：

> 建立一條 Agent 與前端應用程式之間的雙向互動協定，讓兩邊可以共同交換訊息、狀態、操作意圖與執行控制。

也就是說，AG-UI 把 frontend 看成 Agent runtime 的一部分，而不是單純的顯示器。

傳統 chatbot 通常是：

```text
User
  ↓
Frontend
  ↓
POST /chat
  ↓
LLM
  ↓
Text response
  ↓
Frontend
```

Frontend 的角色很被動，只負責：

```text
收文字
顯示文字
```

但 Agent application 通常比 chatbot 複雜很多。

例如 Agent 可能：

```text
查資料
呼叫 tool
等待 approval
更新 application state
產生中間結果
執行長時間 task
要求使用者做選擇
```

這時候如果 frontend 只收到最後一句文字：

```text
"Your order has been processed."
```

其實是不夠的。

Frontend 會不知道：

```text
Agent 現在開始了嗎？
正在做什麼？
用了哪個 tool？
tool 成功了嗎？
現在是不是在等使用者批准？
state 改了嗎？
run 結束了嗎？
```

---


AG-UI 的核心不是「把 Agent 的文字串流到前端」，而是建立一條：

```text
Frontend ↔ Agent Runtime
```

的雙向互動通道。

它同時承載三類東西：

```text
1. 資訊
   文字、tool result、error、progress

2. 狀態
   run started、run finished、shared state

3. 控制
   user action、approve、reject、cancel、retry
```

所以它比單純 SSE 更高一層。SSE 只是 transport；AG-UI 定義「這些事件代表什麼」。

例如 Agent 執行時：

```text
RUN_STARTED
TOOL_CALL_START
TOOL_CALL_RESULT
TEXT_MESSAGE_CONTENT
RUN_FINISHED
```

Frontend 可以據此顯示：

```text
Working...
Checking inventory...
3 lbs in stock
Based on current inventory...
```

反過來，Frontend 也可以告訴 Agent：

```text
使用者輸入了什麼
目前 UI state 是什麼
使用者按了哪個按鈕
是否批准某個操作
是否取消目前 run
```

所以最準確的定位是：

```text
AG-UI = Agent ↔ Frontend interaction protocol
```

而不是：

```text
Agent → Frontend text streaming
```

和其他協定放在一起看：

```text
MCP   = Agent ↔ Tool
A2A   = Agent ↔ Agent
A2UI  = Agent → UI structure
AG-UI = Agent ↔ Frontend runtime
```

一句話總結：

> AG-UI 的設計哲學，是讓 Frontend 不只是顯示 Agent 結果，而是直接參與 Agent 的狀態、互動與控制流程。

## 1. Google Cloud 帳戶、Project 與 Billing 是三件不同的事

「有 Google Cloud 帳戶」只代表你可以登入 Cloud Console。真正的資源都必須建立在一個 Google Cloud Project 裡。

一個 Project 會提供：

```text
Project ID
Project number
IAM 權限邊界
API 啟用狀態
Billing 關聯
Gemini Enterprise Apps
Data Stores
稽核紀錄
配額與費用
```

開始前至少需要：

1. 一個 Google Cloud Project。
2. 該 Project 已綁定 Billing Account。
3. 你的帳號具有啟用 API 和建立 Gemini Enterprise 資源的權限。

Google 目前的 quickstart 要求確認 Project 已啟用 billing，且操作者具備 Gemini Enterprise Admin。啟用 API 則需要 `serviceusage.services.enable` 權限，通常由 Project Owner 或 Service Usage Admin 提供。([Google Cloud Documentation][1])

常見 IAM 角色如下：

| 工作                   | 常見角色                    |
| -------------------- | ----------------------- |
| 建立 Project           | Project Creator         |
| 啟用 API               | Service Usage Admin     |
| 管理 Gemini Enterprise | Gemini Enterprise Admin |
| 建立或修改 Data Store     | Discovery Engine Editor |
| 一般使用 Web App         | Gemini Enterprise User  |

其中 Gemini Enterprise Admin 對應的角色 ID 通常是：

```text
roles/discoveryengine.agentspaceAdmin
```

這個名稱仍保留了舊產品名稱 `agentspace`，所以看到 Agentspace 不代表你進錯產品。

它是提供企業員工使用的搜尋、聊天和 Agent Web App。

## 3. 「Start 30 days free trial」實際做了什麼

影片按下：

```text
Start 30 days free trial
```

```text
Billing account
    ↓
Gemini Enterprise subscription / trial
    ↓
Project license allocation
    ↓
Users receive Gemini Enterprise licenses
```

需要區分兩種試用：

### Google Cloud Free Trial

這是新 Google Cloud 客戶常見的試用金，用於支付 Cloud 資源。

### Gemini Enterprise Free Trial

這是 Gemini Enterprise 的產品授權，通常涉及：

* Subscription
* License count
* 將 licenses 分配到特定 Project
* 將 licenses 指派給使用者

目前 Google 文件指出，Gemini Enterprise 的訂閱通常在 Billing Account 層級管理，但 license 是按 Project 分配；同一使用者若要在不同 Project 使用，可能需要在每個 Project 分別配置。([Google Cloud Documentation][2])

因此，影片中的 free trial 不只是「把 API 打開」，而是在進行產品授權與初始設定。

## 4. 啟用的 Gemini Enterprise API 是什麼

影片畫面說的是：

> activate the required API

Gemini Enterprise 核心使用的 API 是：

```text
Gemini Enterprise (Discovery Engine) API
```

其 service name 是：

```text
discoveryengine.googleapis.com
```

這也是為什麼許多 IAM role、REST resource 和舊文件仍包含 `discoveryengine`。

目前 Google 列出的必要 API 包括：

```text
Gemini Enterprise (Discovery Engine) API
Cloud Storage API
Identity and Access Management API
Vertex AI API
```

不同 onboarding 流程可能只在第一個畫面明確要求啟用 Discovery Engine API，其他 API 可能由精靈自動處理，或在後續功能使用時要求啟用。Google 的安全文件將 Vertex AI、Discovery Engine、Cloud Storage 和 IAM 都列為 Gemini Enterprise 所需 API；目前 quickstart 則明確要求啟用 Discovery Engine、Cloud Storage 和 IAM。([Google Cloud Documentation][3])

你也可以手動確認：

```text
Google Cloud Console
→ APIs & Services
→ Enabled APIs & services
```

搜尋：

```text
Gemini Enterprise
Discovery Engine
Vertex AI
Cloud Storage
Identity and Access Management
```

也可以使用 `gcloud`：

```bash
gcloud services enable \
  discoveryengine.googleapis.com \
  storage.googleapis.com \
  iam.googleapis.com \
  aiplatform.googleapis.com \
  --project=YOUR_PROJECT_ID
```

這段命令只是等價的 CLI 做法。使用 Cloud Console onboarding 時，通常不需要手動執行。

### 啟用 API 的意義

啟用 API 不等於建立了一個 Agent，也不等於產生 API key。

它只是允許這個 Project 建立和呼叫相關服務：

```text
API disabled
→ Project 無法建立 Gemini Enterprise 資源

API enabled
→ Project 可以建立 App、Data Store、Engine 等資源
```

Gemini Enterprise Web App 一般也不是靠你手動建立 API key 來運作。登入者身份、IAM、OAuth 和 Workspace identity 才是主要驗證機制。

## 5. 建立 Gemini Enterprise App

API 啟用後，進入：

```text
Gemini Enterprise
→ Apps
→ Create app
```

現在的建立頁面通常會要求：

```text
App name
App ID
Location
Organization/company display name
```

影片只提到 display name 和 region，是簡化過的說法。Google 目前的文件顯示，建立 App 時可以設定 App name、不可變更的 App ID、multi-region，以及對外顯示的組織名稱。([Google Cloud Documentation][4])

### App name

例如：

```text
Simon AI Workspace
```

這是介面中顯示的名稱，通常可以讓人閱讀。

### App ID

例如系統產生：

```text
simon-ai-workspace_174238
```

App ID 是 API 與資源路徑使用的穩定識別值。

建立後通常不能修改，所以正式環境最好使用有意義的命名，例如：

```text
corp-search-prod
corp-search-dev
ai-assistant-tw
```

不要把 production 和測試環境混在同一 App。

### App 在架構上的含義

App 不是 Agent 本身。它是所有終端使用者功能的容器：

```text
Gemini Enterprise App
├── Web App
├── Assistant
├── Agent Library
├── Agent Designer
├── Identity provider
├── Connected data stores
├── Feature toggles
├── Model settings
└── Access policy
```

因此，同一個 Project 可以有多個 App，例如：

```text
Project: enterprise-ai-prod
├── HR Assistant App
├── Engineering Knowledge App
└── Sales Intelligence App
```

每個 App 可以掛接不同資料來源和開放給不同使用者。

## 6. Region 或 Location 要怎麼選

影片說「choose the region」，但目前 Gemini Enterprise App 主要使用的是 multi-region：

```text
global
us
eu
```

Google 建議，沒有資料駐留需求時使用預設的 `global`；如果公司有法規、合約或資料駐留限制，才考慮 `us` 或 `eu`。([Google Cloud Documentation][4])

三者的概念是：

| Location | 適用情況                             |
| -------- | -------------------------------- |
| `global` | 教學、測試、沒有特定資料駐留要求                 |
| `us`     | 要求資料靜態儲存在美國 multi-region         |
| `eu`     | GDPR 或政策要求資料靜態儲存在歐盟 multi-region |

Location 是重要架構決策，因為它會影響：

* App resource 的位置
* Data Store 可使用的位置
* 部分 connector 的支援情況
* CMEK 設定
* Agent Gateway 或 Agent Registry 相容性
* 資料駐留與合規

通常 App 和所連接的 Data Store 應選擇相容的位置。

教學環境可以選：

```text
global
```

正式企業環境則應先詢問資安、法務或 Cloud 管理團隊。

## 7. 建立 App 後，為什麼要設定 Workforce Identity

App 建立完成後，下一步不是立刻接 Drive，而是設定「誰在使用」。

影片選：

```text
Google Identity
```

這表示 Gemini Enterprise Web App 使用者透過 Google Workspace 或 Cloud Identity 帳號登入。

這一步建立的邏輯是：

```text
Google Workspace user
    ↓ authentication
Gemini Enterprise user identity
    ↓ authorization
App access
    ↓ identity-aware connector query
Google Drive / Gmail / Calendar permissions
```

這裡最重要的是：使用者身份會向下傳遞到資料存取層。

例如：

```text
Alice 可讀 HR-policy.pdf
Bob 不可讀 HR-policy.pdf
```

即使 Alice 和 Bob 使用同一個 Gemini Enterprise App，搜尋結果也應不同：

```text
Alice query → 可以取得 HR-policy.pdf
Bob query   → 不應取得 HR-policy.pdf
```

這叫做 data source access control，而不只是 Web App 登入。

## 8. Google Identity 和 Workforce Identity Federation 的差異

目前主要有兩條路。

### Google Identity

適合：

* 公司已使用 Google Workspace
* 使用者有受管理的 Workspace 帳號
* 要連接 Gmail、Drive、Calendar
* 想快速完成教學

登入身份通常是：

```text
user@company.com
```

並由 Google Workspace 或 Cloud Identity 管理。

### Workforce Identity Federation

適合公司使用：

* Microsoft Entra ID
* Okta
* Ping Identity
* 自建 OIDC
* SAML identity provider

流程大致是：

```text
External IdP
→ Workforce Identity Pool
→ Workforce Identity Provider
→ Gemini Enterprise
```

Workforce Identity Federation 讓外部 IdP 的員工帳號可以登入 Google Cloud 服務，而不必替每個人建立傳統 Google Cloud 使用者身份。([Google Cloud][5])

影片為了簡化，選擇 Google Identity 是合理的。對 Google Workspace Connector 而言，這也是最直接的路徑。

目前介面大致是：

```text
Gemini Enterprise
→ Settings
→ Authentication
→ Add identity provider
```

然後針對 App 所在位置配置 identity provider。([Google Cloud Documentation][6])

## 9. Google Identity 並不等於個人 Gmail

這是一個很容易踩到的問題。

影片說 Google Identity，通常是指組織管理的：

```text
Google Workspace account
或
Cloud Identity account
```

不等於一般消費者帳號：

```text
someone@gmail.com
```

特別是 Google Drive Connector，官方目前要求 Cloud Console 登入帳號必須與要連接的 Google Drive instance 屬於相同的 Google Workspace customer；一般 `@gmail.com` 消費者帳號沒有 Workspace customer ID，因此不受支援。([Google Cloud Documentation][7])

所以做這個教學最好使用：

```text
simon@your-company-domain.com
```

而不是：

```text
simon123@gmail.com
```

即使個人 Gmail 可以建立 GCP Project，也不代表它能完成 Workspace enterprise connector 的完整流程。

## 10. Connected Data Source 實際建立的是什麼

身份設定完成後，影片進入：

```text
Connected data sources
→ Add data source
```

目前的介面更常稱為：

```text
Data stores
→ Create data store
```

Google 文件中的術語關係是：

```text
Data source
    ↓ accessed through
Connector
    ↓ represented inside Gemini Enterprise as
Data Store
    ↓ attached to
App
```

例如：

```text
Google Drive
→ Google Drive connector
→ drive-corporate-docs Data Store
→ Simon AI Workspace App
```

Connector 和 Data Store 有時在介面上會混用，但技術上不是完全相同：

* Connector 定義如何連到來源系統。
* Data Store 是 Gemini Enterprise 中可被 App 使用的資料資源。
* App 決定終端使用者可以搜尋哪些 Data Stores。

Google 將 connectors 定義為把 Google 或第三方資料來源帶入 Gemini Enterprise，並以獨立 Data Stores 管理。([Google Cloud Documentation][8])

## 11. 建立 Google Drive Data Store 的前置條件

對影片使用的 Google Drive connector，開始前應確認：

```text
1. 使用 Google Workspace 帳號
2. Cloud Console 帳號與 Drive Workspace 相符
3. 已設定 Identity Provider
4. 具有建立 Data Store 的 IAM 權限
5. App 與 Data Store 的 location 相容
6. Workspace 管理員沒有封鎖所需功能
```

官方特別要求：

* 使用 Cloud Console 的帳號必須與要連接的 Drive instance 相同。
* 必須是 Google Workspace，而不是一般消費者 Gmail。
* 為了正確執行 access control，應先設定 identity provider。([Google Cloud Documentation][7])

## 12. 建立 Drive Connector 時會填哪些東西

介面大致是：

```text
Gemini Enterprise
→ Data stores
→ Create data store
→ Google Drive
```

接下來通常需要設定：

### 要搜尋的資料範圍

可能包含：

```text
All supported Drive content
Specific folders
Shared drives
Include filters
Exclude filters
```

影片中說：

> Choose what you would like to share

這並不是把檔案權限改成公開，而是定義 Connector 的搜尋範圍。

例如：

```text
Connector scope:
  Include: Shared Drive / AI Study Group
  Exclude: Shared Drive / Finance Confidential
```

實際查詢時仍會再套用終端使用者原本的 Drive 權限。

也就是兩層過濾：

```text
Connector scope
AND
User's Drive permissions
```

某份文件要出現在結果中，通常必須同時滿足：

```text
文件在 Connector 搜尋範圍內
且
登入使用者本來就有權讀取
```

### Data connector name

例如：

```text
company-drive
engineering-shared-drive
workspace-documents-prod
```

這只是管理資源名稱，不是 Workspace 資料夾名稱。

命名最好帶環境：

```text
drive-rnd-dev
drive-rnd-prod
```

### Multi-region

通常應和 App 相同或相容：

```text
App: global
Data Store: global
```

或：

```text
App: eu
Data Store: eu
```

### Encryption

在 `us` 或 `eu` 等配置下，部分 Connector 會讓你選：

```text
Google-managed encryption key
Cloud KMS customer-managed key
```

教學通常選 Google-managed key。企業正式環境若有 CMEK 政策，才使用 Cloud KMS。

## 13. Drive 資料是複製進 Gemini Enterprise 嗎

影片的 Drive 連接通常是 federated retrieval，也就是查詢時從 Drive 取得資料，而不是先將所有 Drive 文件完整複製進 Gemini Enterprise index。

概念上：

```text
User asks question
→ Gemini Enterprise interprets query
→ Connector queries Google Drive
→ Drive returns authorized results
→ Gemini model synthesizes answer
```

官方 Drive 文件指出，這類連接可以直接從 Drive 擷取資訊，不需要你管理另一份資料儲存副本。([Google Cloud Documentation][7])

不過應注意：

* Metadata、設定和必要的服務資料仍可能由 Google Cloud 管理。
* 不同 connector 可能使用 federation 或 ingestion。
* 第三方 connector 未必與 Google Drive 採用完全相同的資料流。

因此不能把所有 Connected Data Sources 都理解成「即時 API 查詢」，也不能全部理解成「複製並建立向量索引」。要看 connector 類型。

## 14. 建立 Data Store 後還要掛到 App

建立 Data Store 和讓 App 使用 Data Store 是兩個動作。

可能的流程有兩種：

```text
先建立 App
→ 在 App 裡新增 Connected Data Source
→ 建立並自動掛載 Data Store
```

或：

```text
先建立 Data Store
→ 開啟 App
→ Connect data store
```

最終需要形成：

```text
App
└── Connected data stores
    └── company-drive
```

如果只建立 Data Store，卻沒有連到目前的 App，Web App 不一定能搜尋它。

可以在 App 的資料來源頁確認：

```text
Data store status: Active
Connected to app: Yes
Connector enabled in web app: Yes
```

# MCP new standard

The 2026-07-28 MCP Specification Release Candidate
https://blog.modelcontextprotocol.io/posts/2026-07-28-release-candidate/

你這裡看到的 **A Stateless Protocol**，最重要的不是「MCP 從此不能有狀態」，而是：

> **MCP 把「Protocol 本身的狀態」拿掉，讓需要保存的狀態回到 Application 層處理。**

2026-07-28 版本最直接造成兩個重要變化。([Model Context Protocol Blog][1])

## 1. Handshake 與 Protocol Session 被拿掉

在 `2025-11-25` 版本中，Client 要先做：

```text
Client
  ↓
initialize
  ↓
Server
  ↓
Mcp-Session-Id
  ↓
後續 tools/call 都帶 Session ID
```

也就是呼叫 Tool 前，先建立一個 MCP session。

例如：

```text
initialize
    ↓
Mcp-Session-Id = ABC123
    ↓
tools/call + ABC123
    ↓
tools/call + ABC123
```

這會產生一個 infrastructure 問題：

```text
Client
   ↓
Load Balancer
   ↓
Server A
```

因為 Session 是 Server A 建立的，後續 request 通常還必須回到 Server A，或者多台 Server 必須共享 Session Store。

所以以前常需要：

```text
Sticky Session
      或
Shared Session Store
```

2026-07-28 改成 Stateless 之後：

```text
Client
   ↓
tools/call
   ↓
任何 Server Instance
```

每一個 Request 自己攜帶：

* Protocol Version
* Client Information
* Method
* 必要的 metadata

不再需要先 `initialize` 建立 protocol session，也沒有 `Mcp-Session-Id`。([Model Context Protocol Blog][1])

因此最大的效果就是：

```text
Before

Client
  ↓
Session
  ↓
固定 Server A


After

Client
  ↓
Request
  ↓
Server A / B / C 都可以處理
```

這使 MCP 更接近一般 Stateless HTTP API，也更容易：

* Load balancing
* Horizontal scaling
* Auto-scaling
* Serverless deployment

---

## 2. Application 仍然可以 Stateful，但 State 要「顯式化」

這是第二個非常重要的改變。

Stateless MCP 並不是說：

```text
MCP 不能有購物車
MCP 不能有 Browser Session
MCP 不能有長任務
```

而是：

> **不要再把這些 application state 偷藏在 MCP transport session 裡。**

例如以前可能是：

```text
Mcp-Session-Id = ABC123

Server 內部知道：

ABC123
→ Shopping Basket #789
```

Model 本身其實不知道真正的 basket identity。

新的方式變成：

```text
create_basket()
      ↓
basket_id = 789
      ↓
add_item(
    basket_id = 789,
    item = "coffee"
)
```

也就是 State 從：

```text
Hidden Transport State
```

變成：

```text
Explicit Application State
```

([Model Context Protocol Blog][1])

這個改變很重要，因為 `basket_id`、`browser_id`、`task_id` 這類 handle 現在是資料的一部分。

Model 可以：

```text
建立 basket_id
     ↓
保存 basket_id
     ↓
下一個 Tool Call 再使用
     ↓
甚至交給另一個步驟或 Agent
```

因此：

```text
Stateless Protocol
        ≠
Stateless Application
```

正確理解應該是：

```text
Protocol
→ Stateless

Application
→ 可以 Stateful
→ 但 State 必須明確表示
```

---

## 為什麼這兩個改變很重要？

可以把整件事情濃縮成：

```text
2025 MCP

Connection / Session
        ↓
藏著 Protocol State
        ↓
Request 必須知道原本的 Session
        ↓
Sticky Routing / Shared Store
```

改成：

```text
2026 MCP

Self-contained Request
        ↓
任何 Server 都可以處理
        ↓
需要的 Application State
用 ID / Handle 明確傳遞
```

所以這次 Stateless 改造真正做的是：

> **把「Transport State」和「Application State」拆開。**

這也是為什麼新的 MCP 更適合大規模 Remote MCP Server。

而且這個改變接著還導致其他設計跟著調整，例如 **Elicitation 不再依賴長時間保持 SSE connection**，以及 `tools/list` 等結果可以真正做 HTTP-style caching。([Model Context Protocol Blog][1])

如果要用一句話記：

> **舊 MCP 是「先建立 Session，再做事情」；新 MCP 是「每個 Request 自己完整，真正需要的狀態由 Application 明確傳遞」。**

[1]: https://blog.modelcontextprotocol.io/posts/2026-07-28-release-candidate/ "The 2026-07-28 MCP Specification Release Candidate | Model Context Protocol Blog"


### Routable, Cacheable, Traceable：Stateless MCP 帶來的三個直接效果

當 MCP 拿掉 protocol-level session 之後，HTTP traffic 變得更接近一般現代 Web infrastructure，因此產生三個很實際的好處：

---

#### 1. Routable：更容易路由與負載平衡

新版 Streamable HTTP 要求加入：

```text
Mcp-Method
Mcp-Name
```

例如：

```text
Mcp-Method: tools/call
Mcp-Name: search
```

這樣 Load Balancer、API Gateway 或 Rate Limiter 不必解析 JSON body，就知道這個 request 是什麼操作。([Model Context Protocol Blog][1])

以前：

```text
Client
  ↓
Session ID
  ↓
固定 Server A
```

現在：

```text
Client
  ↓
Load Balancer
  ↓
Server A / B / C
```

因此可以直接使用：

* Round-robin load balancing
* Horizontal scaling
* Gateway routing
* Per-tool rate limiting

而不再需要 sticky session 或 shared session store。([Model Context Protocol Blog][1])

---

#### 2. Cacheable：工具與資源資訊可以快取

新版 MCP 在 List 與 Resource Read 的結果加入：

```text
ttlMs
cacheScope
```

概念類似 HTTP 的 `Cache-Control`。([Model Context Protocol Blog][1])

例如 Client 呼叫：

```text
tools/list
```

Server 可以告訴 Client：

```text
這份 Tool List
30 秒內仍然有效
而且可以 / 不可以跨使用者共用
```

因此不需要每一次都重新詢問 MCP Server。

流程從：

```text
Client
  ↓
tools/list
  ↓
Server

下一次
  ↓
tools/list
  ↓
Server
```

變成：

```text
Client
  ↓
tools/list
  ↓
Cache

TTL 還有效
  ↓
直接使用
```

這可以降低：

* Network traffic
* Server load
* Latency

尤其大型 MCP Server 有很多 Tool 或 Resource 時會更重要。([Model Context Protocol Blog][1])

---

#### 3. Traceable：整條 Agent Tool Call 可以被追蹤

新版 MCP 正式規範 **W3C Trace Context**，使用：

```text
traceparent
tracestate
baggage
```

並透過 `_meta` 傳遞。([Model Context Protocol Blog][1])

因此一個 request 可以從：

```text
User
 ↓
Host Application
 ↓
MCP Client
 ↓
MCP Server
 ↓
Database / API
```

保持同一條 trace。

最後在 OpenTelemetry 系統中可以看到：

```text
Agent Request
 ├── MCP tools/call
 │    ├── MCP Server
 │    ├── Database Query
 │    └── External API
 └── Result
```

這對 production system 很重要，因為可以回答：

* 哪一個 Agent request 出錯？
* 哪個 Tool 最慢？
* 哪個 downstream API 發生問題？
* 一次任務用了多少時間？

([Model Context Protocol Blog][1])

---

### 三個概念最簡單的記法

```text
Routable
→ Request 可以被正確分流到任何 Server

Cacheable
→ 不需要重複取得相同資訊

Traceable
→ 可以追蹤一個 Agent request 的完整執行鏈
```

因此 Stateless MCP 的意義不只是「拿掉 Session」，而是讓 MCP 更能直接利用現有 HTTP 基礎設施：

> **Load Balancer 負責 Routable，Cache/CDN 負責 Cacheable，OpenTelemetry 負責 Traceable。**

這也是新版 MCP 更適合大規模 production deployment 的關鍵原因。 ([Model Context Protocol Blog][1])
